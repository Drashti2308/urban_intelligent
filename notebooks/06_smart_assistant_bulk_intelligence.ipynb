```python

from pathlib import Path
from io import StringIO
import sys
import pandas as pd

print("PHASE 17.1A.1 — OPTIONAL ZIP NORMALIZATION FIX")
print("=" * 70)

root = Path(PROJECT_ROOT).resolve()

module_path = (
    root / "src/integration/bulk_complaint_contract.py"
)

template_path = (
    root / "outputs/templates/new_complaints_template.csv"
)

assert "module_source" in globals()
assert "sample" in globals()
assert "categories" in globals()
assert "batch_500" in globals()

# --------------------------------------------------
# 1. Fix the reusable validator
# --------------------------------------------------

old_code = "    return pd.DataFrame.from_records(records)"

new_code = """    result = pd.DataFrame.from_records(records)

    # Pandas may convert optional missing ZIPs
    # into NaN. The backend expects Python None.
    result["incident_zip"] = (
        result["incident_zip"]
        .astype(object)
        .where(result["incident_zip"].notna(), None)
    )

    return result"""

assert module_source.count(old_code) == 1, (
    "Expected validator return statement not found."
)

corrected_source = module_source.replace(
    old_code, new_code
)

compile(
    corrected_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        corrected_source,
        str(module_path),
        "exec"
    ),
    namespace
)

validate_batch = namespace["validate_batch"]
read_batch_csv = namespace["read_batch_csv"]

# --------------------------------------------------
# 2. Verify original sample validation
# --------------------------------------------------

validated = validate_batch(sample, categories)

ready = validated.loc[
    validated["validation_status"]
    == "ready_for_geographic_validation"
]

rejected = validated.loc[
    validated["validation_status"]
    == "invalid_input"
]

assert len(validated) == 7
assert len(ready) == 3
assert len(rejected) == 4

first = ready.loc[
    ready["complaint_id"] == "NEW-001"
].iloc[0]

zip_value = first["incident_zip"]

print("\n1. OPTIONAL ZIP DIAGNOSTIC")
print("ZIP value:", repr(zip_value))
print("ZIP Python type:", type(zip_value).__name__)

assert zip_value is None, (
    "Missing ZIP was not normalized to None."
)

# --------------------------------------------------
# 3. Recheck the existing backend
# --------------------------------------------------

print("\n2. BACKEND COMPATIBILITY TEST")

from src.ui.service_factory import get_gui_adapter

adapter = get_gui_adapter()

backend_result = adapter.evaluate(
    float(first["latitude"]),
    float(first["longitude"]),
    str(first["complaint_category"]),
    zip_value
)

assert backend_result["geography"]["borough"] == (
    "MANHATTAN"
)

assert backend_result["dbscan"]["status"] == (
    "established_cluster"
)

assert backend_result["dbscan"]["cluster_id"] == 1421

assert backend_result["kmeans"]["display_allowed"] is False

print("Backend: CONNECTED")
print("Borough: MANHATTAN")
print("DBSCAN Cluster: 1421")
print("K-Means missing ZIP safeguard: PASSED")

# --------------------------------------------------
# 4. Verify all 500 incoming records
# --------------------------------------------------

print("\n3. SIMULATED 500-COMPLAINT VALIDATION")

result_500 = validate_batch(
    batch_500,
    categories
)

assert len(result_500) == 500

assert result_500["complaint_id"].is_unique

assert result_500["validation_status"].eq(
    "ready_for_geographic_validation"
).all()

assert result_500["incident_zip"].map(
    lambda value: value is None
).all()

print("Total complaints:", len(result_500))
print("Input-valid complaints: 500")
print("Missing ZIP values normalized: 500")

# --------------------------------------------------
# 5. Preserve ZIPs with leading zeros
# --------------------------------------------------

csv_example = StringIO(
    "complaint_id,created_date,latitude,"
    "longitude,complaint_category,incident_zip\n"
    "ZIP-001,2026-11-09,40.75,-73.99,"
    "STREET CONDITION,00501\n"
)

zip_test = validate_batch(
    read_batch_csv(csv_example),
    categories
)

assert zip_test.iloc[0]["incident_zip"] == "00501"

print("Leading-zero ZIP preservation: PASSED")

# --------------------------------------------------
# 6. Recheck malformed and duplicate records
# --------------------------------------------------

assert (
    "invalid_latitude"
    in validated.iloc[2]["validation_errors"]
)

assert (
    "invalid_incident_zip"
    in validated.iloc[3]["validation_errors"]
)

assert (
    "unsupported_complaint_category"
    in validated.iloc[5]["validation_errors"]
)

assert (
    "invalid_created_date"
    in validated.iloc[6]["validation_errors"]
)

duplicate_sample = sample.copy()

duplicate_sample.loc[
    6, "complaint_id"
] = "NEW-002"

duplicate_result = validate_batch(
    duplicate_sample,
    categories
)

duplicate_rows = duplicate_result.loc[
    duplicate_result["complaint_id"] == "NEW-002"
]

assert len(duplicate_rows) == 2

assert duplicate_rows["validation_errors"].str.contains(
    "duplicate_complaint_id"
).all()

print("Malformed and duplicate record handling: PASSED")

# --------------------------------------------------
# 7. Prepare reusable template
# --------------------------------------------------

template_columns = [
    "complaint_id",
    "created_date",
    "latitude",
    "longitude",
    "complaint_category",
    "incident_zip"
]

template = sample.iloc[:2][
    template_columns
].copy()

template_csv = template.to_csv(
    index=False,
    lineterminator="\n"
)

# --------------------------------------------------
# 8. Protect existing project files
# --------------------------------------------------

if module_path.exists():
    assert module_path.read_text(
        encoding="utf-8"
    ) == corrected_source, (
        "Existing module differs. No overwrite."
    )

if template_path.exists():
    assert template_path.read_text(
        encoding="utf-8"
    ) == template_csv, (
        "Existing template differs. No overwrite."
    )

# --------------------------------------------------
# 9. Save new module and CSV template
# --------------------------------------------------

module_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

template_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

if not module_path.exists():
    module_path.write_text(
        corrected_source,
        encoding="utf-8"
    )

if not template_path.exists():
    template_path.write_text(
        template_csv,
        encoding="utf-8"
    )

module_source = corrected_source

print("\n4. CREATED / VERIFIED FILES")
print("Validator:", module_path)
print("Template:", template_path)

print("\nSTATUS: PASSED")
print("PHASE 17.1A INPUT CONTRACT VERIFIED")
print("OPTIONAL ZIP NORMALIZATION VERIFIED")
print("500-ROW INPUT VALIDATION VERIFIED")
print("EXISTING MODELS UNMODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import sys
import json
import pandas as pd

print(" BULK COMPLAINT BACKEND PROCESSING")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/bulk_complaint_processor.py"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

from src.integration.bulk_complaint_contract import (
    validate_batch
)
from src.ui.service_factory import get_gui_adapter

taxonomy = pd.read_csv(taxonomy_path)

categories = set(
    taxonomy["complaint_category"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.upper()
)

assert len(categories) == 15

# --------------------------------------------------
# 1. Define the reusable batch processor
# --------------------------------------------------

processor_source = dedent('''
"""Batch processing using saved historical services.

Each input record receives a traceable outcome.
Existing models are never retrained or updated.
"""

import pandas as pd

from src.integration.bulk_complaint_contract import (
    validate_batch
)


def process_complaint_batch(
    frame,
    supported_categories,
    adapter,
    use_cache=True
):
    validated = validate_batch(
        frame,
        supported_categories
    )

    results = []
    responses_by_source_row = {}
    cache = {}
    backend_evaluations = 0

    for record in validated.to_dict("records"):

        output = dict(record)

        output.update({
            "processing_status": None,
            "system_status": None,
            "borough": None,
            "geography_status": None,
            "zip_validation_status": None,
            "dbscan_status": None,
            "dbscan_cluster_id": None,
            "kmeans_status": None,
            "kmeans_area_id": None,
            "kmeans_cluster_id": None,
            "kmeans_reporting_profile": None,
            "kmeans_display_allowed": False,
            "featured_count": 0,
            "supporting_count": 0,
            "limited_evidence_count": 0,
            "warning_count": 0,
            "processing_error": None
        })

        if record["validation_status"] != (
            "ready_for_geographic_validation"
        ):
            output["processing_status"] = (
                "input_rejected"
            )

            results.append(output)
            continue

        zip_value = record["incident_zip"]

        if pd.isna(zip_value):
            zip_value = None

        # The historical backend is deterministic
        # for identical normalized input values.
        # Caching avoids repeating costly lookups.
        key = (
            float(record["latitude"]),
            float(record["longitude"]),
            str(record["complaint_category"]),
            zip_value
        )

        try:

            if use_cache and key in cache:
                response = cache[key]

            else:
                backend_evaluations += 1

                response = adapter.evaluate(
                    key[0],
                    key[1],
                    key[2],
                    key[3]
                )

                if use_cache:
                    cache[key] = response

            geography = response.get(
                "geography"
            ) or {}

            dbscan = response.get(
                "dbscan"
            ) or {}

            kmeans = response.get(
                "kmeans"
            ) or {}

            zip_info = response.get(
                "zip_validation"
            ) or {}

            recommendations = response.get(
                "recommendations"
            ) or {}

            warnings = response.get(
                "warnings"
            ) or []

            needs_review = (
                response.get("status")
                == "geographic_review_required"
            )

            output.update({
                "processing_status": (
                    "geographic_review_required"
                    if needs_review
                    else "processed"
                ),
                "system_status": response.get(
                    "status"
                ),
                "borough": geography.get(
                    "borough"
                ),
                "geography_status": geography.get(
                    "status"
                ),
                "zip_validation_status":
                    zip_info.get("status"),
                "dbscan_status": dbscan.get(
                    "status"
                ),
                "dbscan_cluster_id":
                    dbscan.get("cluster_id"),
                "kmeans_status": kmeans.get(
                    "status"
                ),
                "kmeans_area_id": kmeans.get(
                    "area_id"
                ),
                "kmeans_cluster_id":
                    kmeans.get("cluster_id"),
                "kmeans_reporting_profile":
                    kmeans.get("reporting_profile"),
                "kmeans_display_allowed": (
                    kmeans.get("display_allowed")
                    is True
                ),
                "featured_count": len(
                    recommendations.get(
                        "featured", []
                    )
                ),
                "supporting_count": len(
                    recommendations.get(
                        "supporting", []
                    )
                ),
                "limited_evidence_count": len(
                    recommendations.get(
                        "limited_evidence", []
                    )
                ),
                "warning_count": len(warnings)
            })

            # Preserve full historical evidence
            # separately, indexed by source row.
            responses_by_source_row[
                int(record["source_row"])
            ] = response

        except Exception as exc:

            output["processing_status"] = (
                "backend_error"
            )

            output["processing_error"] = (
                type(exc).__name__
                + ": "
                + str(exc)[:250]
            )

        results.append(output)

    table = pd.DataFrame.from_records(results)

    return {
        "records": table,
        "responses_by_source_row":
            responses_by_source_row,
        "stats": {
            "total": len(table),
            "processed": int(
                table["processing_status"]
                .eq("processed").sum()
            ),
            "geographic_review": int(
                table["processing_status"]
                .eq("geographic_review_required")
                .sum()
            ),
            "input_rejected": int(
                table["processing_status"]
                .eq("input_rejected").sum()
            ),
            "backend_errors": int(
                table["processing_status"]
                .eq("backend_error").sum()
            ),
            "backend_evaluations":
                backend_evaluations,
            "cache_entries": len(cache)
        }
    }
''').lstrip()

compile(
    processor_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        processor_source,
        str(module_path),
        "exec"
    ),
    namespace
)

process_complaint_batch = namespace[
    "process_complaint_batch"
]

# --------------------------------------------------
# 2. Construct 500 synthetic incoming complaints
# --------------------------------------------------

print("\n1. PREPARING MIXED 500-COMPLAINT BATCH")

new_batch = pd.DataFrame([
    {
        "complaint_id": f"NEW-{i:04d}",
        "created_date": "2026-11-09",
        "latitude": "40.702106342862",
        "longitude": "-74.013109867312",
        "complaint_category": "STREET CONDITION",
        "incident_zip": ""
    }
    for i in range(1, 501)
])

# 125 geographically consistent Manhattan records.
new_batch.loc[
    200:324, "latitude"
] = "40.75169000334706"

new_batch.loc[
    200:324, "longitude"
] = "-74.00466101019599"

new_batch.loc[
    200:324, "incident_zip"
] = "10001"

# 100 syntactically valid records outside NYC.
new_batch.loc[
    325:424, "latitude"
] = "0.0"

new_batch.loc[
    325:424, "longitude"
] = "0.0"

# 50 records with a historical ZIP whose
# ZCTA polygon is unavailable.
new_batch.loc[
    425:474, "incident_zip"
] = "10000"

# 20 records with a mismatched Manhattan ZIP.
new_batch.loc[
    475:494, "incident_zip"
] = "10002"

# Five malformed records.
new_batch.loc[495, "latitude"] = "invalid"

new_batch.loc[
    496, "complaint_category"
] = "UNKNOWN CATEGORY"

new_batch.loc[
    497, "incident_zip"
] = "83"

new_batch.loc[
    498, "created_date"
] = "2026-02-30"

new_batch.loc[
    499, "complaint_id"
] = ""

assert len(new_batch) == 500

# --------------------------------------------------
# 3. Process the full incoming batch
# --------------------------------------------------

print("\n2. PROCESSING COMPLAINTS")

adapter = get_gui_adapter()

batch_output = process_complaint_batch(
    new_batch,
    categories,
    adapter,
    use_cache=True
)

batch_results = batch_output["records"]
batch_stats = batch_output["stats"]
batch_responses = batch_output[
    "responses_by_source_row"
]

print("\nBATCH PROCESSING SUMMARY")

for name, value in batch_stats.items():
    print(f"{name}: {value}")

print(
    "\nNote: Identical complaint locations and "
    "categories reuse deterministic historical "
    "lookup results. Every complaint still "
    "receives its own output row."
)

# --------------------------------------------------
# 4. Verify all 500 records are accounted for
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 complaints retained",
    len(batch_results) == 500
)

check(
    "All source rows remain unique",
    batch_results["source_row"].is_unique
)

check(
    "Five invalid complaints rejected",
    batch_stats["input_rejected"] == 5
)

check(
    "100 complaints require geographic review",
    batch_stats["geographic_review"] == 100
)

check(
    "395 complaints processed normally",
    batch_stats["processed"] == 395
)

check(
    "No backend processing errors",
    batch_stats["backend_errors"] == 0
)

check(
    "Every record has a processing outcome",
    batch_results["processing_status"].notna().all()
)

check(
    "All valid complaints have historical responses",
    len(batch_responses) == 495
)

# --------------------------------------------------
# 5. Verify known hotspot association
# --------------------------------------------------

print("\n3. HISTORICAL DBSCAN VALIDATION")

known_group = batch_results.iloc[:200]

check(
    "Known hotspot complaints associate with 1421",
    known_group["dbscan_cluster_id"].eq(
        1421
    ).all()
)

check(
    "Known hotspot complaints show three Featured actions",
    known_group["featured_count"].eq(
        3
    ).all()
)

check(
    "Missing ZIP never invents K-Means profile",
    known_group[
        "kmeans_display_allowed"
    ].eq(False).all()
)

# --------------------------------------------------
# 6. Verify consistent ZIP / historical K-Means
# --------------------------------------------------

print("\n4. HISTORICAL K-MEANS VALIDATION")

consistent_group = batch_results.iloc[
    200:325
]

check(
    "Consistent ZIP group identifies Manhattan",
    consistent_group["borough"].eq(
        "MANHATTAN"
    ).all()
)

check(
    "Consistent ZIP group passes ZCTA validation",
    consistent_group[
        "zip_validation_status"
    ].eq("zcta_consistent").all()
)

check(
    "Historical area MANHATTAN_10001 retrieved",
    consistent_group[
        "kmeans_area_id"
    ].eq("MANHATTAN_10001").all()
)

check(
    "Consistent ZIP permits historical area profile",
    consistent_group[
        "kmeans_display_allowed"
    ].eq(True).all()
)

# --------------------------------------------------
# 7. Verify geographic review safeguards
# --------------------------------------------------

print("\n5. GEOGRAPHIC REVIEW VALIDATION")

outside_group = batch_results.iloc[
    325:425
]

check(
    "Outside-NYC complaints require review",
    outside_group["processing_status"].eq(
        "geographic_review_required"
    ).all()
)

check(
    "Outside-NYC complaints have no DBSCAN cluster",
    outside_group[
        "dbscan_cluster_id"
    ].isna().all()
)

check(
    "Outside-NYC complaints have no K-Means display",
    outside_group[
        "kmeans_display_allowed"
    ].eq(False).all()
)

# --------------------------------------------------
# 8. Verify unavailable and mismatched ZCTAs
# --------------------------------------------------

print("\n6. ZIP REFERENCE SAFEGUARDS")

unavailable_group = batch_results.iloc[
    425:475
]

mismatch_group = batch_results.iloc[
    475:495
]

check(
    "Unavailable ZCTA explicitly identified",
    unavailable_group[
        "zip_validation_status"
    ].eq("zcta_unavailable").all()
)

check(
    "Unavailable ZCTA uses historical lookup",
    unavailable_group[
        "kmeans_display_allowed"
    ].eq(True).all()
)

check(
    "Mismatched ZIP explicitly identified",
    mismatch_group[
        "zip_validation_status"
    ].eq("zcta_mismatch").all()
)

check(
    "Mismatched ZIP K-Means profile withheld",
    mismatch_group[
        "kmeans_display_allowed"
    ].eq(False).all()
)

check(
    "ZIP mismatch does not change DBSCAN hotspot",
    mismatch_group[
        "dbscan_cluster_id"
    ].eq(1421).all()
)

# --------------------------------------------------
# 9. Verify invalid records did not reach backend
# --------------------------------------------------

print("\n7. INVALID RECORD ISOLATION")

invalid_group = batch_results.iloc[
    495:500
]

check(
    "Malformed records remain visible",
    invalid_group["processing_status"].eq(
        "input_rejected"
    ).all()
)

check(
    "Malformed records have no model assignments",
    invalid_group[
        "dbscan_cluster_id"
    ].isna().all()
    and invalid_group[
        "kmeans_cluster_id"
    ].isna().all()
)

# --------------------------------------------------
# 10. Validate response integrity
# --------------------------------------------------

check(
    "All returned backend responses are JSON compatible",
    all(
        isinstance(
            json.dumps(
                response,
                allow_nan=False
            ),
            str
        )
        for response in {
            id(response): response
            for response in batch_responses.values()
        }.values()
    )
)

check(
    "Original historical model period preserved",
    all(
        response["historical_period"]
        == "2023-2025"
        for response in batch_responses.values()
    )
)

# --------------------------------------------------
# 11. Display complaint outcome summaries
# --------------------------------------------------

print("\n8. COMPLAINT PROCESSING STATUS")

display(
    batch_results.groupby(
        "processing_status",
        dropna=False
    ).size().rename(
        "complaints"
    ).reset_index()
)

print("\n9. SAMPLE COMPLAINT OUTPUTS")

sample_rows = [
    0, 200, 325, 425, 475, 495
]

display(
    batch_results.iloc[sample_rows][[
        "complaint_id",
        "processing_status",
        "borough",
        "zip_validation_status",
        "dbscan_cluster_id",
        "kmeans_area_id",
        "kmeans_display_allowed",
        "featured_count",
        "validation_errors"
    ]]
)

# --------------------------------------------------
# 12. Verify test results before writing module
# --------------------------------------------------

results_17_2a = pd.DataFrame(checks)

print("\n10. PHASE 17.2A VALIDATION CHECKLIST")

display(results_17_2a)

passed = int(results_17_2a["passed"].sum())
total = len(results_17_2a)

print(f"\nChecks passed: {passed}/{total}")

failures = results_17_2a.loc[
    ~results_17_2a["passed"],
    "test"
].tolist()

assert not failures, (
    "Phase 17.2A failures: "
    + "; ".join(failures)
)

# --------------------------------------------------
# 13. Save reusable processor safely
# --------------------------------------------------

if module_path.exists():

    assert module_path.read_text(
        encoding="utf-8"
    ) == processor_source, (
        "Existing bulk processor module differs. "
        "No overwrite performed."
    )

else:

    module_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    module_path.write_text(
        processor_source,
        encoding="utf-8"
    )

print("\n11. MODULE CREATED / VERIFIED")

print("Processor:", module_path)

print("\nSTATUS: PASSED")
print("PHASE 17.2A BULK PROCESSING VERIFIED")
print("500 COMPLAINT OUTCOMES GENERATED")
print("HISTORICAL DBSCAN AND K-MEANS SERVICES REUSED")
print("EXISTING MODEL ARTIFACTS UNMODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
from io import BytesIO
import json
import sys
import pandas as pd

print(" BULK EXPORTS AND REVIEW QUEUES")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/bulk_complaint_export.py"
)

export_dir = (
    root / "outputs/tests"
    / "phase_17_2b_synthetic_exports"
)

assert "batch_output" in globals(), (
    "Run Phase 17.2A first."
)

# --------------------------------------------------
# 1. Create reusable export module
# --------------------------------------------------

module_source = dedent('''
"""Safe CSV exports for bulk complaint outcomes.

Exported review queues are not operational
risk rankings or repair assignments.
"""

from io import BytesIO
import json
import pandas as pd


REVIEW_ZIP_STATUSES = {
    "zcta_mismatch",
    "boundary_review_required"
}


def _safe_cell(value):
    if value is None:
        return ""

    try:
        if pd.isna(value):
            return ""
    except (TypeError, ValueError):
        pass

    text = str(value)

    # Prevent spreadsheet formula execution
    # when exported text is opened in Excel.
    if text.lstrip().startswith(
        ("=", "+", "-", "@")
    ) or text.startswith(("\\t", "\\r", "\\n")):
        return "'" + text

    return text


def csv_bytes(frame):
    work = frame.copy()

    for column in work.columns:

        if pd.api.types.is_numeric_dtype(
            work[column]
        ):
            continue

        work[column] = work[column].map(
            _safe_cell
        )

    return work.to_csv(
        index=False,
        na_rep="",
        lineterminator="\\n"
    ).encode("utf-8-sig")


def build_bulk_export_package(batch_output):

    records = batch_output["records"].copy()

    required = {
        "source_row",
        "complaint_id",
        "processing_status",
        "validation_errors",
        "zip_validation_status",
        "dbscan_cluster_id",
        "kmeans_display_allowed"
    }

    if not required.issubset(records.columns):
        raise ValueError(
            "Bulk processor output schema changed."
        )

    if records.empty:
        raise ValueError("No complaint outcomes.")

    if records["source_row"].duplicated().any():
        raise ValueError(
            "Duplicate source-row identifiers."
        )

    statuses = records["processing_status"]

    allowed = {
        "processed",
        "geographic_review_required",
        "input_rejected",
        "backend_error"
    }

    if (
        statuses.isna().any()
        or not set(statuses).issubset(allowed)
    ):
        raise ValueError(
            "Unexpected processing status."
        )

    geo_mask = statuses.eq(
        "geographic_review_required"
    )

    invalid_mask = statuses.eq(
        "input_rejected"
    )

    error_mask = statuses.eq(
        "backend_error"
    )

    processed_mask = statuses.eq(
        "processed"
    )

    zip_mask = records[
        "zip_validation_status"
    ].isin(REVIEW_ZIP_STATUSES)

    # Review queues are subsets. An otherwise
    # processed complaint may need ZIP review.
    review_mask = (
        geo_mask | zip_mask | error_mask
    )

    records["review_reason"] = ""

    records.loc[
        geo_mask, "review_reason"
    ] = "geographic_review"

    records.loc[
        zip_mask & ~geo_mask,
        "review_reason"
    ] = "zip_validation_review"

    records.loc[
        error_mask, "review_reason"
    ] = "backend_processing_error"

    review = records.loc[
        review_mask
    ].copy()

    rejected = records.loc[
        invalid_mask
    ].copy()

    processed = records.loc[
        processed_mask
    ].copy()

    zip_review = records.loc[
        zip_mask
    ].copy()

    errors = records.loc[
        error_mask
    ].copy()

    stats = batch_output.get("stats", {})

    if int(stats.get("total", -1)) != len(records):
        raise ValueError(
            "Batch input count mismatch."
        )

    status_counts = {
        name: int(statuses.eq(name).sum())
        for name in sorted(allowed)
    }

    for status, stat_name in [
        ("processed", "processed"),
        ("geographic_review_required",
         "geographic_review"),
        ("input_rejected", "input_rejected"),
        ("backend_error", "backend_errors")
    ]:
        if status_counts[status] != int(
            stats.get(stat_name, -1)
        ):
            raise ValueError(
                "Batch status totals do not agree."
            )

    summary = {
        "schema_version": "bulk_export_v1",
        "input_type": "synthetic_test_batch",
        "historical_reference_period": "2023-2025",
        "total_complaints": len(records),
        "status_counts": status_counts,
        "review_queue_rows": len(review),
        "zip_review_rows": len(zip_review),
        "rejected_input_rows": len(rejected),
        "backend_error_rows": len(errors),
        "historical_hotspot_associations": int(
            records["dbscan_cluster_id"]
            .notna().sum()
        ),
        "historical_kmeans_profiles_displayed": int(
            records["kmeans_display_allowed"]
            .eq(True).sum()
        ),
        "backend_evaluations": int(
            stats.get("backend_evaluations", 0)
        ),
        "notes": [
            "This batch contains synthetic test records.",
            "Review queues are not safety rankings.",
            "ZIP review can coexist with processed status.",
            "Historical clusters are not retrained.",
            "No new hotspots are discovered in this phase."
        ]
    }

    tables = {
        "all_complaint_outcomes.csv": records,
        "processed_complaints.csv": processed,
        "review_queue.csv": review,
        "zip_review.csv": zip_review,
        "rejected_inputs.csv": rejected,
        "backend_errors.csv": errors
    }

    files = {
        filename: csv_bytes(table)
        for filename, table in tables.items()
    }

    files["batch_summary.json"] = json.dumps(
        summary,
        indent=2,
        allow_nan=False
    ).encode("utf-8")

    return {
        "files": files,
        "tables": tables,
        "summary": summary
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

build_bulk_export_package = namespace[
    "build_bulk_export_package"
]

csv_bytes = namespace["csv_bytes"]

# --------------------------------------------------
# 2. Generate export package in memory
# --------------------------------------------------

print("\n1. BUILDING EXPORT PACKAGE")

package = build_bulk_export_package(
    batch_output
)

files = package["files"]
tables = package["tables"]
summary = package["summary"]

display(
    pd.DataFrame([
        {
            "file": filename,
            "rows": (
                len(tables[filename])
                if filename in tables
                else "-"
            ),
            "size_bytes": len(content)
        }
        for filename, content in files.items()
    ])
)

# --------------------------------------------------
# 3. Validate counts and review logic
# --------------------------------------------------

checks = []

def check(name, passed):
    checks.append({
        "test": name,
        "passed": bool(passed)
    })

check(
    "All 500 outcomes exported",
    len(tables["all_complaint_outcomes.csv"])
    == 500
)

check(
    "395 processed records preserved",
    len(tables["processed_complaints.csv"])
    == 395
)

check(
    "100 geographic review records preserved",
    tables["review_queue.csv"][
        "processing_status"
    ].eq(
        "geographic_review_required"
    ).sum() == 100
)

check(
    "20 ZIP mismatch records identified",
    len(tables["zip_review.csv"]) == 20
)

check(
    "Review queue includes 120 records",
    len(tables["review_queue.csv"]) == 120
)

check(
    "All 5 malformed records exported",
    len(tables["rejected_inputs.csv"]) == 5
)

check(
    "No backend errors recorded",
    len(tables["backend_errors.csv"]) == 0
)

check(
    "Original complaint records preserved",
    set(
        tables["all_complaint_outcomes.csv"][
            "source_row"
        ]
    ) == set(batch_results["source_row"])
)

check(
    "ZIP review does not hide DBSCAN results",
    tables["zip_review.csv"][
        "dbscan_cluster_id"
    ].notna().all()
)

check(
    "Review reasons are traceable",
    tables["review_queue.csv"][
        "review_reason"
    ].ne("").all()
)

# --------------------------------------------------
# 4. Verify CSV export round trips
# --------------------------------------------------

print("\n2. TESTING CSV READABILITY")

roundtrip = {}

for filename, content in files.items():

    if not filename.endswith(".csv"):
        continue

    roundtrip[filename] = pd.read_csv(
        BytesIO(content),
        dtype=str,
        keep_default_na=False,
        encoding="utf-8-sig"
    )

    check(
        f"CSV reopens with correct row count: {filename}",
        len(roundtrip[filename])
        == len(tables[filename])
    )

check(
    "Complaint IDs survive CSV export",
    roundtrip["all_complaint_outcomes.csv"][
        "complaint_id"
    ].iloc[0] == "NEW-0001"
)

check(
    "Historical ZIP 10001 preserved",
    roundtrip["all_complaint_outcomes.csv"][
        "incident_zip"
    ].iloc[200] == "10001"
)

check(
    "Missing ZIP remains blank",
    roundtrip["all_complaint_outcomes.csv"][
        "incident_zip"
    ].iloc[0] == ""
)

# --------------------------------------------------
# 5. Verify spreadsheet formula safety
# --------------------------------------------------

dangerous = pd.DataFrame([
    {
        "complaint_id": "=2+2",
        "created_date": "2026-11-09",
        "latitude": "=2+2",
        "longitude": -74.01
    }
])

dangerous_csv = csv_bytes(
    dangerous
).decode("utf-8-sig")

check(
    "Formula-like complaint IDs safely escaped",
    "'=2+2" in dangerous_csv
)

check(
    "Formula-like invalid coordinates safely escaped",
    dangerous_csv.count("'=2+2") == 2
)

# --------------------------------------------------
# 6. Verify summary JSON
# --------------------------------------------------

loaded_summary = json.loads(
    files["batch_summary.json"].decode("utf-8")
)

check(
    "Summary JSON total is 500",
    loaded_summary["total_complaints"] == 500
)

check(
    "Summary preserves original history period",
    loaded_summary[
        "historical_reference_period"
    ] == "2023-2025"
)

check(
    "Summary distinguishes ZIP review from rejection",
    loaded_summary["zip_review_rows"] == 20
    and loaded_summary["rejected_input_rows"] == 5
)

check(
    "Historical K-Means display count is 175",
    loaded_summary[
        "historical_kmeans_profiles_displayed"
    ] == 175
)

check(
    "Historical DBSCAN association count is 395",
    loaded_summary[
        "historical_hotspot_associations"
    ] == 395
)

# --------------------------------------------------
# 7. Verify all results before saving
# --------------------------------------------------

results_17_2b = pd.DataFrame(checks)

print("\n3. PHASE 17.2B TEST CHECKLIST")

display(results_17_2b)

passed = int(results_17_2b["passed"].sum())
total = len(results_17_2b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_17_2b.loc[
    ~results_17_2b["passed"],
    "test"
].tolist()

assert not failures, (
    "Phase 17.2B failed: "
    + "; ".join(failures)
)

# --------------------------------------------------
# 8. Preflight existing files to avoid overwrites
# --------------------------------------------------

if module_path.exists():
    assert module_path.read_text(
        encoding="utf-8"
    ) == module_source, (
        "Existing export module differs. "
        "No overwrite performed."
    )

for filename, content in files.items():

    target = export_dir / filename

    if target.exists():
        assert target.read_bytes() == content, (
            f"Existing export differs: {filename}. "
            "No overwrite performed."
        )

# --------------------------------------------------
# 9. Save reusable module and synthetic test exports
# --------------------------------------------------

module_path.parent.mkdir(
    parents=True,
    exist_ok=True
)

export_dir.mkdir(
    parents=True,
    exist_ok=True
)

if not module_path.exists():
    module_path.write_text(
        module_source,
        encoding="utf-8"
    )

for filename, content in files.items():

    target = export_dir / filename

    if not target.exists():
        target.write_bytes(content)

print("\n4. SAVED FILES")
print("Exporter module:", module_path)
print("Synthetic export folder:", export_dir)

for filename in files:
    print(" -", filename)

print("\n5. BATCH SUMMARY")
print(json.dumps(summary, indent=2))

print("\nSTATUS: PASSED")
print("PHASE 17.2B CSV EXPORT AND REVIEW QUEUES VERIFIED")
print("ALL 500 SYNTHETIC COMPLAINTS ACCOUNTED FOR")
print("EXISTING MODEL ARTIFACTS UNMODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import sys
import pandas as pd

print(" COMPLAINT GROUPING AND SUMMARIES")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/bulk_complaint_grouping.py"
)

hotspot_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_primary_f_gui_hotspots.csv"
)

assert "batch_output" in globals(), (
    "Phase 17.2A batch output is missing."
)

assert hotspot_path.is_file()

hotspot_reference = pd.read_csv(hotspot_path)

# --------------------------------------------------
# 1. Create reusable grouping module
# --------------------------------------------------

module_source = dedent('''
"""Summaries of incoming complaints and historical context.

These are descriptive reporting groups, not
operational safety priorities or new ML clusters.
"""

import pandas as pd


def summarize_bulk_complaints(
    batch_output,
    hotspot_reference
):

    records = batch_output["records"].copy()
    reference = hotspot_reference.copy()

    required = {
        "source_row",
        "complaint_id",
        "created_date",
        "processing_status",
        "borough",
        "complaint_category",
        "latitude",
        "longitude",
        "dbscan_status",
        "dbscan_cluster_id",
        "zip_validation_status",
        "kmeans_display_allowed",
        "kmeans_area_id",
        "kmeans_cluster_id",
        "kmeans_reporting_profile"
    }

    if not required.issubset(records.columns):
        raise ValueError(
            "Bulk processor output schema changed."
        )

    if records.empty:
        raise ValueError("Empty batch results.")

    if records["source_row"].duplicated().any():
        raise ValueError("Duplicate source rows.")

    if not {
        "dbscan_cluster_id",
        "total_complaints",
        "primary_borough",
        "dominant_category"
    }.issubset(reference.columns):
        raise ValueError(
            "Historical hotspot reference changed."
        )

    if reference[
        "dbscan_cluster_id"
    ].duplicated().any():
        raise ValueError(
            "Duplicate historical hotspot IDs."
        )

    records["dbscan_cluster_id"] = pd.to_numeric(
        records["dbscan_cluster_id"],
        errors="coerce"
    ).astype("Int64")

    records["kmeans_cluster_id"] = pd.to_numeric(
        records["kmeans_cluster_id"],
        errors="coerce"
    ).astype("Int64")

    reference["dbscan_cluster_id"] = pd.to_numeric(
        reference["dbscan_cluster_id"],
        errors="raise"
    ).astype("Int64")

    # ----------------------------------------------
    # 2. Separate input processing outcomes
    # ----------------------------------------------

    processed = records.loc[
        records["processing_status"].eq(
            "processed"
        )
    ].copy()

    geographic_review = records.loc[
        records["processing_status"].eq(
            "geographic_review_required"
        )
    ].copy()

    rejected = records.loc[
        records["processing_status"].eq(
            "input_rejected"
        )
    ].copy()

    backend_errors = records.loc[
        records["processing_status"].eq(
            "backend_error"
        )
    ].copy()

    if (
        len(processed)
        + len(geographic_review)
        + len(rejected)
        + len(backend_errors)
        != len(records)
    ):
        raise ValueError(
            "Batch processing totals do not reconcile."
        )

    # ----------------------------------------------
    # 3. Group processed complaints by borough
    # ----------------------------------------------

    by_borough = (
        processed.groupby("borough")
        .size()
        .reset_index(name="new_complaints")
        .sort_values(
            ["new_complaints", "borough"],
            ascending=[False, True]
        )
        .reset_index(drop=True)
    )

    # ----------------------------------------------
    # 4. Group processed complaints by category
    # ----------------------------------------------

    by_category = (
        processed.groupby("complaint_category")
        .size()
        .reset_index(name="new_complaints")
        .sort_values(
            ["new_complaints", "complaint_category"],
            ascending=[False, True]
        )
        .reset_index(drop=True)
    )

    if len(processed):
        by_category["share_of_processed_pct"] = (
            100
            * by_category["new_complaints"]
            / len(processed)
        ).round(2)
    else:
        by_category["share_of_processed_pct"] = (
            pd.Series(dtype=float)
        )

    # ----------------------------------------------
    # 5. Group new reports by historical DBSCAN ID
    # ----------------------------------------------

    associated = processed.loc[
        processed["dbscan_status"].eq(
            "established_cluster"
        )
    ].copy()

    if associated[
        "dbscan_cluster_id"
    ].isna().any():
        raise ValueError(
            "Established hotspot has no cluster ID."
        )

    # Exact coordinates are descriptive only.
    # Multiple reports at the same coordinate
    # do not prove multiple physical defects.
    associated["location_key"] = [
        repr(float(lat)) + "|" + repr(float(lon))
        for lat, lon in zip(
            associated["latitude"],
            associated["longitude"]
        )
    ]

    hotspot_groups = (
        associated.groupby("dbscan_cluster_id")
        .agg(
            incoming_complaints=(
                "source_row", "size"
            ),
            distinct_submitted_locations=(
                "location_key", "nunique"
            ),
            incoming_categories=(
                "complaint_category", "nunique"
            )
        )
        .reset_index()
    )

    historical_columns = reference[[
        "dbscan_cluster_id",
        "total_complaints",
        "primary_borough",
        "dominant_category"
    ]].rename(columns={
        "total_complaints":
            "historical_complaints_2023_2025",
        "primary_borough":
            "historical_primary_borough",
        "dominant_category":
            "historical_dominant_category"
    })

    hotspot_groups = hotspot_groups.merge(
        historical_columns,
        on="dbscan_cluster_id",
        how="left",
        validate="one_to_one"
    )

    if hotspot_groups[
        "historical_complaints_2023_2025"
    ].isna().any():
        raise ValueError(
            "Associated hotspot missing from "
            "historical reference."
        )

    hotspot_groups = hotspot_groups.sort_values(
        ["incoming_complaints", "dbscan_cluster_id"],
        ascending=[False, True]
    ).reset_index(drop=True)

    # ----------------------------------------------
    # 6. Category distribution within each hotspot
    # ----------------------------------------------

    by_hotspot_category = (
        associated.groupby([
            "dbscan_cluster_id",
            "complaint_category"
        ])
        .size()
        .reset_index(name="new_complaints")
    )

    if not by_hotspot_category.empty:

        totals = (
            hotspot_groups.set_index(
                "dbscan_cluster_id"
            )["incoming_complaints"]
        )

        by_hotspot_category[
            "share_within_incoming_hotspot_pct"
        ] = (
            100
            * by_hotspot_category["new_complaints"]
            / by_hotspot_category[
                "dbscan_cluster_id"
            ].map(totals)
        ).round(2)

    # ----------------------------------------------
    # 7. Group displayed historical K-Means profiles
    # ----------------------------------------------

    shown = processed.loc[
        processed[
            "kmeans_display_allowed"
        ].eq(True)
    ].copy()

    if shown[
        [
            "kmeans_area_id",
            "kmeans_cluster_id",
            "kmeans_reporting_profile"
        ]
    ].isna().any().any():
        raise ValueError(
            "Displayed area profile is incomplete."
        )

    area_groups = (
        shown.groupby([
            "kmeans_area_id",
            "kmeans_cluster_id",
            "kmeans_reporting_profile",
            "zip_validation_status"
        ], dropna=False)
        .size()
        .reset_index(name="new_complaints")
        .sort_values(
            "new_complaints",
            ascending=False
        )
        .reset_index(drop=True)
    )

    # ----------------------------------------------
    # 8. Review cases are not a safety ranking
    # ----------------------------------------------

    zip_review = records.loc[
        records["zip_validation_status"].isin([
            "zcta_mismatch",
            "boundary_review_required"
        ])
    ].copy()

    review_mask = (
        records["processing_status"].isin([
            "geographic_review_required",
            "backend_error"
        ])
        | records["zip_validation_status"].isin([
            "zcta_mismatch",
            "boundary_review_required"
        ])
    )

    review_queue = records.loc[
        review_mask
    ].copy()

    # ----------------------------------------------
    # 9. Daily incoming complaint counts
    # ----------------------------------------------

    by_date = (
        records.groupby([
            "created_date",
            "processing_status"
        ], dropna=False)
        .size()
        .reset_index(name="complaints")
    )

    # ----------------------------------------------
    # 10. Summary with explicit denominators
    # ----------------------------------------------

    summary = {
        "total_input": len(records),
        "processed": len(processed),
        "geographic_review": len(
            geographic_review
        ),
        "input_rejected": len(rejected),
        "backend_errors": len(backend_errors),
        "associated_with_historical_hotspots":
            len(associated),
        "distinct_associated_historical_hotspots":
            len(hotspot_groups),
        "historical_area_profiles_displayed":
            len(shown),
        "zip_review": len(zip_review),
        "review_queue_rows": len(review_queue)
    }

    return {
        "summary": summary,
        "by_borough": by_borough,
        "by_category": by_category,
        "historical_hotspots": hotspot_groups,
        "by_hotspot_category":
            by_hotspot_category,
        "historical_area_profiles":
            area_groups,
        "by_date": by_date,
        "review_queue": review_queue
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

summarize_bulk_complaints = namespace[
    "summarize_bulk_complaints"
]

# --------------------------------------------------
# 2. Generate grouped results
# --------------------------------------------------

print("\n1. GROUPING NEW COMPLAINTS")

grouped = summarize_bulk_complaints(
    batch_output,
    hotspot_reference
)

summary = grouped["summary"]

print("\nBATCH GROUPING SUMMARY")

for key, value in summary.items():
    print(f"{key}: {value}")

# --------------------------------------------------
# 3. Display borough and category summaries
# --------------------------------------------------

print("\n2. NEW COMPLAINTS BY BOROUGH")

display(grouped["by_borough"])

print("\n3. NEW COMPLAINTS BY CATEGORY")

display(grouped["by_category"])

# --------------------------------------------------
# 4. Display historical hotspot groups
# --------------------------------------------------

print("\n4. INCOMING REPORTS BY HISTORICAL HOTSPOT")

display(grouped["historical_hotspots"])

print("\n5. CATEGORY SHARES WITHIN HOTSPOT GROUPS")

display(grouped["by_hotspot_category"])

# --------------------------------------------------
# 5. Display historical K-Means area profiles
# --------------------------------------------------

print("\n6. HISTORICAL AREA PROFILES")

display(grouped["historical_area_profiles"])

# --------------------------------------------------
# 6. Validate all grouping outcomes
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 complaints accounted for",
    summary["total_input"] == 500
)

check(
    "395 complaints processed",
    summary["processed"] == 395
)

check(
    "100 geographic review cases retained",
    summary["geographic_review"] == 100
)

check(
    "Five invalid records retained",
    summary["input_rejected"] == 5
)

check(
    "No backend processing errors",
    summary["backend_errors"] == 0
)

check(
    "395 new reports associated with historical hotspots",
    summary[
        "associated_with_historical_hotspots"
    ] == 395
)

check(
    "Two distinct historical hotspots represented",
    summary[
        "distinct_associated_historical_hotspots"
    ] == 2
)

check(
    "Borough grouping contains 395 complaints",
    int(
        grouped["by_borough"][
            "new_complaints"
        ].sum()
    ) == 395
)

check(
    "Manhattan has all 395 processed complaints",
    int(
        grouped["by_borough"].loc[
            grouped["by_borough"][
                "borough"
            ] == "MANHATTAN",
            "new_complaints"
        ].sum()
    ) == 395
)

check(
    "Category grouping contains 395 complaints",
    int(
        grouped["by_category"][
            "new_complaints"
        ].sum()
    ) == 395
)

check(
    "Hotspot grouping preserves 395 complaints",
    int(
        grouped["historical_hotspots"][
            "incoming_complaints"
        ].sum()
    ) == 395
)

# --------------------------------------------------
# 7. Independently verify exact hotspot grouping
# --------------------------------------------------

hotspot_counts = {
    int(row.dbscan_cluster_id):
        int(row.incoming_complaints)
    for row in grouped[
        "historical_hotspots"
    ].itertuples(index=False)
}

check(
    "Historical hotspot 1421 contains 270 incoming records",
    hotspot_counts.get(1421) == 270
)

check(
    "Historical hotspot 1964 contains 125 incoming records",
    hotspot_counts.get(1964) == 125
)

historical_1421 = hotspot_reference.loc[
    hotspot_reference[
        "dbscan_cluster_id"
    ] == 1421,
    "total_complaints"
].iloc[0]

grouped_1421 = grouped[
    "historical_hotspots"
].loc[
    grouped["historical_hotspots"][
        "dbscan_cluster_id"
    ] == 1421
].iloc[0]

check(
    "Historical 1421 count remains 12773",
    int(historical_1421) == 12773
    and int(
        grouped_1421[
            "historical_complaints_2023_2025"
        ]
    ) == 12773
)

check(
    "Historical and incoming totals remain separate",
    int(
        grouped_1421["incoming_complaints"]
    ) == 270
    and int(
        grouped_1421[
            "historical_complaints_2023_2025"
        ]
    ) == 12773
)

# --------------------------------------------------
# 8. Verify K-Means area grouping
# --------------------------------------------------

area_groups = grouped[
    "historical_area_profiles"
]

area_counts = {
    str(row.kmeans_area_id):
        int(row.new_complaints)
    for row in area_groups.itertuples(index=False)
}

check(
    "175 incoming records have displayed area profiles",
    summary[
        "historical_area_profiles_displayed"
    ] == 175
)

check(
    "Manhattan 10001 has 125 incoming records",
    area_counts.get("MANHATTAN_10001")
    == 125
)

check(
    "Manhattan 10000 has 50 incoming records",
    area_counts.get("MANHATTAN_10000")
    == 50
)

check(
    "Unavailable ZCTA remains distinguishable",
    area_groups.loc[
        area_groups[
            "kmeans_area_id"
        ] == "MANHATTAN_10000",
        "zip_validation_status"
    ].eq("zcta_unavailable").all()
)

# --------------------------------------------------
# 9. Verify review queue accounting
# --------------------------------------------------

check(
    "20 ZIP review records preserved",
    summary["zip_review"] == 20
)

check(
    "Review queue contains 120 records",
    summary["review_queue_rows"] == 120
)

check(
    "Review queue source rows remain unique",
    grouped["review_queue"][
        "source_row"
    ].is_unique
)

check(
    "Incoming category shares sum to 100%",
    abs(
        grouped["by_category"][
            "share_of_processed_pct"
        ].sum() - 100
    ) < 0.01
)

# --------------------------------------------------
# 10. Display verification results
# --------------------------------------------------

results_17_3a = pd.DataFrame(checks)

print("\n7. PHASE 17.3A TEST CHECKLIST")

display(results_17_3a)

passed = int(results_17_3a["passed"].sum())
total = len(results_17_3a)

print(f"\nChecks passed: {passed}/{total}")

failures = results_17_3a.loc[
    ~results_17_3a["passed"],
    "test"
].tolist()

assert not failures, (
    "Phase 17.3A failures: "
    + "; ".join(failures)
)

# --------------------------------------------------
# 11. Safely save the reusable grouping module
# --------------------------------------------------

if module_path.exists():

    assert module_path.read_text(
        encoding="utf-8"
    ) == module_source, (
        "Existing grouping module differs. "
        "No overwrite performed."
    )

else:

    module_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    module_path.write_text(
        module_source,
        encoding="utf-8"
    )

print("\n8. MODULE CREATED / VERIFIED")

print("Grouping module:", module_path)

print("\nSTATUS: PASSED")
print("PHASE 17.3A COMPLAINT GROUPING VERIFIED")
print("INCOMING REPORTS LINKED TO HISTORICAL HOTSPOTS")
print("K-MEANS HISTORICAL AREA SUMMARIES VERIFIED")
print("NO NEW HOTSPOTS OR RISK SCORES CLAIMED")
print("EXISTING MODEL ARTIFACTS UNMODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import sys
import pandas as pd

print(" SMART INVESTIGATION SUMMARIES")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/bulk_investigation_summary.py"
)

assert "batch_output" in globals()
assert "hotspot_reference" in globals()

# --------------------------------------------------
# 1. Build reusable investigation summary module
# --------------------------------------------------

module_source = dedent('''
"""Evidence-based briefs for incoming complaint batches.

Historical hotspot association is not proof of
a defect, repair urgency, or safety risk.
"""

import pandas as pd


def build_investigation_briefs(
    batch_output,
    historical_hotspots
):
    records = batch_output["records"].copy()
    responses = batch_output[
        "responses_by_source_row"
    ]

    reference = historical_hotspots.copy()

    required = {
        "source_row",
        "processing_status",
        "borough",
        "complaint_category",
        "latitude",
        "longitude",
        "dbscan_status",
        "dbscan_cluster_id",
        "zip_validation_status",
        "kmeans_display_allowed",
        "featured_count",
        "supporting_count",
        "limited_evidence_count"
    }

    if not required.issubset(records.columns):
        raise ValueError(
            "Unexpected bulk processor schema."
        )

    if records["source_row"].duplicated().any():
        raise ValueError("Duplicate source rows.")

    reference = reference.set_index(
        "dbscan_cluster_id",
        verify_integrity=True
    )

    # Only established historical associations
    # become historical hotspot briefs.
    established = records.loc[
        records["processing_status"].eq(
            "processed"
        )
        & records["dbscan_status"].eq(
            "established_cluster"
        )
    ].copy()

    established["dbscan_cluster_id"] = (
        pd.to_numeric(
            established["dbscan_cluster_id"],
            errors="raise"
        ).astype(int)
    )

    briefs = []

    for (cluster_id, category), group in (
        established.groupby([
            "dbscan_cluster_id",
            "complaint_category"
        ])
    ):
        cluster_id = int(cluster_id)

        if cluster_id not in reference.index:
            raise ValueError(
                "Historical hotspot reference missing."
            )

        historical = reference.loc[cluster_id]

        incoming_count = len(group)

        location_count = len(set(
            zip(
                group["latitude"].astype(float),
                group["longitude"].astype(float)
            )
        ))

        historical_count = int(
            historical["total_complaints"]
        )

        representative_row = int(
            group["source_row"].iloc[0]
        )

        response = responses.get(
            representative_row
        )

        if response is None:
            raise ValueError(
                "Missing representative backend evidence."
            )

        if (
            response["dbscan"]["cluster_id"]
            != cluster_id
        ):
            raise ValueError(
                "Representative hotspot mismatch."
            )

        category_context = (
            response["dbscan"].get(
                "category_context"
            ) or {}
        )

        historical_category_count = (
            category_context.get(
                "category_complaints"
            )
        )

        if historical_category_count is not None:
            historical_category_count = int(
                historical_category_count
            )

        actions = response.get(
            "recommendations"
        ) or {}

        featured = len(
            actions.get("featured", [])
        )

        supporting = len(
            actions.get("supporting", [])
        )

        limited = len(
            actions.get("limited_evidence", [])
        )

        for column, expected in [
            ("featured_count", featured),
            ("supporting_count", supporting),
            ("limited_evidence_count", limited)
        ]:
            if not group[column].eq(expected).all():
                raise ValueError(
                    "Recommendation count inconsistency."
                )

        zip_review_count = int(
            group["zip_validation_status"].isin([
                "zcta_mismatch",
                "boundary_review_required"
            ]).sum()
        )

        unverified_zip_count = int(
            group["zip_validation_status"].eq(
                "zcta_unavailable"
            ).sum()
        )

        displayed_area_count = int(
            group["kmeans_display_allowed"]
            .eq(True).sum()
        )

        # Avoid asserting that repeated input
        # records represent distinct defects.
        repeated_location_review = (
            incoming_count > location_count
        )

        evidence = (
            f"{incoming_count} incoming complaint "
            f"records in category {category} were "
            f"associated with historical hotspot "
            f"{cluster_id}, across "
            f"{location_count} distinct submitted "
            f"coordinate(s). The hotspot contains "
            f"{historical_count} historical reports "
            f"from 2023-2025."
        )

        if historical_category_count is not None:
            evidence += (
                f" Of these historical reports, "
                f"{historical_category_count} were "
                f"in category {category}."
            )

        safeguards = []

        if repeated_location_review:
            safeguards.append(
                "Check whether repeated coordinates "
                "represent duplicate reports or "
                "different incidents."
            )

        if zip_review_count:
            safeguards.append(
                f"{zip_review_count} incoming records "
                f"require ZIP geographic review."
            )

        if unverified_zip_count:
            safeguards.append(
                f"{unverified_zip_count} incoming "
                f"records use an unavailable ZCTA "
                f"reference."
            )

        safeguards.append(
            "Historical hotspot association does "
            "not establish physical defect severity."
        )

        suggested_review = (
            "Review related complaint descriptions, "
            "maintenance history and potential "
            "duplicate submissions. Consider "
            "coordinated field investigation when "
            "supported by verified operational "
            "information."
        )

        briefs.append({
            "historical_hotspot_id": cluster_id,
            "complaint_category": category,
            "incoming_complaints": incoming_count,
            "distinct_submitted_locations":
                location_count,
            "historical_complaints_2023_2025":
                historical_count,
            "historical_category_complaints":
                historical_category_count,
            "historical_primary_borough":
                historical["primary_borough"],
            "historical_dominant_category":
                historical["dominant_category"],
            "zip_review_records":
                zip_review_count,
            "zcta_unavailable_records":
                unverified_zip_count,
            "historical_area_profiles_displayed":
                displayed_area_count,
            "featured_investigations":
                featured,
            "supporting_investigations":
                supporting,
            "limited_evidence_investigations":
                limited,
            "representative_source_row":
                representative_row,
            "repeat_location_review":
                repeated_location_review,
            "evidence_explanation": evidence,
            "review_safeguards": " ".join(
                safeguards
            ),
            "suggested_investigation":
                suggested_review,
            "operational_priority":
                "not_assessed"
        })

    brief_table = pd.DataFrame.from_records(
        briefs
    )

    if not brief_table.empty:
        brief_table = brief_table.sort_values(
            [
                "incoming_complaints",
                "historical_hotspot_id"
            ],
            ascending=[False, True]
        ).reset_index(drop=True)

    # Complaints outside established historical
    # clusters must not disappear from the queue.
    unmatched = records.loc[
        records["processing_status"].eq(
            "processed"
        )
        & ~records["dbscan_status"].eq(
            "established_cluster"
        )
    ].copy()

    unmatched_groups = (
        unmatched.groupby(
            [
                "borough",
                "complaint_category",
                "dbscan_status"
            ],
            dropna=False
        )
        .size()
        .reset_index(name="incoming_complaints")
    )

    review_counts = {
        "geographic_review": int(
            records["processing_status"].eq(
                "geographic_review_required"
            ).sum()
        ),
        "zip_review": int(
            records["zip_validation_status"]
            .isin([
                "zcta_mismatch",
                "boundary_review_required"
            ]).sum()
        ),
        "invalid_input": int(
            records["processing_status"].eq(
                "input_rejected"
            ).sum()
        ),
        "backend_error": int(
            records["processing_status"].eq(
                "backend_error"
            ).sum()
        )
    }

    summary = {
        "total_complaints": len(records),
        "established_hotspot_reports":
            len(established),
        "historical_hotspot_category_briefs":
            len(brief_table),
        "unmatched_processed_reports":
            len(unmatched),
        "review_counts": review_counts,
        "interpretation": (
            "Briefs are descriptive investigation "
            "aids, not safety rankings or confirmed "
            "repair instructions."
        )
    }

    return {
        "briefs": brief_table,
        "unmatched_groups": unmatched_groups,
        "summary": summary
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

build_investigation_briefs = namespace[
    "build_investigation_briefs"
]

# --------------------------------------------------
# 2. Generate investigation briefs
# --------------------------------------------------

print("\n1. BUILDING INVESTIGATION BRIEFS")

investigation_output = build_investigation_briefs(
    batch_output,
    hotspot_reference
)

briefs = investigation_output["briefs"]
brief_summary = investigation_output["summary"]

display(briefs[[
    "historical_hotspot_id",
    "complaint_category",
    "incoming_complaints",
    "distinct_submitted_locations",
    "historical_complaints_2023_2025",
    "historical_category_complaints",
    "featured_investigations",
    "zip_review_records",
    "operational_priority"
]])

# --------------------------------------------------
# 3. Display the natural-language briefs
# --------------------------------------------------

print("\n2. SMART ASSISTANT INVESTIGATION EXPLANATIONS")

for item in briefs.to_dict("records"):

    print("\n" + "-" * 70)
    print(
        "HISTORICAL HOTSPOT",
        item["historical_hotspot_id"]
    )
    print("Category:", item["complaint_category"])
    print("\nEVIDENCE:")
    print(item["evidence_explanation"])

    print("\nRECOMMENDATION AVAILABILITY:")
    print(
        "Featured:",
        item["featured_investigations"]
    )
    print(
        "Supporting:",
        item["supporting_investigations"]
    )
    print(
        "Limited Evidence:",
        item["limited_evidence_investigations"]
    )

    print("\nSAFEGUARDS:")
    print(item["review_safeguards"])

    print("\nSUGGESTED INVESTIGATION:")
    print(item["suggested_investigation"])

# --------------------------------------------------
# 4. Verify evidence and accounting
# --------------------------------------------------

checks = []

def check(name, passed):
    checks.append({
        "test": name,
        "passed": bool(passed)
    })

check(
    "Two historical hotspot investigation briefs",
    len(briefs) == 2
)

check(
    "Incoming hotspot complaints total 395",
    int(briefs["incoming_complaints"].sum())
    == 395
)

check(
    "No processed complaints lost from summaries",
    (
        brief_summary["established_hotspot_reports"]
        + brief_summary["unmatched_processed_reports"]
    ) == 395
)

by_id = briefs.set_index(
    "historical_hotspot_id"
)

check(
    "Hotspot 1421 contains 270 incoming reports",
    int(
        by_id.loc[1421, "incoming_complaints"]
    ) == 270
)

check(
    "Hotspot 1964 contains 125 incoming reports",
    int(
        by_id.loc[1964, "incoming_complaints"]
    ) == 125
)

check(
    "Hotspot 1421 historical total remains 12773",
    int(
        by_id.loc[
            1421,
            "historical_complaints_2023_2025"
        ]
    ) == 12773
)

check(
    "Hotspot 1964 historical total remains 43",
    int(
        by_id.loc[
            1964,
            "historical_complaints_2023_2025"
        ]
    ) == 43
)

check(
    "Street Condition historical evidence retained",
    int(
        by_id.loc[
            1421,
            "historical_category_complaints"
        ]
    ) == 3870
)

check(
    "Hotspot 1421 has three Featured investigations",
    int(
        by_id.loc[
            1421,
            "featured_investigations"
        ]
    ) == 3
)

check(
    "Both hotspot groups contain one submitted location",
    briefs[
        "distinct_submitted_locations"
    ].eq(1).all()
)

check(
    "Repeated coordinate review is highlighted",
    briefs[
        "repeat_location_review"
    ].eq(True).all()
)

check(
    "20 ZIP mismatch records flagged in hotspot 1421",
    int(
        by_id.loc[
            1421, "zip_review_records"
        ]
    ) == 20
)

check(
    "50 unavailable ZCTA records remain qualified",
    int(
        by_id.loc[
            1421,
            "zcta_unavailable_records"
        ]
    ) == 50
)

check(
    "No automatic repair priorities assigned",
    briefs["operational_priority"].eq(
        "not_assessed"
    ).all()
)

check(
    "Geographic review remains accounted for",
    brief_summary["review_counts"][
        "geographic_review"
    ] == 100
)

check(
    "Five invalid inputs remain accounted for",
    brief_summary["review_counts"][
        "invalid_input"
    ] == 5
)

check(
    "No backend errors remain",
    brief_summary["review_counts"][
        "backend_error"
    ] == 0
)

# --------------------------------------------------
# 5. Display results and save verified module
# --------------------------------------------------

results_17_3b = pd.DataFrame(checks)

print("\n3. PHASE 17.3B TEST CHECKLIST")

display(results_17_3b)

passed = int(results_17_3b["passed"].sum())
total = len(results_17_3b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_17_3b.loc[
    ~results_17_3b["passed"], "test"
].tolist()

assert not failures, (
    "Phase 17.3B failures: "
    + "; ".join(failures)
)

if module_path.exists():

    assert module_path.read_text(
        encoding="utf-8"
    ) == module_source, (
        "Existing investigation module differs. "
        "No overwrite performed."
    )

else:

    module_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    module_path.write_text(
        module_source,
        encoding="utf-8"
    )

print("\n4. MODULE CREATED / VERIFIED")
print("Investigation summaries:", module_path)

print("\nSTATUS: PASSED")
print("PHASE 17.3B INVESTIGATION SUMMARIES VERIFIED")
print("HISTORICAL EVIDENCE PRESERVED")
print("REPEATED LOCATIONS FLAGGED FOR REVIEW")
print("NO AUTOMATIC SAFETY PRIORITY ASSIGNED")
print("EXISTING MODELS UNMODIFIED")
```

```python

from pathlib import Path
from shapely.geometry import shape
import json
import re
import sys
import pandas as pd

print(" DIVERSE NYC AREA REFERENCE AUDIT")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

zcta_path = (
    root / "models/geography"
    / "nyc_zcta_35j5-n34v_snapshot_2026_v1.geojson"
)

profile_path = (
    root / "outputs/tables/kmeans"
    / "kmeans_full_data_k3_descriptive_profiles.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

for path in [zcta_path, profile_path, taxonomy_path]:
    assert path.is_file(), f"Missing: {path}"

# --------------------------------------------------
# 1. Inspect the saved NYC reference data
# --------------------------------------------------

with open(zcta_path, encoding="utf-8") as file:
    zcta_geojson = json.load(file)

features = zcta_geojson["features"]

profiles = pd.read_csv(
    profile_path,
    dtype={"area_id": str}
)

taxonomy = pd.read_csv(taxonomy_path)

categories_17_4a = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

print("\n1. REFERENCE INVENTORY")
print("NYC ZCTA polygons:", len(features))
print("Historical K-Means areas:", len(profiles))
print("Supported categories:", len(categories_17_4a))

print("\nZCTA property fields:")
print(list(features[0]["properties"].keys()))

print("\nSupported complaint categories:")
print(categories_17_4a)

# --------------------------------------------------
# 2. Detect the ZCTA identifier field
# --------------------------------------------------

def is_five_digit_zip(value):
    if value is None:
        return False

    return bool(
        re.fullmatch(
            r"\d{5}",
            str(value).strip()
        )
    )


property_names = list(
    features[0]["properties"].keys()
)

candidate_fields = [
    name
    for name in property_names
    if any(
        token in name.lower()
        for token in ("zcta", "zip", "postal")
    )
]

field_scores = {}

for field in candidate_fields:
    field_scores[field] = sum(
        is_five_digit_zip(
            item["properties"].get(field)
        )
        for item in features
    )

print("\n2. IDENTIFIER FIELD SCORES")
print(field_scores)

assert field_scores, (
    "No ZCTA identifier field found. "
    "Share the property field names."
)

zip_field = max(
    field_scores,
    key=field_scores.get
)

assert field_scores[zip_field] >= 150, (
    "Unable to reliably identify the ZCTA code. "
    "Share field scores before continuing."
)

print("Selected ZCTA field:", zip_field)

# --------------------------------------------------
# 3. Generate candidate points inside ZCTA areas
# --------------------------------------------------

zcta_points = {}

for item in features:

    zip_code = str(
        item["properties"].get(zip_field, "")
    ).strip()

    if not is_five_digit_zip(zip_code):
        continue

    if item.get("geometry") is None:
        continue

    polygon = shape(item["geometry"])

    if polygon.is_empty:
        continue

    # A representative point lies inside the
    # polygon, unlike a centroid in some shapes.
    point = polygon.representative_point()

    zcta_points.setdefault(
        zip_code,
        (float(point.y), float(point.x))
    )

print("\nUsable ZCTA coordinate anchors:",
      len(zcta_points))

# --------------------------------------------------
# 4. Match reference areas to historical profiles
# --------------------------------------------------

candidate_rows = []

for area_id in profiles["area_id"].dropna():

    area_id = str(area_id).strip()

    if "_" not in area_id:
        continue

    borough, zip_code = area_id.rsplit(
        "_", 1
    )

    if not is_five_digit_zip(zip_code):
        continue

    if zip_code not in zcta_points:
        continue

    latitude, longitude = zcta_points[
        zip_code
    ]

    candidate_rows.append({
        "area_id": area_id,
        "expected_borough": borough,
        "incident_zip": zip_code,
        "latitude": latitude,
        "longitude": longitude
    })

area_candidates_17_4a = (
    pd.DataFrame(candidate_rows)
    .drop_duplicates("area_id")
    .sort_values(
        ["expected_borough", "area_id"]
    )
    .reset_index(drop=True)
)

print("\n3. MATCHED AREA COUNTS")

display(
    area_candidates_17_4a.groupby(
        "expected_borough"
    ).size().rename(
        "candidate_areas"
    ).reset_index()
)

assert not area_candidates_17_4a.empty

# Select up to four different ZIP areas
# from each borough for backend verification.
sample_anchors = (
    area_candidates_17_4a.groupby(
        "expected_borough",
        group_keys=False
    )
    .head(4)
    .reset_index(drop=True)
)

print("\n4. CANDIDATE NYC TEST ANCHORS")

display(sample_anchors)

# --------------------------------------------------
# 5. Verify each selected area using existing backend
# --------------------------------------------------

from src.ui.service_factory import get_gui_adapter

adapter = get_gui_adapter()

verification_rows = []

for row in sample_anchors.to_dict("records"):

    result = {
        **row,
        "actual_borough": None,
        "geography_status": None,
        "zcta_status": None,
        "kmeans_display_allowed": False,
        "dbscan_status": None,
        "verified": False,
        "error": None
    }

    try:

        response = adapter.evaluate(
            row["latitude"],
            row["longitude"],
            "STREET CONDITION",
            row["incident_zip"]
        )

        geography = response.get(
            "geography"
        ) or {}

        zip_info = response.get(
            "zip_validation"
        ) or {}

        kmeans = response.get(
            "kmeans"
        ) or {}

        dbscan = response.get(
            "dbscan"
        ) or {}

        result["actual_borough"] = (
            geography.get("borough")
        )

        result["geography_status"] = (
            geography.get("status")
        )

        result["zcta_status"] = (
            zip_info.get("status")
        )

        result["kmeans_display_allowed"] = (
            kmeans.get("display_allowed")
            is True
        )

        result["dbscan_status"] = (
            dbscan.get("status")
        )

        result["verified"] = (
            result["actual_borough"]
            == row["expected_borough"]
            and result["zcta_status"]
            == "zcta_consistent"
            and result["kmeans_display_allowed"]
            and kmeans.get("area_id")
            == row["area_id"]
        )

    except Exception as exc:

        result["error"] = (
            type(exc).__name__
            + ": "
            + str(exc)[:150]
        )

    verification_rows.append(result)

anchor_audit_17_4a = pd.DataFrame(
    verification_rows
)

verified_area_anchors_17_4a = (
    anchor_audit_17_4a.loc[
        anchor_audit_17_4a["verified"]
    ].copy().reset_index(drop=True)
)

print("\n5. BACKEND VERIFICATION RESULTS")

display(anchor_audit_17_4a[[
    "area_id",
    "expected_borough",
    "actual_borough",
    "zcta_status",
    "kmeans_display_allowed",
    "dbscan_status",
    "verified",
    "error"
]])

print("\n6. VERIFIED AREA COVERAGE")

verified_by_borough = (
    verified_area_anchors_17_4a.groupby(
        "expected_borough"
    ).size().rename(
        "verified_areas"
    ).reset_index()
)

display(verified_by_borough)

# --------------------------------------------------
# 6. Summarize readiness for a diverse test batch
# --------------------------------------------------

verified_areas = len(
    verified_area_anchors_17_4a
)

verified_boroughs = (
    verified_area_anchors_17_4a[
        "expected_borough"
    ].nunique()
)

print("\n7. PHASE 17.4A READINESS SUMMARY")
print("Historical K-Means areas:", len(profiles))
print("Geographic area candidates:",
      len(area_candidates_17_4a))
print("Anchors checked:", len(sample_anchors))
print("Verified anchors:", verified_areas)
print("Boroughs represented:", verified_boroughs)
print("Available categories:", len(categories_17_4a))

if (
    verified_areas >= 10
    and verified_boroughs >= 3
    and len(categories_17_4a) >= 3
):
    print("\nSTATUS: PASSED")
    print("DIVERSE SYNTHETIC BATCH ANCHORS READY")
    print("HISTORICAL K-MEANS COMPATIBILITY VERIFIED")
    print("NO TRAINING OR MODEL MODIFICATIONS")
else:
    print("\nSTATUS: NEEDS REVIEW")
    print(
        "Additional valid area anchors may be "
        "needed before generating 500 records."
    )
    print(
        "Share the verification table; "
        "do not assume unverified locations "
        "are suitable for training."
    )
```

```python

from pathlib import Path
from shapely.geometry import shape, Point
from shapely.ops import transform as transform_geometry
from pyproj import Transformer
import numpy as np
import pandas as pd
import json
import sys

from src.integration.bulk_complaint_contract import (
    validate_batch,
    read_batch_csv
)

print(" DIVERSE SYNTHETIC COMPLAINT DATA")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

output_dir = (
    root / "outputs/tests/phase_17_4b_diverse"
)

csv_path = (
    output_dir / "synthetic_500_diverse_complaints.csv"
)

reference_path = (
    output_dir / "synthetic_area_pattern_reference.csv"
)

assert "verified_area_anchors_17_4a" in globals(), (
    "Run Phase 17.4A first."
)

anchors = verified_area_anchors_17_4a.copy()

assert len(anchors) == 20
assert anchors["expected_borough"].nunique() == 5
assert anchors["verified"].all()

# --------------------------------------------------
# 1. Define deliberately different area patterns
# --------------------------------------------------

# Exactly 25 complaints per area.
# First two patterns are intentionally similar.

patterns = {
    "road_a": {
        "STREET CONDITION": 14,
        "SIDEWALK CONDITION": 4,
        "SEWER": 3,
        "STREET LIGHT CONDITION": 2,
        "TRAFFIC SIGNAL CONDITION": 2
    },
    "road_b": {
        "STREET CONDITION": 13,
        "SIDEWALK CONDITION": 5,
        "SEWER": 3,
        "STREET LIGHT CONDITION": 2,
        "TRAFFIC SIGNAL CONDITION": 2
    },
    "sewer": {
        "SEWER": 15,
        "SEWER MAINTENANCE": 5,
        "WATER DRAINAGE": 2,
        "STREET CONDITION": 2,
        "SIDEWALK CONDITION": 1
    },
    "signals": {
        "STREET LIGHT CONDITION": 12,
        "TRAFFIC SIGNAL CONDITION": 7,
        "STREET SIGN - DAMAGED": 2,
        "STREET CONDITION": 2,
        "SEWER": 2
    }
}

assert all(
    sum(counts.values()) == 25
    for counts in patterns.values()
)

allowed_categories = set(categories_17_4a)

assert all(
    category in allowed_categories
    for counts in patterns.values()
    for category in counts
)

print("\n1. SYNTHETIC PATTERN DESIGN")

display(pd.DataFrame([
    {
        "pattern": name,
        "complaints_per_area": sum(counts.values()),
        "street_condition_pct":
            100 * counts.get("STREET CONDITION", 0) / 25,
        "sewer_pct":
            100 * counts.get("SEWER", 0) / 25,
        "street_light_pct":
            100 * counts.get("STREET LIGHT CONDITION", 0) / 25
    }
    for name, counts in patterns.items()
]))

# --------------------------------------------------
# 2. Load actual NYC ZCTA polygons
# --------------------------------------------------

zcta_path = (
    root / "models/geography"
    / "nyc_zcta_35j5-n34v_snapshot_2026_v1.geojson"
)

with zcta_path.open(encoding="utf-8") as file:
    zcta_data = json.load(file)

polygon_by_zip = {
    str(feature["properties"]["zcta5"]): shape(
        feature["geometry"]
    )
    for feature in zcta_data["features"]
    if feature.get("geometry") is not None
}

to_utm = Transformer.from_crs(
    "EPSG:4326",
    "EPSG:32618",
    always_xy=True
)

to_wgs84 = Transformer.from_crs(
    "EPSG:32618",
    "EPSG:4326",
    always_xy=True
)

rng = np.random.default_rng(1742026)

# --------------------------------------------------
# 3. Generate distinct coordinates within each area
# --------------------------------------------------

def generate_area_points(anchor, count=25):
    zip_code = str(anchor["incident_zip"])

    polygon = polygon_by_zip[zip_code]

    polygon_utm = transform_geometry(
        to_utm.transform,
        polygon
    )

    anchor_x, anchor_y = to_utm.transform(
        float(anchor["longitude"]),
        float(anchor["latitude"])
    )

    # Keep generated points inside the ZCTA,
    # away from its boundary, and reasonably
    # close to the previously verified anchor.
    safe_area = polygon_utm.buffer(-12).intersection(
        Point(anchor_x, anchor_y).buffer(180)
    )

    if safe_area.is_empty:
        raise ValueError(
            f"No safe sampling area for {zip_code}"
        )

    min_x, min_y, max_x, max_y = safe_area.bounds

    locations_utm = []
    locations_wgs84 = []

    for attempt in range(20000):

        x = rng.uniform(min_x, max_x)
        y = rng.uniform(min_y, max_y)

        candidate = Point(x, y)

        if not safe_area.contains(candidate):
            continue

        # Avoid producing practically identical
        # coordinate submissions.
        if any(
            candidate.distance(existing) < 8
            for existing in locations_utm
        ):
            continue

        longitude, latitude = to_wgs84.transform(
            x, y
        )

        locations_utm.append(candidate)

        locations_wgs84.append((
            round(latitude, 8),
            round(longitude, 8)
        ))

        if len(locations_wgs84) == count:
            break

    if len(locations_wgs84) != count:
        raise ValueError(
            f"Only {len(locations_wgs84)} valid "
            f"points found for {zip_code}."
        )

    return locations_wgs84


# --------------------------------------------------
# 4. Assign area patterns across five boroughs
# --------------------------------------------------

generated_records = []
area_metadata = []

pattern_order = [
    "road_a",
    "road_b",
    "sewer",
    "signals"
]

for borough, group in anchors.groupby(
    "expected_borough", sort=True
):

    group = group.sort_values(
        "area_id"
    ).reset_index(drop=True)

    assert len(group) == 4, (
        f"Expected four verified areas: {borough}"
    )

    for position, anchor in enumerate(
        group.to_dict("records")
    ):

        pattern_name = pattern_order[position]
        category_counts = patterns[pattern_name]

        locations = generate_area_points(
            anchor, count=25
        )

        category_list = []

        for category, count in category_counts.items():
            category_list.extend(
                [category] * count
            )

        assert len(category_list) == len(locations)

        # Shuffle category assignment without
        # changing the intentionally designed shares.
        rng.shuffle(category_list)

        area_metadata.append({
            "area_id": anchor["area_id"],
            "expected_borough": borough,
            "incident_zip": anchor["incident_zip"],
            "intended_pattern": pattern_name,
            "planned_complaints": 25,
            "street_condition_pct": (
                100 * category_counts.get(
                    "STREET CONDITION", 0
                ) / 25
            ),
            "sewer_pct": (
                100 * category_counts.get(
                    "SEWER", 0
                ) / 25
            )
        })

        for (latitude, longitude), category in zip(
            locations, category_list
        ):

            generated_records.append({
                "complaint_id": (
                    f"SIM-20261109-"
                    f"{len(generated_records)+1:04d}"
                ),
                "created_date": "2026-11-09",
                "latitude": latitude,
                "longitude": longitude,
                "complaint_category": category,
                "incident_zip": anchor["incident_zip"]
            })

synthetic_500_17_4b = pd.DataFrame(
    generated_records
)

area_reference_17_4b = pd.DataFrame(
    area_metadata
)

# --------------------------------------------------
# 5. Validate the generated CSV input
# --------------------------------------------------

print("\n2. VALIDATING ALL 500 COMPLAINT INPUTS")

validated_17_4b = validate_batch(
    synthetic_500_17_4b,
    allowed_categories
)

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Exactly 500 synthetic records",
    len(synthetic_500_17_4b) == 500
)

check(
    "All synthetic complaint IDs are unique",
    synthetic_500_17_4b[
        "complaint_id"
    ].is_unique
)

check(
    "All 500 pass CSV input validation",
    validated_17_4b[
        "validation_status"
    ].eq(
        "ready_for_geographic_validation"
    ).all()
)

check(
    "20 geographic areas represented",
    len(area_reference_17_4b) == 20
)

check(
    "All five NYC boroughs represented",
    area_reference_17_4b[
        "expected_borough"
    ].nunique() == 5
)

check(
    "Each area receives 25 complaints",
    area_reference_17_4b[
        "planned_complaints"
    ].eq(25).all()
)

check(
    "Ten road-heavy areas",
    area_reference_17_4b[
        "intended_pattern"
    ].isin(["road_a", "road_b"]).sum() == 10
)

check(
    "Five sewer-heavy areas",
    area_reference_17_4b[
        "intended_pattern"
    ].eq("sewer").sum() == 5
)

check(
    "Five lighting/signal-heavy areas",
    area_reference_17_4b[
        "intended_pattern"
    ].eq("signals").sum() == 5
)

check(
    "All coordinates are distinct",
    not synthetic_500_17_4b[
        ["latitude", "longitude"]
    ].duplicated().any()
)

check(
    "Multiple complaint categories represented",
    synthetic_500_17_4b[
        "complaint_category"
    ].nunique() >= 7
)

check(
    "Only supported complaint categories used",
    set(
        synthetic_500_17_4b[
            "complaint_category"
        ]
    ).issubset(allowed_categories)
)

check(
    "Road variants have 56% and 52% Street Condition",
    set(
        area_reference_17_4b.loc[
            area_reference_17_4b[
                "intended_pattern"
            ].isin(["road_a", "road_b"]),
            "street_condition_pct"
        ]
    ) == {56.0, 52.0}
)

# --------------------------------------------------
# 6. Independently check ZCTA containment
# --------------------------------------------------

all_inside = True

for row in synthetic_500_17_4b.itertuples(
    index=False
):

    polygon = polygon_by_zip[
        str(row.incident_zip)
    ]

    if not polygon.contains(
        Point(float(row.longitude), float(row.latitude))
    ):
        all_inside = False
        break

check(
    "All generated points lie inside supplied ZCTAs",
    all_inside
)

# --------------------------------------------------
# 7. Display useful summaries
# --------------------------------------------------

print("\n3. COMPLAINT CATEGORY DISTRIBUTION")

display(
    synthetic_500_17_4b.groupby(
        "complaint_category"
    ).size().rename(
        "complaints"
    ).reset_index().sort_values(
        "complaints", ascending=False
    )
)

print("\n4. AREA PATTERN REFERENCE")

display(area_reference_17_4b)

print("\n5. PHASE 17.4B CHECKS")

results_17_4b = pd.DataFrame(checks)

display(results_17_4b)

passed = int(results_17_4b["passed"].sum())
total = len(results_17_4b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_17_4b.loc[
    ~results_17_4b["passed"], "test"
].tolist()

assert not failures, (
    "Phase 17.4B failures: "
    + "; ".join(failures)
)

# --------------------------------------------------
# 8. Save synthetic test files without overwriting
# --------------------------------------------------

csv_content = synthetic_500_17_4b.to_csv(
    index=False,
    lineterminator="\n"
)

reference_content = area_reference_17_4b.to_csv(
    index=False,
    lineterminator="\n"
)

for path, content in [
    (csv_path, csv_content),
    (reference_path, reference_content)
]:
    if path.exists():
        assert path.read_text(
            encoding="utf-8"
        ) == content, (
            f"Existing file differs: {path}. "
            "No overwrite performed."
        )

output_dir.mkdir(
    parents=True,
    exist_ok=True
)

for path, content in [
    (csv_path, csv_content),
    (reference_path, reference_content)
]:
    if not path.exists():
        path.write_text(
            content,
            encoding="utf-8"
        )

# Reopen the saved CSV using our actual CSV reader.
reopened = read_batch_csv(csv_path)

assert len(reopened) == 500

print("\n6. SAVED TEST FILES")
print("Synthetic complaints:", csv_path)
print("Area pattern reference:", reference_path)

print("\nSTATUS: PASSED")
print("PHASE 17.4B DIVERSE SYNTHETIC DATA GENERATED")
print("500 INPUT RECORDS VALIDATED")
print("20 AREAS ACROSS 5 BOROUGHS REPRESENTED")
print("HISTORICAL MODEL ARTIFACTS UNMODIFIED")
print("FULL BACKEND GEOGRAPHIC TEST IS NEXT")
```

```python

from pathlib import Path
from time import perf_counter
import sys
import json
import pandas as pd

from src.integration.bulk_complaint_contract import (
    read_batch_csv,
    validate_batch
)
from src.integration.bulk_complaint_processor import (
    process_complaint_batch
)
from src.integration.bulk_complaint_export import (
    csv_bytes
)
from src.ui.service_factory import get_gui_adapter

print("FULL DIVERSE BATCH VERIFICATION")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

input_dir = (
    root / "outputs/tests/phase_17_4b_diverse"
)

csv_path = (
    input_dir / "synthetic_500_diverse_complaints.csv"
)

reference_path = (
    input_dir / "synthetic_area_pattern_reference.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

for path in [csv_path, reference_path, taxonomy_path]:
    assert path.is_file(), f"Missing file: {path}"

# --------------------------------------------------
# 1. Load the saved synthetic test data
# --------------------------------------------------

incoming = read_batch_csv(csv_path)

area_reference = pd.read_csv(
    reference_path,
    dtype="string"
)

taxonomy = pd.read_csv(taxonomy_path)

allowed_categories = set(
    taxonomy["complaint_category"]
    .dropna()
    .astype(str)
    .str.strip()
    .str.upper()
)

assert len(incoming) == 500
assert len(area_reference) == 20
assert area_reference["incident_zip"].is_unique
assert incoming["complaint_id"].is_unique

print("\n1. INPUT INVENTORY")
print("Incoming complaints:", len(incoming))
print("Expected areas:", len(area_reference))
print(
    "Distinct locations:",
    len(incoming[[
        "latitude", "longitude"
    ]].drop_duplicates())
)
print(
    "Complaint categories:",
    incoming["complaint_category"].nunique()
)

# --------------------------------------------------
# 2. Confirm input validation before processing
# --------------------------------------------------

input_validation = validate_batch(
    incoming,
    allowed_categories
)

assert input_validation["validation_status"].eq(
    "ready_for_geographic_validation"
).all(), (
    "Synthetic CSV no longer passes input validation."
)

# --------------------------------------------------
# 3. Run all 500 through the historical backend
# --------------------------------------------------

print("\n2. RUNNING FULL BACKEND EVALUATION")

adapter = get_gui_adapter()

start = perf_counter()

diverse_batch_output_17_4c = (
    process_complaint_batch(
        incoming,
        allowed_categories,
        adapter,
        use_cache=True
    )
)

elapsed_seconds = perf_counter() - start

diverse_results_17_4c = (
    diverse_batch_output_17_4c["records"]
)

stats = diverse_batch_output_17_4c["stats"]

print("\nBACKEND RESULTS")

for key, value in stats.items():
    print(f"{key}: {value}")

print(
    "Observed processing duration (seconds):",
    round(elapsed_seconds, 2)
)

# --------------------------------------------------
# 4. Add the expected geographic area reference
# --------------------------------------------------

expected_lookup = area_reference.set_index(
    "incident_zip"
)

audit_17_4c = diverse_results_17_4c.copy()

audit_17_4c["expected_area_id"] = (
    audit_17_4c["incident_zip"].map(
        expected_lookup["area_id"]
    )
)

audit_17_4c["expected_borough"] = (
    audit_17_4c["incident_zip"].map(
        expected_lookup["expected_borough"]
    )
)

audit_17_4c["intended_pattern"] = (
    audit_17_4c["incident_zip"].map(
        expected_lookup["intended_pattern"]
    )
)

# A record is fully verified only when every
# important geographic/K-Means check succeeds.
audit_17_4c["fully_verified"] = (
    audit_17_4c["processing_status"].eq(
        "processed"
    )
    & audit_17_4c["borough"].eq(
        audit_17_4c["expected_borough"]
    )
    & audit_17_4c["zip_validation_status"].eq(
        "zcta_consistent"
    )
    & audit_17_4c[
        "kmeans_display_allowed"
    ].eq(True)
    & audit_17_4c["kmeans_area_id"].eq(
        audit_17_4c["expected_area_id"]
    )
)

# --------------------------------------------------
# 5. Verify every submitted record was preserved
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "500 complaint outcomes returned",
    len(audit_17_4c) == 500
)

check(
    "All original complaint IDs preserved",
    audit_17_4c["complaint_id"].tolist()
    == incoming["complaint_id"].astype(str).tolist()
)

check(
    "Source rows remain unique",
    audit_17_4c["source_row"].is_unique
)

check(
    "No input validation failures",
    stats["input_rejected"] == 0
)

check(
    "No backend processing errors",
    stats["backend_errors"] == 0
)

check(
    "All 500 records have processing outcomes",
    audit_17_4c[
        "processing_status"
    ].notna().all()
)

check(
    "500 unique backend evaluations performed",
    stats["backend_evaluations"] == 500
)

check(
    "500 backend responses retained",
    len(
        diverse_batch_output_17_4c[
            "responses_by_source_row"
        ]
    ) == 500
)

check(
    "Every submitted ZIP has an area reference",
    audit_17_4c[
        "expected_area_id"
    ].notna().all()
)

# --------------------------------------------------
# 6. Verify complete geographic compatibility
# --------------------------------------------------

check(
    "All 500 complaints processed",
    stats["processed"] == 500
)

check(
    "All 500 boroughs match expected boroughs",
    audit_17_4c["borough"].eq(
        audit_17_4c["expected_borough"]
    ).all()
)

check(
    "All 500 ZIPs are ZCTA consistent",
    audit_17_4c[
        "zip_validation_status"
    ].eq("zcta_consistent").all()
)

check(
    "All 500 historical K-Means lookups displayed",
    audit_17_4c[
        "kmeans_display_allowed"
    ].eq(True).all()
)

check(
    "All 500 K-Means area IDs are correct",
    audit_17_4c["kmeans_area_id"].eq(
        audit_17_4c["expected_area_id"]
    ).all()
)

check(
    "All 500 geographic checks fully verified",
    audit_17_4c[
        "fully_verified"
    ].all()
)

check(
    "All 20 areas successfully represented",
    audit_17_4c.loc[
        audit_17_4c["fully_verified"],
        "expected_area_id"
    ].nunique() == 20
)

check(
    "Five boroughs successfully represented",
    audit_17_4c.loc[
        audit_17_4c["fully_verified"],
        "borough"
    ].nunique() == 5
)

# --------------------------------------------------
# 7. Verify category distributions were preserved
# --------------------------------------------------

expected_distribution = (
    incoming.groupby([
        "incident_zip",
        "complaint_category"
    ])
    .size()
    .rename("expected_count")
)

actual_distribution = (
    audit_17_4c.loc[
        audit_17_4c["fully_verified"]
    ]
    .groupby([
        "incident_zip",
        "complaint_category"
    ])
    .size()
    .rename("verified_count")
)

distribution_comparison = (
    pd.concat([
        expected_distribution,
        actual_distribution
    ], axis=1)
    .fillna(0)
    .astype(int)
)

check(
    "All area-category counts preserved",
    distribution_comparison[
        "expected_count"
    ].eq(
        distribution_comparison[
            "verified_count"
        ]
    ).all()
)

# --------------------------------------------------
# 8. Summarize DBSCAN results without assumptions
# --------------------------------------------------

print("\n3. HISTORICAL DBSCAN ASSOCIATIONS")

display(
    audit_17_4c.groupby(
        "dbscan_status",
        dropna=False
    )
    .size()
    .rename("complaints")
    .reset_index()
)

check(
    "Every processed record has a DBSCAN status",
    audit_17_4c.loc[
        audit_17_4c[
            "processing_status"
        ].eq("processed"),
        "dbscan_status"
    ].notna().all()
)

# --------------------------------------------------
# 9. Verify each area's coverage
# --------------------------------------------------

area_coverage_17_4c = (
    audit_17_4c.groupby([
        "expected_area_id",
        "expected_borough",
        "intended_pattern"
    ], dropna=False)
    .agg(
        input_complaints=(
            "source_row", "size"
        ),
        fully_verified=(
            "fully_verified", "sum"
        ),
        historical_hotspot_matches=(
            "dbscan_cluster_id",
            "count"
        )
    )
    .reset_index()
    .sort_values([
        "expected_borough",
        "expected_area_id"
    ])
    .reset_index(drop=True)
)

check(
    "Every intended area contains 25 inputs",
    len(area_coverage_17_4c) == 20
    and area_coverage_17_4c[
        "input_complaints"
    ].eq(25).all()
)

check(
    "Every intended area has 25 verified outputs",
    len(area_coverage_17_4c) == 20
    and area_coverage_17_4c[
        "fully_verified"
    ].eq(25).all()
)

print("\n4. VERIFIED AREA COVERAGE")

display(area_coverage_17_4c)

# --------------------------------------------------
# 10. Show errors and unexpected geography clearly
# --------------------------------------------------

problems = audit_17_4c.loc[
    ~audit_17_4c["fully_verified"]
].copy()

print("\n5. GEOGRAPHIC VERIFICATION ISSUES")
print("Records requiring investigation:", len(problems))

if not problems.empty:
    display(
        problems[[
            "complaint_id",
            "expected_area_id",
            "expected_borough",
            "borough",
            "processing_status",
            "geography_status",
            "zip_validation_status",
            "dbscan_status",
            "kmeans_area_id",
            "processing_error"
        ]].head(20)
    )
else:
    print("No mismatches detected.")

# --------------------------------------------------
# 11. Final test checklist
# --------------------------------------------------

results_17_4c = pd.DataFrame(checks)

print("\n6. PHASE 17.4C TEST RESULTS")

display(results_17_4c)

passed = int(results_17_4c["passed"].sum())
total = len(results_17_4c)

print(f"\nChecks passed: {passed}/{total}")

failed_tests = results_17_4c.loc[
    ~results_17_4c["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 12. Save only if geographic verification passes
# --------------------------------------------------

if failed_tests:

    print("\nSTATUS: NEEDS REVIEW")
    print("Failed checks:")

    for name in failed_tests:
        print(" -", name)

    print(
        "\nNo verified output files saved. "
        "Review the mismatched records before "
        "training a new-batch clustering model."
    )

else:

    output_dir = (
        root / "outputs/tests"
        / "phase_17_4c_backend_verified"
    )

    result_path = (
        output_dir
        / "verified_500_backend_results.csv"
    )

    coverage_path = (
        output_dir / "verified_area_coverage.csv"
    )

    summary_path = (
        output_dir / "verification_summary.json"
    )

    summary = {
        "schema_version":
            "phase_17_4c_verification_v1",
        "dataset": "synthetic_500_diverse",
        "input_records": 500,
        "verified_records": int(
            audit_17_4c["fully_verified"].sum()
        ),
        "verified_areas": 20,
        "boroughs": 5,
        "backend_evaluations": int(
            stats["backend_evaluations"]
        ),
        "historical_hotspot_associations": int(
            audit_17_4c[
                "dbscan_cluster_id"
            ].notna().sum()
        ),
        "tests_passed": passed,
        "tests_total": total,
        "historical_reference_period":
            "2023-2025",
        "note": (
            "Synthetic data only. Historical model "
            "lookups are unchanged. No new clustering "
            "model has been trained."
        )
    }

    content_map = {
        result_path: csv_bytes(audit_17_4c),
        coverage_path: csv_bytes(
            area_coverage_17_4c
        ),
        summary_path: json.dumps(
            summary,
            indent=2
        ).encode("utf-8")
    }

    # Protect existing output files.
    for path, content in content_map.items():
        if path.exists():
            assert path.read_bytes() == content, (
                f"Existing file differs: {path}. "
                "No overwrite performed."
            )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    for path, content in content_map.items():
        if not path.exists():
            path.write_bytes(content)

    print("\n7. SAVED VERIFIED TEST RESULTS")
    print("Results:", result_path)
    print("Area coverage:", coverage_path)
    print("Summary:", summary_path)

    print("\nSTATUS: PASSED")
    print("PHASE 17.4C FULL BACKEND VERIFIED")
    print("500 UNIQUE COMPLAINT INPUTS EVALUATED")
    print("20 GEOGRAPHIC AREAS VERIFIED")
    print("READY FOR PHASE 18 NEW-BATCH K-MEANS")
    print("EXISTING HISTORICAL MODELS UNMODIFIED")
```

### New-Batch K-Means: Find Similar Areas
We can now start implementing the K-Means functionality shown in your PDF.

We'll keep two separate K-Means analyses:
Historical K-Means	New-batch K-Means
Existing 2023–2025 model	New model fitted to incoming area-level complaints
Describes historical reporting behavior	Groups areas with similar complaint-category percentages
Existing 238 area profiles	20 geographic areas in this experiment
Existing model unchanged	New model, separately tested

```python

from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.metrics import (
    adjusted_rand_score,
    silhouette_score
)

print("PHASE 18.1A — NEW-BATCH K-MEANS AREA SIMILARITY")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

results_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

for path in [
    results_path,
    taxonomy_path,
    reference_path
]:
    assert path.is_file(), f"Missing: {path}"

# --------------------------------------------------
# 1. Load backend-verified new complaints
# --------------------------------------------------

records = pd.read_csv(
    results_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

category_names = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

required = {
    "complaint_id",
    "borough",
    "incident_zip",
    "complaint_category",
    "processing_status",
    "zip_validation_status",
    "fully_verified",
    "kmeans_area_id",
    "kmeans_cluster_id",
    "kmeans_reporting_profile",
    "kmeans_display_allowed"
}

assert required.issubset(records.columns)

assert len(records) == 500
assert records["complaint_id"].is_unique
assert len(category_names) == 15

# --------------------------------------------------
# 2. Select geographically verified complaints
# --------------------------------------------------

verified_flag = (
    records["fully_verified"]
    .str.lower()
    .eq("true")
)

eligible = records.loc[
    verified_flag
    & records["processing_status"].eq(
        "processed"
    )
    & records["zip_validation_status"].eq(
        "zcta_consistent"
    )
].copy()

# New-batch geographic groups are derived from
# validated borough and ZIP data, not from the
# old K-Means cluster IDs.
eligible["area_id"] = (
    eligible["borough"].str.strip()
    + "_"
    + eligible["incident_zip"].str.strip()
)

assert len(eligible) == 500

assert eligible[
    "complaint_category"
].isin(category_names).all()

print("\n1. ELIGIBLE COMPLAINTS")
print("Geographically verified:", len(eligible))
print("Unique areas:", eligible["area_id"].nunique())
print("Supported categories:", len(category_names))

# --------------------------------------------------
# 3. Build new area-category count matrix
# --------------------------------------------------

area_category_counts_18_1a = pd.crosstab(
    eligible["area_id"],
    eligible["complaint_category"]
).reindex(
    columns=category_names,
    fill_value=0
)

area_totals = area_category_counts_18_1a.sum(
    axis=1
)

# Minimum sample size for this experiment.
# This threshold is a screening safeguard,
# not a guarantee of statistical reliability.
MIN_COMPLAINTS_PER_AREA = 20

eligible_areas = area_totals.loc[
    area_totals >= MIN_COMPLAINTS_PER_AREA
].index

excluded_areas = area_totals.loc[
    area_totals < MIN_COMPLAINTS_PER_AREA
]

feature_matrix_18_1a = (
    area_category_counts_18_1a
    .loc[eligible_areas]
    .div(
        area_totals.loc[eligible_areas],
        axis=0
    )
    .mul(100)
)

assert not feature_matrix_18_1a.isna().any().any()

print("\n2. AREA-CATEGORY FEATURES")
print("Areas eligible for clustering:", len(eligible_areas))
print("Areas below threshold:", len(excluded_areas))
print("Feature matrix shape:", feature_matrix_18_1a.shape)

display(
    feature_matrix_18_1a.loc[:, [
        "STREET CONDITION",
        "SIDEWALK CONDITION",
        "SEWER",
        "SEWER MAINTENANCE",
        "STREET LIGHT CONDITION",
        "TRAFFIC SIGNAL CONDITION"
    ]].round(1)
)

# --------------------------------------------------
# 4. Train separate new-batch K-Means
# --------------------------------------------------

# K=3 is chosen for the PDF-style experiment.
# It is NOT claimed to be optimal for real data.
NEW_BATCH_K = 3

new_batch_kmeans_18_1a = KMeans(
    n_clusters=NEW_BATCH_K,
    random_state=18,
    n_init=20
)

new_labels = new_batch_kmeans_18_1a.fit_predict(
    feature_matrix_18_1a
)

cluster_centers_18_1a = pd.DataFrame(
    new_batch_kmeans_18_1a.cluster_centers_,
    columns=category_names
)

cluster_centers_18_1a.index.name = (
    "new_batch_cluster"
)

# Describe clusters from their learned
# dominant categories, not the numeric IDs.
dominant_categories = (
    cluster_centers_18_1a.idxmax(axis=1)
)

friendly_names = {
    "STREET CONDITION":
        "Road-heavy reporting",
    "SEWER":
        "Sewer-heavy reporting",
    "STREET LIGHT CONDITION":
        "Lighting/signal-heavy reporting"
}

cluster_descriptions = dominant_categories.map(
    lambda category:
        friendly_names.get(
            category,
            f"{category} dominant reporting"
        )
)

print("\n3. NEW-BATCH CLUSTER PROFILES")

cluster_profiles_18_1a = pd.DataFrame({
    "new_batch_cluster":
        cluster_centers_18_1a.index,
    "description":
        cluster_descriptions.values,
    "dominant_category":
        dominant_categories.values,
    "street_condition_pct":
        cluster_centers_18_1a[
            "STREET CONDITION"
        ].round(2).values,
    "sewer_pct":
        cluster_centers_18_1a[
            "SEWER"
        ].round(2).values,
    "street_light_pct":
        cluster_centers_18_1a[
            "STREET LIGHT CONDITION"
        ].round(2).values
})

display(cluster_profiles_18_1a)

# --------------------------------------------------
# 5. Build new vs historical K-Means comparison
# --------------------------------------------------

new_cluster_lookup = pd.Series(
    new_labels,
    index=feature_matrix_18_1a.index,
    name="new_batch_cluster"
)

historical_consistency = (
    eligible.groupby("area_id")
    .agg(
        historical_cluster_variants=(
            "kmeans_cluster_id", "nunique"
        ),
        historical_profile_variants=(
            "kmeans_reporting_profile", "nunique"
        )
    )
)

assert historical_consistency[
    "historical_cluster_variants"
].eq(1).all()

assert historical_consistency[
    "historical_profile_variants"
].eq(1).all()

historical_area_context = (
    eligible.groupby("area_id")
    .agg(
        historical_kmeans_cluster=(
            "kmeans_cluster_id", "first"
        ),
        historical_reporting_profile=(
            "kmeans_reporting_profile", "first"
        )
    )
)

area_similarity_18_1a = pd.DataFrame(
    index=feature_matrix_18_1a.index
)

area_similarity_18_1a[
    "incoming_complaints"
] = area_totals.loc[
    feature_matrix_18_1a.index
]

area_similarity_18_1a[
    "new_batch_cluster"
] = new_cluster_lookup

area_similarity_18_1a[
    "new_batch_description"
] = new_cluster_lookup.map(
    cluster_descriptions
)

for category, output_column in [
    ("STREET CONDITION", "road_pct"),
    ("SEWER", "sewer_pct"),
    ("STREET LIGHT CONDITION", "street_light_pct")
]:
    area_similarity_18_1a[output_column] = (
        feature_matrix_18_1a[category]
    )

area_similarity_18_1a = (
    area_similarity_18_1a.join(
        historical_area_context,
        how="left"
    )
    .reset_index()
    .rename(columns={"index": "area_id"})
)

# Historical and new cluster IDs belong to
# different models and are NOT equivalent.

print("\n4. NEW AND HISTORICAL AREA COMPARISON")

display(
    area_similarity_18_1a[[
        "area_id",
        "incoming_complaints",
        "road_pct",
        "sewer_pct",
        "street_light_pct",
        "new_batch_cluster",
        "new_batch_description",
        "historical_kmeans_cluster",
        "historical_reporting_profile"
    ]].sort_values(
        ["new_batch_cluster", "area_id"]
    ).reset_index(drop=True)
)

# --------------------------------------------------
# 6. Test against intended synthetic patterns
# --------------------------------------------------

# These intended labels are loaded ONLY AFTER
# fitting the model. They were not training inputs.
pattern_reference = pd.read_csv(
    reference_path,
    dtype="string"
)

patterns = pattern_reference.set_index(
    "area_id"
)["intended_pattern"]

area_similarity_18_1a[
    "test_pattern"
] = area_similarity_18_1a[
    "area_id"
].map(patterns)

# The road_a and road_b templates are expected
# to form ONE broad road-reporting group.
expected_groups = (
    area_similarity_18_1a[
        "test_pattern"
    ].replace({
        "road_a": "road",
        "road_b": "road"
    })
)

ari_score = adjusted_rand_score(
    expected_groups,
    area_similarity_18_1a["new_batch_cluster"]
)

silhouette = silhouette_score(
    feature_matrix_18_1a,
    new_labels
)

print("\n5. SYNTHETIC EXPERIMENT QUALITY")

print("K:", NEW_BATCH_K)
print("Silhouette score:", round(silhouette, 4))
print("Synthetic pattern agreement (ARI):", round(ari_score, 4))

print("\n6. LEARNED CLUSTER COUNTS")

display(
    area_similarity_18_1a.groupby([
        "new_batch_cluster",
        "new_batch_description"
    ])
    .agg(
        areas=("area_id", "size"),
        incoming_complaints=(
            "incoming_complaints", "sum"
        )
    )
    .reset_index()
)

# --------------------------------------------------
# 7. Verify the new K-Means experiment
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 new complaints used",
    len(eligible) == 500
)

check(
    "20 areas eligible for analysis",
    len(feature_matrix_18_1a) == 20
)

check(
    "15 supported category features",
    feature_matrix_18_1a.shape[1] == 15
)

check(
    "At least 20 complaints in each area",
    area_totals.loc[
        eligible_areas
    ].ge(20).all()
)

check(
    "Each area has 25 new complaints",
    area_totals.eq(25).all()
)

check(
    "Category percentages total 100 per area",
    np.allclose(
        feature_matrix_18_1a.sum(axis=1),
        100
    )
)

check(
    "Three new K-Means clusters created",
    len(set(new_labels)) == 3
)

check(
    "Road, sewer and lighting cluster types found",
    set(dominant_categories) == {
        "STREET CONDITION",
        "SEWER",
        "STREET LIGHT CONDITION"
    }
)

cluster_sizes = sorted(
    pd.Series(new_labels).value_counts().tolist()
)

check(
    "Synthetic group sizes are 5, 5 and 10 areas",
    cluster_sizes == [5, 5, 10]
)

check(
    "Synthetic intended grouping recovered",
    np.isclose(ari_score, 1.0)
)

check(
    "Clusters show clear synthetic separation",
    silhouette > 0.80
)

check(
    "All 20 historical area profiles preserved",
    area_similarity_18_1a[
        "historical_reporting_profile"
    ].notna().all()
)

# Road variants should be grouped together
# within each borough.
road_pairs_match = True

for borough, group in (
    area_similarity_18_1a.groupby(
        area_similarity_18_1a[
            "area_id"
        ].str.rsplit("_", n=1).str[0]
    )
):
    road_rows = group.loc[
        group["test_pattern"].isin([
            "road_a", "road_b"
        ])
    ]

    if (
        len(road_rows) != 2
        or road_rows[
            "new_batch_cluster"
        ].nunique() != 1
    ):
        road_pairs_match = False

check(
    "Similar road areas grouped together in all boroughs",
    road_pairs_match
)

check(
    "All modeled areas retain unique identifiers",
    area_similarity_18_1a[
        "area_id"
    ].is_unique
)

# --------------------------------------------------
# 8. Final results
# --------------------------------------------------

results_18_1a = pd.DataFrame(checks)

print("\n7. PHASE 18.1A TEST CHECKLIST")

display(results_18_1a)

passed = int(results_18_1a["passed"].sum())
total = len(results_18_1a)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_1a.loc[
    ~results_18_1a["passed"],
    "test"
].tolist()

if failures:
    print("\nSTATUS: NEEDS REVIEW")
    for name in failures:
        print("FAILED:", name)
else:
    print("\nSTATUS: PASSED")
    print("PHASE 18.1A NEW-BATCH K-MEANS VERIFIED")
    print("SIMILAR AREA REPORTING PATTERNS IDENTIFIED")
    print("HISTORICAL AND NEW K-MEANS KEPT SEPARATE")
    print("NO HISTORICAL MODEL MODIFICATIONS")
    print("MODEL NOT YET DEPLOYED")
```

### Build a Reusable New-Batch K-Means Engine

```python

from pathlib import Path
from textwrap import dedent
import sys
import pandas as pd
import numpy as np
from sklearn.metrics import adjusted_rand_score

print(" REUSABLE NEW-BATCH K-MEANS ENGINE")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/models/new_batch_area_kmeans.py"
)

results_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

for path in [results_path, reference_path, taxonomy_path]:
    assert path.is_file(), f"Missing: {path}"

# --------------------------------------------------
# 1. Define reusable K-Means engine
# --------------------------------------------------

module_source = dedent(r'''
"""New-batch area similarity using complaint-category shares.

This module never modifies or retrains historical
K-Means or DBSCAN artifacts.

A cluster identifies similar reporting composition,
not defect severity or operational priority.
"""

import numpy as np
import pandas as pd

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score


REQUIRED_COLUMNS = {
    "source_row",
    "complaint_id",
    "borough",
    "incident_zip",
    "complaint_category",
    "processing_status",
    "zip_validation_status"
}

VALID_BOROUGHS = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}


def analyze_new_batch_areas(
    records,
    supported_categories,
    k=3,
    min_area_size=20,
    random_state=18
):
    """Cluster geographic areas by new complaint shares.

    Input must come from the geographic batch
    processor. Historical labels are context only.
    """

    if not isinstance(records, pd.DataFrame):
        raise TypeError("Expected a DataFrame.")

    missing = REQUIRED_COLUMNS - set(records.columns)

    if missing:
        raise ValueError(
            f"Missing columns: {sorted(missing)}"
        )

    if records["source_row"].duplicated().any():
        raise ValueError("Duplicate source rows.")

    if not isinstance(k, int) or k < 2:
        raise ValueError("k must be an integer >= 2.")

    if (
        not isinstance(min_area_size, int)
        or min_area_size < 1
    ):
        raise ValueError(
            "Minimum area size must be positive."
        )

    categories = sorted({
        str(value).strip().upper()
        for value in supported_categories
        if pd.notna(value)
        and str(value).strip()
    })

    if not categories:
        raise ValueError("No supported categories.")

    # ----------------------------------------------
    # 2. Normalize and screen geographic evidence
    # ----------------------------------------------

    work = records.copy().reset_index(drop=True)

    text_columns = [
        "borough",
        "incident_zip",
        "complaint_category",
        "processing_status",
        "zip_validation_status"
    ]

    for column in text_columns:
        work[column] = (
            work[column]
            .fillna("")
            .astype(str)
            .str.strip()
        )

    work["borough"] = work[
        "borough"
    ].str.upper()

    work["complaint_category"] = work[
        "complaint_category"
    ].str.upper()

    work["area_id"] = pd.NA
    work["eligibility_reason"] = "eligible"

    # Preserve every excluded row with a reason.
    tests = [
        (
            work["processing_status"].ne("processed"),
            "not_processed"
        ),
        (
            work["zip_validation_status"].ne(
                "zcta_consistent"
            ),
            "zip_not_verified"
        ),
        (
            ~work["borough"].isin(VALID_BOROUGHS),
            "invalid_borough"
        ),
        (
            ~work["incident_zip"].str.fullmatch(
                r"\d{5}"
            ),
            "invalid_zip"
        ),
        (
            ~work["complaint_category"].isin(
                categories
            ),
            "unsupported_category"
        )
    ]

    for bad, reason in tests:
        work.loc[
            bad
            & work["eligibility_reason"].eq(
                "eligible"
            ),
            "eligibility_reason"
        ] = reason

    valid = work[
        "eligibility_reason"
    ].eq("eligible")

    work.loc[valid, "area_id"] = (
        work.loc[valid, "borough"]
        + "_"
        + work.loc[valid, "incident_zip"]
    )

    included = work.loc[valid].copy()
    excluded = work.loc[~valid].copy()

    # ----------------------------------------------
    # 3. Create category-count and share matrices
    # ----------------------------------------------

    if included.empty:

        counts = pd.DataFrame(
            index=pd.Index([], name="area_id"),
            columns=categories
        ).fillna(0).astype(int)

    else:

        counts = pd.crosstab(
            included["area_id"],
            included["complaint_category"]
        ).reindex(
            columns=categories,
            fill_value=0
        )

        counts.index.name = "area_id"

    area_sizes = counts.sum(axis=1).astype(int)

    enough = area_sizes.ge(min_area_size)

    features = (
        counts.loc[enough]
        .div(
            area_sizes.loc[enough],
            axis=0
        )
        .mul(100)
    )

    area_summary = pd.DataFrame({
        "area_id": area_sizes.index,
        "incoming_complaints":
            area_sizes.to_numpy(),
        "meets_minimum":
            enough.to_numpy()
    })

    # ----------------------------------------------
    # 4. Add optional historical K-Means context
    # ----------------------------------------------

    area_summary[
        "historical_kmeans_cluster"
    ] = pd.NA

    area_summary[
        "historical_reporting_profile"
    ] = pd.NA

    historical_columns = {
        "kmeans_area_id",
        "kmeans_cluster_id",
        "kmeans_reporting_profile",
        "kmeans_display_allowed"
    }

    if historical_columns.issubset(
        included.columns
    ):

        display_allowed = (
            included["kmeans_display_allowed"]
            .fillna(False)
            .astype(str)
            .str.strip()
            .str.lower()
            .eq("true")
        )

        matching_area = (
            included["kmeans_area_id"]
            .fillna("")
            .astype(str)
            .str.strip()
            .eq(included["area_id"])
        )

        history = included.loc[
            display_allowed & matching_area
        ].copy()

        if not history.empty:

            consistency = (
                history.groupby("area_id")
                .agg(
                    cluster_variants=(
                        "kmeans_cluster_id",
                        "nunique"
                    ),
                    profile_variants=(
                        "kmeans_reporting_profile",
                        "nunique"
                    )
                )
            )

            if (consistency > 1).any().any():
                raise ValueError(
                    "Conflicting historical area profiles."
                )

            lookup = (
                history.drop_duplicates("area_id")
                .set_index("area_id")
            )

            area_summary[
                "historical_kmeans_cluster"
            ] = area_summary["area_id"].map(
                lookup["kmeans_cluster_id"]
            )

            area_summary[
                "historical_reporting_profile"
            ] = area_summary["area_id"].map(
                lookup["kmeans_reporting_profile"]
            )

    # ----------------------------------------------
    # 5. Prepare traceable output and safeguards
    # ----------------------------------------------

    output = {
        "status": None,
        "reason": None,
        "record_eligibility": work[[
            "source_row",
            "complaint_id",
            "area_id",
            "eligibility_reason"
        ]].copy(),
        "area_summary": area_summary,
        "area_category_counts": counts,
        "area_category_percentages": features,
        "cluster_profiles": pd.DataFrame(),
        "model": None,
        "silhouette": None,
        "stats": {
            "total_records": len(work),
            "eligible_records": len(included),
            "excluded_records": len(excluded),
            "eligible_areas": int(enough.sum()),
            "small_sample_areas": int(
                (~enough).sum()
            ),
            "records_in_eligible_areas": int(
                area_sizes.loc[enough].sum()
            ),
            "records_in_small_areas": int(
                area_sizes.loc[~enough].sum()
            )
        }
    }

    if len(features) < k:

        output["status"] = "insufficient_areas"

        output["reason"] = (
            f"Need at least {k} areas with "
            f"{min_area_size}+ complaints."
        )

        return output

    unique_patterns = np.unique(
        features.to_numpy(dtype=float),
        axis=0
    ).shape[0]

    if unique_patterns < k:

        output["status"] = (
            "insufficient_distinct_patterns"
        )

        output["reason"] = (
            f"Need at least {k} different "
            f"area category-share patterns."
        )

        return output

    # ----------------------------------------------
    # 6. Fit only the NEW area-category features
    # ----------------------------------------------

    model = KMeans(
        n_clusters=k,
        random_state=random_state,
        n_init=20
    )

    labels = model.fit_predict(features)

    if len(np.unique(labels)) != k:

        output["status"] = (
            "insufficient_distinct_patterns"
        )

        output["reason"] = (
            "K-Means did not produce the "
            "requested number of clusters."
        )

        return output

    centers = pd.DataFrame(
        model.cluster_centers_,
        columns=categories
    )

    centers.index.name = "new_batch_cluster"

    dominant = centers.idxmax(axis=1)

    readable_names = {
        "STREET CONDITION":
            "Road-heavy reporting",
        "SEWER":
            "Sewer-heavy reporting",
        "STREET LIGHT CONDITION":
            "Lighting/signal-heavy reporting"
    }

    descriptions = dominant.map(
        lambda category: readable_names.get(
            category,
            f"{category.title()}-dominant reporting"
        )
    )

    cluster_profiles = pd.DataFrame({
        "new_batch_cluster": range(k),
        "description": descriptions.to_numpy(),
        "dominant_category": dominant.to_numpy(),
        "areas": (
            pd.Series(labels)
            .value_counts()
            .reindex(range(k), fill_value=0)
            .to_numpy()
        )
    })

    for category in [
        "STREET CONDITION",
        "SEWER",
        "STREET LIGHT CONDITION"
    ]:
        if category in centers:

            cluster_profiles[
                category + "_pct"
            ] = (
                centers[category]
                .round(2)
                .to_numpy()
            )

    assignments = pd.Series(
        labels,
        index=features.index,
        name="new_batch_cluster"
    )

    area_summary["new_batch_cluster"] = (
        pd.to_numeric(
            area_summary["area_id"].map(
                assignments
            ),
            errors="coerce"
        ).astype("Int64")
    )

    area_summary[
        "new_batch_description"
    ] = area_summary[
        "new_batch_cluster"
    ].map(descriptions)

    output.update({
        "status": "clustered",
        "reason": None,
        "model": model,
        "cluster_profiles": cluster_profiles,
        "silhouette": float(
            silhouette_score(features, labels)
        )
    })

    return output
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

analyze_new_batch_areas = namespace[
    "analyze_new_batch_areas"
]

# --------------------------------------------------
# 7. Load verified incoming data
# --------------------------------------------------

records = pd.read_csv(
    results_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

print("\n1. REUSABLE ENGINE — FULL BATCH TEST")

full = analyze_new_batch_areas(
    records,
    categories,
    k=3,
    min_area_size=20
)

print("Status:", full["status"])
print("Silhouette:", full["silhouette"])

display(full["cluster_profiles"])

display(full["area_summary"][[
    "area_id",
    "incoming_complaints",
    "new_batch_cluster",
    "new_batch_description",
    "historical_kmeans_cluster",
    "historical_reporting_profile"
]])

# --------------------------------------------------
# 8. Validate against the known synthetic patterns
# --------------------------------------------------

reference = pd.read_csv(
    reference_path,
    dtype="string"
)

pattern_lookup = reference.set_index(
    "area_id"
)["intended_pattern"]

area_table = full["area_summary"].copy()

actual_labels = area_table[
    "new_batch_cluster"
]

expected_labels = (
    area_table["area_id"]
    .map(pattern_lookup)
    .replace({
        "road_a": "road",
        "road_b": "road"
    })
)

ari_score = adjusted_rand_score(
    expected_labels,
    actual_labels
)

print("\n2. SYNTHETIC QUALITY CHECK")
print("Adjusted Rand Index:", round(ari_score, 4))

# --------------------------------------------------
# 9. Test insufficient per-area complaint counts
# --------------------------------------------------

first_area = (
    records.iloc[0]["borough"]
    + "_"
    + records.iloc[0]["incident_zip"]
)

area_id_series = (
    records["borough"]
    + "_"
    + records["incident_zip"]
)

first_area_mask = area_id_series.eq(
    first_area
)

first_three = records.loc[
    first_area_mask
].index[:3]

small_batch = records.loc[
    ~first_area_mask
    | records.index.isin(first_three)
].copy()

small_result = analyze_new_batch_areas(
    small_batch,
    categories
)

print("\n3. SMALL-SAMPLE SAFEGUARD")
print("Status:", small_result["status"])
print("Stats:", small_result["stats"])

# --------------------------------------------------
# 10. Test insufficient area coverage
# --------------------------------------------------

one_area = records.loc[
    first_area_mask
].copy()

one_area_result = analyze_new_batch_areas(
    one_area,
    categories
)

print("\n4. INSUFFICIENT-AREA SAFEGUARD")
print("Status:", one_area_result["status"])
print("Reason:", one_area_result["reason"])

# --------------------------------------------------
# 11. Test insufficient category-pattern diversity
# --------------------------------------------------

road_area_ids = set(
    reference.loc[
        reference["intended_pattern"].isin([
            "road_a", "road_b"
        ]),
        "area_id"
    ]
)

road_only = records.loc[
    area_id_series.isin(road_area_ids)
].copy()

road_only_result = analyze_new_batch_areas(
    road_only,
    categories
)

print("\n5. INSUFFICIENT-PATTERN SAFEGUARD")
print("Status:", road_only_result["status"])
print("Reason:", road_only_result["reason"])

# --------------------------------------------------
# 12. Test exclusion of geographic mismatches
# --------------------------------------------------

mismatch_data = records.copy()

mismatch_data.loc[
    0, "zip_validation_status"
] = "zcta_mismatch"

mismatch_result = analyze_new_batch_areas(
    mismatch_data,
    categories
)

# --------------------------------------------------
# 13. Test deterministic, reproducible results
# --------------------------------------------------

repeat_result = analyze_new_batch_areas(
    records,
    categories,
    k=3,
    min_area_size=20
)

# --------------------------------------------------
# 14. Comprehensive validation checklist
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 input records accounted for",
    full["stats"]["total_records"] == 500
)

check(
    "500 eligible complaints",
    full["stats"]["eligible_records"] == 500
)

check(
    "20 areas eligible for clustering",
    full["stats"]["eligible_areas"] == 20
)

check(
    "All source complaint rows preserved",
    len(full["record_eligibility"]) == 500
)

check(
    "No complaints excluded in verified batch",
    full["stats"]["excluded_records"] == 0
)

check(
    "Three new-batch clusters created",
    full["status"] == "clustered"
    and len(full["cluster_profiles"]) == 3
)

check(
    "All new area feature percentages total 100",
    np.allclose(
        full[
            "area_category_percentages"
        ].sum(axis=1),
        100
    )
)

check(
    "Intended synthetic groups recovered",
    np.isclose(ari_score, 1.0)
)

check(
    "Synthetic silhouette exceeds 0.80",
    full["silhouette"] > 0.80
)

check(
    "Three broad reporting patterns recovered",
    set(
        full["cluster_profiles"][
            "description"
        ]
    ) == {
        "Road-heavy reporting",
        "Sewer-heavy reporting",
        "Lighting/signal-heavy reporting"
    }
)

check(
    "Expected synthetic cluster sizes recovered",
    sorted(
        full["cluster_profiles"]["areas"].tolist()
    ) == [5, 5, 10]
)

check(
    "All historical area profiles preserved",
    full["area_summary"][
        "historical_reporting_profile"
    ].notna().all()
)

check(
    "Small-sample area excluded from clustering",
    small_result["status"] == "clustered"
    and small_result["stats"][
        "small_sample_areas"
    ] == 1
    and small_result["stats"][
        "records_in_small_areas"
    ] == 3
)

check(
    "Single-area input does not fit K-Means",
    one_area_result["status"]
    == "insufficient_areas"
    and one_area_result["model"] is None
)

check(
    "Insufficient pattern diversity does not fit K-Means",
    road_only_result["status"]
    == "insufficient_distinct_patterns"
    and road_only_result["model"] is None
)

check(
    "ZIP-mismatched complaint excluded with reason",
    mismatch_result["stats"][
        "excluded_records"
    ] == 1
    and mismatch_result[
        "record_eligibility"
    ].iloc[0]["eligibility_reason"]
    == "zip_not_verified"
)

check(
    "Repeated analysis is reproducible",
    full["area_summary"][
        "new_batch_cluster"
    ].tolist()
    == repeat_result["area_summary"][
        "new_batch_cluster"
    ].tolist()
)

check(
    "Existing historical models not fitted by new engine",
    full["model"].n_features_in_
    == len(categories)
)

# --------------------------------------------------
# 15. Display validation results
# --------------------------------------------------

results_18_1b = pd.DataFrame(checks)

print("\n6. PHASE 18.1B VALIDATION")

display(results_18_1b)

passed = int(results_18_1b["passed"].sum())
total = len(results_18_1b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_1b.loc[
    ~results_18_1b["passed"],
    "test"
].tolist()

if failures:

    print("\nSTATUS: NEEDS REVIEW")

    for failure in failures:
        print("FAILED:", failure)

    print("No new module saved.")

else:

    # ----------------------------------------------
    # 16. Save only the validated reusable module
    # ----------------------------------------------

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == module_source, (
            "Existing new-batch K-Means module "
            "differs. No overwrite performed."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        module_path.write_text(
            module_source,
            encoding="utf-8"
        )

    print("\n7. VERIFIED MODULE")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 18.1B REUSABLE K-MEANS ENGINE VERIFIED")
    print("SMALL-SAMPLE SAFEGUARDS VERIFIED")
    print("INSUFFICIENT-DIVERSITY SAFEGUARDS VERIFIED")
    print("HISTORICAL MODELS UNMODIFIED")
    print("REAL-DATA VALIDATION STILL REQUIRED")
```

### K Selection and Stability Testing

Should we always use K=3, or should the system evaluate whether another number of groups is better?

Our synthetic dataset deliberately has four exact category-distribution templates, but we want three broader reporting groups. A model with K=4 could separate the two very similar road-heavy templates unnecessarily.

We'll compare K=2, K=3 and K=4, then test how the assignments change when complaint-category counts vary slightly. This will help prevent the assistant from choosing a misleading number of clusters based only on a silhouette score.

```python

from pathlib import Path
from threadpoolctl import threadpool_limits
from sklearn.metrics import adjusted_rand_score
import hashlib
import numpy as np
import pandas as pd

print(" K-SELECTION AND STABILITY TESTS")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

module_path = (
    root / "src/models/new_batch_area_kmeans.py"
)

results_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

for path in [
    module_path,
    results_path,
    reference_path,
    taxonomy_path
]:
    assert path.is_file(), f"Missing: {path}"

from src.models.new_batch_area_kmeans import (
    analyze_new_batch_areas
)

source_hash_before = hashlib.sha256(
    module_path.read_bytes()
).hexdigest()

records = pd.read_csv(
    results_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

assert len(records) == 500
assert len(categories) == 15

# --------------------------------------------------
# 1. Compare possible numbers of clusters
# --------------------------------------------------

print("\n1. COMPARING K=2, K=3, K=4 AND K=5")

candidate_results = {}
comparison_rows = []

with threadpool_limits(
    limits=1,
    user_api="openmp"
):

    for k in [2, 3, 4, 5]:

        result = analyze_new_batch_areas(
            records,
            categories,
            k=k,
            min_area_size=20,
            random_state=18
        )

        candidate_results[k] = result

        cluster_sizes = []

        if result["status"] == "clustered":
            cluster_sizes = sorted(
                result["cluster_profiles"][
                    "areas"
                ].astype(int).tolist()
            )

        comparison_rows.append({
            "k": k,
            "status": result["status"],
            "silhouette": (
                round(result["silhouette"], 4)
                if result["silhouette"] is not None
                else np.nan
            ),
            "eligible_areas":
                result["stats"]["eligible_areas"],
            "cluster_sizes": str(cluster_sizes)
        })

k_comparison_18_1c = pd.DataFrame(
    comparison_rows
)

display(k_comparison_18_1c)

# --------------------------------------------------
# 2. Examine the distinct category-share patterns
# --------------------------------------------------

features = candidate_results[3][
    "area_category_percentages"
]

unique_patterns = np.unique(
    features.to_numpy(dtype=float),
    axis=0
).shape[0]

print("\n2. DATASET CHARACTERISTICS")

print("Areas:", len(features))
print("Category features:", features.shape[1])
print("Distinct category-share patterns:", unique_patterns)

# --------------------------------------------------
# 3. Evaluate agreement with test patterns
# --------------------------------------------------

# Intended patterns are used only for evaluation,
# not as inputs to the K-Means models.

reference = pd.read_csv(
    reference_path,
    dtype="string"
)

pattern_lookup = reference.set_index(
    "area_id"
)["intended_pattern"]

def get_assignments(result):
    return (
        result["area_summary"]
        .set_index("area_id")[
            "new_batch_cluster"
        ]
        .dropna()
        .astype(int)
        .sort_index()
    )

baseline_k3 = get_assignments(
    candidate_results[3]
)

baseline_k4 = get_assignments(
    candidate_results[4]
)

expected_four = (
    baseline_k3.index.to_series()
    .map(pattern_lookup)
)

expected_three = expected_four.replace({
    "road_a": "road",
    "road_b": "road"
})

k3_broad_ari = adjusted_rand_score(
    expected_three,
    baseline_k3
)

k4_broad_ari = adjusted_rand_score(
    expected_three,
    baseline_k4
)

k4_four_pattern_ari = adjusted_rand_score(
    expected_four,
    baseline_k4
)

print("\n3. SYNTHETIC PATTERN AGREEMENT")

print(
    "K=3 vs three broad groups:",
    round(k3_broad_ari, 4)
)

print(
    "K=4 vs three broad groups:",
    round(k4_broad_ari, 4)
)

print(
    "K=4 vs four exact templates:",
    round(k4_four_pattern_ari, 4)
)

# --------------------------------------------------
# 4. Check whether K=4 splits similar road areas
# --------------------------------------------------

road_a_ids = pattern_lookup.loc[
    pattern_lookup.eq("road_a")
].index

road_b_ids = pattern_lookup.loc[
    pattern_lookup.eq("road_b")
].index

road_a_clusters = set(
    baseline_k4.loc[road_a_ids].tolist()
)

road_b_clusters = set(
    baseline_k4.loc[road_b_ids].tolist()
)

k4_splits_road_variants = (
    len(road_a_clusters) == 1
    and len(road_b_clusters) == 1
    and road_a_clusters != road_b_clusters
)

print("\n4. SIMILAR ROAD-AREA COMPARISON")
print("K=4 road_a cluster(s):", road_a_clusters)
print("K=4 road_b cluster(s):", road_b_clusters)
print(
    "K=4 separates road variants:",
    k4_splits_road_variants
)

# --------------------------------------------------
# 5. Check K=3 consistency across random seeds
# --------------------------------------------------

print("\n5. RANDOM-SEED STABILITY")

seed_rows = []

with threadpool_limits(
    limits=1,
    user_api="openmp"
):

    for seed in [0, 7, 18, 42, 99]:

        result = analyze_new_batch_areas(
            records,
            categories,
            k=3,
            min_area_size=20,
            random_state=seed
        )

        if result["status"] != "clustered":
            agreement = np.nan
        else:
            labels = get_assignments(result)

            agreement = adjusted_rand_score(
                baseline_k3.loc[labels.index],
                labels
            )

        seed_rows.append({
            "random_state": seed,
            "status": result["status"],
            "agreement_with_baseline": agreement
        })

seed_stability_18_1c = pd.DataFrame(
    seed_rows
)

display(seed_stability_18_1c)

# --------------------------------------------------
# 6. Simulate variation in category reporting
# --------------------------------------------------

# Keep the same 500 records and validated areas,
# but resample complaint categories within each
# area to assess sampling sensitivity.
#
# These are synthetic perturbations, NOT
# additional real complaint records.

def resample_area_categories(frame, seed):

    rng = np.random.default_rng(seed)

    sampled = frame.copy().reset_index(
        drop=True
    )

    grouped_indices = sampled.groupby([
        "borough",
        "incident_zip"
    ]).groups

    for _, indices in grouped_indices.items():

        original_categories = (
            sampled.loc[
                indices,
                "complaint_category"
            ]
            .astype(str)
            .to_numpy()
        )

        sampled.loc[
            indices,
            "complaint_category"
        ] = rng.choice(
            original_categories,
            size=len(indices),
            replace=True
        )

    return sampled


print("\n6. CATEGORY-SAMPLING STABILITY")

bootstrap_rows = []

N_TRIALS = 12

with threadpool_limits(
    limits=1,
    user_api="openmp"
):

    for trial in range(N_TRIALS):

        perturbed = resample_area_categories(
            records,
            seed=18100 + trial
        )

        trial_row = {"trial": trial + 1}

        for k in [3, 4]:

            result = analyze_new_batch_areas(
                perturbed,
                categories,
                k=k,
                min_area_size=20,
                random_state=18
            )

            trial_row[
                f"k{k}_status"
            ] = result["status"]

            if result["status"] == "clustered":

                assignments = get_assignments(
                    result
                )

                reference_labels = (
                    baseline_k3
                    if k == 3
                    else baseline_k4
                )

                trial_row[
                    f"k{k}_agreement"
                ] = adjusted_rand_score(
                    reference_labels.loc[
                        assignments.index
                    ],
                    assignments
                )

            else:

                trial_row[
                    f"k{k}_agreement"
                ] = np.nan

        bootstrap_rows.append(trial_row)

bootstrap_stability_18_1c = pd.DataFrame(
    bootstrap_rows
)

display(bootstrap_stability_18_1c)

# --------------------------------------------------
# 7. Summarize the stability findings
# --------------------------------------------------

stability_summary_18_1c = pd.DataFrame([
    {
        "model": "K=3",
        "median_agreement": (
            bootstrap_stability_18_1c[
                "k3_agreement"
            ].median()
        ),
        "minimum_agreement": (
            bootstrap_stability_18_1c[
                "k3_agreement"
            ].min()
        ),
        "successful_trials": (
            bootstrap_stability_18_1c[
                "k3_status"
            ].eq("clustered").sum()
        )
    },
    {
        "model": "K=4",
        "median_agreement": (
            bootstrap_stability_18_1c[
                "k4_agreement"
            ].median()
        ),
        "minimum_agreement": (
            bootstrap_stability_18_1c[
                "k4_agreement"
            ].min()
        ),
        "successful_trials": (
            bootstrap_stability_18_1c[
                "k4_status"
            ].eq("clustered").sum()
        )
    }
])

print("\n7. STABILITY SUMMARY")

display(
    stability_summary_18_1c.round(4)
)

k3_median = float(
    stability_summary_18_1c.iloc[0][
        "median_agreement"
    ]
)

k4_median = float(
    stability_summary_18_1c.iloc[1][
        "median_agreement"
    ]
)

# --------------------------------------------------
# 8. Validate the selection experiment
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "500 complaint records preserved",
    len(records) == 500
)

check(
    "20 area feature vectors available",
    len(features) == 20
)

check(
    "Four distinct synthetic templates detected",
    unique_patterns == 4
)

for k in [2, 3, 4]:
    check(
        f"K={k} clustering completed",
        candidate_results[k]["status"]
        == "clustered"
    )

check(
    "K=5 rejected due to insufficient patterns",
    candidate_results[5]["status"]
    == "insufficient_distinct_patterns"
)

check(
    "K=3 recovers three intended broad groups",
    np.isclose(k3_broad_ari, 1.0)
)

check(
    "K=4 recovers four exact patterns",
    np.isclose(k4_four_pattern_ari, 1.0)
)

check(
    "K=4 separates similar road variants",
    k4_splits_road_variants
)

check(
    "K=4 has higher synthetic silhouette than K=3",
    candidate_results[4]["silhouette"]
    > candidate_results[3]["silhouette"]
)

check(
    "K=3 has higher broad-group agreement",
    k3_broad_ari > k4_broad_ari
)

check(
    "All five K=3 seeds are reproducible",
    seed_stability_18_1c[
        "agreement_with_baseline"
    ].eq(1.0).all()
)

check(
    "12 category-sampling trials completed",
    len(bootstrap_stability_18_1c)
    == N_TRIALS
)

check(
    "All K=3 sampling trials produced clusters",
    bootstrap_stability_18_1c[
        "k3_status"
    ].eq("clustered").all()
)

check(
    "All K=4 sampling trials produced clusters",
    bootstrap_stability_18_1c[
        "k4_status"
    ].eq("clustered").all()
)

check(
    "K=3 median sampling agreement exceeds 0.90",
    k3_median >= 0.90
)

check(
    "K=3 more stable than K=4 in this experiment",
    k3_median > k4_median
)

check(
    "Historical and new-batch models remain separate",
    candidate_results[3][
        "area_summary"
    ]["historical_reporting_profile"]
    .notna().all()
)

source_hash_after = hashlib.sha256(
    module_path.read_bytes()
).hexdigest()

check(
    "Reusable model source remains unchanged",
    source_hash_before == source_hash_after
)

# --------------------------------------------------
# 9. Report results
# --------------------------------------------------

results_18_1c = pd.DataFrame(checks)

print("\n8. PHASE 18.1C TEST CHECKLIST")

display(results_18_1c)

passed = int(
    results_18_1c["passed"].sum()
)

total = len(results_18_1c)

print(f"\nChecks passed: {passed}/{total}")

failed = results_18_1c.loc[
    ~results_18_1c["passed"],
    "test"
].tolist()

if failed:

    print("\nSTATUS: NEEDS REVIEW")

    for name in failed:
        print("FAILED:", name)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 18.1C K-SELECTION EXPERIMENT VERIFIED")
    print("K=3 BROAD REPORTING GROUPS RECOVERED")
    print("CATEGORY-SAMPLING STABILITY EVALUATED")
    print("MODEL SELECTION REMAINS HUMAN-REVIEWED")
    print("HISTORICAL MODELS UNMODIFIED")
    print("NO PRODUCTION MODEL DEPLOYED")
```

### Robustness Hardening
Before adding the new K-Means engine to the GUI, we should fix two details in the reusable module:
1. Silhouette safeguard: When the number of qualifying areas equals K, scikit-learn cannot calculate a silhouette score. The engine should return a clear status instead of raising an error.
2. Small-area traceability: When an area has fewer than 20 complaints, its individual records should explicitly say that the area was excluded from clustering.

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import ast
import hashlib
import importlib
import sys
import pandas as pd
import numpy as np

print("PHASE 18.1D.1 — SAFE K-MEANS REPAIR")
print("=" * 70)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/models/new_batch_area_kmeans.py"
)

results_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

backup_path = (
    root / "outputs/tests/phase_18_1d"
    / "new_batch_area_kmeans_before_hardening.txt"
)

for path in [
    module_path,
    results_path,
    reference_path,
    taxonomy_path
]:
    assert path.is_file(), f"Missing file: {path}"

# --------------------------------------------------
# 1. Read the real module without changing it
# --------------------------------------------------

original_bytes = module_path.read_bytes()
original_source = original_bytes.decode("utf-8")

print("\n1. ORIGINAL MODULE")
print("File:", module_path)
print("Bytes:", len(original_bytes))

tree = ast.parse(original_source)

# --------------------------------------------------
# 2. Locate assignments by Python syntax
#    instead of exact whitespace
# --------------------------------------------------

def find_assignment(variable):
    matches = []

    for node in ast.walk(tree):
        if not isinstance(node, ast.Assign):
            continue

        if any(
            isinstance(target, ast.Name)
            and target.id == variable
            for target in node.targets
        ):
            matches.append(node)

    if len(matches) != 1:
        raise ValueError(
            f"Expected one assignment for "
            f"{variable}, found {len(matches)}. "
            "Module unchanged."
        )

    return matches[0]


enough_node = find_assignment("enough")
patterns_node = find_assignment(
    "unique_patterns"
)

enough_expression = ast.get_source_segment(
    original_source, enough_node
)

patterns_expression = ast.get_source_segment(
    original_source, patterns_node
)

assert "area_sizes" in enough_expression
assert "min_area_size" in enough_expression
assert "np.unique" in patterns_expression

print("\n2. PYTHON STATEMENTS FOUND")
print("Area threshold:", enough_expression)
print("Pattern count:", patterns_expression)

# --------------------------------------------------
# 3. Prepare fixes using syntax locations
# --------------------------------------------------

newline = (
    "\r\n"
    if "\r\n" in original_source
    else "\n"
)

lines = original_source.splitlines(
    keepends=True
)

def indentation_of(node):
    line = lines[node.lineno - 1]
    return line[:len(line) - len(line.lstrip())]


def make_block(body, indentation):
    body_lines = dedent(body).strip(
        "\n"
    ).splitlines()

    return newline + "".join(
        (
            indentation + line + newline
            if line.strip()
            else newline
        )
        for line in body_lines
    )


small_area_fix = """
# Mark records from areas below the minimum
# sample threshold without discarding them.
small_area_mask = (
    valid
    & work["area_id"].isin(
        area_sizes.index[~enough]
    )
)

work.loc[
    small_area_mask,
    "eligibility_reason"
] = "below_minimum_area_size"
"""

silhouette_fix = """
# Silhouette is undefined when the number of
# qualifying areas equals the number of clusters.
if len(features) == k:

    output["status"] = (
        "insufficient_areas_for_validation"
    )

    output["reason"] = (
        f"Need at least {k + 1} qualifying "
        f"areas to evaluate K={k} safely."
    )

    return output
"""

has_small_fix = (
    '"below_minimum_area_size"'
    in original_source
)

has_silhouette_fix = (
    '"insufficient_areas_for_validation"'
    in original_source
)

insertions = []

if not has_small_fix:
    insertions.append((
        enough_node.end_lineno,
        make_block(
            small_area_fix,
            indentation_of(enough_node)
        )
    ))

if not has_silhouette_fix:
    insertions.append((
        patterns_node.lineno - 1,
        make_block(
            silhouette_fix,
            indentation_of(patterns_node)
        )
    ))

# Insert from bottom to top so original line
# numbers remain correct during both insertions.
for position, block in sorted(
    insertions,
    key=lambda item: item[0],
    reverse=True
):
    lines.insert(position, block)

revised_source = "".join(lines)

compile(
    revised_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        revised_source,
        str(module_path),
        "exec"
    ),
    namespace
)

analyze_fixed = namespace[
    "analyze_new_batch_areas"
]

print("\n3. PATCH PREPARATION")
print("New corrections required:", len(insertions))
print("Corrected module compiles: YES")
print("Original file still unchanged: YES")

# --------------------------------------------------
# 4. Load previously verified synthetic data
# --------------------------------------------------

records = pd.read_csv(
    results_path,
    dtype="string",
    keep_default_na=False
)

reference = pd.read_csv(
    reference_path,
    dtype="string"
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(x).strip().upper()
    for x in taxonomy[
        "complaint_category"
    ].dropna()
})

assert len(records) == 500

area_ids = (
    records["borough"]
    + "_"
    + records["incident_zip"]
)

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

# --------------------------------------------------
# 5. Regression test with all 500 complaints
# --------------------------------------------------

print("\n4. FULL BATCH REGRESSION")

with threadpool_limits(limits=1):
    full = analyze_fixed(
        records,
        categories,
        k=3,
        min_area_size=20
    )

print("Status:", full["status"])
print("Silhouette:", full["silhouette"])

check(
    "500 records retained",
    full["stats"]["total_records"] == 500
)

check(
    "20 areas eligible",
    full["stats"]["eligible_areas"] == 20
)

check(
    "K=3 clustering succeeds",
    full["status"] == "clustered"
)

check(
    "Original silhouette preserved",
    np.isclose(
        full["silhouette"],
        0.97664289820422,
        atol=0.00001
    )
)

check(
    "Expected cluster sizes preserved",
    sorted(
        full["cluster_profiles"][
            "areas"
        ].astype(int).tolist()
    ) == [5, 5, 10]
)

check(
    "Historical K-Means profiles retained",
    full["area_summary"][
        "historical_reporting_profile"
    ].notna().all()
)

# --------------------------------------------------
# 6. Test exactly K qualifying areas
# --------------------------------------------------

print("\n5. EXACTLY-K SAFEGUARD")

selected_areas = [
    reference.loc[
        reference[
            "intended_pattern"
        ].eq(pattern),
        "area_id"
    ].iloc[0]
    for pattern in [
        "road_a",
        "sewer",
        "signals"
    ]
]

three_area_data = records.loc[
    area_ids.isin(selected_areas)
].copy()

with threadpool_limits(limits=1):
    exactly_three = analyze_fixed(
        three_area_data,
        categories,
        k=3
    )

print("Status:", exactly_three["status"])
print("Reason:", exactly_three["reason"])

check(
    "Exactly K areas handled safely",
    exactly_three["status"]
    == "insufficient_areas_for_validation"
)

check(
    "No model fitted for exactly K areas",
    exactly_three["model"] is None
)

check(
    "Three areas retained in the summary",
    len(exactly_three["area_summary"]) == 3
)

# --------------------------------------------------
# 7. Test small-area complaint traceability
# --------------------------------------------------

print("\n6. SMALL-AREA SAFEGUARD")

first_area = area_ids.iloc[0]

first_indices = records.index[
    area_ids.eq(first_area)
]

small_data = records.drop(
    index=first_indices[3:]
).copy()

with threadpool_limits(limits=1):
    small_result = analyze_fixed(
        small_data,
        categories,
        k=3,
        min_area_size=20
    )

small_flagged = small_result[
    "record_eligibility"
].loc[
    lambda df: df[
        "eligibility_reason"
    ].eq("below_minimum_area_size")
]

display(small_flagged)

check(
    "Three small-area records flagged",
    len(small_flagged) == 3
)

check(
    "Small-area records retain correct IDs",
    set(
        small_flagged[
            "complaint_id"
        ].astype(str)
    ) == set(
        records.loc[
            first_indices[:3],
            "complaint_id"
        ].astype(str)
    )
)

check(
    "Remaining 19 areas cluster",
    small_result["status"] == "clustered"
    and small_result["stats"][
        "eligible_areas"
    ] == 19
)

check(
    "Small-area records not silently removed",
    len(
        small_result["record_eligibility"]
    ) == len(small_data)
)

# --------------------------------------------------
# 8. Verify no source modification yet
# --------------------------------------------------

check(
    "Existing module unchanged during testing",
    module_path.read_bytes() == original_bytes
)

# --------------------------------------------------
# 9. Report tests before saving
# --------------------------------------------------

results_18_1d_fix = pd.DataFrame(checks)

print("\n7. REPAIR VALIDATION")

display(results_18_1d_fix)

passed = int(
    results_18_1d_fix["passed"].sum()
)

total = len(results_18_1d_fix)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_1d_fix.loc[
    ~results_18_1d_fix["passed"],
    "test"
].tolist()

assert not failures, (
    "Repair validation failed: "
    + "; ".join(failures)
)

# --------------------------------------------------
# 10. Back up and save only after tests pass
# --------------------------------------------------

assert module_path.read_bytes() == original_bytes

if revised_source != original_source:

    backup_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    if backup_path.exists():
        assert backup_path.read_bytes() == (
            original_bytes
        ), (
            "Existing backup differs. "
            "No overwrite performed."
        )
    else:
        backup_path.write_bytes(
            original_bytes
        )

    temp_path = module_path.with_suffix(
        ".phase18_1d.tmp"
    )

    assert not temp_path.exists(), (
        "Temporary file already exists. "
        "No overwrite performed."
    )

    temp_path.write_bytes(
        revised_source.encode("utf-8")
    )

    assert temp_path.read_bytes() == (
        revised_source.encode("utf-8")
    )

    temp_path.replace(module_path)

# --------------------------------------------------
# 11. Reload the corrected module
# --------------------------------------------------

import src.models.new_batch_area_kmeans as km_module

importlib.invalidate_caches()
importlib.reload(km_module)

assert hasattr(
    km_module,
    "analyze_new_batch_areas"
)

# Confirm the saved import has the new behavior.
saved_check = km_module.analyze_new_batch_areas(
    three_area_data,
    categories,
    k=3
)

assert saved_check["status"] == (
    "insufficient_areas_for_validation"
)

print("\n8. VERIFIED REPAIR")
print("Module:", module_path)

if backup_path.exists():
    print("Backup:", backup_path)

print("\nSTATUS: PASSED")
print("PHASE 18.1D ROBUSTNESS REPAIR VERIFIED")
print("EXACTLY-K SILHOUETTE SAFEGUARD WORKING")
print("SMALL-AREA REASONS PRESERVED")
print("EXISTING HISTORICAL MODELS UNMODIFIED")
```

### now move to dbscan functionality

```python

from pathlib import Path
import hashlib
import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN

print(" NEW-BATCH DBSCAN")
print("SAME-CATEGORY SPATIAL CONCENTRATIONS")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

assert input_path.is_file()

original_hash = hashlib.sha256(
    input_path.read_bytes()
).hexdigest()

# --------------------------------------------------
# 1. Load 500 verified incoming complaints
# --------------------------------------------------

records = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
).reset_index(drop=True)

assert len(records) == 500
assert records["complaint_id"].is_unique
assert records["source_row"].is_unique

assert records["fully_verified"].str.lower().eq(
    "true"
).all()

assert records["processing_status"].eq(
    "processed"
).all()

assert records["zip_validation_status"].eq(
    "zcta_consistent"
).all()

records["lat_num"] = pd.to_numeric(
    records["latitude"],
    errors="coerce"
)

records["lon_num"] = pd.to_numeric(
    records["longitude"],
    errors="coerce"
)

assert np.isfinite(
    records[["lat_num", "lon_num"]]
    .to_numpy(dtype=float)
).all()

assert records[
    ["lat_num", "lon_num"]
].drop_duplicates().shape[0] == 500

print("\n1. DATASET INFORMATION")
print("Complaints:", len(records))
print("Boroughs:", records["borough"].nunique())
print("Categories:",
      records["complaint_category"].nunique())

print("\nHistorical DBSCAN associations:")
print(records["dbscan_status"].value_counts())

# --------------------------------------------------
# 2. Reusable experimental DBSCAN function
# --------------------------------------------------

EARTH_RADIUS_M = 6371008.8

def run_spatial_dbscan(
    frame,
    eps_m,
    min_samples,
    same_category=True
):
    """Identify candidate new-batch concentrations.

    IDs are local to this batch and configuration.
    They are not historical hotspot identifiers.
    """

    frame = frame.reset_index(drop=True).copy()

    candidate_ids = np.full(
        len(frame),
        None,
        dtype=object
    )

    if same_category:

        groups = [
            (category, group.index.to_numpy())
            for category, group in frame.groupby(
                "complaint_category",
                sort=True
            )
        ]

    else:

        groups = [(
            "ALL_CATEGORIES",
            frame.index.to_numpy()
        )]

    for category_number, (
        category, indices
    ) in enumerate(groups):

        if len(indices) < min_samples:
            continue

        coords = frame.loc[
            indices,
            ["lat_num", "lon_num"]
        ].to_numpy(dtype=float)

        coords_rad = np.radians(coords)

        model = DBSCAN(
            eps=eps_m / EARTH_RADIUS_M,
            min_samples=min_samples,
            metric="haversine",
            algorithm="ball_tree"
        )

        labels = model.fit_predict(
            coords_rad
        )

        for cluster_label in sorted(
            set(labels) - {-1}
        ):

            selected = indices[
                labels == cluster_label
            ]

            identifier = (
                f"NB-{category_number:02d}"
                f"-{int(cluster_label):03d}"
            )

            candidate_ids[selected] = identifier

    frame["new_batch_candidate_id"] = (
        candidate_ids
    )

    frame["new_batch_status"] = np.where(
        pd.notna(candidate_ids),
        "candidate_concentration",
        "unclustered"
    )

    return frame


# --------------------------------------------------
# 3. Compare experimental parameters
# --------------------------------------------------

print("\n2. PARAMETER SENSITIVITY AUDIT")

settings = [
    (75, 5),
    (100, 5),
    (125, 5),
    (150, 5),
    (125, 8),
    (150, 8)
]

experiment_rows = []

for eps_m, min_samples in settings:

    result = run_spatial_dbscan(
        records,
        eps_m=eps_m,
        min_samples=min_samples,
        same_category=True
    )

    counts = (
        result["new_batch_candidate_id"]
        .value_counts()
    )

    experiment_rows.append({
        "eps_m": eps_m,
        "min_samples": min_samples,
        "candidate_concentrations": len(counts),
        "clustered_complaints": int(
            result["new_batch_candidate_id"]
            .notna().sum()
        ),
        "unclustered_complaints": int(
            result["new_batch_candidate_id"]
            .isna().sum()
        ),
        "largest_candidate": (
            int(counts.max())
            if len(counts) else 0
        )
    })

parameter_audit_18_2a = pd.DataFrame(
    experiment_rows
)

display(parameter_audit_18_2a)

# --------------------------------------------------
# 4. Select an exploratory configuration
# --------------------------------------------------

# These are testing parameters, NOT validated
# operational thresholds.

EPS_M = 125
MIN_SAMPLES = 5

spatial_results_18_2a = run_spatial_dbscan(
    records,
    eps_m=EPS_M,
    min_samples=MIN_SAMPLES,
    same_category=True
)

print("\n3. SELECTED EXPERIMENT")
print("Radius:", EPS_M, "metres")
print("Minimum reports:", MIN_SAMPLES)

display(
    spatial_results_18_2a[
        "new_batch_status"
    ].value_counts().rename_axis(
        "status"
    ).reset_index(name="complaints")
)

# --------------------------------------------------
# 5. Compare historical and incoming results
# --------------------------------------------------

print("\n4. HISTORICAL VS NEW-BATCH DBSCAN")

comparison_18_2a = pd.crosstab(
    spatial_results_18_2a[
        "dbscan_status"
    ],
    spatial_results_18_2a[
        "new_batch_status"
    ]
)

display(comparison_18_2a)

# --------------------------------------------------
# 6. Summarize local candidate concentrations
# --------------------------------------------------

candidates = spatial_results_18_2a.loc[
    spatial_results_18_2a[
        "new_batch_status"
    ].eq("candidate_concentration")
].copy()

candidates["coordinate_key"] = (
    candidates["latitude"].astype(str)
    + "|"
    + candidates["longitude"].astype(str)
)

candidates["historical_established"] = (
    candidates["dbscan_status"].eq(
        "established_cluster"
    )
)

candidates["historical_outside"] = (
    candidates["dbscan_status"].eq(
        "outside_existing_clusters"
    )
)

candidates["historical_ambiguous"] = (
    candidates["dbscan_status"].eq(
        "ambiguous_association"
    )
)

if candidates.empty:

    candidate_profiles_18_2a = pd.DataFrame(
        columns=[
            "new_batch_candidate_id",
            "complaint_category",
            "incoming_complaints",
            "distinct_locations",
            "boroughs",
            "historical_established",
            "historical_outside",
            "historical_ambiguous",
            "center_latitude",
            "center_longitude"
        ]
    )

else:

    candidate_profiles_18_2a = (
        candidates.groupby(
            "new_batch_candidate_id"
        )
        .agg(
            complaint_category=(
                "complaint_category",
                "first"
            ),
            incoming_complaints=(
                "complaint_id",
                "size"
            ),
            distinct_locations=(
                "coordinate_key",
                "nunique"
            ),
            boroughs=(
                "borough",
                "nunique"
            ),
            historical_established=(
                "historical_established",
                "sum"
            ),
            historical_outside=(
                "historical_outside",
                "sum"
            ),
            historical_ambiguous=(
                "historical_ambiguous",
                "sum"
            ),
            center_latitude=(
                "lat_num",
                "mean"
            ),
            center_longitude=(
                "lon_num",
                "mean"
            )
        )
        .reset_index()
        .sort_values(
            [
                "incoming_complaints",
                "new_batch_candidate_id"
            ],
            ascending=[False, True]
        )
        .reset_index(drop=True)
    )

print("\n5. NEW-BATCH CANDIDATE SUMMARIES")

display(
    candidate_profiles_18_2a.head(25)
)

# --------------------------------------------------
# 7. Describe locations outside historical hotspots
# --------------------------------------------------

print("\n6. CANDIDATES OUTSIDE HISTORICAL HOTSPOTS")

outside_candidates = (
    candidate_profiles_18_2a.loc[
        candidate_profiles_18_2a[
            "historical_outside"
        ].gt(0)
    ]
)

display(
    outside_candidates[[
        "new_batch_candidate_id",
        "complaint_category",
        "incoming_complaints",
        "historical_outside",
        "historical_established",
        "center_latitude",
        "center_longitude"
    ]]
)

# --------------------------------------------------
# 8. Validate all outputs
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 complaints retained",
    len(spatial_results_18_2a) == 500
)

check(
    "All complaint IDs remain unique",
    spatial_results_18_2a[
        "complaint_id"
    ].is_unique
)

check(
    "Six parameter configurations tested",
    len(parameter_audit_18_2a) == 6
)

check(
    "Every configuration accounts for 500 inputs",
    (
        parameter_audit_18_2a[
            "clustered_complaints"
        ]
        + parameter_audit_18_2a[
            "unclustered_complaints"
        ]
    ).eq(500).all()
)

check(
    "No complaints disappear in selected DBSCAN",
    spatial_results_18_2a[
        "new_batch_status"
    ].value_counts().sum() == 500
)

check(
    "Candidate assignments are traceable",
    spatial_results_18_2a.loc[
        spatial_results_18_2a[
            "new_batch_status"
        ].eq("candidate_concentration"),
        "new_batch_candidate_id"
    ].notna().all()
)

check(
    "Unclustered records have no candidate ID",
    spatial_results_18_2a.loc[
        spatial_results_18_2a[
            "new_batch_status"
        ].eq("unclustered"),
        "new_batch_candidate_id"
    ].isna().all()
)

check(
    "All candidates meet minimum report count",
    candidate_profiles_18_2a[
        "incoming_complaints"
    ].ge(MIN_SAMPLES).all()
)

check(
    "No candidate combines different categories",
    (
        candidates.groupby(
            "new_batch_candidate_id"
        )["complaint_category"]
        .nunique()
        .le(1)
        .all()
    )
)

check(
    "Candidate counts reconcile with records",
    int(
        candidate_profiles_18_2a[
            "incoming_complaints"
        ].sum()
    ) == len(candidates)
)

check(
    "Historical established count preserved",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("established_cluster").sum() == 137
)

check(
    "Historical ambiguous count preserved",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("ambiguous_association").sum() == 9
)

check(
    "Historical outside-cluster count preserved",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("outside_existing_clusters").sum() == 354
)

check(
    "Comparison table accounts for all 500",
    int(comparison_18_2a.to_numpy().sum()) == 500
)

check(
    "Candidate location counts cannot exceed reports",
    (
        candidate_profiles_18_2a[
            "distinct_locations"
        ]
        <= candidate_profiles_18_2a[
            "incoming_complaints"
        ]
    ).all()
)

check(
    "Verified source CSV remains unchanged",
    hashlib.sha256(
        input_path.read_bytes()
    ).hexdigest() == original_hash
)

# --------------------------------------------------
# 9. Final verification
# --------------------------------------------------

results_18_2a = pd.DataFrame(checks)

print("\n7. PHASE 18.2A VALIDATION")

display(results_18_2a)

passed = int(
    results_18_2a["passed"].sum()
)

total = len(results_18_2a)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_2a.loc[
    ~results_18_2a["passed"],
    "test"
].tolist()

if failures:

    print("\nSTATUS: NEEDS REVIEW")

    for failure in failures:
        print("FAILED:", failure)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 18.2A NEW-BATCH DBSCAN VERIFIED")
    print("SAME-CATEGORY SPATIAL CONCENTRATIONS ANALYZED")
    print("HISTORICAL ASSOCIATIONS PRESERVED")
    print("ALL 500 COMPLAINTS ACCOUNTED FOR")
    print("NO MODEL ARTIFACTS MODIFIED")
    print("NOT YET DEPLOYED")
```

esults are encouraging: 15 of 16 checks passed, and the new-batch DBSCAN detected 25 raw spatial clusters containing 236 simulated complaints.

The failed test is:

All candidates meet minimum report count

Two clusters caused it:
Candidate	Category	Reports
NB-00-002	SEWER	4
NB-03-001	STREET CONDITION	4

Why did this happen?
In DBSCAN, min_samples=5 means a core point must have at least five points in its neighborhood, including itself. It does not strictly guarantee that every final cluster contains five members. Border-point assignments can result in a smaller cluster. 
scikit-learn 1.9.1 documentation+1


So the DBSCAN algorithm is not necessarily wrong. Our test incorrectly assumed that min_samples was also a minimum final cluster size.



This preserves the raw DBSCAN result while preventing four-report groups from being presented as five-report investigation candidates.

Under this adjustment, we expect:
Outcome	Reports
Reportable candidate concentrations	228
Below reporting threshold	8
Unclustered by raw DBSCAN	264
Total	500

These are still synthetic experimental results, not confirmed infrastructure incidents

```python

from pathlib import Path
import hashlib
import numpy as np
import pandas as pd

print(" DBSCAN REPORTING THRESHOLD REPAIR")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

assert input_path.is_file()

assert "run_spatial_dbscan" in globals(), (
    "Run the original Phase 18.2A cell first."
)

source_hash = hashlib.sha256(
    input_path.read_bytes()
).hexdigest()

# --------------------------------------------------
# 1. Load the verified synthetic complaints
# --------------------------------------------------

records = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

records["lat_num"] = pd.to_numeric(
    records["latitude"]
)

records["lon_num"] = pd.to_numeric(
    records["longitude"]
)

assert len(records) == 500
assert records["complaint_id"].is_unique
assert records["fully_verified"].str.lower().eq(
    "true"
).all()

# --------------------------------------------------
# 2. Preserve raw DBSCAN, filter reporting candidates
# --------------------------------------------------

def run_reportable_dbscan(
    frame,
    eps_m,
    min_samples,
    min_report_count=5
):
    """
    Preserve raw DBSCAN memberships.

    A separate minimum group size determines
    which clusters become reportable candidates.
    """

    if min_report_count < 2:
        raise ValueError(
            "Minimum reporting size must be >= 2."
        )

    result = run_spatial_dbscan(
        frame,
        eps_m=eps_m,
        min_samples=min_samples,
        same_category=True
    ).copy()

    # Keep original cluster assignments.
    result["raw_candidate_id"] = (
        result["new_batch_candidate_id"]
        .copy()
    )

    raw_counts = (
        result["raw_candidate_id"]
        .value_counts()
    )

    undersized_ids = set(
        raw_counts.loc[
            raw_counts < min_report_count
        ].index
    )

    undersized = result[
        "raw_candidate_id"
    ].isin(undersized_ids)

    # Suppress these from reported candidates
    # without calling them original DBSCAN noise.
    result.loc[
        undersized,
        "new_batch_candidate_id"
    ] = None

    result["new_batch_status"] = "unclustered"

    result.loc[
        undersized,
        "new_batch_status"
    ] = "below_reporting_threshold"

    result.loc[
        result["new_batch_candidate_id"]
        .notna(),
        "new_batch_status"
    ] = "candidate_concentration"

    return result


# --------------------------------------------------
# 3. Re-evaluate selected configuration
# --------------------------------------------------

EPS_M = 125
MIN_SAMPLES = 5
MIN_REPORT_COUNT = 5

spatial_results_18_2a = run_reportable_dbscan(
    records,
    eps_m=EPS_M,
    min_samples=MIN_SAMPLES,
    min_report_count=MIN_REPORT_COUNT
)

print("\n1. CORRECTED DBSCAN OUTCOMES")

status_counts = (
    spatial_results_18_2a[
        "new_batch_status"
    ].value_counts()
)

display(
    status_counts.rename_axis(
        "status"
    ).reset_index(name="complaints")
)

print("\n2. ORIGINAL VS REPORTABLE CLUSTERS")

raw_counts = (
    spatial_results_18_2a[
        "raw_candidate_id"
    ].value_counts()
)

reportable_counts = (
    spatial_results_18_2a[
        "new_batch_candidate_id"
    ].value_counts()
)

print("Raw DBSCAN clusters:", len(raw_counts))
print("Reportable candidates:", len(reportable_counts))
print(
    "Clusters below reporting threshold:",
    int((raw_counts < MIN_REPORT_COUNT).sum())
)

print(
    "Reports below reporting threshold:",
    int(raw_counts[
        raw_counts < MIN_REPORT_COUNT
    ].sum())
)

# --------------------------------------------------
# 4. Inspect undersized raw DBSCAN clusters
# --------------------------------------------------

below_threshold = (
    spatial_results_18_2a.loc[
        spatial_results_18_2a[
            "new_batch_status"
        ].eq("below_reporting_threshold")
    ].copy()
)

print("\n3. SUPPRESSED REPORTING CANDIDATES")

display(
    below_threshold.groupby([
        "raw_candidate_id",
        "complaint_category"
    ])
    .size()
    .rename("reports")
    .reset_index()
)

# --------------------------------------------------
# 5. Rebuild reportable candidate summaries
# --------------------------------------------------

reportable = spatial_results_18_2a.loc[
    spatial_results_18_2a[
        "new_batch_status"
    ].eq("candidate_concentration")
].copy()

reportable["location_key"] = (
    reportable["latitude"].astype(str)
    + "|"
    + reportable["longitude"].astype(str)
)

reportable["historical_outside"] = (
    reportable["dbscan_status"].eq(
        "outside_existing_clusters"
    )
)

reportable["historical_established"] = (
    reportable["dbscan_status"].eq(
        "established_cluster"
    )
)

reportable["historical_ambiguous"] = (
    reportable["dbscan_status"].eq(
        "ambiguous_association"
    )
)

candidate_profiles_18_2a = (
    reportable.groupby(
        "new_batch_candidate_id"
    )
    .agg(
        complaint_category=(
            "complaint_category", "first"
        ),
        incoming_complaints=(
            "complaint_id", "size"
        ),
        distinct_locations=(
            "location_key", "nunique"
        ),
        historical_outside=(
            "historical_outside", "sum"
        ),
        historical_established=(
            "historical_established", "sum"
        ),
        historical_ambiguous=(
            "historical_ambiguous", "sum"
        ),
        center_latitude=(
            "lat_num", "mean"
        ),
        center_longitude=(
            "lon_num", "mean"
        )
    )
    .reset_index()
    .sort_values(
        [
            "incoming_complaints",
            "new_batch_candidate_id"
        ],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

candidate_profiles_18_2a[
    "entirely_outside_historical_clusters"
] = (
    candidate_profiles_18_2a[
        "historical_outside"
    ].eq(
        candidate_profiles_18_2a[
            "incoming_complaints"
        ]
    )
)

print("\n4. REPORTABLE CONCENTRATIONS")

display(candidate_profiles_18_2a.head(25))

# --------------------------------------------------
# 6. Preserve historical DBSCAN comparison
# --------------------------------------------------

print("\n5. HISTORICAL VS NEW DBSCAN")

comparison_18_2a = pd.crosstab(
    spatial_results_18_2a[
        "dbscan_status"
    ],
    spatial_results_18_2a[
        "new_batch_status"
    ]
)

display(comparison_18_2a)

# --------------------------------------------------
# 7. Re-evaluate all parameter settings
# --------------------------------------------------

print("\n6. CORRECTED PARAMETER COMPARISON")

settings = [
    (75, 5),
    (100, 5),
    (125, 5),
    (150, 5),
    (125, 8),
    (150, 8)
]

parameter_rows = []

for eps_m, min_samples in settings:

    experiment = run_reportable_dbscan(
        records,
        eps_m=eps_m,
        min_samples=min_samples,
        min_report_count=min_samples
    )

    statuses = experiment[
        "new_batch_status"
    ]

    parameter_rows.append({
        "eps_m": eps_m,
        "min_samples": min_samples,
        "reporting_minimum": min_samples,
        "raw_clusters": experiment[
            "raw_candidate_id"
        ].nunique(),
        "reportable_clusters": experiment[
            "new_batch_candidate_id"
        ].nunique(),
        "reportable_complaints": int(
            statuses.eq(
                "candidate_concentration"
            ).sum()
        ),
        "below_reporting_threshold": int(
            statuses.eq(
                "below_reporting_threshold"
            ).sum()
        ),
        "raw_unclustered": int(
            statuses.eq("unclustered").sum()
        )
    })

parameter_audit_18_2a = pd.DataFrame(
    parameter_rows
)

display(parameter_audit_18_2a)

# --------------------------------------------------
# 8. Verify corrected results
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 records retained",
    len(spatial_results_18_2a) == 500
)

check(
    "Original complaint IDs preserved",
    spatial_results_18_2a[
        "complaint_id"
    ].tolist() == records[
        "complaint_id"
    ].tolist()
)

check(
    "25 raw DBSCAN clusters preserved",
    len(raw_counts) == 25
)

check(
    "Two undersized clusters identified",
    (raw_counts < 5).sum() == 2
)

check(
    "Eight reports flagged below threshold",
    len(below_threshold) == 8
)

check(
    "23 reportable candidates retained",
    len(candidate_profiles_18_2a) == 23
)

check(
    "228 complaints in reportable candidates",
    len(reportable) == 228
)

check(
    "264 original DBSCAN unclustered reports",
    status_counts.get("unclustered", 0) == 264
)

check(
    "Every reportable candidate has >= 5 reports",
    candidate_profiles_18_2a[
        "incoming_complaints"
    ].ge(5).all()
)

check(
    "Suppressed clusters retain raw IDs",
    below_threshold[
        "raw_candidate_id"
    ].notna().all()
)

check(
    "Suppressed clusters have no reportable IDs",
    below_threshold[
        "new_batch_candidate_id"
    ].isna().all()
)

check(
    "All reportable groups contain one category",
    reportable.groupby(
        "new_batch_candidate_id"
    )["complaint_category"].nunique().eq(1).all()
)

check(
    "All reporting statuses account for 500",
    int(status_counts.sum()) == 500
)

check(
    "137 historical established associations retained",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("established_cluster").sum() == 137
)

check(
    "Nine historical ambiguous associations retained",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("ambiguous_association").sum() == 9
)

check(
    "354 historical outside-cluster reports retained",
    spatial_results_18_2a[
        "dbscan_status"
    ].eq("outside_existing_clusters").sum() == 354
)

check(
    "All six parameter experiments reconciled",
    len(parameter_audit_18_2a) == 6
    and (
        parameter_audit_18_2a[
            "reportable_complaints"
        ]
        + parameter_audit_18_2a[
            "below_reporting_threshold"
        ]
        + parameter_audit_18_2a[
            "raw_unclustered"
        ]
    ).eq(500).all()
)

check(
    "Original verified input remains unchanged",
    hashlib.sha256(
        input_path.read_bytes()
    ).hexdigest() == source_hash
)

# --------------------------------------------------
# 9. Final results
# --------------------------------------------------

results_18_2a_fix = pd.DataFrame(checks)

print("\n7. PHASE 18.2A.1 VALIDATION")

display(results_18_2a_fix)

passed = int(
    results_18_2a_fix["passed"].sum()
)

total = len(results_18_2a_fix)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_2a_fix.loc[
    ~results_18_2a_fix["passed"],
    "test"
].tolist()

if failures:
    print("\nSTATUS: NEEDS REVIEW")

    for failure in failures:
        print("FAILED:", failure)

else:
    print("\nSTATUS: PASSED")
    print("PHASE 18.2A DBSCAN REPORTING FIX VERIFIED")
    print("RAW DBSCAN CLUSTERS PRESERVED")
    print("MINIMUM REPORTABLE SIZE ENFORCED")
    print("HISTORICAL ASSOCIATIONS PRESERVED")
    print("NO MODEL FILES MODIFIED")
```

### REUSABLE NEW-BATCH DBSCAN ENGINE

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import sys
import hashlib
import numpy as np
import pandas as pd

print(" REUSABLE NEW-BATCH DBSCAN ENGINE")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/models/new_batch_dbscan.py"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

assert input_path.is_file()
assert taxonomy_path.is_file()

# --------------------------------------------------
# 1. Create reusable new-batch DBSCAN engine
# --------------------------------------------------

module_source = dedent(r'''
"""New-batch same-category spatial concentration analysis.

This engine never retrains historical DBSCAN.
Candidates are not confirmed infrastructure defects.

Candidate IDs are local to the batch and parameters.
"""

import numpy as np
import pandas as pd
from sklearn.cluster import DBSCAN

EARTH_RADIUS_M = 6371008.8

BOROUGHS = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}

REQUIRED = {
    "source_row",
    "complaint_id",
    "latitude",
    "longitude",
    "borough",
    "complaint_category",
    "processing_status",
    "dbscan_status"
}


def analyze_new_batch_spatial(
    records,
    supported_categories,
    eps_m=125,
    min_samples=5,
    min_report_count=5,
    min_distinct_locations=2
):
    """Analyze already processed geographic complaint records.

    ZIP consistency is not required for spatial
    analysis when geographic processing has succeeded.

    Time-window selection must be done upstream.
    """

    if not isinstance(records, pd.DataFrame):
        raise TypeError("Expected pandas DataFrame.")

    if records.columns.duplicated().any():
        raise ValueError("Duplicate input columns.")

    missing = REQUIRED - set(records.columns)

    if missing:
        raise ValueError(
            f"Missing columns: {sorted(missing)}"
        )

    if records["source_row"].duplicated().any():
        raise ValueError("Duplicate source rows.")

    if (
        isinstance(eps_m, bool)
        or not isinstance(eps_m, (int, float))
        or not np.isfinite(eps_m)
        or eps_m <= 0
    ):
        raise ValueError(
            "eps_m must be finite and positive."
        )

    parameters = [
        ("min_samples", min_samples, 2),
        ("min_report_count", min_report_count, 2),
        (
            "min_distinct_locations",
            min_distinct_locations,
            1
        )
    ]

    for name, value, minimum in parameters:
        if (
            isinstance(value, bool)
            or not isinstance(value, int)
            or value < minimum
        ):
            raise ValueError(
                f"{name} must be an integer >= {minimum}."
            )

    categories = {
        str(category).strip().upper()
        for category in supported_categories
        if pd.notna(category)
        and str(category).strip()
    }

    if not categories:
        raise ValueError(
            "Supported categories are empty."
        )

    # ----------------------------------------------
    # 2. Normalize geographic inputs
    # ----------------------------------------------

    work = records.copy().reset_index(drop=True)

    for column in [
        "borough",
        "complaint_category",
        "processing_status",
        "dbscan_status"
    ]:
        work[column] = (
            work[column]
            .fillna("")
            .astype(str)
            .str.strip()
        )

    work["borough"] = work[
        "borough"
    ].str.upper()

    work["complaint_category"] = work[
        "complaint_category"
    ].str.upper()

    work["lat_num"] = pd.to_numeric(
        work["latitude"],
        errors="coerce"
    )

    work["lon_num"] = pd.to_numeric(
        work["longitude"],
        errors="coerce"
    )

    lat = work["lat_num"].to_numpy(
        dtype=float,
        na_value=np.nan
    )

    lon = work["lon_num"].to_numpy(
        dtype=float,
        na_value=np.nan
    )

    invalid_coords = (
        ~np.isfinite(lat)
        | ~np.isfinite(lon)
        | (lat < -90)
        | (lat > 90)
        | (lon < -180)
        | (lon > 180)
    )

    # ----------------------------------------------
    # 3. Preserve all excluded records with reasons
    # ----------------------------------------------

    work["record_eligibility"] = "eligible"

    rules = [
        (
            work["processing_status"].ne(
                "processed"
            ),
            "not_processed"
        ),
        (
            ~work["borough"].isin(BOROUGHS),
            "invalid_borough"
        ),
        (
            pd.Series(invalid_coords),
            "invalid_coordinates"
        ),
        (
            ~work["complaint_category"].isin(
                categories
            ),
            "unsupported_category"
        )
    ]

    for mask, reason in rules:
        work.loc[
            mask
            & work["record_eligibility"].eq(
                "eligible"
            ),
            "record_eligibility"
        ] = reason

    work["raw_candidate_id"] = pd.Series(
        [None] * len(work),
        dtype="object"
    )

    work["new_batch_candidate_id"] = pd.Series(
        [None] * len(work),
        dtype="object"
    )

    work["new_batch_status"] = "excluded_input"

    eligible = work[
        "record_eligibility"
    ].eq("eligible")

    work.loc[
        eligible, "new_batch_status"
    ] = "unclustered"

    # ----------------------------------------------
    # 4. Run DBSCAN separately for each category
    # ----------------------------------------------

    groups = (
        work.loc[eligible]
        .groupby(
            "complaint_category",
            sort=True
        ).groups
    )

    for category_index, (
        category, group_indices
    ) in enumerate(groups.items()):

        indices = np.asarray(
            list(group_indices),
            dtype=int
        )

        if len(indices) < min_samples:
            continue

        coordinates = work.loc[
            indices,
            ["lat_num", "lon_num"]
        ].to_numpy(dtype=float)

        model = DBSCAN(
            eps=float(eps_m) / EARTH_RADIUS_M,
            min_samples=min_samples,
            metric="haversine",
            algorithm="ball_tree"
        )

        labels = model.fit_predict(
            np.radians(coordinates)
        )

        for label in sorted(
            set(labels) - {-1}
        ):

            members = indices[
                labels == label
            ]

            candidate_id = (
                f"NB-{category_index:02d}"
                f"-{int(label):03d}"
            )

            work.loc[
                members,
                "raw_candidate_id"
            ] = candidate_id

    # ----------------------------------------------
    # 5. Apply independent reporting safeguards
    # ----------------------------------------------

    raw_mask = work[
        "raw_candidate_id"
    ].notna()

    raw = work.loc[raw_mask].copy()

    if len(raw):

        raw["location_pair"] = list(zip(
            raw["lat_num"],
            raw["lon_num"]
        ))

        raw_groups = (
            raw.groupby("raw_candidate_id")
            .agg(
                report_count=(
                    "source_row", "size"
                ),
                distinct_locations=(
                    "location_pair", "nunique"
                )
            )
        )

        reportable_ids = set(
            raw_groups.index[
                (
                    raw_groups["report_count"]
                    >= min_report_count
                )
                & (
                    raw_groups["distinct_locations"]
                    >= min_distinct_locations
                )
            ]
        )

    else:
        reportable_ids = set()

    # Raw clusters remain traceable even when
    # they do not meet reporting requirements.

    work.loc[
        raw_mask,
        "new_batch_status"
    ] = "below_reporting_threshold"

    reportable_mask = (
        raw_mask
        & work["raw_candidate_id"].isin(
            reportable_ids
        )
    )

    work.loc[
        reportable_mask,
        "new_batch_status"
    ] = "candidate_concentration"

    work.loc[
        reportable_mask,
        "new_batch_candidate_id"
    ] = work.loc[
        reportable_mask,
        "raw_candidate_id"
    ]

    # ----------------------------------------------
    # 6. Summarize reportable concentrations
    # ----------------------------------------------

    candidate = work.loc[
        reportable_mask
    ].copy()

    profile_columns = [
        "new_batch_candidate_id",
        "complaint_category",
        "incoming_complaints",
        "distinct_locations",
        "boroughs",
        "historical_established",
        "historical_outside",
        "historical_ambiguous",
        "center_latitude",
        "center_longitude",
        "entirely_outside_historical_clusters"
    ]

    if candidate.empty:
        profiles = pd.DataFrame(
            columns=profile_columns
        )

    else:

        candidate["location_pair"] = list(zip(
            candidate["lat_num"],
            candidate["lon_num"]
        ))

        for column, status in [
            (
                "historical_established",
                "established_cluster"
            ),
            (
                "historical_outside",
                "outside_existing_clusters"
            ),
            (
                "historical_ambiguous",
                "ambiguous_association"
            )
        ]:
            candidate[column] = (
                candidate["dbscan_status"].eq(
                    status
                )
            )

        profiles = (
            candidate.groupby(
                "new_batch_candidate_id",
                sort=True
            )
            .agg(
                complaint_category=(
                    "complaint_category",
                    "first"
                ),
                incoming_complaints=(
                    "source_row", "size"
                ),
                distinct_locations=(
                    "location_pair", "nunique"
                ),
                boroughs=(
                    "borough", "nunique"
                ),
                historical_established=(
                    "historical_established", "sum"
                ),
                historical_outside=(
                    "historical_outside", "sum"
                ),
                historical_ambiguous=(
                    "historical_ambiguous", "sum"
                ),
                center_latitude=(
                    "lat_num", "mean"
                ),
                center_longitude=(
                    "lon_num", "mean"
                )
            )
            .reset_index()
        )

        profiles[
            "entirely_outside_historical_clusters"
        ] = profiles[
            "historical_outside"
        ].eq(
            profiles["incoming_complaints"]
        )

        profiles = profiles.sort_values(
            [
                "incoming_complaints",
                "new_batch_candidate_id"
            ],
            ascending=[False, True]
        ).reset_index(drop=True)

    # ----------------------------------------------
    # 7. Account for every original record
    # ----------------------------------------------

    stats = {
        "total_records": len(work),
        "eligible_records": int(
            eligible.sum()
        ),
        "excluded_records": int(
            (~eligible).sum()
        ),
        "raw_clusters": int(
            work["raw_candidate_id"].nunique()
        ),
        "reportable_candidates": len(profiles),
        "candidate_complaints": int(
            work["new_batch_status"].eq(
                "candidate_concentration"
            ).sum()
        ),
        "below_reporting_threshold": int(
            work["new_batch_status"].eq(
                "below_reporting_threshold"
            ).sum()
        ),
        "unclustered": int(
            work["new_batch_status"].eq(
                "unclustered"
            ).sum()
        )
    }

    total_outcomes = sum([
        stats["candidate_complaints"],
        stats["below_reporting_threshold"],
        stats["unclustered"],
        stats["excluded_records"]
    ])

    if total_outcomes != len(work):
        raise AssertionError(
            "Spatial results did not reconcile."
        )

    return {
        "records": work,
        "candidate_profiles": profiles,
        "stats": stats,
        "parameters": {
            "eps_m": eps_m,
            "min_samples": min_samples,
            "min_report_count": min_report_count,
            "min_distinct_locations":
                min_distinct_locations
        }
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

analyze_new_batch_spatial = namespace[
    "analyze_new_batch_spatial"
]

# --------------------------------------------------
# 2. Load verified synthetic test inputs
# --------------------------------------------------

records = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

original_records = records.copy(deep=True)

original_hash = hashlib.sha256(
    input_path.read_bytes()
).hexdigest()

# --------------------------------------------------
# 3. Run complete 500-complaint experiment
# --------------------------------------------------

print("\n1. FULL BATCH TEST")

with threadpool_limits(limits=1):

    spatial_engine_output_18_2b = (
        analyze_new_batch_spatial(
            records,
            categories,
            eps_m=125,
            min_samples=5,
            min_report_count=5,
            min_distinct_locations=2
        )
    )

spatial_records_18_2b = (
    spatial_engine_output_18_2b["records"]
)

spatial_profiles_18_2b = (
    spatial_engine_output_18_2b[
        "candidate_profiles"
    ]
)

stats = spatial_engine_output_18_2b["stats"]

print("\nENGINE SUMMARY")

for key, value in stats.items():
    print(f"{key}: {value}")

print("\n2. REPORTABLE CANDIDATES")

display(spatial_profiles_18_2b.head(25))

# --------------------------------------------------
# 4. Test safeguards and output integrity
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "500 records retained",
    stats["total_records"] == 500
)

check(
    "All 500 geographic inputs eligible",
    stats["eligible_records"] == 500
)

check(
    "No records unexpectedly excluded",
    stats["excluded_records"] == 0
)

check(
    "25 raw DBSCAN clusters retained",
    stats["raw_clusters"] == 25
)

check(
    "23 reportable candidates retained",
    stats["reportable_candidates"] == 23
)

check(
    "228 complaints in reportable candidates",
    stats["candidate_complaints"] == 228
)

check(
    "Eight reports below threshold",
    stats["below_reporting_threshold"] == 8
)

check(
    "264 complaints remain unclustered",
    stats["unclustered"] == 264
)

check(
    "Every candidate has at least five reports",
    spatial_profiles_18_2b[
        "incoming_complaints"
    ].ge(5).all()
)

check(
    "Every reported candidate has distinct locations",
    spatial_profiles_18_2b[
        "distinct_locations"
    ].ge(2).all()
)

check(
    "Every candidate belongs to one complaint category",
    spatial_records_18_2b.loc[
        lambda df: df[
            "new_batch_candidate_id"
        ].notna()
    ].groupby(
        "new_batch_candidate_id"
    )["complaint_category"].nunique().eq(1).all()
)

check(
    "All 500 statuses reconciled",
    (
        stats["candidate_complaints"]
        + stats["below_reporting_threshold"]
        + stats["unclustered"]
        + stats["excluded_records"]
    ) == 500
)

# --------------------------------------------------
# 5. Test repeated-coordinate suppression
# --------------------------------------------------

print("\n3. REPEATED-COORDINATE SAFEGUARD")

same_location = records.iloc[:6].copy()

same_location["latitude"] = (
    records.iloc[0]["latitude"]
)

same_location["longitude"] = (
    records.iloc[0]["longitude"]
)

same_location["complaint_category"] = (
    "STREET CONDITION"
)

with threadpool_limits(limits=1):

    repeated_result = analyze_new_batch_spatial(
        same_location,
        categories
    )

print("Repeated-location stats:", repeated_result["stats"])

check(
    "Identical coordinates can form a raw cluster",
    repeated_result["stats"]["raw_clusters"] == 1
)

check(
    "Repeated coordinates alone are not reported",
    repeated_result["stats"][
        "reportable_candidates"
    ] == 0
    and repeated_result["stats"][
        "below_reporting_threshold"
    ] == 6
)

# --------------------------------------------------
# 6. Verify multiple distinct nearby coordinates
# --------------------------------------------------

distinct_nearby = same_location.copy()

base_lat = float(
    same_location.iloc[0]["latitude"]
)

distinct_nearby["latitude"] = [
    base_lat + i * 0.00001
    for i in range(6)
]

with threadpool_limits(limits=1):

    distinct_result = analyze_new_batch_spatial(
        distinct_nearby,
        categories
    )

check(
    "Six nearby distinct locations are reportable",
    distinct_result["stats"][
        "reportable_candidates"
    ] == 1
    and distinct_result["stats"][
        "candidate_complaints"
    ] == 6
)

# --------------------------------------------------
# 7. Geographic review exclusion test
# --------------------------------------------------

geographic_review = records.iloc[:6].copy()

geographic_review[
    "processing_status"
] = "geographic_review_required"

excluded_result = analyze_new_batch_spatial(
    geographic_review,
    categories
)

check(
    "Geographic review records excluded safely",
    excluded_result["stats"][
        "excluded_records"
    ] == 6
    and excluded_result["stats"][
        "reportable_candidates"
    ] == 0
)

check(
    "Excluded records remain individually traceable",
    excluded_result["records"][
        "record_eligibility"
    ].eq("not_processed").all()
)

# --------------------------------------------------
# 8. Missing or mismatched ZIP does not block
#    otherwise valid geographic spatial analysis
# --------------------------------------------------

zip_test = records.copy()

zip_test.loc[
    0, "zip_validation_status"
] = "zcta_mismatch"

with threadpool_limits(limits=1):

    zip_result = analyze_new_batch_spatial(
        zip_test,
        categories
    )

check(
    "ZIP mismatch alone does not suppress valid coordinates",
    zip_result["stats"][
        "eligible_records"
    ] == 500
)

# --------------------------------------------------
# 9. Invalid coordinates handled without crashing
# --------------------------------------------------

bad_coordinates = records.iloc[:2].copy()

bad_coordinates.loc[
    bad_coordinates.index[0],
    "latitude"
] = "not-a-coordinate"

bad_result = analyze_new_batch_spatial(
    bad_coordinates,
    categories
)

check(
    "Invalid coordinates isolated with a reason",
    bad_result["records"].iloc[0][
        "record_eligibility"
    ] == "invalid_coordinates"
)

# --------------------------------------------------
# 10. Empty input should not crash
# --------------------------------------------------

empty_result = analyze_new_batch_spatial(
    records.iloc[:0],
    categories
)

check(
    "Empty input produces safe empty results",
    empty_result["stats"][
        "total_records"
    ] == 0
    and empty_result["candidate_profiles"].empty
)

# --------------------------------------------------
# 11. Verify original data and models untouched
# --------------------------------------------------

check(
    "Input DataFrame remains unchanged",
    records.equals(original_records)
)

check(
    "Verified CSV remains unchanged",
    hashlib.sha256(
        input_path.read_bytes()
    ).hexdigest() == original_hash
)

# --------------------------------------------------
# 12. Display validation results
# --------------------------------------------------

results_18_2b = pd.DataFrame(checks)

print("\n4. PHASE 18.2B VALIDATION")

display(results_18_2b)

passed = int(results_18_2b["passed"].sum())
total = len(results_18_2b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_2b.loc[
    ~results_18_2b["passed"],
    "test"
].tolist()

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No module saved.")

    for name in failures:
        print("FAILED:", name)

else:

    # ----------------------------------------------
    # 13. Save new module without overwriting work
    # ----------------------------------------------

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == module_source, (
            "Existing new-batch DBSCAN module "
            "differs. No overwrite performed."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        module_path.write_text(
            module_source,
            encoding="utf-8"
        )

    print("\n5. MODULE CREATED / VERIFIED")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 18.2B REUSABLE DBSCAN ENGINE VERIFIED")
    print("MINIMUM REPORTING SAFEGUARDS VERIFIED")
    print("REPEATED-COORDINATE SAFEGUARDS VERIFIED")
    print("HISTORICAL DBSCAN PRESERVED")
    print("NO AUTOMATIC REPAIR PRIORITIES")
```

### Explain Concentrations and Prepare the Map

```python

from pathlib import Path
from threadpoolctl import threadpool_limits
import hashlib
import json
import sys
import numpy as np
import pandas as pd

print(" SPATIAL EXPLANATIONS AND MAP PREPARATION")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.models.new_batch_dbscan import (
    analyze_new_batch_spatial
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

module_path = (
    root / "src/models/new_batch_dbscan.py"
)

output_dir = (
    root / "outputs/tests"
    / "phase_18_2c_spatial_explanations"
)

briefs_path = output_dir / "candidate_briefs.json"
map_path = output_dir / "candidate_map_points.geojson"

for path in [input_path, taxonomy_path, module_path]:
    assert path.is_file(), f"Missing: {path}"

source_hash = hashlib.sha256(
    input_path.read_bytes()
).hexdigest()

module_hash = hashlib.sha256(
    module_path.read_bytes()
).hexdigest()

# --------------------------------------------------
# 1. Reload and analyze the same verified batch
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

assert len(source) == 500
assert source["fully_verified"].str.lower().eq(
    "true"
).all()

with threadpool_limits(limits=1):
    output = analyze_new_batch_spatial(
        source,
        categories,
        eps_m=125,
        min_samples=5,
        min_report_count=5,
        min_distinct_locations=2
    )

records = output["records"]
profiles = output["candidate_profiles"]
stats = output["stats"]
parameters = output["parameters"]

print("\n1. ENGINE RECONCILIATION")
print("Total complaints:", stats["total_records"])
print("Raw clusters:", stats["raw_clusters"])
print(
    "Reportable candidates:",
    stats["reportable_candidates"]
)
print(
    "Complaints in candidates:",
    stats["candidate_complaints"]
)

# --------------------------------------------------
# 2. Distance calculation for explanatory locations
# --------------------------------------------------

EARTH_RADIUS_M = 6371008.8

def distance_metres(lat1, lon1, lat2, lon2):
    """Vectorized Haversine distance in metres."""

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlat = lat2 - lat1
    dlon = lon2 - lon1

    a = (
        np.sin(dlat / 2) ** 2
        + np.cos(lat1)
        * np.cos(lat2)
        * np.sin(dlon / 2) ** 2
    )

    a = np.clip(a, 0.0, 1.0)

    return 2 * EARTH_RADIUS_M * np.arcsin(
        np.sqrt(a)
    )

# --------------------------------------------------
# 3. Category-specific investigation suggestions
# --------------------------------------------------

guidance = {
    "SEWER": (
        "Review sewer complaint descriptions, "
        "location accuracy and relevant "
        "maintenance history."
    ),
    "STREET CONDITION": (
        "Review street-condition descriptions "
        "to determine whether reports concern "
        "the same roadway issue."
    ),
    "STREET LIGHT CONDITION": (
        "Review lighting complaints and check "
        "whether reports refer to the same "
        "lighting assets or nearby locations."
    ),
    "TRAFFIC SIGNAL CONDITION": (
        "Review traffic-signal complaints and "
        "verify the reported intersections "
        "and equipment."
    )
}

# --------------------------------------------------
# 4. Build explanations and map-ready features
# --------------------------------------------------

reported = records.loc[
    records["new_batch_status"].eq(
        "candidate_concentration"
    )
].copy()

brief_rows = []

geojson = {
    "type": "FeatureCollection",
    "name": "synthetic_new_batch_dbscan_candidates",
    "metadata": {
        "synthetic_data": True,
        "eps_m": float(parameters["eps_m"]),
        "min_samples": int(parameters["min_samples"]),
        "min_report_count": int(
            parameters["min_report_count"]
        ),
        "coordinate_note": (
            "Each point is one submitted complaint "
            "coordinate nearest its candidate's "
            "mean location. It is not a verified "
            "infrastructure asset or defect."
        )
    },
    "features": []
}

for profile in profiles.to_dict(orient="records"):

    candidate_id = str(
        profile["new_batch_candidate_id"]
    )

    group = reported.loc[
        reported["new_batch_candidate_id"].eq(
            candidate_id
        )
    ].copy()

    count = len(group)

    if count != int(profile["incoming_complaints"]):
        raise ValueError(
            f"Count mismatch: {candidate_id}"
        )

    category_values = group[
        "complaint_category"
    ].unique()

    if len(category_values) != 1:
        raise ValueError(
            f"Mixed categories: {candidate_id}"
        )

    category = str(category_values[0])

    coords = group[
        ["lat_num", "lon_num"]
    ].to_numpy(dtype=float)

    center_lat = float(coords[:, 0].mean())
    center_lon = float(coords[:, 1].mean())

    # Find an actual submitted coordinate closest
    # to the average, rather than inventing a
    # street address or asset location.
    distances = distance_metres(
        center_lat,
        center_lon,
        coords[:, 0],
        coords[:, 1]
    )

    representative_index = int(
        np.argmin(distances)
    )

    representative_lat = float(
        coords[representative_index, 0]
    )

    representative_lon = float(
        coords[representative_index, 1]
    )

    max_distance = float(np.max(
        distance_metres(
            representative_lat,
            representative_lon,
            coords[:, 0],
            coords[:, 1]
        )
    ))

    distinct_locations = len(
        np.unique(coords, axis=0)
    )

    boroughs = sorted(
        group["borough"].astype(str).unique()
    )

    zip_areas = sorted({
        value
        for value in group[
            "incident_zip"
        ].astype(str)
        if len(value) == 5 and value.isdigit()
    })

    historical_counts = (
        group["dbscan_status"]
        .value_counts()
    )

    established = int(
        historical_counts.get(
            "established_cluster", 0
        )
    )

    outside = int(
        historical_counts.get(
            "outside_existing_clusters", 0
        )
    )

    ambiguous = int(
        historical_counts.get(
            "ambiguous_association", 0
        )
    )

    unclassified = (
        count - established - outside - ambiguous
    )

    if outside == count:
        relationship = (
            "entirely_outside_historical_clusters"
        )
        relationship_text = (
            "All incoming reports are outside "
            "the saved historical cluster associations."
        )

    elif established == count:
        relationship = (
            "entirely_associated_with_historical_clusters"
        )
        relationship_text = (
            "All incoming reports are associated "
            "with established historical clusters."
        )

    else:
        relationship = (
            "mixed_or_uncertain_historical_associations"
        )
        relationship_text = (
            f"Historical association results: "
            f"{established} established, "
            f"{outside} outside, "
            f"{ambiguous} ambiguous, "
            f"{unclassified} other."
        )

    explanation = (
        f"{count} incoming {category} complaints "
        f"at {distinct_locations} distinct submitted "
        f"coordinates form a candidate spatial "
        f"concentration in this synthetic batch. "
        f"{relationship_text} "
        f"The representative map point "
        f"({representative_lat:.6f}, "
        f"{representative_lon:.6f}) "
        f"is a submitted complaint coordinate, "
        f"not a confirmed defect location."
    )

    action = (
        guidance.get(
            category,
            "Review the complaint descriptions "
            "and their reported locations."
        )
        + " Check for duplicate reports and "
        + "confirm the situation before considering "
        + "a field investigation."
    )

    brief = {
        "candidate_id": candidate_id,
        "complaint_category": category,
        "incoming_complaints": int(count),
        "distinct_locations": int(
            distinct_locations
        ),
        "boroughs": boroughs,
        "incident_zip_areas": zip_areas,
        "representative_latitude": round(
            representative_lat, 7
        ),
        "representative_longitude": round(
            representative_lon, 7
        ),
        "mean_latitude": round(center_lat, 7),
        "mean_longitude": round(center_lon, 7),
        "max_distance_from_representative_m":
            round(max_distance, 1),
        "historical_established": established,
        "historical_outside": outside,
        "historical_ambiguous": ambiguous,
        "historical_unclassified": unclassified,
        "historical_relationship": relationship,
        "explanation": explanation,
        "suggested_review": action,
        "synthetic_data": True
    }

    brief_rows.append(brief)

    geojson["features"].append({
        "type": "Feature",
        "id": candidate_id,
        "geometry": {
            "type": "Point",
            "coordinates": [
                representative_lon,
                representative_lat
            ]
        },
        "properties": {
            "candidate_id": candidate_id,
            "complaint_category": category,
            "incoming_complaints": int(count),
            "distinct_locations": int(
                distinct_locations
            ),
            "historical_relationship":
                relationship,
            "historical_outside": outside,
            "historical_established": established,
            "historical_ambiguous": ambiguous,
            "synthetic_data": True
        }
    })

briefs_18_2c = pd.DataFrame(brief_rows)

# --------------------------------------------------
# 5. Preview spatial intelligence explanations
# --------------------------------------------------

print("\n2. MAP-READY CANDIDATE SUMMARY")

display(
    briefs_18_2c[[
        "candidate_id",
        "complaint_category",
        "incoming_complaints",
        "distinct_locations",
        "historical_relationship",
        "representative_latitude",
        "representative_longitude"
    ]].head(15)
)

print("\n3. HISTORICAL RELATIONSHIP SUMMARY")

display(
    briefs_18_2c[
        "historical_relationship"
    ].value_counts().rename_axis(
        "relationship"
    ).reset_index(name="candidates")
)

print("\n4. SAMPLE INVESTIGATION BRIEF")

example = briefs_18_2c.loc[
    briefs_18_2c["candidate_id"].eq(
        "NB-00-004"
    )
]

if not example.empty:
    sample = example.iloc[0]
    print(sample["explanation"])
    print("\nSuggested review:")
    print(sample["suggested_review"])

# --------------------------------------------------
# 6. Independence from historical DBSCAN
# --------------------------------------------------

# Change historical labels only.
# New cluster memberships must remain identical.

historical_blinded = source.copy()

historical_blinded["dbscan_status"] = (
    "outside_existing_clusters"
)

with threadpool_limits(limits=1):
    blinded_output = analyze_new_batch_spatial(
        historical_blinded,
        categories,
        eps_m=125,
        min_samples=5,
        min_report_count=5,
        min_distinct_locations=2
    )

independent_assignments = (
    records["new_batch_candidate_id"]
    .fillna("NONE")
    .astype(str)
    .tolist()
    ==
    blinded_output["records"][
        "new_batch_candidate_id"
    ]
    .fillna("NONE")
    .astype(str)
    .tolist()
)

# --------------------------------------------------
# 7. Validation checks
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "500 incoming complaints retained",
    stats["total_records"] == 500
)

check(
    "23 reportable candidate concentrations",
    len(briefs_18_2c) == 23
)

check(
    "228 complaints represented in briefs",
    int(
        briefs_18_2c[
            "incoming_complaints"
        ].sum()
    ) == 228
)

check(
    "Every candidate has a unique ID",
    briefs_18_2c["candidate_id"].is_unique
)

check(
    "Each brief matches one reportable candidate",
    set(briefs_18_2c["candidate_id"])
    == set(
        profiles["new_batch_candidate_id"]
    )
)

check(
    "All reported candidate counts reconcile",
    all(
        int(
            reported["new_batch_candidate_id"]
            .eq(row["candidate_id"]).sum()
        ) == row["incoming_complaints"]
        for row in brief_rows
    )
)

check(
    "Each candidate has a single complaint category",
    reported.groupby(
        "new_batch_candidate_id"
    )["complaint_category"].nunique().eq(1).all()
)

check(
    "Every brief has a representative location",
    briefs_18_2c[[
        "representative_latitude",
        "representative_longitude"
    ]].notna().all().all()
)

check(
    "Representative coordinates are submitted locations",
    all(
        any(
            np.isclose(
                group["lat_num"].astype(float),
                row["representative_latitude"],
                atol=0.0000001
            )
            &
            np.isclose(
                group["lon_num"].astype(float),
                row["representative_longitude"],
                atol=0.0000001
            )
        )
        for row in brief_rows
        for group in [
            reported.loc[
                reported["new_batch_candidate_id"]
                .eq(row["candidate_id"])
            ]
        ]
    )
)

check(
    "Historical outcomes reconcile in every brief",
    all(
        row["historical_established"]
        + row["historical_outside"]
        + row["historical_ambiguous"]
        + row["historical_unclassified"]
        == row["incoming_complaints"]
        for row in brief_rows
    )
)

check(
    "New clusters independent of historical labels",
    independent_assignments
)

check(
    "Sewer example has 15 outside historical clusters",
    not example.empty
    and int(
        example.iloc[0]["incoming_complaints"]
    ) == 15
    and int(
        example.iloc[0]["historical_outside"]
    ) == 15
)

check(
    "Map contains 23 candidate point features",
    len(geojson["features"]) == 23
)

check(
    "All map geometries are valid Point structures",
    all(
        feature["geometry"]["type"] == "Point"
        and len(
            feature["geometry"]["coordinates"]
        ) == 2
        for feature in geojson["features"]
    )
)

check(
    "GeoJSON uses longitude-latitude coordinate order",
    all(
        -180 <= feature["geometry"][
            "coordinates"
        ][0] <= 180
        and -90 <= feature["geometry"][
            "coordinates"
        ][1] <= 90
        for feature in geojson["features"]
    )
)

check(
    "All map features explicitly identify synthetic data",
    all(
        feature["properties"]["synthetic_data"]
        for feature in geojson["features"]
    )
)

check(
    "Every brief contains a suggested review",
    briefs_18_2c[
        "suggested_review"
    ].str.len().gt(20).all()
)

check(
    "DBSCAN parameter metadata preserved",
    geojson["metadata"]["eps_m"] == 125
    and geojson["metadata"]["min_samples"] == 5
)

check(
    "All representative distances are finite",
    np.isfinite(
        briefs_18_2c[
            "max_distance_from_representative_m"
        ].to_numpy(dtype=float)
    ).all()
)

check(
    "Original source CSV unchanged",
    hashlib.sha256(
        input_path.read_bytes()
    ).hexdigest() == source_hash
)

check(
    "Existing DBSCAN engine unchanged",
    hashlib.sha256(
        module_path.read_bytes()
    ).hexdigest() == module_hash
)

# Verify JSON serialization before writing.
geojson_text = json.dumps(
    geojson,
    indent=2,
    allow_nan=False
) + "\n"

brief_document = {
    "synthetic_data": True,
    "model": "new_batch_same_category_dbscan",
    "parameters": {
        "eps_m": 125,
        "min_samples": 5,
        "min_report_count": 5,
        "min_distinct_locations": 2
    },
    "limitations": [
        "Candidates are not confirmed defects.",
        "Historical association is not proof of onset.",
        "Representative points are not street addresses.",
        "The test batch is synthetic.",
        "Production parameter selection remains pending."
    ],
    "candidate_briefs": brief_rows
}

briefs_text = json.dumps(
    brief_document,
    indent=2,
    allow_nan=False
) + "\n"

check(
    "Both output files are valid JSON",
    len(json.loads(geojson_text)["features"]) == 23
    and len(
        json.loads(briefs_text)["candidate_briefs"]
    ) == 23
)

# --------------------------------------------------
# 8. Save only after successful validation
# --------------------------------------------------

results_18_2c = pd.DataFrame(checks)

print("\n5. PHASE 18.2C VALIDATION")

display(results_18_2c)

passed = int(results_18_2c["passed"].sum())
total = len(results_18_2c)

print(f"\nChecks passed: {passed}/{total}")

failed = results_18_2c.loc[
    ~results_18_2c["passed"],
    "test"
].tolist()

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No output files saved.")

    for failure in failed:
        print("FAILED:", failure)

else:

    outputs_to_save = {
        briefs_path: briefs_text,
        map_path: geojson_text
    }

    # Check ALL existing files before
    # writing either new output.
    for path, content in outputs_to_save.items():

        if path.exists():
            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing output differs: {path}. "
                "No files overwritten."
            )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    for path, content in outputs_to_save.items():

        if not path.exists():

            path.write_text(
                content,
                encoding="utf-8"
            )

    print("\n6. OUTPUT FILES")

    print("Investigation briefs:", briefs_path)
    print("Map-ready GeoJSON:", map_path)

    print("\nSTATUS: PASSED")
    print("PHASE 18.2C SPATIAL EXPLANATIONS VERIFIED")
    print("23 SYNTHETIC CANDIDATE BRIEFS PREPARED")
    print("MAP-READY GEOJSON PREPARED")
    print("NEW CLUSTERS INDEPENDENT OF HISTORICAL LABELS")
    print("NO AUTOMATIC INFRASTRUCTURE RISK CLAIMS")
    print("HISTORICAL MODEL ARTIFACTS UNMODIFIED")
```

### COMBINED BATCH INTELLIGENCE

```python

from pathlib import Path
from threadpoolctl import threadpool_limits
import hashlib
import json
import sys

import pandas as pd

print(" COMBINED BATCH INTELLIGENCE")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.models.new_batch_area_kmeans import (
    analyze_new_batch_areas
)
from src.models.new_batch_dbscan import (
    analyze_new_batch_spatial
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

briefs_path = (
    root / "outputs/tests"
    / "phase_18_2c_spatial_explanations"
    / "candidate_briefs.json"
)

map_path = (
    root / "outputs/tests"
    / "phase_18_2c_spatial_explanations"
    / "candidate_map_points.geojson"
)

kmeans_module = (
    root / "src/models/new_batch_area_kmeans.py"
)

dbscan_module = (
    root / "src/models/new_batch_dbscan.py"
)

protected_files = [
    input_path,
    taxonomy_path,
    briefs_path,
    map_path,
    kmeans_module,
    dbscan_module
]

for path in protected_files:
    assert path.is_file(), f"Missing: {path}"

original_hashes = {
    str(path): hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected_files
}

# --------------------------------------------------
# 1. Load existing verified evidence
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(x).strip().upper()
    for x in taxonomy[
        "complaint_category"
    ].dropna()
})

with briefs_path.open(
    encoding="utf-8"
) as file:
    brief_document = json.load(file)

with map_path.open(
    encoding="utf-8"
) as file:
    candidate_map = json.load(file)

existing_briefs = pd.DataFrame(
    brief_document["candidate_briefs"]
)

assert len(source) == 500
assert source["source_row"].is_unique
assert source["complaint_id"].is_unique

# --------------------------------------------------
# 2. Run both independent new-batch models
# --------------------------------------------------

print("\n1. RUNNING INDEPENDENT NEW-BATCH MODELS")

with threadpool_limits(limits=1):

    kmeans_output = analyze_new_batch_areas(
        source,
        categories,
        k=3,
        min_area_size=20,
        random_state=18
    )

    dbscan_output = analyze_new_batch_spatial(
        source,
        categories,
        eps_m=125,
        min_samples=5,
        min_report_count=5,
        min_distinct_locations=2
    )

assert kmeans_output["status"] == "clustered"

print(
    "New K-Means areas:",
    kmeans_output["stats"]["eligible_areas"]
)
print(
    "New K-Means clusters:",
    len(kmeans_output["cluster_profiles"])
)
print(
    "New DBSCAN candidates:",
    dbscan_output["stats"]["reportable_candidates"]
)

# --------------------------------------------------
# 3. Prepare K-Means area assignments
# --------------------------------------------------

kmeans_records = (
    kmeans_output["record_eligibility"][[
        "source_row",
        "area_id",
        "eligibility_reason"
    ]]
    .rename(columns={
        "area_id": "new_kmeans_area_id",
        "eligibility_reason":
            "new_kmeans_eligibility"
    })
)

kmeans_areas = (
    kmeans_output["area_summary"][[
        "area_id",
        "incoming_complaints",
        "meets_minimum",
        "new_batch_cluster",
        "new_batch_description"
    ]]
    .rename(columns={
        "area_id": "new_kmeans_area_id",
        "incoming_complaints":
            "kmeans_area_complaint_count",
        "meets_minimum":
            "new_kmeans_area_eligible",
        "new_batch_cluster":
            "new_kmeans_cluster",
        "new_batch_description":
            "new_kmeans_description"
    })
)

# --------------------------------------------------
# 4. Prepare DBSCAN complaint assignments
# --------------------------------------------------

dbscan_records = (
    dbscan_output["records"][[
        "source_row",
        "record_eligibility",
        "new_batch_status",
        "raw_candidate_id",
        "new_batch_candidate_id"
    ]]
    .rename(columns={
        "record_eligibility":
            "new_dbscan_eligibility",
        "new_batch_status":
            "new_dbscan_status",
        "raw_candidate_id":
            "new_dbscan_raw_candidate_id",
        "new_batch_candidate_id":
            "new_dbscan_candidate_id"
    })
)

# --------------------------------------------------
# 5. Combine by original complaint identity
# --------------------------------------------------

# source_row is the explicit join key.
# Never join on cluster number: historical and
# new clusters use unrelated ID systems.

fusion_records_18_3a = (
    source
    .merge(
        kmeans_records,
        on="source_row",
        how="left",
        validate="one_to_one",
        sort=False
    )
    .merge(
        kmeans_areas,
        on="new_kmeans_area_id",
        how="left",
        validate="many_to_one",
        sort=False
    )
    .merge(
        dbscan_records,
        on="source_row",
        how="left",
        validate="one_to_one",
        sort=False
    )
)

fused = fusion_records_18_3a

fused["is_new_spatial_candidate"] = (
    fused["new_dbscan_status"].eq(
        "candidate_concentration"
    )
)

fused["is_historical_established"] = (
    fused["dbscan_status"].eq(
        "established_cluster"
    )
)

fused["is_historical_outside"] = (
    fused["dbscan_status"].eq(
        "outside_existing_clusters"
    )
)

print("\n2. INTEGRATED COMPLAINT COVERAGE")

print("Input complaints:", len(source))
print("Integrated complaints:", len(fused))

print(
    "New K-Means area assignments:",
    fused["new_kmeans_cluster"].notna().sum()
)

print(
    "New DBSCAN candidate complaints:",
    fused["is_new_spatial_candidate"].sum()
)

# --------------------------------------------------
# 6. Summarize intelligence by new K-Means group
# --------------------------------------------------

area_pattern_summary_18_3a = (
    fused.groupby(
        "new_kmeans_description",
        dropna=False
    )
    .agg(
        geographic_areas=(
            "new_kmeans_area_id",
            "nunique"
        ),
        incoming_complaints=(
            "complaint_id",
            "size"
        ),
        spatial_candidate_complaints=(
            "is_new_spatial_candidate",
            "sum"
        ),
        historically_established=(
            "is_historical_established",
            "sum"
        ),
        outside_historical_hotspots=(
            "is_historical_outside",
            "sum"
        )
    )
    .reset_index()
    .sort_values(
        "incoming_complaints",
        ascending=False
    )
)

print("\n3. K-MEANS AND DBSCAN COMBINED SUMMARY")

display(area_pattern_summary_18_3a)

# --------------------------------------------------
# 7. Summarize evidence for all 20 areas
# --------------------------------------------------

area_intelligence_18_3a = (
    fused.groupby(
        [
            "new_kmeans_area_id",
            "new_kmeans_description"
        ],
        dropna=False
    )
    .agg(
        incoming_complaints=(
            "complaint_id",
            "size"
        ),
        new_spatial_candidate_complaints=(
            "is_new_spatial_candidate",
            "sum"
        ),
        distinct_spatial_candidates=(
            "new_dbscan_candidate_id",
            "nunique"
        ),
        historical_established_complaints=(
            "is_historical_established",
            "sum"
        ),
        historical_outside_complaints=(
            "is_historical_outside",
            "sum"
        )
    )
    .reset_index()
    .sort_values("new_kmeans_area_id")
)

print("\n4. AREA-LEVEL INTELLIGENCE")

display(area_intelligence_18_3a)

# --------------------------------------------------
# 8. Connect candidate briefs to area patterns
# --------------------------------------------------

candidate_rows = fused.loc[
    fused["is_new_spatial_candidate"]
].copy()

def joined_values(series):
    return " | ".join(
        sorted({
            str(value)
            for value in series.dropna()
            if str(value).strip()
        })
    )

candidate_area_context = (
    candidate_rows.groupby(
        "new_dbscan_candidate_id"
    )
    .agg(
        kmeans_area_count=(
            "new_kmeans_area_id",
            "nunique"
        ),
        kmeans_pattern_count=(
            "new_kmeans_description",
            "nunique"
        ),
        linked_area_patterns=(
            "new_kmeans_description",
            joined_values
        ),
        linked_area_ids=(
            "new_kmeans_area_id",
            joined_values
        )
    )
    .reset_index()
    .rename(columns={
        "new_dbscan_candidate_id":
            "candidate_id"
    })
)

integrated_candidates_18_3a = (
    existing_briefs.merge(
        candidate_area_context,
        on="candidate_id",
        how="left",
        validate="one_to_one"
    )
)

print("\n5. INTEGRATED CANDIDATE EXPLANATIONS")

display(
    integrated_candidates_18_3a[[
        "candidate_id",
        "complaint_category",
        "incoming_complaints",
        "historical_relationship",
        "kmeans_area_count",
        "kmeans_pattern_count",
        "linked_area_patterns"
    ]].head(15)
)

# --------------------------------------------------
# 9. Show a combined example
# --------------------------------------------------

print("\n6. EXAMPLE — SYNTHETIC SEWER CONCENTRATION")

example = integrated_candidates_18_3a.loc[
    integrated_candidates_18_3a[
        "candidate_id"
    ].eq("NB-00-004")
]

if not example.empty:

    row = example.iloc[0]

    print("Candidate:", row["candidate_id"])
    print("Category:", row["complaint_category"])
    print(
        "Concentrated complaints:",
        row["incoming_complaints"]
    )
    print(
        "Historical context:",
        row["historical_relationship"]
    )
    print(
        "Area reporting pattern:",
        row["linked_area_patterns"]
    )
    print("\nSuggested investigation:")
    print(row["suggested_review"])

# --------------------------------------------------
# 10. Inspect incoming complaint dates
# --------------------------------------------------

print("\n7. INCOMING DATE AUDIT")

if "created_date" in source.columns:

    date_text = (
        source["created_date"]
        .fillna("")
        .astype(str)
        .str.strip()
        .str[:10]
    )

    parsed_days = pd.to_datetime(
        date_text,
        format="%Y-%m-%d",
        errors="coerce"
    )

    day_labels = (
        parsed_days
        .dt.strftime("%Y-%m-%d")
        .fillna("INVALID_OR_MISSING")
    )

    day_audit_18_3a = (
        day_labels
        .value_counts()
        .rename_axis("created_day")
        .reset_index(name="complaints")
    )

    single_valid_day = (
        parsed_days.notna().all()
        and parsed_days.nunique() == 1
    )

else:

    day_audit_18_3a = pd.DataFrame([{
        "created_day": "COLUMN_MISSING",
        "complaints": len(source)
    }])

    single_valid_day = False

display(day_audit_18_3a)

print(
    "Single valid reporting day:",
    single_valid_day
)

print(
    "NOTE: This audit does not yet enforce "
    "a production daily ingestion window."
)

# --------------------------------------------------
# 11. Verify new K-Means ignores historical labels
# --------------------------------------------------

history_blinded = source.copy()

for column in [
    "kmeans_area_id",
    "kmeans_cluster_id",
    "kmeans_reporting_profile"
]:
    if column in history_blinded.columns:
        history_blinded[column] = (
            "HISTORICAL_LABEL_HIDDEN"
        )

if "kmeans_display_allowed" in (
    history_blinded.columns
):
    history_blinded[
        "kmeans_display_allowed"
    ] = "False"

with threadpool_limits(limits=1):

    blinded_kmeans = analyze_new_batch_areas(
        history_blinded,
        categories,
        k=3,
        min_area_size=20,
        random_state=18
    )

original_assignments = (
    kmeans_output["area_summary"]
    .set_index("area_id")[
        "new_batch_cluster"
    ]
    .sort_index()
)

blinded_assignments = (
    blinded_kmeans["area_summary"]
    .set_index("area_id")[
        "new_batch_cluster"
    ]
    .sort_index()
)

kmeans_independent = (
    original_assignments.equals(
        blinded_assignments
    )
)

print("\n8. MODEL INDEPENDENCE")

print(
    "New K-Means independent of historical labels:",
    kmeans_independent
)

# --------------------------------------------------
# 12. Verify reconciliation and output integrity
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 original complaints retained",
    len(fused) == 500
)

check(
    "Every source row appears exactly once",
    fused["source_row"].is_unique
    and set(fused["source_row"])
    == set(source["source_row"])
)

check(
    "Complaint identities preserved in original order",
    fused["complaint_id"].tolist()
    == source["complaint_id"].tolist()
)

check(
    "All 500 new K-Means assignments present",
    fused["new_kmeans_cluster"].notna().sum()
    == 500
)

check(
    "Exactly 20 new-batch geographic areas",
    fused["new_kmeans_area_id"].nunique() == 20
)

check(
    "Three new K-Means groups retained",
    fused["new_kmeans_cluster"].nunique() == 3
)

check(
    "New K-Means groups contain 10, 5, 5 areas",
    sorted(
        area_pattern_summary_18_3a[
            "geographic_areas"
        ].astype(int).tolist()
    ) == [5, 5, 10]
)

check(
    "Each area receives only one new K-Means label",
    fused.groupby(
        "new_kmeans_area_id"
    )["new_kmeans_cluster"].nunique().eq(1).all()
)

check(
    "All 20 area complaint totals reconcile",
    len(area_intelligence_18_3a) == 20
    and int(
        area_intelligence_18_3a[
            "incoming_complaints"
        ].sum()
    ) == 500
)

check(
    "228 candidate complaints preserved",
    int(
        fused["is_new_spatial_candidate"].sum()
    ) == 228
)

check(
    "Eight reports remain below threshold",
    fused["new_dbscan_status"].eq(
        "below_reporting_threshold"
    ).sum() == 8
)

check(
    "264 reports remain unclustered",
    fused["new_dbscan_status"].eq(
        "unclustered"
    ).sum() == 264
)

check(
    "All 23 DBSCAN candidate IDs retained",
    fused[
        "new_dbscan_candidate_id"
    ].nunique() == 23
)

check(
    "Every candidate remains category-specific",
    candidate_rows.groupby(
        "new_dbscan_candidate_id"
    )["complaint_category"].nunique().eq(1).all()
)

check(
    "All 23 existing briefs successfully integrated",
    len(integrated_candidates_18_3a) == 23
    and integrated_candidates_18_3a[
        "kmeans_area_count"
    ].notna().all()
)

model_counts = (
    candidate_rows.groupby(
        "new_dbscan_candidate_id"
    )
    .size()
    .rename("model_count")
    .reset_index()
    .rename(columns={
        "new_dbscan_candidate_id":
            "candidate_id"
    })
)

brief_count_check = (
    integrated_candidates_18_3a.merge(
        model_counts,
        on="candidate_id",
        validate="one_to_one"
    )
)

check(
    "Integrated brief counts match DBSCAN records",
    len(brief_count_check) == 23
    and brief_count_check[
        "incoming_complaints"
    ].astype(int).eq(
        brief_count_check["model_count"]
    ).all()
)

map_ids = {
    feature["properties"]["candidate_id"]
    for feature in candidate_map["features"]
}

check(
    "Map and integrated brief candidates match",
    len(candidate_map["features"]) == 23
    and map_ids == set(
        integrated_candidates_18_3a[
            "candidate_id"
        ]
    )
)

check(
    "Historical established associations preserved",
    fused["dbscan_status"].eq(
        "established_cluster"
    ).sum() == 137
)

check(
    "Historical ambiguous associations preserved",
    fused["dbscan_status"].eq(
        "ambiguous_association"
    ).sum() == 9
)

check(
    "Historical outside-cluster outcomes preserved",
    fused["dbscan_status"].eq(
        "outside_existing_clusters"
    ).sum() == 354
)

check(
    "Original historical K-Means profiles retained",
    fused["kmeans_reporting_profile"].tolist()
    == source["kmeans_reporting_profile"].tolist()
)

check(
    "New K-Means does not use historical labels",
    kmeans_independent
)

check(
    "Date audit accounts for all 500 reports",
    int(
        day_audit_18_3a[
            "complaints"
        ].sum()
    ) == 500
)

check(
    "No protected model or evidence file changed",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == original_hashes[str(path)]
        for path in protected_files
    )
)

# --------------------------------------------------
# 13. Final validation
# --------------------------------------------------

results_18_3a = pd.DataFrame(checks)

print("\n9. PHASE 18.3A VALIDATION")

display(results_18_3a)

passed = int(results_18_3a["passed"].sum())
total = len(results_18_3a)

print(f"\nChecks passed: {passed}/{total}")

failed = results_18_3a.loc[
    ~results_18_3a["passed"],
    "test"
].tolist()

if failed:

    print("\nSTATUS: NEEDS REVIEW")

    for name in failed:
        print("FAILED:", name)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 18.3A COMBINED INTELLIGENCE VERIFIED")
    print("NEW K-MEANS AND DBSCAN RESULTS INTEGRATED")
    print("HISTORICAL MODEL CONTEXT PRESERVED")
    print("23 SPATIAL BRIEFS LINKED TO AREA PATTERNS")
    print("INCOMING DATE COVERAGE AUDITED")
    print("NO MODEL OR EVIDENCE FILES MODIFIED")
    print("NO AUTOMATIC INFRASTRUCTURE RISK CLAIMS")
```

### Enforce Daily Complaint Windows
if the next CSV contains 500 reports from November 9 and 100 from November 10, the system should analyze only the selected reporting day, while retaining the other 100 in an auditable exclusion record.

We'll also interpret timestamps using New York's local time zone, rather than your computer's time zone, so an incoming UTC timestamp near midnight is assigned to the correct NYC reporting day.

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import hashlib
import sys
import pandas as pd

print(" DAILY BATCH WINDOW ENFORCEMENT")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/daily_batch_window.py"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

kmeans_path = (
    root / "src/models/new_batch_area_kmeans.py"
)

dbscan_path = (
    root / "src/models/new_batch_dbscan.py"
)

protected = [
    input_path,
    taxonomy_path,
    kmeans_path,
    dbscan_path
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

original_hashes = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Define reusable date-window module
# --------------------------------------------------

module_source = dedent(r'''
"""Select one NYC-local reporting day from an incoming batch.

Naive timestamps are interpreted as New York local
time. Explicit timestamp offsets are converted to
New York time.

Ambiguous/nonexistent local DST times are rejected.
All original records remain auditable.
"""

from datetime import date, datetime
import pandas as pd


NYC_TIMEZONE = "America/New_York"


def _parse_nyc_day(value):
    """Return (NYC day, parsing status)."""

    if pd.isna(value):
        return None, "invalid_or_missing_date"

    text = str(value).strip()

    if not text:
        return None, "invalid_or_missing_date"

    parsed = None

    try:
        parsed = datetime.fromisoformat(
            text.replace("Z", "+00:00")
        )
    except ValueError:
        pass

    if parsed is None:

        for fmt in [
            "%m/%d/%Y %I:%M:%S %p",
            "%m/%d/%Y %H:%M:%S"
        ]:
            try:
                parsed = datetime.strptime(
                    text, fmt
                )
                break
            except ValueError:
                continue

    if parsed is None:
        return None, "invalid_or_missing_date"

    try:
        timestamp = pd.Timestamp(parsed)

        if timestamp.tzinfo is None:

            timestamp = timestamp.tz_localize(
                NYC_TIMEZONE,
                ambiguous="NaT",
                nonexistent="NaT"
            )

            parse_status = "assumed_nyc_local"

        else:

            timestamp = timestamp.tz_convert(
                NYC_TIMEZONE
            )

            parse_status = "converted_timezone"

        if pd.isna(timestamp):
            return None, "invalid_or_missing_date"

        return (
            timestamp.date().isoformat(),
            parse_status
        )

    except (ValueError, TypeError, OverflowError):
        return None, "invalid_or_missing_date"


def select_daily_complaints(
    records,
    reporting_day
):
    """Select an explicit NYC reporting day.

    Returns selected records plus a row-level audit.
    Does not mutate the input DataFrame.
    """

    if not isinstance(records, pd.DataFrame):
        raise TypeError("Expected a DataFrame.")

    required = {
        "source_row",
        "complaint_id",
        "created_date"
    }

    missing = required - set(records.columns)

    if missing:
        raise ValueError(
            f"Missing columns: {sorted(missing)}"
        )

    if records["source_row"].duplicated().any():
        raise ValueError(
            "Duplicate source_row values."
        )

    if not isinstance(reporting_day, str):
        raise ValueError(
            "reporting_day must be YYYY-MM-DD."
        )

    try:
        target = date.fromisoformat(
            reporting_day
        ).isoformat()
    except ValueError as exc:
        raise ValueError(
            "Invalid reporting day."
        ) from exc

    if target != reporting_day:
        raise ValueError(
            "Use YYYY-MM-DD reporting day."
        )

    work = records.copy().reset_index(drop=True)

    parsed = [
        _parse_nyc_day(value)
        for value in work["created_date"]
    ]

    work["nyc_reporting_day"] = [
        item[0] for item in parsed
    ]

    work["date_parse_status"] = [
        item[1] for item in parsed
    ]

    work["daily_window_status"] = (
        "outside_selected_day"
    )

    invalid = work["nyc_reporting_day"].isna()

    work.loc[
        invalid,
        "daily_window_status"
    ] = "invalid_or_missing_date"

    selected = work[
        "nyc_reporting_day"
    ].eq(target).fillna(False)

    work.loc[
        selected,
        "daily_window_status"
    ] = "selected_day"

    # Select from the original records, preserving
    # original model input columns and row order.
    selected_records = (
        records.reset_index(drop=True)
        .loc[selected.to_numpy()]
        .copy()
    )

    audit_columns = [
        "source_row",
        "complaint_id",
        "created_date",
        "nyc_reporting_day",
        "date_parse_status",
        "daily_window_status"
    ]

    audit = work[audit_columns].copy()

    stats = {
        "reporting_day": target,
        "timezone": NYC_TIMEZONE,
        "input_records": int(len(work)),
        "selected_records": int(selected.sum()),
        "outside_selected_day": int(
            work["daily_window_status"].eq(
                "outside_selected_day"
            ).sum()
        ),
        "invalid_or_missing_date": int(
            invalid.sum()
        ),
        "naive_local_assumptions": int(
            work["date_parse_status"].eq(
                "assumed_nyc_local"
            ).sum()
        ),
        "timezone_conversions": int(
            work["date_parse_status"].eq(
                "converted_timezone"
            ).sum()
        )
    }

    assert (
        stats["selected_records"]
        + stats["outside_selected_day"]
        + stats["invalid_or_missing_date"]
    ) == stats["input_records"]

    return {
        "selected_records": selected_records,
        "audit": audit,
        "stats": stats
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

# Test entirely in memory before saving.
namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

select_daily_complaints = namespace[
    "select_daily_complaints"
]

parse_nyc_day = namespace[
    "_parse_nyc_day"
]

# --------------------------------------------------
# 2. Load the verified 500 complaint batch
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

assert len(source) == 500

source_snapshot = source.copy(deep=True)

REPORTING_DAY = "2026-11-09"

# --------------------------------------------------
# 3. Verify original single-day batch
# --------------------------------------------------

print("\n1. ORIGINAL DAILY BATCH")

daily_output_18_3b = select_daily_complaints(
    source,
    REPORTING_DAY
)

print(daily_output_18_3b["stats"])

display(
    daily_output_18_3b["audit"][
        "daily_window_status"
    ].value_counts().rename_axis(
        "status"
    ).reset_index(name="complaints")
)

# --------------------------------------------------
# 4. Construct a controlled mixed-day batch
# --------------------------------------------------

mixed_source = source.copy()

# Two complaints explicitly dated November 10.
mixed_source.loc[
    [0, 1], "created_date"
] = "2026-11-10"

# Two invalid timestamps.
mixed_source.loc[
    [2, 3], "created_date"
] = "not-a-valid-date"

# UTC timestamps converted to NYC local dates.
# November 9, 22:30 UTC = November 9 NYC.
mixed_source.loc[
    4, "created_date"
] = "2026-11-09T22:30:00+00:00"

# November 10, 02:30 UTC = November 9 NYC.
mixed_source.loc[
    5, "created_date"
] = "2026-11-10T02:30:00+00:00"

# November 9, 01:00 UTC = November 8 NYC.
mixed_source.loc[
    6, "created_date"
] = "2026-11-09T01:00:00+00:00"

mixed_input_snapshot = mixed_source.copy(
    deep=True
)

mixed_output_18_3b = select_daily_complaints(
    mixed_source,
    REPORTING_DAY
)

mixed_stats = mixed_output_18_3b["stats"]

print("\n2. MIXED-DAY BATCH TEST")
print(mixed_stats)

display(
    mixed_output_18_3b["audit"].head(10)
)

# --------------------------------------------------
# 5. Verify timezone-specific edge cases
# --------------------------------------------------

print("\n3. TIMEZONE EDGE CASES")

timezone_examples = [
    "2026-11-09T12:00:00",
    "2026-11-10T02:30:00+00:00",
    "2026-11-09T01:00:00+00:00",
    "2026-11-09",
    "11/09/2026 10:30:00 AM",
    "2026-11-01T01:30:00"
]

timezone_rows = []

for value in timezone_examples:

    nyc_day, status = parse_nyc_day(value)

    timezone_rows.append({
        "input": value,
        "nyc_day": nyc_day,
        "parse_status": status
    })

timezone_audit_18_3b = pd.DataFrame(
    timezone_rows
)

display(timezone_audit_18_3b)

# --------------------------------------------------
# 6. Run both models on the filtered input
# --------------------------------------------------

from src.models.new_batch_area_kmeans import (
    analyze_new_batch_areas
)

from src.models.new_batch_dbscan import (
    analyze_new_batch_spatial
)

selected_records_18_3b = (
    mixed_output_18_3b["selected_records"]
)

print("\n4. MODEL TEST WITH FILTERED DAILY COMPLAINTS")
print(
    "Selected complaints:",
    len(selected_records_18_3b)
)

with threadpool_limits(limits=1):

    filtered_kmeans_18_3b = (
        analyze_new_batch_areas(
            selected_records_18_3b,
            categories,
            k=3,
            min_area_size=20,
            random_state=18
        )
    )

    filtered_dbscan_18_3b = (
        analyze_new_batch_spatial(
            selected_records_18_3b,
            categories,
            eps_m=125,
            min_samples=5,
            min_report_count=5,
            min_distinct_locations=2
        )
    )

print(
    "K-Means status:",
    filtered_kmeans_18_3b["status"]
)

print(
    "DBSCAN total processed:",
    filtered_dbscan_18_3b["stats"][
        "total_records"
    ]
)

print(
    "DBSCAN candidates:",
    filtered_dbscan_18_3b["stats"][
        "reportable_candidates"
    ]
)

# --------------------------------------------------
# 7. Test invalid input safeguards
# --------------------------------------------------

invalid_target_rejected = False

try:
    select_daily_complaints(
        source,
        "2026-02-30"
    )
except ValueError:
    invalid_target_rejected = True

duplicated_rows = pd.concat([
    source.iloc[:2],
    source.iloc[[0]]
], ignore_index=True)

duplicate_rejected = False

try:
    select_daily_complaints(
        duplicated_rows,
        REPORTING_DAY
    )
except ValueError:
    duplicate_rejected = True

empty_output = select_daily_complaints(
    source.iloc[:0],
    REPORTING_DAY
)

# --------------------------------------------------
# 8. Final validation checklist
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Original 500 reports retained",
    daily_output_18_3b["stats"][
        "input_records"
    ] == 500
)

check(
    "Original 500 reports selected for correct day",
    daily_output_18_3b["stats"][
        "selected_records"
    ] == 500
)

check(
    "Original batch contains no date exclusions",
    daily_output_18_3b["stats"][
        "outside_selected_day"
    ] == 0
    and daily_output_18_3b["stats"][
        "invalid_or_missing_date"
    ] == 0
)

check(
    "Original audit preserves source identities",
    daily_output_18_3b["audit"][
        "source_row"
    ].tolist()
    == source["source_row"].tolist()
)

check(
    "Mixed batch preserves 500 original records",
    mixed_stats["input_records"] == 500
)

check(
    "Mixed batch selects 495 NYC-day complaints",
    mixed_stats["selected_records"] == 495
)

check(
    "Three other-day complaints excluded",
    mixed_stats["outside_selected_day"] == 3
)

check(
    "Two invalid dates excluded with reasons",
    mixed_stats["invalid_or_missing_date"] == 2
)

check(
    "UTC November 10 timestamp maps to NYC November 9",
    mixed_output_18_3b["audit"].iloc[5][
        "nyc_reporting_day"
    ] == "2026-11-09"
)

check(
    "UTC November 9 early timestamp maps to NYC November 8",
    mixed_output_18_3b["audit"].iloc[6][
        "nyc_reporting_day"
    ] == "2026-11-08"
)

check(
    "Both explicit UTC conversions recorded",
    mixed_stats["timezone_conversions"] == 3
)

check(
    "Ambiguous DST timestamp rejected safely",
    timezone_audit_18_3b.iloc[5][
        "parse_status"
    ] == "invalid_or_missing_date"
)

check(
    "Selected batch has no mixed NYC reporting days",
    mixed_output_18_3b["audit"].loc[
        lambda df: df[
            "daily_window_status"
        ].eq("selected_day"),
        "nyc_reporting_day"
    ].eq(REPORTING_DAY).all()
)

check(
    "Selected records keep original source IDs",
    set(selected_records_18_3b["source_row"])
    == set(
        mixed_output_18_3b["audit"].loc[
            lambda df: df[
                "daily_window_status"
            ].eq("selected_day"),
            "source_row"
        ]
    )
)

check(
    "K-Means receives only selected 495 records",
    filtered_kmeans_18_3b["stats"][
        "total_records"
    ] == 495
)

check(
    "Filtered K-Means still identifies 3 clusters",
    filtered_kmeans_18_3b["status"]
    == "clustered"
    and len(
        filtered_kmeans_18_3b[
            "cluster_profiles"
        ]
    ) == 3
)

check(
    "DBSCAN receives only selected 495 records",
    filtered_dbscan_18_3b["stats"][
        "total_records"
    ] == 495
)

check(
    "Filtered DBSCAN statuses reconcile",
    sum([
        filtered_dbscan_18_3b["stats"][
            key
        ]
        for key in [
            "candidate_complaints",
            "below_reporting_threshold",
            "unclustered",
            "excluded_records"
        ]
    ]) == 495
)

check(
    "Invalid requested reporting day rejected",
    invalid_target_rejected
)

check(
    "Duplicate source rows rejected",
    duplicate_rejected
)

check(
    "Empty batch handled safely",
    empty_output["stats"][
        "input_records"
    ] == 0
    and empty_output[
        "selected_records"
    ].empty
)

check(
    "Original input DataFrame unchanged",
    source.equals(source_snapshot)
)

check(
    "Mixed-day test input unchanged",
    mixed_source.equals(
        mixed_input_snapshot
    )
)

check(
    "All protected files remain unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == original_hashes[path]
        for path in protected
    )
)

results_18_3b = pd.DataFrame(checks)

print("\n5. PHASE 18.3B VALIDATION")

display(results_18_3b)

passed = int(results_18_3b["passed"].sum())
total = len(results_18_3b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_18_3b.loc[
    ~results_18_3b["passed"], "test"
].tolist()

# --------------------------------------------------
# 9. Save only after all tests pass
# --------------------------------------------------

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new module saved.")

    for failure in failures:
        print("FAILED:", failure)

else:

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == module_source, (
            "An existing daily-window module differs. "
            "No overwrite performed."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        module_path.write_text(
            module_source,
            encoding="utf-8"
        )

    print("\n6. VERIFIED REUSABLE MODULE")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 18.3B DAILY WINDOW VERIFIED")
    print("NYC LOCAL REPORTING DAY ENFORCED")
    print("CROSS-DAY RECORDS EXCLUDED FROM MODELS")
    print("INVALID DATES RETAINED IN AUDIT")
    print("K-MEANS AND DBSCAN FILTERED CORRECTLY")
    print("HISTORICAL MODELS UNMODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import hashlib
import sys
import pandas as pd

print(" REUSABLE COMBINED INTELLIGENCE PIPELINE")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/integration/combined_batch_intelligence.py"
)

window_path = (
    root / "src/integration/daily_batch_window.py"
)

kmeans_path = (
    root / "src/models/new_batch_area_kmeans.py"
)

dbscan_path = (
    root / "src/models/new_batch_dbscan.py"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

protected = [
    window_path,
    kmeans_path,
    dbscan_path,
    input_path,
    taxonomy_path
]

# Check exact daily-window filename, because the
# previous output displayed an unusual path.
if not window_path.is_file():
    print("Expected:", window_path)
    print(
        "Available daily-window files:",
        [
            repr(p.name)
            for p in window_path.parent.glob(
                "daily_batch*"
            )
        ]
    )
    raise FileNotFoundError(
        "Daily-window module filename needs review."
    )

for path in protected:
    assert path.is_file(), f"Missing: {path}"

protected_hashes = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Define reusable combined-intelligence module
# --------------------------------------------------

module_source = dedent(r'''
"""Combine daily new-batch and historical complaint context.

Historical model results are used as existing
record-level context, never as new model inputs.

No infrastructure defects or priorities are inferred.
"""

import pandas as pd

from src.integration.daily_batch_window import (
    select_daily_complaints
)

from src.models.new_batch_area_kmeans import (
    analyze_new_batch_areas
)

from src.models.new_batch_dbscan import (
    analyze_new_batch_spatial
)


def _joined_unique(values):
    return " | ".join(
        sorted({
            str(value)
            for value in values.dropna()
            if str(value).strip()
        })
    )


def analyze_combined_daily_batch(
    records,
    supported_categories,
    reporting_day,
    k=3,
    min_area_size=20,
    eps_m=125,
    min_samples=5,
    min_report_count=5,
    min_distinct_locations=2,
    random_state=18
):
    """Analyze one NYC reporting day and retain full audit."""

    if not isinstance(records, pd.DataFrame):
        raise TypeError("records must be a DataFrame.")

    if records.columns.duplicated().any():
        raise ValueError("Duplicate column names.")

    source = records.copy().reset_index(drop=True)

    # 1. Select one NYC-local day.
    daily = select_daily_complaints(
        source,
        reporting_day
    )

    selected = daily["selected_records"]

    # 2. Fit the two independent new-batch models.
    km = analyze_new_batch_areas(
        selected,
        supported_categories,
        k=k,
        min_area_size=min_area_size,
        random_state=random_state
    )

    spatial = analyze_new_batch_spatial(
        selected,
        supported_categories,
        eps_m=eps_m,
        min_samples=min_samples,
        min_report_count=min_report_count,
        min_distinct_locations=min_distinct_locations
    )

    # 3. Prepare per-complaint K-Means eligibility.
    km_rows = (
        km["record_eligibility"][[
            "source_row",
            "area_id",
            "eligibility_reason"
        ]]
        .rename(columns={
            "area_id": "new_kmeans_area_id",
            "eligibility_reason":
                "new_kmeans_eligibility"
        })
    )

    # 4. Prepare area assignments. These columns
    # may not exist if K-Means could not fit.
    km_areas = km["area_summary"].copy()

    for column in [
        "new_batch_cluster",
        "new_batch_description"
    ]:
        if column not in km_areas.columns:
            km_areas[column] = pd.NA

    km_areas = km_areas[[
        "area_id",
        "incoming_complaints",
        "meets_minimum",
        "new_batch_cluster",
        "new_batch_description"
    ]].rename(columns={
        "area_id": "new_kmeans_area_id",
        "incoming_complaints":
            "new_kmeans_area_complaints",
        "meets_minimum":
            "new_kmeans_area_meets_minimum",
        "new_batch_cluster":
            "new_kmeans_cluster",
        "new_batch_description":
            "new_kmeans_description"
    })

    # 5. Prepare per-complaint DBSCAN results.
    spatial_rows = (
        spatial["records"][[
            "source_row",
            "record_eligibility",
            "new_batch_status",
            "raw_candidate_id",
            "new_batch_candidate_id"
        ]]
        .rename(columns={
            "record_eligibility":
                "new_dbscan_eligibility",
            "new_batch_status":
                "new_dbscan_status",
            "raw_candidate_id":
                "new_dbscan_raw_candidate_id",
            "new_batch_candidate_id":
                "new_dbscan_candidate_id"
        })
    )

    date_rows = daily["audit"][[
        "source_row",
        "nyc_reporting_day",
        "date_parse_status",
        "daily_window_status"
    ]]

    # 6. Use source_row, never model cluster IDs,
    # as the join key.
    combined = (
        source
        .merge(
            date_rows,
            on="source_row",
            how="left",
            validate="one_to_one",
            sort=False
        )
        .merge(
            km_rows,
            on="source_row",
            how="left",
            validate="one_to_one",
            sort=False
        )
        .merge(
            km_areas,
            on="new_kmeans_area_id",
            how="left",
            validate="many_to_one",
            sort=False
        )
        .merge(
            spatial_rows,
            on="source_row",
            how="left",
            validate="one_to_one",
            sort=False
        )
    )

    selected_mask = combined[
        "daily_window_status"
    ].eq("selected_day")

    # 7. Preserve excluded date records explicitly.
    for column in [
        "new_kmeans_eligibility",
        "new_dbscan_eligibility",
        "new_dbscan_status"
    ]:
        combined.loc[
            ~selected_mask, column
        ] = "not_in_reporting_day"

    combined["new_kmeans_fit_status"] = (
        "not_in_reporting_day"
    )

    combined.loc[
        selected_mask,
        "new_kmeans_fit_status"
    ] = km["status"]

    # 8. Build fresh candidate context. Do not
    # reuse briefs from a different reporting day.
    candidate_rows = combined.loc[
        combined["new_dbscan_status"].eq(
            "candidate_concentration"
        )
    ].copy()

    context_columns = [
        "new_dbscan_candidate_id",
        "associated_kmeans_areas",
        "linked_kmeans_area_ids",
        "linked_kmeans_patterns"
    ]

    if candidate_rows.empty:

        context = pd.DataFrame(
            columns=context_columns
        )

    else:

        context = (
            candidate_rows.groupby(
                "new_dbscan_candidate_id"
            )
            .agg(
                associated_kmeans_areas=(
                    "new_kmeans_area_id",
                    "nunique"
                ),
                linked_kmeans_area_ids=(
                    "new_kmeans_area_id",
                    _joined_unique
                ),
                linked_kmeans_patterns=(
                    "new_kmeans_description",
                    _joined_unique
                )
            )
            .reset_index()
        )

    candidate_intelligence = (
        spatial["candidate_profiles"]
        .merge(
            context,
            left_on="new_batch_candidate_id",
            right_on="new_dbscan_candidate_id",
            how="left",
            validate="one_to_one"
        )
        .drop(
            columns=["new_dbscan_candidate_id"]
        )
    )

    guidance = {
        "SEWER":
            "Review sewer complaint details and "
            "relevant maintenance history.",
        "STREET CONDITION":
            "Review roadway complaint descriptions "
            "and reported locations.",
        "STREET LIGHT CONDITION":
            "Verify whether reports describe "
            "the same lighting assets.",
        "TRAFFIC SIGNAL CONDITION":
            "Verify reported intersections "
            "and traffic-signal equipment."
    }

    candidate_intelligence[
        "suggested_review"
    ] = (
        candidate_intelligence[
            "complaint_category"
        ]
        .map(guidance)
        .fillna(
            "Review the complaint descriptions "
            "and submitted locations."
        )
        + " Check for duplicate reports and "
        + "confirm evidence before field action."
    )

    # 9. Reconcile full original input coverage.
    if len(combined) != len(source):
        raise AssertionError(
            "Combined output changed record count."
        )

    if (
        combined["source_row"].tolist()
        != source["source_row"].tolist()
    ):
        raise AssertionError(
            "Complaint order or identity changed."
        )

    if (
        spatial["stats"]["total_records"]
        != daily["stats"]["selected_records"]
    ):
        raise AssertionError(
            "DBSCAN did not receive the selected day."
        )

    if (
        km["stats"]["total_records"]
        != daily["stats"]["selected_records"]
    ):
        raise AssertionError(
            "K-Means did not receive the selected day."
        )

    if (
        candidate_intelligence[
            "incoming_complaints"
        ].sum() != spatial["stats"][
            "candidate_complaints"
        ]
    ):
        raise AssertionError(
            "Candidate complaint counts disagree."
        )

    stats = {
        "input_records": len(source),
        "reporting_day": reporting_day,
        "selected_records":
            daily["stats"]["selected_records"],
        "date_exclusions": (
            daily["stats"]["outside_selected_day"]
            + daily["stats"][
                "invalid_or_missing_date"
            ]
        ),
        "kmeans_status": km["status"],
        "kmeans_eligible_areas":
            km["stats"]["eligible_areas"],
        "kmeans_assigned_complaints": int(
            combined[
                "new_kmeans_cluster"
            ].notna().sum()
        ),
        "dbscan_reportable_candidates":
            spatial["stats"][
                "reportable_candidates"
            ],
        "dbscan_candidate_complaints":
            spatial["stats"][
                "candidate_complaints"
            ],
        "dbscan_below_reporting_threshold":
            spatial["stats"][
                "below_reporting_threshold"
            ],
        "dbscan_unclustered":
            spatial["stats"]["unclustered"],
        "dbscan_excluded":
            spatial["stats"]["excluded_records"]
    }

    return {
        "combined_records": combined,
        "selected_records": selected.copy(),
        "date_audit": daily["audit"],
        "area_summary": km["area_summary"],
        "candidate_intelligence":
            candidate_intelligence,
        "kmeans_output": km,
        "dbscan_output": spatial,
        "stats": stats
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

# Test in memory before writing a new module.
namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

analyze_combined_daily_batch = namespace[
    "analyze_combined_daily_batch"
]

# --------------------------------------------------
# 2. Load existing synthetic test evidence
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

taxonomy = pd.read_csv(
    taxonomy_path
)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

source_snapshot = source.copy(deep=True)

DAY = "2026-11-09"

# --------------------------------------------------
# 3. Run complete 500-record integration
# --------------------------------------------------

print("\n1. COMPLETE DAILY INTEGRATION")

with threadpool_limits(limits=1):

    full = analyze_combined_daily_batch(
        source,
        categories,
        DAY
    )

print("\nSUMMARY")

for key, value in full["stats"].items():
    print(f"{key}: {value}")

print("\n2. COMBINED RECORD PREVIEW")

display(
    full["combined_records"][[
        "complaint_id",
        "nyc_reporting_day",
        "new_kmeans_area_id",
        "new_kmeans_description",
        "new_dbscan_status",
        "new_dbscan_candidate_id",
        "dbscan_status"
    ]].head(10)
)

print("\n3. COMBINED CANDIDATE INTELLIGENCE")

display(
    full["candidate_intelligence"][[
        "new_batch_candidate_id",
        "complaint_category",
        "incoming_complaints",
        "historical_outside",
        "associated_kmeans_areas",
        "linked_kmeans_patterns"
    ]].head(15)
)

# --------------------------------------------------
# 4. Mixed-day input with audit preservation
# --------------------------------------------------

print("\n4. MIXED-DAY INTEGRATION")

mixed = source.copy()

mixed.loc[[0, 1], "created_date"] = (
    "2026-11-10"
)

mixed.loc[[2, 3], "created_date"] = (
    "invalid-date"
)

mixed.loc[
    4, "created_date"
] = "2026-11-09T22:30:00+00:00"

mixed.loc[
    5, "created_date"
] = "2026-11-10T02:30:00+00:00"

mixed.loc[
    6, "created_date"
] = "2026-11-09T01:00:00+00:00"

with threadpool_limits(limits=1):

    mixed_result = analyze_combined_daily_batch(
        mixed,
        categories,
        DAY
    )

print(mixed_result["stats"])

display(
    mixed_result["combined_records"][
        "daily_window_status"
    ]
    .value_counts()
    .rename_axis("status")
    .reset_index(name="complaints")
)

# --------------------------------------------------
# 5. Small-day and empty-day safeguards
# --------------------------------------------------

print("\n5. SMALL-BATCH SAFEGUARD")

single_area = (
    source["borough"]
    + "_"
    + source["incident_zip"]
)

first_area = single_area.iloc[0]

small_day = source.loc[
    single_area.eq(first_area)
].copy()

with threadpool_limits(limits=1):

    small_result = analyze_combined_daily_batch(
        small_day,
        categories,
        DAY
    )

print(
    "Small-day K-Means:",
    small_result["stats"]["kmeans_status"]
)

print(
    "Small-day DBSCAN records:",
    small_result["dbscan_output"][
        "stats"
    ]["total_records"]
)

empty_day = analyze_combined_daily_batch(
    source.iloc[:0].copy(),
    categories,
    DAY
)

# --------------------------------------------------
# 6. Check all key integration requirements
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

full_records = full["combined_records"]
mixed_records = mixed_result["combined_records"]

check(
    "Original 500 records preserved",
    len(full_records) == 500
)

check(
    "All 500 source rows remain unique",
    full_records["source_row"].is_unique
)

check(
    "Original complaint order preserved",
    full_records["complaint_id"].tolist()
    == source["complaint_id"].tolist()
)

check(
    "All 500 belong to selected NYC day",
    full["stats"]["selected_records"] == 500
)

check(
    "All 500 receive new K-Means assignments",
    full["stats"][
        "kmeans_assigned_complaints"
    ] == 500
)

check(
    "K-Means identifies 20 qualifying areas",
    full["stats"]["kmeans_eligible_areas"]
    == 20
)

check(
    "New K-Means model fits three clusters",
    full["stats"]["kmeans_status"]
    == "clustered"
    and len(
        full["kmeans_output"][
            "cluster_profiles"
        ]
    ) == 3
)

check(
    "23 reportable DBSCAN candidates preserved",
    full["stats"][
        "dbscan_reportable_candidates"
    ] == 23
)

check(
    "228 candidate complaints preserved",
    full["stats"][
        "dbscan_candidate_complaints"
    ] == 228
)

check(
    "Eight below reporting threshold",
    full["stats"][
        "dbscan_below_reporting_threshold"
    ] == 8
)

check(
    "264 DBSCAN unclustered complaints",
    full["stats"]["dbscan_unclustered"]
    == 264
)

check(
    "All 23 candidates have combined context",
    len(full["candidate_intelligence"]) == 23
    and full["candidate_intelligence"][
        "linked_kmeans_patterns"
    ].str.len().gt(0).all()
)

check(
    "Historical DBSCAN outcomes preserved",
    full_records["dbscan_status"]
    .value_counts().to_dict()
    == source["dbscan_status"]
    .value_counts().to_dict()
)

check(
    "Historical K-Means outcomes preserved",
    full_records["kmeans_reporting_profile"]
    .tolist()
    == source["kmeans_reporting_profile"]
    .tolist()
)

check(
    "Mixed input still retains all 500 records",
    len(mixed_records) == 500
)

check(
    "Mixed input selects exactly 495 records",
    mixed_result["stats"][
        "selected_records"
    ] == 495
)

check(
    "Five date exclusions remain auditable",
    mixed_result["stats"][
        "date_exclusions"
    ] == 5
)

excluded = mixed_records.loc[
    mixed_records["daily_window_status"].ne(
        "selected_day"
    )
]

check(
    "Other-day complaints have no new model status",
    len(excluded) == 5
    and excluded["new_dbscan_status"].eq(
        "not_in_reporting_day"
    ).all()
    and excluded["new_kmeans_fit_status"].eq(
        "not_in_reporting_day"
    ).all()
)

check(
    "Other-day complaints have no new clusters",
    excluded["new_kmeans_cluster"].isna().all()
    and excluded[
        "new_dbscan_candidate_id"
    ].isna().all()
)

check(
    "Mixed-day K-Means uses only 495 records",
    mixed_result["kmeans_output"]["stats"][
        "total_records"
    ] == 495
)

check(
    "Mixed-day DBSCAN uses only 495 records",
    mixed_result["dbscan_output"]["stats"][
        "total_records"
    ] == 495
)

check(
    "Single-area K-Means safely declines fitting",
    small_result["stats"]["kmeans_status"]
    == "insufficient_areas"
)

check(
    "Single-area DBSCAN still analyzes 25 reports",
    small_result["dbscan_output"]["stats"][
        "total_records"
    ] == 25
)

check(
    "Empty day handled safely",
    empty_day["stats"]["input_records"] == 0
    and empty_day["combined_records"].empty
)

check(
    "Original DataFrame remains unchanged",
    source.equals(source_snapshot)
)

check(
    "All protected files remain unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == protected_hashes[path]
        for path in protected
    )
)

# --------------------------------------------------
# 7. Report test results
# --------------------------------------------------

results_18_3c = pd.DataFrame(checks)

print("\n6. PHASE 18.3C VALIDATION")

display(results_18_3c)

passed = int(
    results_18_3c["passed"].sum()
)

total = len(results_18_3c)

print(f"\nChecks passed: {passed}/{total}")

failed = results_18_3c.loc[
    ~results_18_3c["passed"], "test"
].tolist()

# --------------------------------------------------
# 8. Save only the fully validated new module
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No combined module saved.")

    for failure in failed:
        print("FAILED:", failure)

else:

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == module_source, (
            "Existing combined module differs. "
            "No overwrite performed."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        module_path.write_text(
            module_source,
            encoding="utf-8"
        )

    print("\n7. VERIFIED MODULE")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 18.3C COMBINED ENGINE VERIFIED")
    print("DAILY NYC REPORTING WINDOW ENFORCED")
    print("NEW K-MEANS AND DBSCAN INTEGRATED")
    print("HISTORICAL MODEL CONTEXT PRESERVED")
    print("DATE-EXCLUDED COMPLAINTS RETAINED")
    print("SMALL-DAY SAFEGUARDS VERIFIED")
    print("NO HISTORICAL MODEL MODIFICATIONS")
```

Anomaly Detection
The next capability is identifying unusual changes in complaint reporting, rather than only identifying similar areas or spatial concentrations.

For example, the assistant should eventually distinguish between these two situations:
Historical baseline	New reporting day	Interpretation
Typically 2–4 sewer complaints per day	3 complaints	Within the usual range
Typically 2–4 sewer complaints per day	20 complaints	Potentially unusual increase worth reviewing

These numbers are illustrative, not findings from your dataset.

We must use actual historical daily complaint records to establish the baseline. Historical K-Means labels and DBSCAN cluster IDs alone cannot tell us the normal daily complaint count.

Our next step is to find and verify the appropriate historical dataset before building the anomaly detection model.

### Historical Baseline Dataset Audit

```python

from pathlib import Path
import os
import re
import hashlib
import pandas as pd

print(" HISTORICAL ANOMALY BASELINE AUDIT")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

assert taxonomy_path.is_file()

taxonomy_hash = hashlib.sha256(
    taxonomy_path.read_bytes()
).hexdigest()

taxonomy = pd.read_csv(taxonomy_path)

supported_categories = {
    str(value).strip().upper()
    for value in taxonomy["complaint_category"].dropna()
}

print("\n1. BASELINE REQUIREMENTS")
print("Historical period expected: 2023–2025")
print("Incoming example day: 2026-11-09")
print("Supported categories:", len(supported_categories))
print("Analysis: historical daily counts by area/category")

# --------------------------------------------------
# 2. Search existing files without modifying them
# --------------------------------------------------

SKIP_DIRECTORIES = {
    ".git",
    ".venv",
    "venv",
    "env",
    "node_modules",
    "__pycache__",
    ".ipynb_checkpoints",
    ".pytest_cache",
    "site-packages",
    ".mypy_cache"
}

EXTENSIONS = (
    ".csv",
    ".csv.gz",
    ".parquet",
    ".pq"
)

files = []

for current_dir, folders, filenames in os.walk(root):

    current_path = Path(current_dir)

    relative_parts = current_path.relative_to(
        root
    ).parts

    # Do not use synthetic test outputs
    # as the historical anomaly baseline.
    if (
        len(relative_parts) >= 2
        and relative_parts[0] == "outputs"
        and relative_parts[1] == "tests"
    ):
        folders[:] = []
        continue

    folders[:] = [
        folder
        for folder in folders
        if folder not in SKIP_DIRECTORIES
        and not folder.startswith(".")
    ]

    for filename in filenames:

        if not filename.lower().endswith(
            EXTENSIONS
        ):
            continue

        path = current_path / filename

        try:
            size_bytes = path.stat().st_size
        except OSError:
            continue

        if size_bytes == 0:
            continue

        files.append({
            "path": path,
            "size_mb": size_bytes / (1024 ** 2)
        })

print("\n2. FILE DISCOVERY")
print("Potential tabular files discovered:", len(files))

# --------------------------------------------------
# 3. Rank likely historical complaint sources
# --------------------------------------------------

def file_priority(path):

    name = str(
        path.relative_to(root)
    ).lower().replace("\\", "/")

    score = 0

    for word, points in [
        ("311", 7),
        ("complaint", 6),
        ("service_request", 5),
        ("clean", 5),
        ("processed", 4),
        ("historical", 4),
        ("2023", 2),
        ("2024", 2),
        ("2025", 2),
        ("geocoded", 2)
    ]:
        if word in name:
            score += points

    for word, penalty in [
        ("synthetic", 15),
        ("kmeans", 10),
        ("dbscan", 10),
        ("hotspot", 10),
        ("profile", 8),
        ("summary", 6),
        ("taxonomy", 20),
        ("cluster", 8),
        ("export", 5)
    ]:
        if word in name:
            score -= penalty

    return score


files = sorted(
    files,
    key=lambda item: (
        -file_priority(item["path"]),
        -item["size_mb"],
        str(item["path"])
    )
)

# Inspect only the strongest candidates.
# Read at most 500 rows per file.
inspection_files = files[:50]

# --------------------------------------------------
# 4. Identify required source columns
# --------------------------------------------------

ALIASES = {
    "created_date": [
        "created_date",
        "created_at",
        "created_datetime",
        "created_date_time",
        "date_created",
        "createddate"
    ],
    "complaint_category": [
        "complaint_category",
        "complaint_type",
        "complaint_type_name",
        "category",
        "issue_type"
    ],
    "borough": [
        "borough",
        "boro",
        "borough_name"
    ],
    "incident_zip": [
        "incident_zip",
        "incident_zip_code",
        "incidentzip",
        "zip",
        "zip_code",
        "zipcode"
    ]
}


def normalize_column(name):
    return re.sub(
        r"[^a-z0-9]+",
        "_",
        str(name).strip().lower()
    ).strip("_")


def detect_columns(frame):

    lookup = {
        normalize_column(column): column
        for column in frame.columns
    }

    found = {}

    for target, choices in ALIASES.items():

        found[target] = next(
            (
                lookup[name]
                for name in choices
                if name in lookup
            ),
            None
        )

    return found


def preview_dataset(path):

    lower_name = path.name.lower()

    if lower_name.endswith((".parquet", ".pq")):

        import pyarrow.parquet as pq

        parquet_file = pq.ParquetFile(str(path))

        batches = parquet_file.iter_batches(
            batch_size=500
        )

        first_batch = next(batches, None)

        if first_batch is None:
            sample = pd.DataFrame(
                columns=parquet_file.schema_arrow.names
            )
        else:
            sample = first_batch.to_pandas()

        row_count = parquet_file.metadata.num_rows

    else:

        try:

            sample = pd.read_csv(
                path,
                nrows=500,
                dtype="string",
                keep_default_na=False,
                encoding="utf-8-sig"
            )

        except UnicodeError:

            sample = pd.read_csv(
                path,
                nrows=500,
                dtype="string",
                keep_default_na=False,
                encoding="latin1"
            )

        row_count = None

    return sample, row_count


# --------------------------------------------------
# 5. Audit the source schemas and sample values
# --------------------------------------------------

audit_rows = []

for item in inspection_files:

    path = item["path"]

    try:

        sample, known_rows = preview_dataset(
            path
        )

        columns = detect_columns(sample)

        required_found = sum(
            value is not None
            for value in columns.values()
        )

        category_overlap = 0
        sample_years = []
        sample_categories = []

        category_col = columns[
            "complaint_category"
        ]

        if category_col is not None:

            values = {
                str(value).strip().upper()
                for value in sample[
                    category_col
                ].dropna()
            }

            category_overlap = len(
                values & supported_categories
            )

            sample_categories = sorted(
                values
            )[:5]

        date_col = columns["created_date"]

        if date_col is not None:

            years = (
                sample[date_col]
                .astype(str)
                .str.extract(
                    r"(20\d{2})",
                    expand=False
                )
                .dropna()
                .unique()
            )

            sample_years = sorted(
                str(year) for year in years
            )

        audit_rows.append({
            "file": str(
                path.relative_to(root)
            ),
            "size_mb": round(
                item["size_mb"], 2
            ),
            "known_rows": known_rows,
            "required_columns_found":
                required_found,
            "created_date_col":
                columns["created_date"],
            "category_col":
                columns["complaint_category"],
            "borough_col":
                columns["borough"],
            "zip_col":
                columns["incident_zip"],
            "taxonomy_sample_overlap":
                category_overlap,
            "sample_years":
                ", ".join(sample_years),
            "sample_categories":
                ", ".join(sample_categories),
            "file_priority":
                file_priority(path),
            "inspection_error": ""
        })

    except Exception as exc:

        audit_rows.append({
            "file": str(
                path.relative_to(root)
            ),
            "size_mb": round(
                item["size_mb"], 2
            ),
            "known_rows": None,
            "required_columns_found": 0,
            "created_date_col": None,
            "category_col": None,
            "borough_col": None,
            "zip_col": None,
            "taxonomy_sample_overlap": 0,
            "sample_years": "",
            "sample_categories": "",
            "file_priority":
                file_priority(path),
            "inspection_error": str(exc)[:120]
        })


audit_19_1a = pd.DataFrame(audit_rows)

# --------------------------------------------------
# 6. Identify credible baseline candidates
# --------------------------------------------------

if audit_19_1a.empty:

    baseline_candidates_19_1a = (
        audit_19_1a.copy()
    )

else:

    audit_19_1a = audit_19_1a.sort_values(
        [
            "required_columns_found",
            "taxonomy_sample_overlap",
            "file_priority",
            "size_mb"
        ],
        ascending=[False, False, False, False]
    ).reset_index(drop=True)

    baseline_candidates_19_1a = (
        audit_19_1a.loc[
            audit_19_1a[
                "required_columns_found"
            ].eq(4)
            & audit_19_1a[
                "taxonomy_sample_overlap"
            ].gt(0)
            & audit_19_1a[
                "inspection_error"
            ].eq("")
        ].copy()
    )

print("\n3. TOP INSPECTED DATASETS")

if audit_19_1a.empty:
    print("No CSV or Parquet candidates discovered.")
else:
    display(
        audit_19_1a[[
            "file",
            "size_mb",
            "known_rows",
            "required_columns_found",
            "taxonomy_sample_overlap",
            "sample_years",
            "inspection_error"
        ]].head(20)
    )

print("\n4. POSSIBLE HISTORICAL BASELINE SOURCES")

if baseline_candidates_19_1a.empty:

    print(
        "No source yet satisfies all four "
        "schema requirements with a sample "
        "category match."
    )

else:

    display(
        baseline_candidates_19_1a[[
            "file",
            "size_mb",
            "known_rows",
            "created_date_col",
            "category_col",
            "borough_col",
            "zip_col",
            "sample_years",
            "sample_categories"
        ]].head(10)
    )

# --------------------------------------------------
# 7. Check large in-memory notebook DataFrames
# --------------------------------------------------

print("\n5. EXISTING NOTEBOOK DATAFRAMES")

large_frames = []

for name, value in list(globals().items()):

    if (
        isinstance(value, pd.DataFrame)
        and len(value) >= 100_000
    ):

        large_frames.append({
            "variable": name,
            "rows": len(value),
            "column_count": len(value.columns),
            "columns_preview": ", ".join(
                map(str, value.columns[:8])
            )
        })

if large_frames:

    display(
        pd.DataFrame(
            large_frames
        ).sort_values(
            "rows",
            ascending=False
        ).head(10)
    )

else:

    print(
        "No large historical DataFrame is "
        "currently loaded in the notebook."
    )

# --------------------------------------------------
# 8. Verify audit did not modify evidence
# --------------------------------------------------

assert hashlib.sha256(
    taxonomy_path.read_bytes()
).hexdigest() == taxonomy_hash

assert len(files) >= len(inspection_files)

print("\n6. AUDIT RESULT")

print("Files discovered:", len(files))
print("Files inspected:", len(audit_rows))

print(
    "Possible historical sources:",
    len(baseline_candidates_19_1a)
)

print("\nSTATUS: AUDIT COMPLETED")
print("PHASE 19.1A HISTORICAL SOURCE DISCOVERY FINISHED")
print("NO HISTORICAL FILES MODIFIED")
print("NO ANOMALY SCORES GENERATED")
print("BASELINE COVERAGE STILL REQUIRES VERIFICATION")
```

```python

from pathlib import Path
from collections import Counter
import re
import pandas as pd

print(" HISTORICAL BASELINE COVERAGE AUDIT")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

primary_path = (
    root / "data/processed"
    / "nyc311_infrastructure_cleaned.csv"
)

comparison_candidates = [
    root / "data/processed"
    / "nyc311_all_15_complaints_2023_2025.csv",
    root / "data/raw"
    / "nyc311_all_15_complaints_2023_2025.csv",
    root / "nyc311_all_15_complaints_2023_2025.csv"
]

comparison_path = next(
    (
        path for path in comparison_candidates
        if path.is_file()
    ),
    None
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

assert primary_path.is_file(), (
    f"Primary historical dataset missing: {primary_path}"
)

assert reference_path.is_file()
assert taxonomy_path.is_file()

source_paths = {
    "processed_cleaned": primary_path
}

if comparison_path is not None:
    source_paths["all_15_complaints"] = comparison_path

# Record filesystem metadata before reading.
file_signatures_before = {
    name: (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for name, path in source_paths.items()
}

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

reference = pd.read_csv(
    reference_path,
    dtype="string"
)

assert "area_id" in reference.columns

reference_areas = set(
    reference["area_id"].astype(str)
)

assert len(reference_areas) == 20

START_DAY = "2023-01-01"
END_DAY = "2025-12-31"

calendar = pd.date_range(
    START_DAY,
    END_DAY,
    freq="D"
)

VALID_BOROUGHS = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}

print("\n1. HISTORICAL SOURCE CANDIDATES")

for name, path in source_paths.items():
    print(name, ":", path)

print("Expected calendar days:", len(calendar))
print("Supported categories:", len(categories))
print("Test ZIP/borough areas:", len(reference_areas))

# --------------------------------------------------
# 2. Recognize existing source column names
# --------------------------------------------------

ALIASES = {
    "created_date": [
        "created_date",
        "created_at",
        "created_datetime"
    ],
    "complaint_category": [
        "complaint_type",
        "complaint_category"
    ],
    "borough": [
        "borough",
        "boro"
    ],
    "incident_zip": [
        "incident_zip",
        "zip_code",
        "zip"
    ],
    "complaint_id": [
        "unique_key",
        "unique_key_id",
        "complaint_id"
    ]
}

def identify_columns(path):

    header = pd.read_csv(
        path,
        nrows=0,
        encoding="utf-8-sig"
    )

    available = {
        re.sub(
            r"[^a-z0-9]+",
            "_",
            str(column).lower()
        ).strip("_"): column
        for column in header.columns
    }

    found = {}

    for field, choices in ALIASES.items():
        found[field] = next(
            (
                available[choice]
                for choice in choices
                if choice in available
            ),
            None
        )

    missing = [
        name
        for name in [
            "created_date",
            "complaint_category",
            "borough",
            "incident_zip"
        ]
        if found[name] is None
    ]

    if missing:
        raise ValueError(
            f"{path.name}: missing columns {missing}"
        )

    return found

# --------------------------------------------------
# 3. Audit full historical data in chunks
# --------------------------------------------------

def audit_historical_csv(
    path,
    supported_categories,
    focus_areas
):

    columns = identify_columns(path)

    use_columns = sorted({
        column
        for column in columns.values()
        if column is not None
    })

    stats = Counter()
    years = Counter()
    eligible_years = Counter()
    daily_totals = Counter()
    category_totals = Counter()
    unsupported_totals = Counter()
    area_totals = Counter()
    area_daily_totals = Counter()
    area_category_totals = Counter()

    observed_days = set()
    seen_ids = set()

    date_review_examples = []
    duplicate_id_examples = []

    reader = pd.read_csv(
        path,
        usecols=use_columns,
        chunksize=60000,
        dtype="string",
        keep_default_na=False,
        encoding="utf-8-sig"
    )

    for chunk_number, chunk in enumerate(
        reader,
        start=1
    ):

        stats["source_rows"] += len(chunk)

        # ------------------------------------------
        # Historical date recognition
        # ------------------------------------------

        raw_dates = (
            chunk[columns["created_date"]]
            .fillna("")
            .astype("string")
            .str.strip()
        )

        # Explicit timezone offsets require
        # proper conversion before use in a
        # production NYC-local baseline.
        # Exclude them from this preliminary audit.
        has_offset = raw_dates.str.contains(
            r"(?:[Zz]|[+-]\d{2}:?\d{2})$",
            regex=True,
            na=False
        )

        iso_text = raw_dates.str.extract(
            r"^(\d{4}-\d{2}-\d{2})(?:$|[ T])",
            expand=False
        )

        us_text = raw_dates.str.extract(
            r"^(\d{1,2}/\d{1,2}/\d{4})(?:$|[ T])",
            expand=False
        )

        iso_days = pd.to_datetime(
            iso_text,
            format="%Y-%m-%d",
            errors="coerce"
        ).dt.strftime("%Y-%m-%d")

        us_days = pd.to_datetime(
            us_text,
            format="%m/%d/%Y",
            errors="coerce"
        ).dt.strftime("%Y-%m-%d")

        days = (
            iso_days
            .fillna(us_days)
            .fillna("")
        )

        parseable = (
            days.ne("")
            & ~has_offset
        )

        in_period = (
            parseable
            & days.ge(START_DAY)
            & days.le(END_DAY)
        )

        stats["timestamp_offset_review"] += int(
            has_offset.sum()
        )

        stats["invalid_or_unparsed_dates"] += int(
            (~parseable).sum()
        )

        stats["out_of_period_dates"] += int(
            (parseable & ~in_period).sum()
        )

        stats["historical_date_rows"] += int(
            in_period.sum()
        )

        if len(date_review_examples) < 5:

            date_review_examples.extend(
                raw_dates.loc[
                    ~parseable
                ].head(
                    5 - len(date_review_examples)
                ).tolist()
            )

        years.update(
            days.loc[in_period]
            .str[:4]
            .value_counts()
            .to_dict()
        )

        # ------------------------------------------
        # Complaint category compatibility
        # ------------------------------------------

        complaint_categories = (
            chunk[
                columns["complaint_category"]
            ]
            .fillna("")
            .astype("string")
            .str.strip()
            .str.upper()
        )

        supported = complaint_categories.isin(
            supported_categories
        )

        stats["in_period_supported_category"] += (
            int((in_period & supported).sum())
        )

        category_totals.update(
            complaint_categories.loc[
                in_period & supported
            ].value_counts().to_dict()
        )

        unsupported_totals.update(
            complaint_categories.loc[
                in_period & ~supported
            ].value_counts().to_dict()
        )

        # ------------------------------------------
        # Area identifier quality
        # ------------------------------------------

        borough = (
            chunk[columns["borough"]]
            .fillna("")
            .astype("string")
            .str.strip()
            .str.upper()
        )

        zip_code = (
            chunk[columns["incident_zip"]]
            .fillna("")
            .astype("string")
            .str.strip()
        )

        valid_area = (
            borough.isin(VALID_BOROUGHS)
            & zip_code.str.fullmatch(
                r"\d{5}",
                na=False
            )
        )

        stats[
            "in_period_valid_borough_zip_syntax"
        ] += int(
            (in_period & valid_area).sum()
        )

        # This validates area ID syntax, not
        # geographic point-in-polygon consistency.
        eligible = (
            in_period
            & supported
            & valid_area
        )

        stats["eligible_area_category_rows"] += (
            int(eligible.sum())
        )

        eligible_days = days.loc[eligible]

        observed_days.update(
            eligible_days.unique().tolist()
        )

        daily_totals.update(
            eligible_days.value_counts().to_dict()
        )

        eligible_years.update(
            eligible_days.str[:4]
            .value_counts()
            .to_dict()
        )

        area_ids = borough + "_" + zip_code

        area_totals.update(
            area_ids.loc[
                eligible
            ].value_counts().to_dict()
        )

        # ------------------------------------------
        # Examine 20 known synthetic test areas
        # against the real historical records
        # ------------------------------------------

        in_focus = (
            eligible
            & area_ids.isin(focus_areas)
        )

        if in_focus.any():

            detail = pd.DataFrame({
                "area_id": area_ids.loc[
                    in_focus
                ],
                "day": days.loc[in_focus],
                "category":
                    complaint_categories.loc[
                        in_focus
                    ]
            })

            by_area_day = detail.groupby(
                ["area_id", "day"]
            ).size()

            area_daily_totals.update({
                (area, day): int(count)
                for (area, day), count
                in by_area_day.items()
            })

            by_area_category = detail.groupby(
                ["area_id", "category"]
            ).size()

            area_category_totals.update({
                (area, category): int(count)
                for (area, category), count
                in by_area_category.items()
            })

        # ------------------------------------------
        # Check source complaint ID duplication
        # ------------------------------------------

        id_column = columns["complaint_id"]

        if id_column is not None:

            ids = (
                chunk[id_column]
                .fillna("")
                .astype("string")
                .str.strip()
            )

            stats["missing_ids"] += int(
                ids.eq("").sum()
            )

            for identifier in ids.loc[
                ids.ne("")
            ].tolist():

                if identifier in seen_ids:

                    stats["duplicate_ids"] += 1

                    if len(
                        duplicate_id_examples
                    ) < 3:
                        duplicate_id_examples.append(
                            identifier
                        )

                else:

                    seen_ids.add(identifier)

    # ----------------------------------------------
    # Full-file summary
    # ----------------------------------------------

    if columns["complaint_id"] is not None:
        stats["unique_nonempty_ids"] = len(
            seen_ids
        )

    stats["days_with_eligible_reports"] = len(
        observed_days
    )

    stats["reference_areas_with_reports"] = sum(
        area_totals.get(area_id, 0) > 0
        for area_id in focus_areas
    )

    return {
        "path": path,
        "columns": columns,
        "stats": stats,
        "years": years,
        "eligible_years": eligible_years,
        "daily_totals": daily_totals,
        "category_totals": category_totals,
        "unsupported_totals": unsupported_totals,
        "area_totals": area_totals,
        "area_daily_totals": area_daily_totals,
        "area_category_totals": area_category_totals,
        "date_review_examples":
            date_review_examples,
        "duplicate_id_examples":
            duplicate_id_examples
    }

# --------------------------------------------------
# 4. Scan the source candidates separately
# --------------------------------------------------

historical_audits_19_1b = {}

for name, path in source_paths.items():

    print(f"\nScanning historical source: {name}")

    audit = audit_historical_csv(
        path,
        set(categories),
        reference_areas
    )

    historical_audits_19_1b[name] = audit

    print(
        "Total rows:",
        audit["stats"]["source_rows"]
    )

    print(
        "Eligible area/category rows:",
        audit["stats"][
            "eligible_area_category_rows"
        ]
    )

    print(
        "Historical years:",
        dict(audit["years"])
    )

# --------------------------------------------------
# 5. Full-file quality comparison
# --------------------------------------------------

comparison_rows = []

for name, audit in historical_audits_19_1b.items():

    stats = audit["stats"]

    comparison_rows.append({
        "source": name,
        "total_rows": stats["source_rows"],
        "historical_date_rows":
            stats["historical_date_rows"],
        "eligible_area_category_rows":
            stats["eligible_area_category_rows"],
        "days_with_reports":
            stats["days_with_eligible_reports"],
        "supported_category_count":
            sum(
                count > 0
                for count in audit[
                    "category_totals"
                ].values()
            ),
        "distinct_area_ids": len(
            audit["area_totals"]
        ),
        "reference_areas_covered":
            stats["reference_areas_with_reports"],
        "duplicate_complaint_ids":
            (
                stats["duplicate_ids"]
                if audit["columns"]["complaint_id"]
                is not None
                else None
            ),
        "invalid_or_unparsed_dates":
            stats["invalid_or_unparsed_dates"],
        "out_of_period_dates":
            stats["out_of_period_dates"],
        "offset_timestamps_to_review":
            stats["timestamp_offset_review"]
    })

historical_comparison_19_1b = pd.DataFrame(
    comparison_rows
)

print("\n2. COMPLETE SOURCE COMPARISON")

display(historical_comparison_19_1b)

# --------------------------------------------------
# 6. Verify coverage across all three years
# --------------------------------------------------

year_rows = []

for name, audit in historical_audits_19_1b.items():

    for year in [2023, 2024, 2025]:

        days_in_year = sum(
            calendar.year == year
        )

        observed_year_days = sum(
            day.startswith(str(year))
            for day in audit[
                "daily_totals"
            ].keys()
        )

        year_rows.append({
            "source": name,
            "year": year,
            "historical_date_rows":
                audit["years"].get(
                    str(year), 0
                ),
            "eligible_area_category_rows":
                audit["eligible_years"].get(
                    str(year), 0
                ),
            "calendar_days":
                days_in_year,
            "days_with_eligible_reports":
                observed_year_days,
            "calendar_coverage_pct": round(
                100
                * observed_year_days
                / days_in_year,
                2
            )
        })

historical_years_19_1b = pd.DataFrame(
    year_rows
)

print("\n3. YEAR-BY-YEAR COVERAGE")

display(historical_years_19_1b)

# --------------------------------------------------
# 7. Inspect the primary source's 20 test areas
# --------------------------------------------------

primary_audit_19_1b = (
    historical_audits_19_1b[
        "processed_cleaned"
    ]
)

primary_area_counts = primary_audit_19_1b[
    "area_totals"
]

primary_area_days = primary_audit_19_1b[
    "area_daily_totals"
]

primary_area_categories = primary_audit_19_1b[
    "area_category_totals"
]

reference_rows = []

for area in sorted(reference_areas):

    total = int(
        primary_area_counts.get(area, 0)
    )

    active_days = sum(
        1
        for area_id, day in primary_area_days
        if area_id == area
    )

    reference_rows.append({
        "area_id": area,
        "historical_reports": total,
        "days_with_reports": active_days,
        "mean_reports_per_calendar_day": round(
            total / len(calendar),
            3
        ),
        "street_condition_reports": int(
            primary_area_categories.get(
                (area, "STREET CONDITION"),
                0
            )
        ),
        "sewer_reports": int(
            primary_area_categories.get(
                (area, "SEWER"),
                0
            )
        ),
        "street_light_reports": int(
            primary_area_categories.get(
                (area, "STREET LIGHT CONDITION"),
                0
            )
        )
    })

reference_coverage_19_1b = pd.DataFrame(
    reference_rows
)

print("\n4. HISTORICAL COVERAGE OF THE 20 TEST AREAS")

display(reference_coverage_19_1b)

# --------------------------------------------------
# 8. Inspect taxonomy mismatches and sparse dates
# --------------------------------------------------

print("\n5. CATEGORY COMPATIBILITY")

category_coverage_19_1b = pd.DataFrame([
    {
        "complaint_category": category,
        "historical_records": int(
            primary_audit_19_1b[
                "category_totals"
            ].get(category, 0)
        )
    }
    for category in categories
])

display(category_coverage_19_1b)

unsupported = (
    primary_audit_19_1b[
        "unsupported_totals"
    ]
)

if unsupported:

    print("\nUnmatched historical category names:")

    display(
        pd.DataFrame(
            unsupported.most_common(15),
            columns=[
                "source_category",
                "historical_records"
            ]
        )
    )

# Daily counts with calendar zeros are useful
# for detecting sparse reporting patterns,
# but are not yet anomaly thresholds.
citywide_daily = pd.Series(
    [
        primary_audit_19_1b[
            "daily_totals"
        ].get(
            day.strftime("%Y-%m-%d"),
            0
        )
        for day in calendar
    ],
    index=calendar,
    dtype="int64"
)

print("\n6. DAILY COVERAGE SUMMARY")

print(
    "Days with eligible reports:",
    int(citywide_daily.gt(0).sum())
)

print(
    "Zero-report calendar days:",
    int(citywide_daily.eq(0).sum())
)

print(
    "Median eligible reports/day:",
    float(citywide_daily.median())
)

print(
    "95th-percentile reports/day:",
    float(citywide_daily.quantile(0.95))
)

# --------------------------------------------------
# 9. Check source integrity
# --------------------------------------------------

file_signatures_after = {
    name: (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for name, path in source_paths.items()
}

source_files_unchanged = (
    file_signatures_before
    == file_signatures_after
)

assert source_files_unchanged, (
    "A source file changed during the audit. "
    "Review the input before continuing."
)

# --------------------------------------------------
# 10. Validate audit coverage
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Primary historical dataset available",
    "processed_cleaned"
    in historical_audits_19_1b
)

check(
    "Primary dataset contains historical rows",
    primary_audit_19_1b["stats"][
        "historical_date_rows"
    ] > 0
)

check(
    "Primary dataset has supported categories",
    sum(
        primary_audit_19_1b[
            "category_totals"
        ].values()
    ) > 0
)

check(
    "Primary dataset has valid area-category records",
    primary_audit_19_1b["stats"][
        "eligible_area_category_rows"
    ] > 0
)

check(
    "Year audit includes 2023, 2024 and 2025",
    set(
        historical_years_19_1b.loc[
            historical_years_19_1b[
                "source"
            ].eq("processed_cleaned"),
            "year"
        ]
    ) == {2023, 2024, 2025}
)

check(
    "All 20 test areas individually audited",
    len(reference_coverage_19_1b) == 20
)

check(
    "All 15 supported categories individually audited",
    len(category_coverage_19_1b) == 15
)

check(
    "Calendar includes 1096 historical days",
    len(calendar) == 1096
)

check(
    "Citywide daily counts reconcile with eligible rows",
    int(citywide_daily.sum())
    == primary_audit_19_1b["stats"][
        "eligible_area_category_rows"
    ]
)

check(
    "No historical sources modified",
    source_files_unchanged
)

results_19_1b = pd.DataFrame(checks)

print("\n7. PHASE 19.1B VALIDATION")

display(results_19_1b)

passed = int(results_19_1b["passed"].sum())
total = len(results_19_1b)

print(f"\nChecks passed: {passed}/{total}")

# These are additional source-quality indicators,
# NOT hard-coded assertions about unknown data.
primary_years_present = all(
    primary_audit_19_1b[
        "eligible_years"
    ].get(str(year), 0) > 0
    for year in [2023, 2024, 2025]
)

historical_day_coverage = (
    primary_audit_19_1b["stats"][
        "days_with_eligible_reports"
    ] / len(calendar)
)

reference_areas_covered = int(
    reference_coverage_19_1b[
        "historical_reports"
    ].gt(0).sum()
)

print("\n8. BASELINE READINESS INDICATORS")

print(
    "All 3 historical years represented:",
    primary_years_present
)

print(
    "Historical calendar coverage:",
    f"{historical_day_coverage:.1%}"
)

print(
    "Test areas with historical reports:",
    f"{reference_areas_covered}/20"
)

print(
    "Primary duplicate IDs:",
    primary_audit_19_1b[
        "stats"
    ]["duplicate_ids"]
)

print(
    "Primary timestamp-offset records needing review:",
    primary_audit_19_1b[
        "stats"
    ]["timestamp_offset_review"]
)

if passed != total:

    print("\nSTATUS: NEEDS REVIEW")
    print("AUDIT CHECKS DID NOT ALL PASS")

elif not primary_years_present:

    print("\nSTATUS: BASELINE COVERAGE NEEDS REVIEW")
    print("ONE OR MORE HISTORICAL YEARS MISSING")

else:

    print("\nSTATUS: AUDIT COMPLETED")
    print("PHASE 19.1B HISTORICAL COVERAGE VERIFIED")
    print("HISTORICAL YEARS AND CATEGORIES INSPECTED")
    print("20 TEST AREAS AUDITED")
    print("SOURCE SELECTION STILL REQUIRES REVIEW")
    print("NO ANOMALY SCORES GENERATED")
    print("NO HISTORICAL FILES MODIFIED")
```

Build Historical Daily Baselines
We'll construct historical daily counts separately for each of the 20 test areas and each of the 15 complaint categories.

For the November 9, 2026 incoming batch, we'll inspect comparable historical Mondays from October–December of 2023–2025. This is an exploratory seasonal and weekday-matched baseline, not a finalized statistical anomaly model.

We will explicitly record insufficient historical evidence rather than treating every zero as reliable proof of an unusual increase.

```python

from pathlib import Path
from collections import Counter
import pandas as pd
import numpy as np

print("PHASE 19.1C — HISTORICAL DAILY BASELINE CONSTRUCTION")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

historical_path = (
    root / "data/processed"
    / "nyc311_infrastructure_cleaned.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

incoming_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

protected_paths = [
    historical_path,
    taxonomy_path,
    reference_path,
    incoming_path
]

for path in protected_paths:
    assert path.is_file(), f"Missing: {path}"

file_signatures_before = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected_paths
}

# --------------------------------------------------
# 1. Define the comparison period
# --------------------------------------------------

HISTORY_START = "2023-01-01"
HISTORY_END = "2025-12-31"
REPORTING_DAY = "2026-11-09"

historical_calendar = pd.date_range(
    HISTORY_START,
    HISTORY_END,
    freq="D"
)

target_date = pd.Timestamp(REPORTING_DAY)

# Pilot comparison:
# Same weekday in October, November or December.
#
# This is exploratory season matching,
# not an approved production baseline.

SEASONAL_MONTHS = {10, 11, 12}

matched_dates = [
    date
    for date in historical_calendar
    if (
        date.weekday() == target_date.weekday()
        and date.month in SEASONAL_MONTHS
    )
]

matched_days = [
    date.strftime("%Y-%m-%d")
    for date in matched_dates
]

matched_day_years = {
    day: day[:4]
    for day in matched_days
}

# Provisional evidence requirements, not
# statistical significance thresholds.
MIN_HISTORICAL_MATCHED_REPORTS = 8
MIN_ACTIVE_MATCHED_DAYS = 5
MIN_YEARS_REPRESENTED = 2

taxonomy = pd.read_csv(taxonomy_path)

categories = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

reference = pd.read_csv(
    reference_path,
    dtype="string",
    keep_default_na=False
)

focus_areas = sorted(
    reference["area_id"]
    .astype(str)
    .unique()
)

assert len(categories) == 15
assert len(focus_areas) == 20

print("\n1. HISTORICAL COMPARISON DESIGN")
print("Historical days:", len(historical_calendar))
print("Incoming day:", REPORTING_DAY)
print("Incoming weekday:", target_date.day_name())
print("Historical matched weekdays:", len(matched_days))
print("Seasonal months:", sorted(SEASONAL_MONTHS))
print("Pilot geographic areas:", len(focus_areas))
print("Complaint categories:", len(categories))

# --------------------------------------------------
# 2. Scan the full historical source in chunks
# --------------------------------------------------

historical_daily_counts_19_1c = Counter()

total_source_rows = 0
total_eligible_rows = 0
total_focus_rows = 0

valid_boroughs = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}

source_columns = [
    "created_date",
    "complaint_type",
    "borough",
    "incident_zip"
]

reader = pd.read_csv(
    historical_path,
    usecols=source_columns,
    dtype="string",
    keep_default_na=False,
    chunksize=60000,
    encoding="utf-8-sig"
)

print("\n2. CONSTRUCTING HISTORICAL DAILY COUNTS")

for chunk_number, chunk in enumerate(
    reader,
    start=1
):

    total_source_rows += len(chunk)

    raw_date = (
        chunk["created_date"]
        .fillna("")
        .astype("string")
        .str.strip()
    )

    # Same two historical date families as
    # the verified Phase 19.1B audit.
    iso_text = raw_date.str.extract(
        r"^(\d{4}-\d{2}-\d{2})(?:$|[ T])",
        expand=False
    )

    us_text = raw_date.str.extract(
        r"^(\d{1,2}/\d{1,2}/\d{4})(?:$|[ T])",
        expand=False
    )

    iso_parsed = pd.to_datetime(
        iso_text,
        format="%Y-%m-%d",
        errors="coerce"
    )

    us_parsed = pd.to_datetime(
        us_text,
        format="%m/%d/%Y",
        errors="coerce"
    )

    days = (
        iso_parsed
        .fillna(us_parsed)
        .dt.strftime("%Y-%m-%d")
        .fillna("")
    )

    in_history = (
        days.ge(HISTORY_START)
        & days.le(HISTORY_END)
    )

    category = (
        chunk["complaint_type"]
        .fillna("")
        .astype("string")
        .str.strip()
        .str.upper()
    )

    borough = (
        chunk["borough"]
        .fillna("")
        .astype("string")
        .str.strip()
        .str.upper()
    )

    zip_code = (
        chunk["incident_zip"]
        .fillna("")
        .astype("string")
        .str.strip()
    )

    valid_area = (
        borough.isin(valid_boroughs)
        & zip_code.str.fullmatch(
            r"\d{5}",
            na=False
        )
    )

    eligible = (
        in_history
        & category.isin(categories)
        & valid_area
    )

    total_eligible_rows += int(
        eligible.sum()
    )

    area_ids = borough + "_" + zip_code

    in_focus = (
        eligible
        & area_ids.isin(focus_areas)
    )

    total_focus_rows += int(
        in_focus.sum()
    )

    if not in_focus.any():
        continue

    detail = pd.DataFrame({
        "area_id": area_ids.loc[
            in_focus
        ].to_numpy(),
        "complaint_category": category.loc[
            in_focus
        ].to_numpy(),
        "historical_day": days.loc[
            in_focus
        ].to_numpy()
    })

    grouped = detail.groupby([
        "area_id",
        "complaint_category",
        "historical_day"
    ]).size()

    historical_daily_counts_19_1c.update({
        (
            str(area),
            str(complaint_category),
            str(day)
        ): int(count)
        for (
            area,
            complaint_category,
            day
        ), count in grouped.items()
    })

print("Full historical source rows:", total_source_rows)
print("Eligible historical rows:", total_eligible_rows)
print("Rows in 20 focus areas:", total_focus_rows)

print(
    "Observed area-category-day combinations:",
    len(historical_daily_counts_19_1c)
)

# --------------------------------------------------
# 3. Build matched-day baseline statistics
# --------------------------------------------------

print("\n3. CONSTRUCTING MATCHED-DAY BASELINES")

baseline_rows = []

for area_id in focus_areas:

    for complaint_category in categories:

        counts = np.array([
            historical_daily_counts_19_1c.get(
                (area_id, complaint_category, day),
                0
            )
            for day in matched_days
        ], dtype=int)

        total_matched_reports = int(
            counts.sum()
        )

        active_days = int(
            (counts > 0).sum()
        )

        represented_years = sum(
            any(
                historical_daily_counts_19_1c.get(
                    (
                        area_id,
                        complaint_category,
                        day
                    ),
                    0
                ) > 0
                for day in matched_days
                if matched_day_years[day] == str(year)
            )
            for year in [2023, 2024, 2025]
        )

        sufficient_history = (
            total_matched_reports
            >= MIN_HISTORICAL_MATCHED_REPORTS
            and active_days
            >= MIN_ACTIVE_MATCHED_DAYS
            and represented_years
            >= MIN_YEARS_REPRESENTED
        )

        baseline_rows.append({
            "area_id": area_id,
            "complaint_category":
                complaint_category,
            "matched_calendar_days":
                len(matched_days),
            "matched_historical_reports":
                total_matched_reports,
            "matched_active_days":
                active_days,
            "matched_zero_days":
                int((counts == 0).sum()),
            "years_with_reports":
                represented_years,
            "historical_daily_mean":
                float(counts.mean()),
            "historical_daily_median":
                float(np.median(counts)),
            "historical_daily_p95":
                float(np.quantile(counts, 0.95)),
            "historical_daily_max":
                int(counts.max()),
            "baseline_status": (
                "exploratory_baseline_available"
                if sufficient_history
                else "insufficient_historical_support"
            )
        })

historical_baselines_19_1c = pd.DataFrame(
    baseline_rows
)

print("Baseline area-category combinations:",
      len(historical_baselines_19_1c))

display(
    historical_baselines_19_1c[
        "baseline_status"
    ]
    .value_counts()
    .rename_axis("status")
    .reset_index(name="area_category_combinations")
)

# --------------------------------------------------
# 4. Connect incoming reports to historical
#    baseline coverage (NO anomaly scores yet)
# --------------------------------------------------

incoming = pd.read_csv(
    incoming_path,
    dtype="string",
    keep_default_na=False
)

incoming["area_id"] = (
    incoming["borough"]
    .str.strip()
    .str.upper()
    + "_"
    + incoming["incident_zip"].str.strip()
)

incoming["complaint_category"] = (
    incoming["complaint_category"]
    .str.strip()
    .str.upper()
)

incoming_counts = (
    incoming.groupby([
        "area_id",
        "complaint_category"
    ])
    .size()
    .rename("incoming_reports")
    .reset_index()
)

incoming_baseline_19_1c = (
    incoming_counts.merge(
        historical_baselines_19_1c,
        on=[
            "area_id",
            "complaint_category"
        ],
        how="left",
        validate="many_to_one"
    )
)

print("\n4. INCOMING REPORTS VS HISTORICAL COVERAGE")

display(
    incoming_baseline_19_1c[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "matched_historical_reports",
        "matched_active_days",
        "historical_daily_mean",
        "historical_daily_p95",
        "baseline_status"
    ]]
    .sort_values(
        "incoming_reports",
        ascending=False
    )
    .head(25)
)

# --------------------------------------------------
# 5. Inspect sparse-history area explicitly
# --------------------------------------------------

print("\n5. SPARSE HISTORICAL COVERAGE EXAMPLE")

sparse_example = historical_baselines_19_1c.loc[
    historical_baselines_19_1c[
        "area_id"
    ].eq("QUEENS_11005")
    & historical_baselines_19_1c[
        "complaint_category"
    ].eq("SEWER")
]

display(sparse_example)

# --------------------------------------------------
# 6. Summarize baseline availability by category
# --------------------------------------------------

category_readiness_19_1c = (
    historical_baselines_19_1c
    .assign(
        has_exploratory_baseline=lambda df:
            df["baseline_status"].eq(
                "exploratory_baseline_available"
            )
    )
    .groupby("complaint_category")
    .agg(
        areas_evaluated=("area_id", "nunique"),
        areas_with_exploratory_baseline=(
            "has_exploratory_baseline",
            "sum"
        ),
        matched_historical_reports=(
            "matched_historical_reports",
            "sum"
        )
    )
    .reset_index()
)

print("\n6. CATEGORY BASELINE READINESS")

display(category_readiness_19_1c)

# --------------------------------------------------
# 7. Compare against previous historical audit
# --------------------------------------------------

previous_audit_match = None

if "reference_coverage_19_1b" in globals():

    previous = reference_coverage_19_1b[
        ["area_id", "historical_reports"]
    ].copy()

    calculated_area_totals = Counter()

    for (
        area_id,
        complaint_category,
        historical_day
    ), count in historical_daily_counts_19_1c.items():

        calculated_area_totals[area_id] += count

    current = pd.DataFrame([
        {
            "area_id": area_id,
            "historical_reports_new":
                calculated_area_totals.get(area_id, 0)
        }
        for area_id in focus_areas
    ])

    comparison = previous.merge(
        current,
        on="area_id",
        how="outer",
        validate="one_to_one"
    )

    previous_audit_match = bool(
        len(comparison) == 20
        and comparison[
            "historical_reports"
        ].eq(
            comparison[
                "historical_reports_new"
            ]
        ).all()
    )

    print("\n7. CROSS-CHECK WITH PHASE 19.1B")
    print(
        "All 20 historical area totals identical:",
        previous_audit_match
    )

# --------------------------------------------------
# 8. Validation and reconciliation
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Historical source contains 685284 rows",
    total_source_rows == 685284
)

check(
    "Eligible historical records reconcile",
    total_eligible_rows == 668440
)

check(
    "Historical calendar has 1096 days",
    len(historical_calendar) == 1096
)

check(
    "Matched historical weekdays span all three years",
    all(
        any(
            date.year == year
            for date in matched_dates
        )
        for year in [2023, 2024, 2025]
    )
)

check(
    "At least 30 matched historical days",
    len(matched_days) >= 30
)

check(
    "All 20 focus areas included",
    historical_baselines_19_1c[
        "area_id"
    ].nunique() == 20
)

check(
    "All 15 categories included",
    historical_baselines_19_1c[
        "complaint_category"
    ].nunique() == 15
)

check(
    "300 area-category baselines created",
    len(historical_baselines_19_1c) == 300
)

check(
    "Historical focus records reconcile",
    sum(
        historical_daily_counts_19_1c.values()
    ) == total_focus_rows
)

check(
    "All 500 incoming complaints accounted for",
    int(incoming_counts[
        "incoming_reports"
    ].sum()) == 500
)

check(
    "Every incoming area-category has baseline metadata",
    incoming_baseline_19_1c[
        "baseline_status"
    ].notna().all()
)

check(
    "Matched calendar days fully accounted for",
    (
        historical_baselines_19_1c[
            "matched_active_days"
        ]
        + historical_baselines_19_1c[
            "matched_zero_days"
        ]
    ).eq(len(matched_days)).all()
)

check(
    "All baseline means and percentiles finite",
    np.isfinite(
        historical_baselines_19_1c[[
            "historical_daily_mean",
            "historical_daily_p95"
        ]].to_numpy(dtype=float)
    ).all()
)

check(
    "Sparse sewer example does not get trusted baseline",
    len(sparse_example) == 1
    and sparse_example.iloc[0][
        "baseline_status"
    ] == "insufficient_historical_support"
)

if previous_audit_match is not None:
    check(
        "Area totals match previous historical audit",
        previous_audit_match
    )

file_signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected_paths
}

check(
    "No protected files modified",
    file_signatures_before == file_signatures_after
)

results_19_1c = pd.DataFrame(checks)

print("\n8. PHASE 19.1C VALIDATION")

display(results_19_1c)

passed = int(
    results_19_1c["passed"].sum()
)

total = len(results_19_1c)

print(f"\nChecks passed: {passed}/{total}")

failed = results_19_1c.loc[
    ~results_19_1c["passed"],
    "test"
].tolist()

if failed:

    print("\nSTATUS: NEEDS REVIEW")

    for failure in failed:
        print("FAILED:", failure)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 19.1C HISTORICAL DAILY BASELINES VERIFIED")
    print("WEEKDAY AND SEASONAL COMPARISON PREPARED")
    print("SPARSE HISTORICAL EVIDENCE FLAGGED")
    print("ALL 500 INCOMING COMPLAINTS COVERED")
    print("NO ANOMALY SCORES GENERATED")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

### Detect Potentially Unusual Reporting Increases
We'll now compare the November 9, 2026 synthetic complaint counts with the 39 comparable historical Mondays.

Our first anomaly detector will identify:
- Candidate unusual increase: Enough historical evidence, at least five incoming reports and a count above every observed matched historical day.
- Elevated reporting: Above the historical 95th percentile, but not above the recorded maximum.
- Insufficient historical support: Cannot reliably determine whether the count is unusual.
- Within historical range: No increase beyond the defined exploratory thresholds.

```python

from pathlib import Path
import hashlib
import numpy as np
import pandas as pd

from src.integration.daily_batch_window import (
    select_daily_complaints
)

print(" EXPLORATORY ANOMALY DETECTION")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

# --------------------------------------------------
# 1. Verify the previous baseline is available
# --------------------------------------------------

required_variables = [
    "historical_daily_counts_19_1c",
    "historical_baselines_19_1c",
    "matched_days"
]

missing = [
    name for name in required_variables
    if name not in globals()
]

assert not missing, (
    "Run Phase 19.1C first. Missing: "
    + ", ".join(missing)
)

history_counts = historical_daily_counts_19_1c

baselines = historical_baselines_19_1c.copy(
    deep=True
)

historical_days = list(matched_days)

REPORTING_DAY = "2026-11-09"
MIN_REPORTS_FOR_CANDIDATE = 5

assert len(historical_days) == 39
assert len(baselines) == 300

assert baselines[
    ["area_id", "complaint_category"]
].duplicated().sum() == 0

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

historical_path = (
    root / "data/processed"
    / "nyc311_infrastructure_cleaned.csv"
)

assert input_path.is_file()
assert historical_path.is_file()

protected = [input_path, historical_path]

signatures_before = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

input_hash_before = hashlib.sha256(
    input_path.read_bytes()
).hexdigest()

print("\n1. BASELINE EVIDENCE")

print("Historical matched days:", len(historical_days))
print("Area-category combinations:", len(baselines))

print(
    "Available exploratory baselines:",
    baselines["baseline_status"].eq(
        "exploratory_baseline_available"
    ).sum()
)

print(
    "Insufficient historical support:",
    baselines["baseline_status"].eq(
        "insufficient_historical_support"
    ).sum()
)

# --------------------------------------------------
# 2. Select only the requested NYC reporting day
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

daily = select_daily_complaints(
    source,
    REPORTING_DAY
)

selected = daily["selected_records"].copy()

assert daily["stats"]["selected_records"] == 500
assert len(selected) == 500

assert selected["processing_status"].eq(
    "processed"
).all()

assert selected["zip_validation_status"].eq(
    "zcta_consistent"
).all()

selected["area_id"] = (
    selected["borough"]
    .astype(str)
    .str.strip()
    .str.upper()
    + "_"
    + selected["incident_zip"]
    .astype(str)
    .str.strip()
)

selected["complaint_category"] = (
    selected["complaint_category"]
    .astype(str)
    .str.strip()
    .str.upper()
)

incoming_counts = (
    selected.groupby(
        ["area_id", "complaint_category"]
    )
    .size()
    .rename("incoming_reports")
    .reset_index()
)

# Do not silently discard an incoming area-category
# that has no matching historical baseline.
baseline_keys = set(zip(
    baselines["area_id"].astype(str),
    baselines["complaint_category"].astype(str)
))

incoming_keys = set(zip(
    incoming_counts["area_id"].astype(str),
    incoming_counts[
        "complaint_category"
    ].astype(str)
))

unmatched_keys = incoming_keys - baseline_keys

assert not unmatched_keys, (
    "New area-category combinations require "
    f"separate baseline handling: {unmatched_keys}"
)

# --------------------------------------------------
# 3. Construct all 300 daily comparisons
# --------------------------------------------------

anomaly_screen_19_1d = baselines.merge(
    incoming_counts,
    on=["area_id", "complaint_category"],
    how="left",
    validate="one_to_one"
)

anomaly_screen_19_1d[
    "incoming_reports"
] = (
    anomaly_screen_19_1d[
        "incoming_reports"
    ]
    .fillna(0)
    .astype(int)
)

anomaly_screen_19_1d[
    "reporting_day"
] = REPORTING_DAY

# --------------------------------------------------
# 4. Define a transparent exploratory rule
# --------------------------------------------------

def classify_daily_increase(
    incoming_count,
    baseline_status,
    historical_max,
    historical_p95,
    min_report_count=5
):

    if incoming_count == 0:
        return "no_incoming_reports"

    if baseline_status != (
        "exploratory_baseline_available"
    ):
        return "insufficient_historical_support"

    if incoming_count > historical_max:

        if incoming_count >= min_report_count:
            return "candidate_unusual_increase"

        return (
            "above_historical_max_below_report_floor"
        )

    if incoming_count > historical_p95:
        return "above_historical_p95"

    return "within_matched_historical_range"


# --------------------------------------------------
# 5. Calculate exploratory comparison evidence
# --------------------------------------------------

status_values = []
historical_tail_values = []
mean_ratio_values = []
historical_consistency = []

for row in anomaly_screen_19_1d.itertuples(
    index=False
):

    area_id = str(row.area_id)

    category = str(row.complaint_category)

    incoming_count = int(
        row.incoming_reports
    )

    history = np.array([
        history_counts.get(
            (area_id, category, day),
            0
        )
        for day in historical_days
    ], dtype=int)

    # Verify the detailed historical counts
    # against the Phase 19.1C summaries.
    consistent = (
        int(history.sum())
        == int(row.matched_historical_reports)
        and np.isclose(
            history.mean(),
            row.historical_daily_mean
        )
        and np.isclose(
            np.quantile(history, 0.95),
            row.historical_daily_p95
        )
        and int(history.max())
        == int(row.historical_daily_max)
    )

    historical_consistency.append(
        bool(consistent)
    )

    status = classify_daily_increase(
        incoming_count,
        row.baseline_status,
        row.historical_daily_max,
        row.historical_daily_p95,
        MIN_REPORTS_FOR_CANDIDATE
    )

    status_values.append(status)

    # Smoothed empirical upper-tail frequency.
    # Descriptive only: not a significance claim.
    if (
        row.baseline_status
        == "exploratory_baseline_available"
        and incoming_count > 0
    ):

        upper_tail = (
            1 + int(
                (history >= incoming_count).sum()
            )
        ) / (len(history) + 1)

    else:
        upper_tail = np.nan

    historical_tail_values.append(
        upper_tail
    )

    if row.historical_daily_mean > 0:

        mean_ratio_values.append(
            incoming_count
            / row.historical_daily_mean
        )

    else:
        mean_ratio_values.append(np.nan)

anomaly_screen_19_1d[
    "anomaly_review_status"
] = status_values

anomaly_screen_19_1d[
    "empirical_upper_tail_fraction"
] = historical_tail_values

anomaly_screen_19_1d[
    "incoming_to_historical_mean_ratio"
] = mean_ratio_values

anomaly_screen_19_1d[
    "above_historical_max"
] = (
    anomaly_screen_19_1d[
        "incoming_reports"
    ]
    > anomaly_screen_19_1d[
        "historical_daily_max"
    ]
)

anomaly_screen_19_1d[
    "above_historical_p95"
] = (
    anomaly_screen_19_1d[
        "incoming_reports"
    ]
    > anomaly_screen_19_1d[
        "historical_daily_p95"
    ]
)

# --------------------------------------------------
# 6. Summarize the complete anomaly screen
# --------------------------------------------------

print("\n2. EXPLORATORY ANOMALY SCREEN")

status_summary_19_1d = (
    anomaly_screen_19_1d.groupby(
        "anomaly_review_status"
    )
    .agg(
        area_category_combinations=(
            "area_id", "size"
        ),
        incoming_complaints=(
            "incoming_reports", "sum"
        )
    )
    .reset_index()
    .sort_values(
        "incoming_complaints",
        ascending=False
    )
)

display(status_summary_19_1d)

# --------------------------------------------------
# 7. Inspect candidate unusual increases
# --------------------------------------------------

candidate_increases_19_1d = (
    anomaly_screen_19_1d.loc[
        anomaly_screen_19_1d[
            "anomaly_review_status"
        ].eq("candidate_unusual_increase")
    ]
    .sort_values(
        [
            "incoming_reports",
            "area_id"
        ],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("\n3. CANDIDATE UNUSUAL INCREASES")

print(
    "Exploratory candidates:",
    len(candidate_increases_19_1d)
)

display(
    candidate_increases_19_1d[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "matched_historical_reports",
        "historical_daily_mean",
        "historical_daily_p95",
        "historical_daily_max",
        "empirical_upper_tail_fraction",
        "anomaly_review_status"
    ]].head(30)
)

# --------------------------------------------------
# 8. Inspect insufficient-evidence cases
# --------------------------------------------------

insufficient_19_1d = (
    anomaly_screen_19_1d.loc[
        anomaly_screen_19_1d[
            "anomaly_review_status"
        ].eq("insufficient_historical_support")
    ]
    .sort_values(
        "incoming_reports",
        ascending=False
    )
)

print("\n4. INCOMING REPORTS WITH WEAK HISTORY")

display(
    insufficient_19_1d[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "matched_historical_reports",
        "matched_active_days",
        "years_with_reports",
        "historical_daily_max",
        "anomaly_review_status"
    ]].head(20)
)

# --------------------------------------------------
# 9. Specific sewer examples from the PDF workflow
# --------------------------------------------------

print("\n5. THREE SEWER-REPORTING EXAMPLES")

example_areas = [
    "BROOKLYN_11204",
    "BRONX_10453",
    "QUEENS_11005"
]

sewer_examples = (
    anomaly_screen_19_1d.loc[
        anomaly_screen_19_1d[
            "area_id"
        ].isin(example_areas)
        & anomaly_screen_19_1d[
            "complaint_category"
        ].eq("SEWER")
    ]
)

display(
    sewer_examples[[
        "area_id",
        "incoming_reports",
        "historical_daily_mean",
        "historical_daily_max",
        "matched_historical_reports",
        "baseline_status",
        "anomaly_review_status"
    ]]
)

# --------------------------------------------------
# 10. Controlled tests of the anomaly rules
# --------------------------------------------------

supported_example = baselines.loc[
    baselines["area_id"].eq("BRONX_10453")
    & baselines["complaint_category"].eq("SEWER")
]

assert len(supported_example) == 1

reference_row = supported_example.iloc[0]

probe_max = int(
    reference_row["historical_daily_max"]
)

probe_count = max(
    5,
    probe_max + 1
)

strong_probe = classify_daily_increase(
    probe_count,
    reference_row["baseline_status"],
    probe_max,
    reference_row["historical_daily_p95"]
)

ordinary_probe = classify_daily_increase(
    probe_max,
    reference_row["baseline_status"],
    probe_max,
    reference_row["historical_daily_p95"]
)

sparse_probe = classify_daily_increase(
    100,
    "insufficient_historical_support",
    0,
    0
)

zero_probe = classify_daily_increase(
    0,
    "exploratory_baseline_available",
    10,
    5
)

# --------------------------------------------------
# 11. Verify statistical reporting safeguards
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 39 historical comparison days retained",
    len(historical_days) == 39
)

check(
    "All 300 baseline combinations retained",
    len(anomaly_screen_19_1d) == 300
)

check(
    "All 20 areas represented",
    anomaly_screen_19_1d[
        "area_id"
    ].nunique() == 20
)

check(
    "All 15 categories represented",
    anomaly_screen_19_1d[
        "complaint_category"
    ].nunique() == 15
)

check(
    "All 500 incoming records selected",
    len(selected) == 500
)

check(
    "All 500 incoming reports reconciled",
    anomaly_screen_19_1d[
        "incoming_reports"
    ].sum() == 500
)

check(
    "Every incoming category has matching baseline",
    not unmatched_keys
)

check(
    "83 exploratory baselines preserved",
    anomaly_screen_19_1d[
        "baseline_status"
    ].eq(
        "exploratory_baseline_available"
    ).sum() == 83
)

check(
    "217 insufficient baselines preserved",
    anomaly_screen_19_1d[
        "baseline_status"
    ].eq(
        "insufficient_historical_support"
    ).sum() == 217
)

check(
    "Detailed historical counts match baseline summaries",
    all(historical_consistency)
)

check(
    "Every area-category has a review status",
    anomaly_screen_19_1d[
        "anomaly_review_status"
    ].notna().all()
)

check(
    "Review-status totals reconcile",
    status_summary_19_1d[
        "area_category_combinations"
    ].sum() == 300
    and status_summary_19_1d[
        "incoming_complaints"
    ].sum() == 500
)

check(
    "Candidates require supported history",
    candidate_increases_19_1d[
        "baseline_status"
    ].eq(
        "exploratory_baseline_available"
    ).all()
)

check(
    "Candidates meet minimum report count",
    candidate_increases_19_1d[
        "incoming_reports"
    ].ge(5).all()
)

check(
    "Candidates exceed recorded historical maximum",
    candidate_increases_19_1d[
        "above_historical_max"
    ].all()
)

check(
    "Sparse sewer history never generates candidate",
    anomaly_screen_19_1d.loc[
        anomaly_screen_19_1d[
            "area_id"
        ].eq("QUEENS_11005")
        & anomaly_screen_19_1d[
            "complaint_category"
        ].eq("SEWER"),
        "anomaly_review_status"
    ].eq(
        "insufficient_historical_support"
    ).all()
)

check(
    "Controlled large increase is flagged",
    strong_probe == "candidate_unusual_increase"
)

check(
    "Count equal to historical max is not a candidate",
    ordinary_probe != "candidate_unusual_increase"
)

check(
    "Large count with sparse history remains insufficient",
    sparse_probe
    == "insufficient_historical_support"
)

check(
    "Zero incoming reports never trigger alert",
    zero_probe == "no_incoming_reports"
)

nonmissing_tail = anomaly_screen_19_1d[
    "empirical_upper_tail_fraction"
].dropna()

check(
    "Empirical tail fractions remain valid",
    nonmissing_tail.between(
        1 / 40,
        1.0
    ).all()
)

check(
    "Only supported active comparisons have tail fractions",
    anomaly_screen_19_1d.loc[
        anomaly_screen_19_1d[
            "empirical_upper_tail_fraction"
        ].notna(),
        "baseline_status"
    ].eq(
        "exploratory_baseline_available"
    ).all()
)

signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

check(
    "Historical and incoming files not modified",
    signatures_before == signatures_after
    and hashlib.sha256(
        input_path.read_bytes()
    ).hexdigest() == input_hash_before
)

# --------------------------------------------------
# 12. Final results
# --------------------------------------------------

results_19_1d = pd.DataFrame(checks)

print("\n6. PHASE 19.1D VALIDATION")

display(results_19_1d)

passed = int(
    results_19_1d["passed"].sum()
)

total = len(results_19_1d)

print(f"\nChecks passed: {passed}/{total}")

failures = results_19_1d.loc[
    ~results_19_1d["passed"],
    "test"
].tolist()

if failures:

    print("\nSTATUS: NEEDS REVIEW")

    for failure in failures:
        print("FAILED:", failure)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 19.1D EXPLORATORY ANOMALY SCREEN VERIFIED")
    print("HISTORICAL DAILY COUNTS COMPARED WITH NEW BATCH")
    print("POTENTIAL UNUSUAL INCREASES IDENTIFIED")
    print("SPARSE HISTORICAL EVIDENCE SAFELY FLAGGED")
    print("NO STATISTICAL SIGNIFICANCE CLAIMS")
    print("NO PRODUCTION ANOMALY MODEL DEPLOYED")
    print("NO HISTORICAL FILES MODIFIED")
```

### Anomaly Calibration and Stability
This phase will answer three questions: Do the same candidates remain unusual when we use more historical Mondays? How much does a minimum count of 5 versus 8 or 10 change the results? And how many alerts would this rule have produced when applied retrospectively to historical data?

We'll keep this experiment read-only and separate screening candidates from statistically validated anomalies.

```python

from pathlib import Path
import numpy as np
import pandas as pd

print(" ANOMALY THRESHOLD CALIBRATION")
print("=" * 75)

# --------------------------------------------------
# 1. Verify existing historical evidence
# --------------------------------------------------

required = [
    "historical_daily_counts_19_1c",
    "anomaly_screen_19_1d",
    "matched_days"
]

missing = [
    name for name in required
    if name not in globals()
]

assert not missing, (
    "Run Phases 19.1C and 19.1D first. "
    f"Missing: {missing}"
)

root = Path(PROJECT_ROOT).resolve()

protected_paths = [
    root / "data/processed"
    / "nyc311_infrastructure_cleaned.csv",
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
]

for path in protected_paths:
    assert path.is_file(), f"Missing: {path}"

signatures_before = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected_paths
}

screen = anomaly_screen_19_1d.copy(deep=True)
screen_original = screen.copy(deep=True)

history_counts = historical_daily_counts_19_1c

REPORTING_DAY = "2026-11-09"
REPORT_FLOORS = [5, 8, 10]

assert len(screen) == 300

keys = list(zip(
    screen["area_id"].astype(str),
    screen["complaint_category"].astype(str)
))

assert len(keys) == len(set(keys))

incoming = screen[
    "incoming_reports"
].to_numpy(dtype=int)

matched_index = pd.DatetimeIndex(
    pd.to_datetime(matched_days)
).sort_values()

target = pd.Timestamp(REPORTING_DAY)

all_mondays = pd.DatetimeIndex([
    day
    for day in pd.date_range(
        "2023-01-01",
        "2025-12-31",
        freq="D"
    )
    if day.weekday() == target.weekday()
])

print("\n1. REFERENCE WINDOW DESIGN")
print("Incoming day:", REPORTING_DAY)
print("Matched autumn Mondays:", len(matched_index))
print("All historical Mondays:", len(all_mondays))
print("Area-category comparisons:", len(keys))

# --------------------------------------------------
# 2. Reconstruct the historical count matrices
# --------------------------------------------------

def build_count_matrix(days):

    labels = [
        day.strftime("%Y-%m-%d")
        for day in days
    ]

    values = np.fromiter(
        (
            int(history_counts.get(
                (area, category, day),
                0
            ))
            for area, category in keys
            for day in labels
        ),
        dtype=np.int64,
        count=len(keys) * len(labels)
    )

    return values.reshape(
        len(keys),
        len(labels)
    )


history_39 = build_count_matrix(
    matched_index
)

history_all_mondays = build_count_matrix(
    all_mondays
)

print("\n2. HISTORICAL MATRICES")
print("Matched:", history_39.shape)
print("All Mondays:", history_all_mondays.shape)

# --------------------------------------------------
# 3. Apply the same historical support rules
# --------------------------------------------------

MIN_HISTORICAL_REPORTS = 8
MIN_ACTIVE_DAYS = 5
MIN_REPRESENTED_YEARS = 2

def evaluate_support(matrix, dates):

    historical_reports = matrix.sum(axis=1)

    active_days = (
        matrix > 0
    ).sum(axis=1)

    represented_years = np.zeros(
        len(matrix),
        dtype=int
    )

    years = np.asarray(dates.year)

    for year in np.unique(years):

        represented_years += (
            matrix[:, years == year] > 0
        ).any(axis=1).astype(int)

    supported = (
        (historical_reports >= MIN_HISTORICAL_REPORTS)
        & (active_days >= MIN_ACTIVE_DAYS)
        & (represented_years >= MIN_REPRESENTED_YEARS)
    )

    return {
        "supported": supported,
        "historical_reports": historical_reports,
        "active_days": active_days,
        "represented_years": represented_years
    }


support_39 = evaluate_support(
    history_39,
    matched_index
)

support_all = evaluate_support(
    history_all_mondays,
    all_mondays
)

max_39 = history_39.max(axis=1)
max_all = history_all_mondays.max(axis=1)

print("\n3. BASELINE COVERAGE COMPARISON")
print(
    "39-day supported comparisons:",
    int(support_39["supported"].sum())
)

print(
    "All-Monday supported comparisons:",
    int(support_all["supported"].sum())
)

# --------------------------------------------------
# 4. Compare current anomaly candidate rules
# --------------------------------------------------

def candidate_flags(
    incoming_counts,
    support,
    historical_max,
    report_floor
):

    return (
        (incoming_counts >= report_floor)
        & support
        & (incoming_counts > historical_max)
    )


original_flags = screen[
    "anomaly_review_status"
].eq("candidate_unusual_increase").to_numpy()

recreated_flags = candidate_flags(
    incoming,
    support_39["supported"],
    max_39,
    5
)

assert np.array_equal(
    original_flags,
    recreated_flags
), (
    "Original Phase 19.1D candidate rules "
    "did not reconcile."
)

sensitivity_rows = []

for window_name, support, historical_max in [
    (
        "39 seasonal Mondays",
        support_39["supported"],
        max_39
    ),
    (
        "All historical Mondays",
        support_all["supported"],
        max_all
    )
]:

    for floor in REPORT_FLOORS:

        flags = candidate_flags(
            incoming,
            support,
            historical_max,
            floor
        )

        sensitivity_rows.append({
            "reference_window": window_name,
            "minimum_incoming_reports": floor,
            "supported_comparisons": int(
                support.sum()
            ),
            "candidate_combinations": int(
                flags.sum()
            ),
            "incoming_reports_in_candidates": int(
                incoming[flags].sum()
            )
        })

sensitivity_19_1e = pd.DataFrame(
    sensitivity_rows
)

print("\n4. CURRENT CANDIDATE SENSITIVITY")

display(sensitivity_19_1e)

# --------------------------------------------------
# 5. Determine reference-window consistency
# --------------------------------------------------

all_monday_flags = candidate_flags(
    incoming,
    support_all["supported"],
    max_all,
    5
)

candidate_evidence_19_1e = (
    screen.loc[
        original_flags,
        [
            "area_id",
            "complaint_category",
            "incoming_reports",
            "matched_historical_reports",
            "historical_daily_max"
        ]
    ].copy()
)

candidate_evidence_19_1e[
    "all_monday_historical_max"
] = max_all[original_flags]

candidate_evidence_19_1e[
    "all_monday_support_available"
] = support_all[
    "supported"
][original_flags]

candidate_evidence_19_1e[
    "remains_candidate_with_all_mondays"
] = all_monday_flags[original_flags]

candidate_evidence_19_1e[
    "reference_window_consistency"
] = np.where(
    candidate_evidence_19_1e[
        "remains_candidate_with_all_mondays"
    ],
    "candidate_in_both_windows",
    "sensitive_to_reference_window"
)

candidate_evidence_19_1e = (
    candidate_evidence_19_1e
    .sort_values(
        [
            "incoming_reports",
            "area_id"
        ],
        ascending=[False, True]
    )
    .reset_index(drop=True)
)

print("\n5. ORIGINAL 22 CANDIDATE EVIDENCE REVIEW")

display(candidate_evidence_19_1e)

print("\nReference consistency summary:")

display(
    candidate_evidence_19_1e[
        "reference_window_consistency"
    ]
    .value_counts()
    .rename_axis("assessment")
    .reset_index(name="candidates")
)

# --------------------------------------------------
# 6. Retrospective historical replay
# --------------------------------------------------

# Two complementary historical tests:
#
# Leave-one-day-out:
#   Evaluate a historical day against the other
#   38 matched days.
#
# Leave-one-year-out:
#   Evaluate a historical day against matched
#   days from the other two historical years.
#
# Neither is a formal false-alarm-rate estimate
# for operational 2026 data.

print("\n6. HISTORICAL HOLDOUT REPLAY")

replay_rows = []

for day_position, heldout_day in enumerate(
    matched_index
):

    heldout_year = heldout_day.year

    test_counts = history_39[
        :, day_position
    ]

    for method in [
        "leave_one_day_out",
        "leave_one_year_out"
    ]:

        if method == "leave_one_day_out":

            train_mask = np.ones(
                len(matched_index),
                dtype=bool
            )

            train_mask[day_position] = False

        else:

            train_mask = (
                matched_index.year
                != heldout_year
            )

        train_dates = matched_index[
            train_mask
        ]

        train_matrix = history_39[
            :, train_mask
        ]

        train_evidence = evaluate_support(
            train_matrix,
            train_dates
        )

        historical_max = train_matrix.max(
            axis=1
        )

        result_row = {
            "method": method,
            "heldout_day": (
                heldout_day.strftime(
                    "%Y-%m-%d"
                )
            ),
            "heldout_year": heldout_year,
            "training_days": len(train_dates),
            "supported_comparisons": int(
                train_evidence[
                    "supported"
                ].sum()
            ),
            "heldout_year_excluded": (
                heldout_year
                not in set(
                    train_dates.year
                )
                if method
                == "leave_one_year_out"
                else True
            )
        }

        for floor in REPORT_FLOORS:

            flags = candidate_flags(
                test_counts,
                train_evidence["supported"],
                historical_max,
                floor
            )

            result_row[
                f"candidates_floor_{floor}"
            ] = int(flags.sum())

            result_row[
                f"reports_in_candidates_floor_{floor}"
            ] = int(
                test_counts[flags].sum()
            )

        replay_rows.append(result_row)

historical_replay_19_1e = pd.DataFrame(
    replay_rows
)

print("Historical replay evaluations:",
      len(historical_replay_19_1e))

display(
    historical_replay_19_1e.head(12)
)

# --------------------------------------------------
# 7. Summarize historical alert burden
# --------------------------------------------------

replay_summary_rows = []

for method in [
    "leave_one_day_out",
    "leave_one_year_out"
]:

    subset = historical_replay_19_1e.loc[
        historical_replay_19_1e[
            "method"
        ].eq(method)
    ]

    for floor in REPORT_FLOORS:

        values = subset[
            f"candidates_floor_{floor}"
        ]

        replay_summary_rows.append({
            "method": method,
            "minimum_reports": floor,
            "historical_test_days": len(values),
            "median_candidate_count":
                float(values.median()),
            "p95_candidate_count":
                float(values.quantile(0.95)),
            "maximum_candidate_count":
                int(values.max()),
            "days_with_candidates":
                int(values.gt(0).sum()),
            "total_candidate_flags":
                int(values.sum())
        })

historical_replay_summary_19_1e = (
    pd.DataFrame(replay_summary_rows)
)

print("\n7. HISTORICAL REPLAY SUMMARY")

display(
    historical_replay_summary_19_1e
)

# --------------------------------------------------
# 8. Compare current screen with replayed history
# --------------------------------------------------

current_candidate_count = int(
    original_flags.sum()
)

print("\n8. CURRENT VS HISTORICAL REPLAY")

comparison_rows = []

for method in [
    "leave_one_day_out",
    "leave_one_year_out"
]:

    subset = historical_replay_19_1e.loc[
        historical_replay_19_1e[
            "method"
        ].eq(method)
    ]

    replay_counts = subset[
        "candidates_floor_5"
    ]

    comparison_rows.append({
        "method": method,
        "current_synthetic_candidates":
            current_candidate_count,
        "historical_replay_maximum":
            int(replay_counts.max()),
        "historical_days_with_at_least_"
        "current_candidate_count": int(
            replay_counts.ge(
                current_candidate_count
            ).sum()
        ),
        "historical_days_tested":
            len(subset)
    })

current_vs_replay_19_1e = pd.DataFrame(
    comparison_rows
)

display(current_vs_replay_19_1e)

print(
    "\nInterpretation: retrospective comparison "
    "only. It does not establish a predictive "
    "false-alarm probability or significance."
)

# --------------------------------------------------
# 9. Sparse-history safeguard
# --------------------------------------------------

print("\n9. SPARSE HISTORY SAFEGUARD")

sparse_mask = (
    screen["area_id"].eq("QUEENS_11005")
    & screen["complaint_category"].eq("SEWER")
)

assert sparse_mask.sum() == 1

sparse_index = np.flatnonzero(
    sparse_mask.to_numpy()
)[0]

sparse_result = pd.DataFrame([{
    "area_id": "QUEENS_11005",
    "complaint_category": "SEWER",
    "incoming_reports":
        int(incoming[sparse_index]),
    "matched_reports":
        int(history_39[
            sparse_index
        ].sum()),
    "39_day_support":
        bool(support_39[
            "supported"
        ][sparse_index]),
    "all_monday_support":
        bool(support_all[
            "supported"
        ][sparse_index]),
    "original_review_status":
        screen.iloc[sparse_index][
            "anomaly_review_status"
        ]
}])

display(sparse_result)

# --------------------------------------------------
# 10. Validation checks
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "300 area-category comparisons preserved",
    len(screen) == 300
)

check(
    "All 500 synthetic incoming reports reconciled",
    incoming.sum() == 500
)

check(
    "39 matched historical Mondays preserved",
    len(matched_index) == 39
)

check(
    "All historical Mondays available",
    len(all_mondays) >= 150
)

check(
    "Matched dates are included in all Mondays",
    set(matched_index).issubset(
        set(all_mondays)
    )
)

check(
    "Historical 39-day report totals reconcile",
    np.array_equal(
        history_39.sum(axis=1),
        screen[
            "matched_historical_reports"
        ].to_numpy(dtype=int)
    )
)

check(
    "Historical daily maxima reconcile",
    np.array_equal(
        max_39,
        screen[
            "historical_daily_max"
        ].to_numpy(dtype=int)
    )
)

check(
    "Historical 95th percentiles reconcile",
    np.allclose(
        np.quantile(
            history_39, 0.95, axis=1
        ),
        screen[
            "historical_daily_p95"
        ].to_numpy(dtype=float)
    )
)

check(
    "Original baseline eligibility reproduced",
    np.array_equal(
        support_39["supported"],
        screen["baseline_status"].eq(
            "exploratory_baseline_available"
        ).to_numpy()
    )
)

check(
    "Original 22 candidates reproduced",
    current_candidate_count == 22
    and np.array_equal(
        original_flags,
        recreated_flags
    )
)

check(
    "Original 251 candidate complaints reproduced",
    incoming[original_flags].sum() == 251
)

check(
    "All-Monday window has no fewer matched observations",
    len(all_mondays) > len(matched_index)
)

check(
    "Additional reference days do not reduce observed maxima",
    np.all(max_all >= max_39)
)

check(
    "Additional Mondays preserve previously sufficient evidence",
    np.all(
        support_all["supported"][
            support_39["supported"]
        ]
    )
)

check(
    "All six current sensitivity configurations evaluated",
    len(sensitivity_19_1e) == 6
)

floor_order_valid = True

for window in sensitivity_19_1e[
    "reference_window"
].unique():

    values = (
        sensitivity_19_1e.loc[
            sensitivity_19_1e[
                "reference_window"
            ].eq(window)
        ]
        .sort_values(
            "minimum_incoming_reports"
        )["candidate_combinations"]
        .tolist()
    )

    floor_order_valid &= (
        values[0] >= values[1] >= values[2]
    )

check(
    "Raising report floor never increases candidate count",
    floor_order_valid
)

check(
    "All 22 original candidates reviewed for window consistency",
    len(candidate_evidence_19_1e) == 22
)

check(
    "Cross-window candidates remain original candidates",
    int(all_monday_flags[
        original_flags
    ].sum()) <= 22
)

check(
    "39 matched days evaluated in both replay methods",
    len(historical_replay_19_1e) == 78
    and historical_replay_19_1e.groupby(
        "method"
    ).size().eq(39).all()
)

check(
    "Year-held-out tests use only other historical years",
    historical_replay_19_1e.loc[
        historical_replay_19_1e[
            "method"
        ].eq("leave_one_year_out"),
        "heldout_year_excluded"
    ].all()
)

check(
    "Every replay has enough reference days",
    historical_replay_19_1e[
        "training_days"
    ].ge(20).all()
)

check(
    "Replay candidates are nonnegative and finite",
    historical_replay_19_1e[[
        f"candidates_floor_{floor}"
        for floor in REPORT_FLOORS
    ]].ge(0).all().all()
)

check(
    "Historical replay candidate thresholds are monotonic",
    (
        historical_replay_19_1e[
            "candidates_floor_5"
        ]
        >= historical_replay_19_1e[
            "candidates_floor_8"
        ]
    ).all()
    and (
        historical_replay_19_1e[
            "candidates_floor_8"
        ]
        >= historical_replay_19_1e[
            "candidates_floor_10"
        ]
    ).all()
)

check(
    "Sparse sewer history remains unsupported",
    not bool(
        support_39["supported"][
            sparse_index
        ]
    )
    and screen.iloc[sparse_index][
        "anomaly_review_status"
    ] == "insufficient_historical_support"
)

check(
    "Previous anomaly screen unchanged",
    anomaly_screen_19_1d.equals(
        screen_original
    )
)

signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected_paths
}

check(
    "Historical and incoming source files unchanged",
    signatures_before == signatures_after
)

# --------------------------------------------------
# 11. Final assessment
# --------------------------------------------------

results_19_1e = pd.DataFrame(checks)

print("\n10. PHASE 19.1E VALIDATION")

display(results_19_1e)

passed = int(
    results_19_1e["passed"].sum()
)

total = len(results_19_1e)

print(f"\nChecks passed: {passed}/{total}")

failures = results_19_1e.loc[
    ~results_19_1e["passed"],
    "test"
].tolist()

if failures:

    print("\nSTATUS: NEEDS REVIEW")

    for failure in failures:
        print("FAILED:", failure)

else:

    print("\nSTATUS: PASSED")
    print("PHASE 19.1E ANOMALY CALIBRATION VERIFIED")
    print("HISTORICAL REFERENCE SENSITIVITY TESTED")
    print("REPORT COUNT THRESHOLDS COMPARED")
    print("RETROSPECTIVE HISTORICAL REPLAY COMPLETED")
    print("MULTIPLE-COMPARISON RISK RETAINED AS LIMITATION")
    print("NO STATISTICAL SIGNIFICANCE CLAIMS")
    print("NO PRODUCTION ANOMALY MODEL DEPLOYED")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

### Create the Reusable Anomaly Screening Engine

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import hashlib
import sys
import pandas as pd

print(" REUSABLE ANOMALY SCREENING ENGINE")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

module_path = (
    root / "src/models/new_batch_anomaly_screen.py"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

window_path = (
    root / "src/integration/daily_batch_window.py"
)

for path in [input_path, taxonomy_path, window_path]:
    assert path.is_file(), f"Missing: {path}"

assert "historical_daily_counts_19_1c" in globals(), (
    "Run Phase 19.1C to load historical daily counts."
)

history_counts = historical_daily_counts_19_1c

historical_calendar_19_2a = pd.date_range(
    "2023-01-01",
    "2025-12-31",
    freq="D"
)

supported_categories_19_2a = sorted({
    str(x).strip().upper()
    for x in pd.read_csv(taxonomy_path)[
        "complaint_category"
    ].dropna()
})

historical_area_ids_19_2a = sorted({
    str(area)
    for area, category, day
    in history_counts.keys()
})

assert len(historical_area_ids_19_2a) == 20

protected = [input_path, taxonomy_path, window_path]

original_hashes = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Define reusable anomaly screening module
# --------------------------------------------------

module_source = dedent(r'''
"""Exploratory historical-versus-incoming complaint screen.

Uses weekday-matched daily count references.
Screening candidates are not confirmed anomalies,
statistical significance claims, or repair priorities.

Historical area coverage must be explicitly supplied.
"""

import re
import numpy as np
import pandas as pd

from src.integration.daily_batch_window import (
    select_daily_complaints
)


VALID_BOROUGHS = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}


def _window_evidence(
    daily_counts,
    area_id,
    category,
    dates,
    min_historical_reports,
    min_active_days,
    min_years
):
    labels = dates.strftime("%Y-%m-%d")

    values = np.array([
        int(daily_counts.get(
            (area_id, category, day),
            0
        ))
        for day in labels
    ], dtype=np.int64)

    active = values > 0

    years = len(set(
        dates.year[active].tolist()
    ))

    total = int(values.sum())
    active_days = int(active.sum())

    supported = (
        total >= min_historical_reports
        and active_days >= min_active_days
        and years >= min_years
    )

    return {
        "calendar_days": len(values),
        "reports": total,
        "active_days": active_days,
        "years_with_reports": years,
        "supported": bool(supported),
        "mean": (
            float(values.mean())
            if len(values) else 0.0
        ),
        "p95": (
            float(np.quantile(values, 0.95))
            if len(values) else 0.0
        ),
        "maximum": (
            int(values.max())
            if len(values) else 0
        )
    }


def screen_daily_anomalies(
    records,
    supported_categories,
    historical_daily_counts,
    historical_calendar,
    historical_area_ids,
    reporting_day,
    min_incoming_reports=8,
    min_historical_reports=8,
    min_active_days=5,
    min_years=2
):
    """Return exploratory anomaly evidence and row audit.

    Only processed, ZIP-consistent complaints
    belonging to the selected NYC reporting day
    enter the area/category count comparison.
    """

    required = {
        "source_row",
        "complaint_id",
        "created_date",
        "processing_status",
        "zip_validation_status",
        "borough",
        "incident_zip",
        "complaint_category"
    }

    if not isinstance(records, pd.DataFrame):
        raise TypeError("records must be a DataFrame")

    if records.columns.duplicated().any():
        raise ValueError("Duplicate column names")

    missing = required - set(records.columns)

    if missing:
        raise ValueError(
            f"Missing columns: {sorted(missing)}"
        )

    if min_incoming_reports < 1:
        raise ValueError(
            "min_incoming_reports must be positive"
        )

    categories = {
        str(x).strip().upper()
        for x in supported_categories
    }

    known_areas = {
        str(x).strip().upper()
        for x in historical_area_ids
    }

    if not re.fullmatch(
        r"\d{4}-\d{2}-\d{2}",
        str(reporting_day)
    ):
        raise ValueError("Use YYYY-MM-DD")

    daily = select_daily_complaints(
        records,
        reporting_day
    )

    selected = daily[
        "selected_records"
    ].copy().reset_index(drop=True)

    borough = (
        selected["borough"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    zip_code = (
        selected["incident_zip"]
        .astype(str)
        .str.strip()
    )

    category = (
        selected["complaint_category"]
        .astype(str)
        .str.strip()
        .str.upper()
    )

    processed = selected[
        "processing_status"
    ].astype(str).eq("processed")

    zip_consistent = selected[
        "zip_validation_status"
    ].astype(str).eq("zcta_consistent")

    valid_borough = borough.isin(
        VALID_BOROUGHS
    )

    valid_zip = zip_code.str.fullmatch(
        r"\d{5}",
        na=False
    )

    supported = category.isin(
        categories
    )

    # Order of conditions gives a single
    # auditable reason per excluded record.
    reasons = np.select(
        [
            ~processed,
            ~zip_consistent,
            ~valid_borough,
            ~valid_zip,
            ~supported
        ],
        [
            "not_processed",
            "zip_not_verified",
            "invalid_borough",
            "invalid_zip",
            "unsupported_category"
        ],
        default="eligible"
    )

    selected["anomaly_input_status"] = reasons

    selected["area_id"] = (
        borough + "_" + zip_code
    )

    selected["complaint_category"] = category

    eligible = selected.loc[
        selected["anomaly_input_status"].eq(
            "eligible"
        )
    ].copy()

    incoming_counts = (
        eligible.groupby(
            ["area_id", "complaint_category"]
        )
        .size()
        .rename("incoming_reports")
        .reset_index()
    )

    # Do not include historical reference days
    # that occur on or after the reporting day.
    dates = pd.DatetimeIndex(
        pd.to_datetime(
            list(historical_calendar)
        )
    ).normalize().unique().sort_values()

    if dates.tz is not None:
        raise ValueError(
            "Historical calendar must use local dates"
        )

    target = pd.Timestamp(reporting_day)

    all_weekdays = dates[
        (dates < target)
        & (dates.weekday == target.weekday())
    ]

    # Three-month comparison window centered
    # on the reporting month, wrapping years.
    seasonal_months = {
        ((target.month - 2) % 12) + 1,
        target.month,
        (target.month % 12) + 1
    }

    seasonal = all_weekdays[
        np.isin(
            all_weekdays.month,
            list(seasonal_months)
        )
    ]

    rows = []

    for item in incoming_counts.itertuples(
        index=False
    ):

        area = str(item.area_id)
        complaint_category = str(
            item.complaint_category
        )

        count = int(item.incoming_reports)

        seasonal_ev = _window_evidence(
            historical_daily_counts,
            area,
            complaint_category,
            seasonal,
            min_historical_reports,
            min_active_days,
            min_years
        )

        broad_ev = _window_evidence(
            historical_daily_counts,
            area,
            complaint_category,
            all_weekdays,
            min_historical_reports,
            min_active_days,
            min_years
        )

        area_known = area in known_areas

        # Raw candidate flags are retained solely
        # for audit and threshold comparison.
        seasonal_raw_5 = (
            area_known
            and seasonal_ev["supported"]
            and count >= 5
            and count > seasonal_ev["maximum"]
        )

        broad_raw_5 = (
            area_known
            and broad_ev["supported"]
            and count >= 5
            and count > broad_ev["maximum"]
        )

        seasonal_above_max = (
            count > seasonal_ev["maximum"]
        )

        broad_above_max = (
            count > broad_ev["maximum"]
        )

        if not area_known:

            review_status = (
                "no_historical_area_coverage"
            )

        elif not seasonal_ev["supported"]:

            review_status = (
                "insufficient_historical_support"
            )

        elif (
            count >= min_incoming_reports
            and broad_ev["supported"]
            and seasonal_above_max
            and broad_above_max
        ):

            review_status = (
                "cross_window_consistent_candidate"
            )

        elif (
            count >= min_incoming_reports
            and seasonal_above_max
        ):

            review_status = (
                "reference_window_sensitive_review"
            )

        elif seasonal_above_max:

            review_status = (
                "above_seasonal_max_below_report_floor"
            )

        elif count > seasonal_ev["p95"]:

            review_status = "above_seasonal_p95"

        else:

            review_status = (
                "within_matched_historical_range"
            )

        rows.append({
            "reporting_day": reporting_day,
            "area_id": area,
            "complaint_category":
                complaint_category,
            "incoming_reports": count,
            "historical_area_covered": area_known,
            "seasonal_days":
                seasonal_ev["calendar_days"],
            "seasonal_historical_reports":
                seasonal_ev["reports"],
            "seasonal_active_days":
                seasonal_ev["active_days"],
            "seasonal_years_with_reports":
                seasonal_ev["years_with_reports"],
            "seasonal_support":
                seasonal_ev["supported"],
            "seasonal_daily_mean":
                seasonal_ev["mean"],
            "seasonal_daily_p95":
                seasonal_ev["p95"],
            "seasonal_daily_max":
                seasonal_ev["maximum"],
            "all_weekday_days":
                broad_ev["calendar_days"],
            "all_weekday_support":
                broad_ev["supported"],
            "all_weekday_daily_max":
                broad_ev["maximum"],
            "seasonal_candidate_floor_5":
                bool(seasonal_raw_5),
            "all_weekday_candidate_floor_5":
                bool(broad_raw_5),
            "anomaly_review_status":
                review_status
        })

    comparisons = pd.DataFrame(rows)

    # Ensure a stable schema when input is empty.
    if comparisons.empty:
        comparisons = pd.DataFrame(columns=[
            "reporting_day",
            "area_id",
            "complaint_category",
            "incoming_reports",
            "historical_area_covered",
            "seasonal_days",
            "seasonal_historical_reports",
            "seasonal_active_days",
            "seasonal_years_with_reports",
            "seasonal_support",
            "seasonal_daily_mean",
            "seasonal_daily_p95",
            "seasonal_daily_max",
            "all_weekday_days",
            "all_weekday_support",
            "all_weekday_daily_max",
            "seasonal_candidate_floor_5",
            "all_weekday_candidate_floor_5",
            "anomaly_review_status"
        ])

    quality_audit = selected[[
        "source_row",
        "anomaly_input_status"
    ]]

    eligible_review = (
        eligible[[
            "source_row",
            "area_id",
            "complaint_category"
        ]]
        .merge(
            comparisons[[
                "area_id",
                "complaint_category",
                "anomaly_review_status"
            ]],
            on=[
                "area_id",
                "complaint_category"
            ],
            how="left",
            validate="many_to_one"
        )[[
            "source_row",
            "anomaly_review_status"
        ]]
    )

    record_audit = (
        daily["audit"]
        .merge(
            quality_audit,
            on="source_row",
            how="left",
            validate="one_to_one"
        )
        .merge(
            eligible_review,
            on="source_row",
            how="left",
            validate="one_to_one"
        )
    )

    record_audit[
        "anomaly_input_status"
    ] = record_audit[
        "anomaly_input_status"
    ].fillna("not_in_reporting_day")

    record_audit[
        "anomaly_review_status"
    ] = record_audit[
        "anomaly_review_status"
    ].fillna("not_analyzed")

    candidates = comparisons.loc[
        comparisons["anomaly_review_status"].eq(
            "cross_window_consistent_candidate"
        )
    ].copy()

    stats = {
        "input_records": int(len(records)),
        "selected_records": int(len(selected)),
        "date_exclusions": int(
            len(records) - len(selected)
        ),
        "eligible_records": int(len(eligible)),
        "quality_exclusions": int(
            len(selected) - len(eligible)
        ),
        "active_area_category_combinations":
            int(len(comparisons)),
        "candidate_combinations":
            int(len(candidates)),
        "candidate_complaints": (
            int(candidates[
                "incoming_reports"
            ].sum())
            if not candidates.empty else 0
        ),
        "seasonal_reference_days":
            int(len(seasonal)),
        "all_weekday_reference_days":
            int(len(all_weekdays))
    }

    if (
        stats["date_exclusions"]
        + stats["quality_exclusions"]
        + stats["eligible_records"]
        != stats["input_records"]
    ):
        raise AssertionError(
            "Input coverage did not reconcile"
        )

    if (
        int(comparisons[
            "incoming_reports"
        ].sum()) != len(eligible)
    ):
        raise AssertionError(
            "Incoming comparison counts disagree"
        )

    if len(record_audit) != len(records):
        raise AssertionError(
            "Record audit lost source records"
        )

    return {
        "comparisons": comparisons,
        "candidates": candidates,
        "record_audit": record_audit,
        "stats": stats,
        "parameters": {
            "reporting_day": reporting_day,
            "min_incoming_reports":
                min_incoming_reports,
            "min_historical_reports":
                min_historical_reports,
            "min_active_days": min_active_days,
            "min_years": min_years,
            "seasonal_months":
                sorted(seasonal_months),
            "screening_only": True
        }
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

# Execute in memory before creating any file.
namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

screen_daily_anomalies = namespace[
    "screen_daily_anomalies"
]

# --------------------------------------------------
# 2. Test the complete synthetic batch
# --------------------------------------------------

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

source_snapshot = source.copy(deep=True)

def run_screen(frame, floor=8):

    return screen_daily_anomalies(
        records=frame,
        supported_categories=
            supported_categories_19_2a,
        historical_daily_counts=history_counts,
        historical_calendar=
            historical_calendar_19_2a,
        historical_area_ids=
            historical_area_ids_19_2a,
        reporting_day="2026-11-09",
        min_incoming_reports=floor
    )

print("\n1. CALIBRATED ANOMALY SCREEN")

with threadpool_limits(limits=1):
    result = run_screen(source)

comparisons = result["comparisons"]
candidates = result["candidates"]

print(result["stats"])

display(
    comparisons[
        "anomaly_review_status"
    ]
    .value_counts()
    .rename_axis("review_status")
    .reset_index(name="area_category_combinations")
)

# --------------------------------------------------
# 3. Inspect cross-window candidates
# --------------------------------------------------

print("\n2. CROSS-WINDOW CANDIDATES")

display(
    candidates[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "seasonal_historical_reports",
        "seasonal_daily_mean",
        "seasonal_daily_max",
        "all_weekday_daily_max",
        "anomaly_review_status"
    ]]
    .sort_values(
        "incoming_reports",
        ascending=False
    )
)

# --------------------------------------------------
# 4. Inspect records requiring extra review
# --------------------------------------------------

print("\n3. REFERENCE-SENSITIVE OR WEAK-HISTORY CASES")

review_cases = comparisons.loc[
    comparisons["anomaly_review_status"].isin([
        "reference_window_sensitive_review",
        "insufficient_historical_support",
        "no_historical_area_coverage"
    ])
]

display(
    review_cases[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "seasonal_support",
        "all_weekday_support",
        "anomaly_review_status"
    ]]
    .sort_values(
        "incoming_reports",
        ascending=False
    )
    .head(20)
)

# --------------------------------------------------
# 5. Compare report-count thresholds
# --------------------------------------------------

with threadpool_limits(limits=1):

    floor_5 = run_screen(source, floor=5)
    floor_10 = run_screen(source, floor=10)

print("\n4. THRESHOLD COMPARISON")

threshold_summary = pd.DataFrame([
    {
        "minimum_reports": 5,
        "candidates":
            floor_5["stats"][
                "candidate_combinations"
            ],
        "candidate_complaints":
            floor_5["stats"][
                "candidate_complaints"
            ]
    },
    {
        "minimum_reports": 8,
        "candidates":
            result["stats"][
                "candidate_combinations"
            ],
        "candidate_complaints":
            result["stats"][
                "candidate_complaints"
            ]
    },
    {
        "minimum_reports": 10,
        "candidates":
            floor_10["stats"][
                "candidate_combinations"
            ],
        "candidate_complaints":
            floor_10["stats"][
                "candidate_complaints"
            ]
    }
])

display(threshold_summary)

# --------------------------------------------------
# 6. Test mixed-day and exclusion safeguards
# --------------------------------------------------

mixed = source.copy()

mixed.loc[[0, 1], "created_date"] = (
    "2026-11-10"
)

mixed.loc[[2, 3], "created_date"] = (
    "invalid-date"
)

mixed.loc[
    4, "created_date"
] = "2026-11-09T01:00:00+00:00"

with threadpool_limits(limits=1):
    mixed_result = run_screen(mixed)

unverified = source.copy()

unverified.loc[
    0, "processing_status"
] = "geographic_review_required"

with threadpool_limits(limits=1):
    unverified_result = run_screen(unverified)

unknown = source.iloc[[0]].copy()

# Controlled uncovered-area test only;
# this is not a real ZIP-geography verification.
unknown["incident_zip"] = "99999"

with threadpool_limits(limits=1):
    unknown_result = run_screen(unknown)

empty_result = run_screen(
    source.iloc[:0].copy()
)

# --------------------------------------------------
# 7. Validation checks
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 synthetic records retained",
    result["stats"]["input_records"] == 500
)

check(
    "All 500 belong to selected reporting day",
    result["stats"]["selected_records"] == 500
)

check(
    "All 500 inputs eligible for screening",
    result["stats"]["eligible_records"] == 500
)

check(
    "39 seasonal Mondays reconstructed",
    result["stats"][
        "seasonal_reference_days"
    ] == 39
)

check(
    "157 historical Mondays reconstructed",
    result["stats"][
        "all_weekday_reference_days"
    ] == 157
)

check(
    "100 active area-category combinations retained",
    len(comparisons) == 100
)

check(
    "Original 22 seasonal candidates reproduced",
    int(
        comparisons[
            "seasonal_candidate_floor_5"
        ].sum()
    ) == 22
)

check(
    "Original 23 all-Monday candidates reproduced",
    int(
        comparisons[
            "all_weekday_candidate_floor_5"
        ].sum()
    ) == 23
)

check(
    "20 original candidates agree across windows",
    int((
        comparisons[
            "seasonal_candidate_floor_5"
        ]
        & comparisons[
            "all_weekday_candidate_floor_5"
        ]
    ).sum()) == 20
)

check(
    "Minimum-five rule produces 20 consistent candidates",
    floor_5["stats"][
        "candidate_combinations"
    ] == 20
)

check(
    "Calibrated minimum-eight rule produces 15 candidates",
    result["stats"][
        "candidate_combinations"
    ] == 15
)

check(
    "15 calibrated candidates contain 204 reports",
    result["stats"][
        "candidate_complaints"
    ] == 204
)

check(
    "Every candidate has at least eight incoming reports",
    candidates[
        "incoming_reports"
    ].ge(8).all()
)

check(
    "Every candidate has supported seasonal history",
    candidates[
        "seasonal_support"
    ].all()
)

check(
    "Every candidate has supported all-Monday history",
    candidates[
        "all_weekday_support"
    ].all()
)

check(
    "Every candidate exceeds both observed historical maxima",
    (
        candidates["incoming_reports"]
        > candidates["seasonal_daily_max"]
    ).all()
    and (
        candidates["incoming_reports"]
        > candidates["all_weekday_daily_max"]
    ).all()
)

check(
    "Raising report threshold cannot add candidates",
    floor_5["stats"]["candidate_combinations"]
    >= result["stats"]["candidate_combinations"]
    >= floor_10["stats"][
        "candidate_combinations"
    ]
)

sparse = comparisons.loc[
    comparisons["area_id"].eq(
        "QUEENS_11005"
    )
    & comparisons["complaint_category"].eq(
        "SEWER"
    )
]

check(
    "Sparse Queens sewer history does not generate candidate",
    len(sparse) == 1
    and sparse.iloc[0][
        "anomaly_review_status"
    ] == "insufficient_historical_support"
)

sensitive = comparisons.loc[
    comparisons["area_id"].eq(
        "BROOKLYN_11203"
    )
    & comparisons["complaint_category"].eq(
        "STREET CONDITION"
    )
]

check(
    "Reference-sensitive street reports are not automatic candidates",
    len(sensitive) == 1
    and sensitive.iloc[0][
        "anomaly_review_status"
    ] == "reference_window_sensitive_review"
)

check(
    "Mixed-day input retains all original records in audit",
    len(mixed_result["record_audit"]) == 500
)

check(
    "Mixed-day input selects 495 reports",
    mixed_result["stats"][
        "selected_records"
    ] == 495
)

check(
    "Five mixed-day records excluded from screening",
    mixed_result["stats"][
        "date_exclusions"
    ] == 5
)

check(
    "Unverified geography excluded from anomaly counts",
    unverified_result["stats"][
        "eligible_records"
    ] == 499
    and unverified_result["stats"][
        "quality_exclusions"
    ] == 1
)

check(
    "Uncovered area cannot generate an anomaly candidate",
    len(unknown_result["comparisons"]) == 1
    and unknown_result["comparisons"].iloc[0][
        "anomaly_review_status"
    ] == "no_historical_area_coverage"
)

check(
    "Empty incoming batch handled safely",
    empty_result["stats"][
        "input_records"
    ] == 0
    and empty_result[
        "comparisons"
    ].empty
)

check(
    "Record-level audit preserves source order",
    result["record_audit"][
        "source_row"
    ].tolist()
    == source["source_row"].tolist()
)

check(
    "Original source DataFrame unchanged",
    source.equals(source_snapshot)
)

check(
    "Existing input and daily-window module unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == original_hashes[path]
        for path in protected
    )
)

results_19_2a = pd.DataFrame(checks)

print("\n5. PHASE 19.2A VALIDATION")

display(results_19_2a)

passed = int(results_19_2a["passed"].sum())
total = len(results_19_2a)

print(f"\nChecks passed: {passed}/{total}")

failed = results_19_2a.loc[
    ~results_19_2a["passed"], "test"
].tolist()

# --------------------------------------------------
# 8. Save only after successful validation
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No anomaly module saved.")

    for failure in failed:
        print("FAILED:", failure)

else:

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == module_source, (
            "Existing anomaly module differs. "
            "No overwrite performed."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        module_path.write_text(
            module_source,
            encoding="utf-8"
        )

    print("\n6. VERIFIED MODULE")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 19.2A ANOMALY SCREEN ENGINE VERIFIED")
    print("DUAL HISTORICAL WINDOWS SUPPORTED")
    print("SPARSE HISTORY SAFEGUARD VERIFIED")
    print("DATE AND GEOGRAPHIC SAFEGUARDS VERIFIED")
    print("NO STATISTICAL SIGNIFICANCE CLAIMS")
    print("NO AUTOMATIC REPAIR PRIORITIES")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

```python

from pathlib import Path
from collections import Counter
import hashlib
import json
import sys

import numpy as np
import pandas as pd
from threadpoolctl import threadpool_limits

print("FULL HISTORICAL BASELINE SNAPSHOT")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.models.new_batch_anomaly_screen import (
    screen_daily_anomalies
)

historical_path = (
    root / "data/processed"
    / "nyc311_infrastructure_cleaned.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

incoming_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

reference_path = (
    root / "outputs/tests"
    / "phase_17_4b_diverse"
    / "synthetic_area_pattern_reference.csv"
)

anomaly_module_path = (
    root / "src/models"
    / "new_batch_anomaly_screen.py"
)

output_dir = (
    root / "outputs/tables/anomaly"
)

snapshot_path = (
    output_dir
    / "historical_daily_counts_2023_2025_v1.csv.gz"
)

manifest_path = (
    output_dir
    / "historical_daily_counts_2023_2025_v1_manifest.json"
)

protected = [
    historical_path,
    taxonomy_path,
    incoming_path,
    reference_path,
    anomaly_module_path
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

signatures_before = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

# --------------------------------------------------
# 1. Historical coverage configuration
# --------------------------------------------------

START_DAY = "2023-01-01"
END_DAY = "2025-12-31"
TEST_DAY = "2026-11-09"

calendar_19_2b = pd.date_range(
    START_DAY,
    END_DAY,
    freq="D"
)

taxonomy = pd.read_csv(taxonomy_path)

categories_19_2b = sorted({
    str(value).strip().upper()
    for value in taxonomy[
        "complaint_category"
    ].dropna()
})

reference = pd.read_csv(
    reference_path,
    dtype="string",
    keep_default_na=False
)

focus_areas = set(
    reference["area_id"].astype(str)
)

valid_boroughs = {
    "BRONX",
    "BROOKLYN",
    "MANHATTAN",
    "QUEENS",
    "STATEN ISLAND"
}

print("\n1. SNAPSHOT CONFIGURATION")
print("Historical period:", START_DAY, "to", END_DAY)
print("Calendar days:", len(calendar_19_2b))
print("Supported categories:", len(categories_19_2b))
print("Existing test areas:", len(focus_areas))

# --------------------------------------------------
# 2. Construct counts for ALL historical areas
# --------------------------------------------------

all_daily_counts_19_2b = Counter()

source_rows = 0
eligible_rows = 0
invalid_dates = 0
offset_timestamps = 0

eligible_by_year = Counter()
eligible_by_day = Counter()

columns = [
    "created_date",
    "complaint_type",
    "borough",
    "incident_zip"
]

reader = pd.read_csv(
    historical_path,
    usecols=columns,
    dtype="string",
    keep_default_na=False,
    chunksize=60000,
    encoding="utf-8-sig"
)

print("\n2. SCANNING HISTORICAL COMPLAINTS")

for chunk_number, chunk in enumerate(
    reader, start=1
):

    source_rows += len(chunk)

    raw_dates = (
        chunk["created_date"]
        .fillna("")
        .astype("string")
        .str.strip()
    )

    has_offset = raw_dates.str.contains(
        r"(?:[Zz]|[+-]\d{2}:?\d{2})$",
        regex=True,
        na=False
    )

    offset_timestamps += int(
        has_offset.sum()
    )

    iso_dates = raw_dates.str.extract(
        r"^(\d{4}-\d{2}-\d{2})(?:$|[ T])",
        expand=False
    )

    us_dates = raw_dates.str.extract(
        r"^(\d{1,2}/\d{1,2}/\d{4})(?:$|[ T])",
        expand=False
    )

    parsed_iso = pd.to_datetime(
        iso_dates,
        format="%Y-%m-%d",
        errors="coerce"
    )

    parsed_us = pd.to_datetime(
        us_dates,
        format="%m/%d/%Y",
        errors="coerce"
    )

    days = (
        parsed_iso
        .fillna(parsed_us)
        .dt.strftime("%Y-%m-%d")
        .fillna("")
    )

    valid_date = (
        days.ne("")
        & ~has_offset
    )

    invalid_dates += int(
        (~valid_date).sum()
    )

    within_period = (
        valid_date
        & days.ge(START_DAY)
        & days.le(END_DAY)
    )

    category = (
        chunk["complaint_type"]
        .fillna("")
        .astype("string")
        .str.strip()
        .str.upper()
    )

    borough = (
        chunk["borough"]
        .fillna("")
        .astype("string")
        .str.strip()
        .str.upper()
    )

    zip_code = (
        chunk["incident_zip"]
        .fillna("")
        .astype("string")
        .str.strip()
    )

    valid_area = (
        borough.isin(valid_boroughs)
        & zip_code.str.fullmatch(
            r"\d{5}",
            na=False
        )
    )

    eligible = (
        within_period
        & category.isin(categories_19_2b)
        & valid_area
    )

    eligible_rows += int(
        eligible.sum()
    )

    if not eligible.any():
        continue

    area_ids = borough + "_" + zip_code

    detail = pd.DataFrame({
        "area_id": area_ids.loc[
            eligible
        ].to_numpy(),
        "complaint_category": category.loc[
            eligible
        ].to_numpy(),
        "historical_day": days.loc[
            eligible
        ].to_numpy()
    })

    grouped = detail.groupby(
        [
            "area_id",
            "complaint_category",
            "historical_day"
        ],
        sort=False
    ).size()

    all_daily_counts_19_2b.update({
        (
            str(area),
            str(complaint_category),
            str(day)
        ): int(count)
        for (
            area,
            complaint_category,
            day
        ), count in grouped.items()
    })

    eligible_by_year.update(
        detail["historical_day"]
        .str[:4]
        .value_counts()
        .to_dict()
    )

    eligible_by_day.update(
        detail["historical_day"]
        .value_counts()
        .to_dict()
    )

print("Historical source rows:", source_rows)
print("Eligible complaint rows:", eligible_rows)
print("Distinct daily-count entries:",
      len(all_daily_counts_19_2b))
print("Days with eligible reports:",
      len(eligible_by_day))

# --------------------------------------------------
# 3. Create the compact sparse snapshot
# --------------------------------------------------

snapshot_19_2b = pd.DataFrame(
    [
        (area, category, day, count)
        for (
            area,
            category,
            day
        ), count in all_daily_counts_19_2b.items()
    ],
    columns=[
        "area_id",
        "complaint_category",
        "historical_day",
        "complaint_count"
    ]
)

snapshot_19_2b = (
    snapshot_19_2b
    .sort_values([
        "area_id",
        "complaint_category",
        "historical_day"
    ])
    .reset_index(drop=True)
)

snapshot_19_2b[
    "complaint_count"
] = snapshot_19_2b[
    "complaint_count"
].astype("int64")

historical_areas_19_2b = sorted(
    snapshot_19_2b["area_id"].unique()
)

print("\n3. FULL SNAPSHOT SUMMARY")
print("Historical area IDs:",
      len(historical_areas_19_2b))
print("Supported categories represented:",
      snapshot_19_2b[
          "complaint_category"
      ].nunique())
print("Snapshot records:",
      len(snapshot_19_2b))

display(snapshot_19_2b.head(12))

# --------------------------------------------------
# 4. Historical area-coverage summary
# --------------------------------------------------

area_coverage_19_2b = (
    snapshot_19_2b.groupby("area_id")
    .agg(
        historical_reports=(
            "complaint_count",
            "sum"
        ),
        active_days=(
            "historical_day",
            "nunique"
        ),
        observed_categories=(
            "complaint_category",
            "nunique"
        )
    )
    .reset_index()
)

area_coverage_19_2b[
    "mean_reports_per_calendar_day"
] = (
    area_coverage_19_2b[
        "historical_reports"
    ] / len(calendar_19_2b)
)

print("\n4. AREA COVERAGE EXAMPLES")

display(
    area_coverage_19_2b.loc[
        area_coverage_19_2b[
            "area_id"
        ].isin(focus_areas)
    ].sort_values(
        "area_id"
    ).head(20)
)

# --------------------------------------------------
# 5. Reconcile against the previous 20-area audit
# --------------------------------------------------

previous_counts_match = None

if "historical_daily_counts_19_1c" in globals():

    pilot_counts = (
        historical_daily_counts_19_1c
    )

    new_focus_counts = {
        key: int(count)
        for key, count
        in all_daily_counts_19_2b.items()
        if key[0] in focus_areas
    }

    previous_counts_match = (
        new_focus_counts
        == dict(pilot_counts)
    )

print("\n5. PREVIOUS BASELINE RECONCILIATION")

print(
    "20-area daily counts match Phase 19.1C:",
    previous_counts_match
)

# --------------------------------------------------
# 6. Test existing anomaly engine using ALL areas
# --------------------------------------------------

print("\n6. FULL-SNAPSHOT ANOMALY REGRESSION")

incoming = pd.read_csv(
    incoming_path,
    dtype="string",
    keep_default_na=False
)

with threadpool_limits(limits=1):

    full_snapshot_screen_19_2b = (
        screen_daily_anomalies(
            records=incoming,
            supported_categories=categories_19_2b,
            historical_daily_counts=
                all_daily_counts_19_2b,
            historical_calendar=
                calendar_19_2b,
            historical_area_ids=
                historical_areas_19_2b,
            reporting_day=TEST_DAY,
            min_incoming_reports=8
        )
    )

screen_stats = (
    full_snapshot_screen_19_2b["stats"]
)

print(screen_stats)

display(
    full_snapshot_screen_19_2b[
        "candidates"
    ][[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "seasonal_daily_max",
        "all_weekday_daily_max",
        "anomaly_review_status"
    ]].sort_values(
        "incoming_reports",
        ascending=False
    )
)

# --------------------------------------------------
# 7. Create reproducible snapshot metadata
# --------------------------------------------------

print("\n7. SOURCE FINGERPRINT")

source_digest = hashlib.sha256()

with historical_path.open("rb") as file:

    for block in iter(
        lambda: file.read(4 * 1024 * 1024),
        b""
    ):
        source_digest.update(block)

manifest_19_2b = {
    "snapshot_version": 1,
    "source_path": str(
        historical_path.relative_to(root)
    ).replace("\\", "/"),
    "source_sha256": source_digest.hexdigest(),
    "historical_start": START_DAY,
    "historical_end": END_DAY,
    "calendar_days": len(calendar_19_2b),
    "source_rows": int(source_rows),
    "eligible_rows": int(eligible_rows),
    "area_count": len(historical_areas_19_2b),
    "category_count": len(categories_19_2b),
    "sparse_daily_entries": len(snapshot_19_2b),
    "schema": [
        "area_id",
        "complaint_category",
        "historical_day",
        "complaint_count"
    ],
    "zero_count_policy": (
        "Missing area-category-day records represent "
        "zero observed complaints within the historical "
        "calendar; sparse categories require an "
        "independent support check."
    ),
    "date_policy": (
        "Source date prefixes interpreted as NYC-local "
        "calendar dates; timezone-offset timestamps "
        "are not accepted by this snapshot builder."
    ),
    "area_policy": (
        "Borough and five-digit ZIP syntax validated; "
        "not a point-in-polygon geographic verification."
    ),
    "intended_use": (
        "Exploratory area-category anomaly screening; "
        "not confirmed anomalies or repair priorities."
    )
}

manifest_text = json.dumps(
    manifest_19_2b,
    indent=2,
    ensure_ascii=False
) + "\n"

print(
    "Source SHA-256:",
    manifest_19_2b["source_sha256"][:20] + "..."
)

# --------------------------------------------------
# 8. Validation checks before saving
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 685284 historical records scanned",
    source_rows == 685284
)

check(
    "668440 eligible historical records reproduced",
    eligible_rows == 668440
)

check(
    "All 1096 historical calendar days included",
    len(calendar_19_2b) == 1096
)

check(
    "All historical calendar days have citywide reports",
    len(eligible_by_day) == 1096
)

check(
    "All 238 historical area IDs represented",
    len(historical_areas_19_2b) == 238
)

check(
    "All 15 supported categories represented",
    snapshot_19_2b[
        "complaint_category"
    ].nunique() == 15
)

check(
    "Historical daily counts are positive integers",
    snapshot_19_2b[
        "complaint_count"
    ].gt(0).all()
    and pd.api.types.is_integer_dtype(
        snapshot_19_2b[
            "complaint_count"
        ]
    )
)

check(
    "No duplicate daily-count keys",
    not snapshot_19_2b.duplicated(
        [
            "area_id",
            "complaint_category",
            "historical_day"
        ]
    ).any()
)

check(
    "Snapshot counts sum to eligible history",
    int(
        snapshot_19_2b[
            "complaint_count"
        ].sum()
    ) == 668440
)

check(
    "All three historical years have eligible records",
    all(
        eligible_by_year[str(year)] > 0
        for year in [2023, 2024, 2025]
    )
)

check(
    "Historical year totals reproduce prior audit",
    dict(eligible_by_year) == {
        "2023": 218362,
        "2024": 220152,
        "2025": 229926
    }
)

check(
    "No invalid dates encountered",
    invalid_dates == 0
)

check(
    "No timezone-offset timestamps silently interpreted",
    offset_timestamps == 0
)

check(
    "All 20 synthetic test areas have historical coverage",
    focus_areas.issubset(
        set(historical_areas_19_2b)
    )
)

check(
    "20-area detailed historical counts reconcile",
    previous_counts_match is True
)

check(
    "Snapshot covers same 63123 focus-area reports",
    int(
        snapshot_19_2b.loc[
            snapshot_19_2b[
                "area_id"
            ].isin(focus_areas),
            "complaint_count"
        ].sum()
    ) == 63123
)

check(
    "All 500 incoming synthetic records retained",
    screen_stats["input_records"] == 500
)

check(
    "Full snapshot retains 100 active comparisons",
    screen_stats[
        "active_area_category_combinations"
    ] == 100
)

check(
    "Existing anomaly results reproduce 15 candidates",
    screen_stats["candidate_combinations"] == 15
)

check(
    "Existing anomaly results reproduce 204 complaints",
    screen_stats["candidate_complaints"] == 204
)

check(
    "All 500 incoming records remain auditable",
    len(
        full_snapshot_screen_19_2b[
            "record_audit"
        ]
    ) == 500
)

sparse = full_snapshot_screen_19_2b[
    "comparisons"
].loc[
    lambda df:
        df["area_id"].eq("QUEENS_11005")
        & df["complaint_category"].eq("SEWER")
]

check(
    "Sparse Queens sewer case remains unsupported",
    len(sparse) == 1
    and sparse.iloc[0][
        "anomaly_review_status"
    ] == "insufficient_historical_support"
)

check(
    "Snapshot metadata is valid JSON",
    json.loads(manifest_text)[
        "eligible_rows"
    ] == 668440
)

signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

check(
    "Historical sources and model files unchanged",
    signatures_before == signatures_after
)

results_19_2b = pd.DataFrame(checks)

print("\n8. PHASE 19.2B VALIDATION")

display(results_19_2b)

passed = int(
    results_19_2b["passed"].sum()
)

total = len(results_19_2b)

print(f"\nChecks passed: {passed}/{total}")

failed = results_19_2b.loc[
    ~results_19_2b["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 9. Save only if all validation checks pass
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new snapshot files saved.")

    for failure in failed:
        print("FAILED:", failure)

else:

    # Existing files are never overwritten.
    # Compare their contents before proceeding.
    if snapshot_path.exists():

        existing = pd.read_csv(
            snapshot_path,
            compression="gzip",
            dtype={
                "area_id": "string",
                "complaint_category": "string",
                "historical_day": "string",
                "complaint_count": "int64"
            }
        )

        expected = snapshot_19_2b.copy()

        for column in [
            "area_id",
            "complaint_category",
            "historical_day"
        ]:
            expected[column] = expected[
                column
            ].astype("string")

        assert existing.equals(expected), (
            "Existing snapshot differs. "
            "No files overwritten."
        )

    if manifest_path.exists():

        assert manifest_path.read_text(
            encoding="utf-8"
        ) == manifest_text, (
            "Existing manifest differs. "
            "No files overwritten."
        )

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    if not snapshot_path.exists():

        snapshot_19_2b.to_csv(
            snapshot_path,
            index=False,
            compression={
                "method": "gzip",
                "compresslevel": 6,
                "mtime": 0
            }
        )

    if not manifest_path.exists():

        manifest_path.write_text(
            manifest_text,
            encoding="utf-8"
        )

    print("\n9. VERIFIED SNAPSHOT FILES")

    print("Daily-count snapshot:", snapshot_path)
    print("Provenance manifest:", manifest_path)

    print(
        "Compressed snapshot size (MB):",
        round(
            snapshot_path.stat().st_size
            / (1024 ** 2),
            2
        )
    )

    print("\nSTATUS: PASSED")
    print("PHASE 19.2B FULL HISTORICAL SNAPSHOT VERIFIED")
    print("238 HISTORICAL AREAS PRESERVED")
    print("15 COMPLAINT CATEGORIES PRESERVED")
    print("2023–2025 DAILY COUNTS PRESERVED")
    print("15 ANOMALY CANDIDATES REPRODUCED")
    print("SPARSE HISTORY SAFEGUARDS PRESERVED")
    print("ORIGINAL HISTORICAL FILES UNMODIFIED")
    print("NO STATISTICAL SIGNIFICANCE CLAIMS")
```

```python

from pathlib import Path
import hashlib
import json
import pandas as pd

print(" CATEGORY COVERAGE CORRECTION")
print("=" * 75)

# --------------------------------------------------
# 1. Require previously verified Phase 19.2B data
# --------------------------------------------------

required = [
    "snapshot_19_2b",
    "categories_19_2b",
    "historical_areas_19_2b",
    "manifest_19_2b",
    "results_19_2b",
    "full_snapshot_screen_19_2b",
    "snapshot_path",
    "manifest_path",
    "protected",
    "signatures_before"
]

missing_variables = [
    name for name in required
    if name not in globals()
]

assert not missing_variables, (
    "Run Phase 19.2B first. Missing variables: "
    f"{missing_variables}"
)

snapshot = snapshot_19_2b.copy(deep=True)

supported_categories = set(
    map(str, categories_19_2b)
)

observed_categories = set(
    snapshot[
        "complaint_category"
    ].astype(str)
)

missing_categories = sorted(
    supported_categories - observed_categories
)

unexpected_categories = sorted(
    observed_categories - supported_categories
)

print("\n1. CATEGORY COVERAGE DIAGNOSIS")

print("Taxonomy-supported categories:",
      len(supported_categories))

print("Observed eligible categories:",
      len(observed_categories))

print("Absent after area eligibility:",
      missing_categories)

print("Unexpected categories:",
      unexpected_categories)

# --------------------------------------------------
# 2. Compare with earlier original-source audit
# --------------------------------------------------

coverage_rows = []

original_counts = {}

if "primary_audit_19_1b" in globals():

    original_counts = dict(
        primary_audit_19_1b[
            "category_totals"
        ]
    )

eligible_category_counts = (
    snapshot.groupby(
        "complaint_category"
    )["complaint_count"]
    .sum()
    .to_dict()
)

for category in sorted(supported_categories):

    eligible_count = int(
        eligible_category_counts.get(
            category, 0
        )
    )

    original_count = original_counts.get(
        category
    )

    coverage_rows.append({
        "complaint_category": category,
        "source_historical_reports": (
            int(original_count)
            if original_count is not None
            else None
        ),
        "eligible_area_reports":
            eligible_count,
        "coverage_status": (
            "represented"
            if eligible_count > 0
            else "no_eligible_area_records"
        )
    })

category_coverage_19_2b1 = pd.DataFrame(
    coverage_rows
)

print("\n2. ALL SUPPORTED CATEGORY COVERAGE")

display(category_coverage_19_2b1)

print("\n3. MISSING CATEGORY EXPLANATION")

display(
    category_coverage_19_2b1.loc[
        category_coverage_19_2b1[
            "coverage_status"
        ].eq("no_eligible_area_records")
    ]
)

print(
    "\nNote: Source counts include records before "
    "borough-ZIP eligibility filtering. An absent "
    "eligible category must not be interpreted as "
    "proof that no real complaints occurred."
)

# --------------------------------------------------
# 3. Correct the provenance manifest
# --------------------------------------------------

corrected_manifest_19_2b1 = dict(
    manifest_19_2b
)

# Retain the original category_count field,
# explicitly defining what it represents.
corrected_manifest_19_2b1[
    "category_count"
] = len(supported_categories)

corrected_manifest_19_2b1[
    "category_count_semantics"
] = "taxonomy_supported_categories"

corrected_manifest_19_2b1[
    "supported_category_count"
] = len(supported_categories)

corrected_manifest_19_2b1[
    "observed_eligible_category_count"
] = len(observed_categories)

corrected_manifest_19_2b1[
    "supported_categories"
] = sorted(supported_categories)

corrected_manifest_19_2b1[
    "observed_eligible_categories"
] = sorted(observed_categories)

corrected_manifest_19_2b1[
    "categories_without_eligible_area_records"
] = missing_categories

corrected_manifest_19_2b1[
    "category_coverage"
] = [
    {
        "complaint_category":
            row["complaint_category"],
        "eligible_area_reports":
            int(row["eligible_area_reports"]),
        "coverage_status":
            row["coverage_status"]
    }
    for row in coverage_rows
]

corrected_manifest_19_2b1[
    "zero_count_policy"
] = (
    "Missing area-category-day keys represent zero "
    "observed ELIGIBLE complaint records within "
    "the defined historical calendar. They do not "
    "prove zero actual complaints or complete "
    "geographic coverage. Sparse-history safeguards "
    "must be applied before anomaly screening."
)

corrected_manifest_19_2b1[
    "category_coverage_limitation"
] = (
    "The taxonomy supports 15 categories, but "
    "not every category has historical records "
    "passing borough-ZIP eligibility filtering. "
    "Do not manufacture zero-count complaint rows "
    "or classify absent categories as confirmed "
    "unusual increases."
)

manifest_text = json.dumps(
    corrected_manifest_19_2b1,
    indent=2,
    ensure_ascii=False,
    allow_nan=False
) + "\n"

# --------------------------------------------------
# 4. Correct validation, not the underlying data
# --------------------------------------------------

previous = results_19_2b.copy()

failed_before = previous.loc[
    ~previous["passed"],
    "test"
].tolist()

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Original 23 noncategory checks passed",
    len(previous) == 24
    and int(previous["passed"].sum()) == 23
)

check(
    "Only previous failure was category representation",
    failed_before
    == ["All 15 supported categories represented"]
)

check(
    "Exactly 15 categories supported by taxonomy",
    len(supported_categories) == 15
)

check(
    "14 categories observed in eligible records",
    len(observed_categories) == 14
)

check(
    "Exactly one supported category absent",
    len(missing_categories) == 1
)

check(
    "No category exists outside supported taxonomy",
    len(unexpected_categories) == 0
)

check(
    "No synthetic zero-count rows inserted",
    snapshot["complaint_count"].gt(0).all()
)

check(
    "All 668440 eligible reports preserved",
    int(
        snapshot["complaint_count"].sum()
    ) == 668440
)

check(
    "All 238 historical geographic areas preserved",
    len(historical_areas_19_2b) == 238
)

check(
    "All 1096 historical dates have eligible reports",
    snapshot[
        "historical_day"
    ].nunique() == 1096
)

check(
    "Daily-count snapshot keys remain unique",
    not snapshot.duplicated([
        "area_id",
        "complaint_category",
        "historical_day"
    ]).any()
)

check(
    "All 15 taxonomy categories documented",
    len(corrected_manifest_19_2b1[
        "supported_categories"
    ]) == 15
)

check(
    "Observed category count documented correctly",
    corrected_manifest_19_2b1[
        "observed_eligible_category_count"
    ] == 14
)

check(
    "Missing category explicitly documented",
    corrected_manifest_19_2b1[
        "categories_without_eligible_area_records"
    ] == missing_categories
)

check(
    "Missing category has no eligible snapshot records",
    all(
        int(eligible_category_counts.get(
            category, 0
        )) == 0
        for category in missing_categories
    )
)

check(
    "Original anomaly screen still has 15 candidates",
    full_snapshot_screen_19_2b[
        "stats"
    ]["candidate_combinations"] == 15
)

check(
    "Original anomaly screen still covers 204 reports",
    full_snapshot_screen_19_2b[
        "stats"
    ]["candidate_complaints"] == 204
)

check(
    "Manifest JSON is valid",
    json.loads(manifest_text)[
        "eligible_rows"
    ] == 668440
)

signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

check(
    "No protected historical or model files changed",
    signatures_before == signatures_after
)

validation_19_2b1 = pd.DataFrame(checks)

print("\n4. CORRECTED VALIDATION")

display(validation_19_2b1)

passed = int(
    validation_19_2b1["passed"].sum()
)

total = len(validation_19_2b1)

print(f"\nChecks passed: {passed}/{total}")

failures = validation_19_2b1.loc[
    ~validation_19_2b1["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 5. Save only after validation succeeds
# --------------------------------------------------

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No snapshot or manifest saved.")

    for name in failures:
        print("FAILED:", name)

else:

    # Check both destinations BEFORE writing.
    # Existing files will never be overwritten.

    if snapshot_path.exists():

        existing_snapshot = pd.read_csv(
            snapshot_path,
            compression="gzip",
            dtype={
                "area_id": "string",
                "complaint_category": "string",
                "historical_day": "string",
                "complaint_count": "int64"
            }
        )

        expected = snapshot.copy()

        for column in [
            "area_id",
            "complaint_category",
            "historical_day"
        ]:
            expected[column] = (
                expected[column].astype("string")
            )

        assert existing_snapshot.equals(expected), (
            "Existing snapshot differs. "
            "No files overwritten."
        )

    if manifest_path.exists():

        assert manifest_path.read_text(
            encoding="utf-8"
        ) == manifest_text, (
            "Existing manifest differs. "
            "No files overwritten."
        )

    snapshot_path.parent.mkdir(
        parents=True,
        exist_ok=True
    )

    if not snapshot_path.exists():

        snapshot.to_csv(
            snapshot_path,
            index=False,
            mode="x",
            compression={
                "method": "gzip",
                "compresslevel": 6,
                "mtime": 0
            }
        )

    if not manifest_path.exists():

        with manifest_path.open(
            "x",
            encoding="utf-8"
        ) as file:
            file.write(manifest_text)

    # Verify saved data can be reopened.
    loaded = pd.read_csv(
        snapshot_path,
        compression="gzip"
    )

    saved_manifest = json.loads(
        manifest_path.read_text(
            encoding="utf-8"
        )
    )

    assert len(loaded) == len(snapshot)

    assert int(
        loaded["complaint_count"].sum()
    ) == 668440

    assert saved_manifest[
        "supported_category_count"
    ] == 15

    assert saved_manifest[
        "observed_eligible_category_count"
    ] == 14

    print("\n5. VERIFIED SNAPSHOT FILES")

    print("Snapshot:", snapshot_path)
    print("Manifest:", manifest_path)

    print(
        "Compressed size (MB):",
        round(
            snapshot_path.stat().st_size
            / (1024 ** 2),
            2
        )
    )

    print("\nSTATUS: PASSED")
    print("PHASE 19.2B.1 CATEGORY COVERAGE CORRECTED")
    print("FULL HISTORICAL SNAPSHOT SAVED")
    print("238 AREAS PRESERVED")
    print("15 SUPPORTED CATEGORIES DOCUMENTED")
    print("14 HISTORICALLY OBSERVED ELIGIBLE CATEGORIES")
    print("MISSING CATEGORY EXPLICITLY DOCUMENTED")
    print("15 ANOMALY CANDIDATES REPRODUCED")
    print("204 CANDIDATE COMPLAINTS REPRODUCED")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

```python

from pathlib import Path
from threadpoolctl import threadpool_limits
import hashlib
import pandas as pd

print("PANDAS NULLABLE DTYPE CORRECTION")
print("=" * 75)

# --------------------------------------------------
# 1. Recover the module from the previous cell
# --------------------------------------------------

required_variables = [
    "module_source",
    "module_path",
    "baseline_19_2c",
    "source"
]

missing = [
    name for name in required_variables
    if name not in globals()
]

assert not missing, (
    "Previous Phase 19.2C variables missing: "
    f"{missing}"
)

module_path = Path(module_path)

old_code = """    fused.loc[
        ~eligible_mask,
        evidence_columns
    ] = pd.NA"""

new_code = """    # Pandas nullable dtypes are required because
    # excluded complaints must have missing evidence.
    # Ordinary numpy bool/int columns cannot reliably
    # accept pd.NA through DataFrame assignment.

    boolean_columns = [
        "anomaly_seasonal_support",
        "anomaly_broad_support"
    ]

    integer_columns = [
        "anomaly_area_category_reports",
        "anomaly_seasonal_max",
        "anomaly_broad_max"
    ]

    float_columns = [
        "anomaly_historical_daily_mean"
    ]

    for column in boolean_columns:
        fused[column] = fused[column].astype(
            "boolean"
        )

    for column in integer_columns:
        fused[column] = pd.to_numeric(
            fused[column],
            errors="coerce"
        ).astype("Int64")

    for column in float_columns:
        fused[column] = pd.to_numeric(
            fused[column],
            errors="coerce"
        ).astype("Float64")

    # Assign one column at a time to avoid
    # mixed-dtype block assignment problems.
    excluded_mask = ~eligible_mask

    if excluded_mask.any():
        for column in evidence_columns:
            fused.loc[
                excluded_mask, column
            ] = pd.NA"""

assert module_source.count(old_code) == 1, (
    "The expected assignment was not found "
    "exactly once. No patch applied."
)

patched_source_19_2c1 = (
    module_source.replace(
        old_code,
        new_code,
        1
    )
)

compile(
    patched_source_19_2c1,
    str(module_path),
    "exec"
)

print("\n1. PATCH COMPILED SUCCESSFULLY")
print("Original problematic assignment replaced.")
print("Nullable Boolean, Int64 and Float64 supported.")

# --------------------------------------------------
# 2. Test the patched module in memory
# --------------------------------------------------

namespace = {}

exec(
    compile(
        patched_source_19_2c1,
        str(module_path),
        "exec"
    ),
    namespace
)

analyze_fixed = namespace[
    "analyze_batch_with_anomaly_context"
]

load_fixed = namespace[
    "load_anomaly_snapshot"
]

baseline = baseline_19_2c
source_before = source.copy(deep=True)

DAY = "2026-11-09"

print("\n2. FULL 500-RECORD INTEGRATION")

with threadpool_limits(limits=1):
    full_fixed = analyze_fixed(
        source,
        baseline,
        DAY
    )

for name, value in full_fixed["stats"].items():
    print(f"{name}: {value}")

fused = full_fixed["combined_records"]
candidates = full_fixed["anomaly_candidates"]

print("\n3. CANDIDATE SUMMARY")

display(
    candidates[[
        "area_id",
        "complaint_category",
        "incoming_reports",
        "linked_kmeans_patterns",
        "dbscan_candidate_complaints",
        "seasonal_daily_max",
        "all_weekday_daily_max"
    ]].sort_values(
        "incoming_reports",
        ascending=False
    )
)

# --------------------------------------------------
# 3. Verify the original error is resolved
# --------------------------------------------------

print("\n4. NULLABLE EVIDENCE COLUMN TYPES")

evidence_columns = [
    "anomaly_area_category_reports",
    "anomaly_seasonal_support",
    "anomaly_historical_daily_mean",
    "anomaly_seasonal_max",
    "anomaly_broad_support",
    "anomaly_broad_max"
]

dtype_audit = pd.DataFrame([
    {
        "column": column,
        "dtype": str(fused[column].dtype)
    }
    for column in evidence_columns
])

display(dtype_audit)

# --------------------------------------------------
# 4. Mixed reporting-day test
# --------------------------------------------------

mixed = source.copy()

mixed.loc[
    [0, 1], "created_date"
] = "2026-11-10"

mixed.loc[
    [2, 3], "created_date"
] = "invalid-date"

mixed.loc[
    4, "created_date"
] = "2026-11-09T01:00:00+00:00"

print("\n5. MIXED-DAY SAFEGUARD")

with threadpool_limits(limits=1):
    mixed_fixed = analyze_fixed(
        mixed,
        baseline,
        DAY
    )

mixed_fused = mixed_fixed[
    "combined_records"
]

excluded = mixed_fused.loc[
    mixed_fused[
        "daily_window_status"
    ].ne("selected_day")
]

print("Selected records:",
      mixed_fixed["stats"]["selected_records"])

print("Excluded records:", len(excluded))

print(
    "Excluded evidence fully missing:",
    bool(
        excluded[
            evidence_columns
        ].isna().all().all()
    )
)

# --------------------------------------------------
# 5. Geographic review exclusion
# --------------------------------------------------

print("\n6. GEOGRAPHIC REVIEW SAFEGUARD")

quality_test = source.copy()

quality_test.loc[
    0, "processing_status"
] = "geographic_review_required"

with threadpool_limits(limits=1):
    quality_fixed = analyze_fixed(
        quality_test,
        baseline,
        DAY
    )

quality_fused = quality_fixed[
    "combined_records"
]

quality_excluded = quality_fused.loc[
    quality_fused[
        "anomaly_input_status"
    ].ne("eligible")
]

print(
    "Anomaly eligible records:",
    quality_fixed[
        "stats"
    ]["anomaly_eligible_records"]
)

print(
    "Geographically excluded records:",
    len(quality_excluded)
)

# --------------------------------------------------
# 6. Small and empty batch regression
# --------------------------------------------------

print("\n7. SMALL AND EMPTY BATCH SAFEGUARDS")

area_ids = (
    source["borough"].astype(str)
    + "_"
    + source["incident_zip"].astype(str)
)

first_area = area_ids.iloc[0]

small_source = source.loc[
    area_ids.eq(first_area)
].copy()

with threadpool_limits(limits=1):

    small_fixed = analyze_fixed(
        small_source,
        baseline,
        DAY
    )

    empty_fixed = analyze_fixed(
        source.iloc[:0].copy(),
        baseline,
        DAY
    )

print(
    "Small batch K-Means:",
    small_fixed["stats"]["kmeans_status"]
)

print(
    "Small batch DBSCAN records:",
    small_fixed[
        "kmeans_dbscan_output"
    ]["dbscan_output"]["stats"]["total_records"]
)

print(
    "Empty batch records:",
    empty_fixed["stats"]["input_records"]
)

# --------------------------------------------------
# 7. Validation
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Patched source compiles",
    len(patched_source_19_2c1) > 0
)

check(
    "All 500 records preserved",
    len(fused) == 500
)

check(
    "Original record identities retained",
    fused["source_row"].tolist()
    == source["source_row"].tolist()
)

check(
    "500 records selected for correct reporting day",
    full_fixed["stats"]["selected_records"] == 500
)

check(
    "500 K-Means assignments retained",
    full_fixed["stats"][
        "kmeans_assigned_complaints"
    ] == 500
)

check(
    "Three K-Means groups retained",
    fused["new_kmeans_cluster"].nunique() == 3
)

check(
    "23 spatial candidates retained",
    full_fixed["stats"][
        "dbscan_reportable_candidates"
    ] == 23
)

check(
    "228 DBSCAN candidate complaints retained",
    full_fixed["stats"][
        "dbscan_candidate_complaints"
    ] == 228
)

check(
    "15 anomaly candidate groups retained",
    len(candidates) == 15
)

check(
    "204 anomaly candidate complaints retained",
    int(fused["anomaly_candidate"].sum()) == 204
)

check(
    "Candidate counts reconcile",
    int(candidates[
        "incoming_reports"
    ].sum()) == 204
)

check(
    "All candidates have area-pattern context",
    candidates[
        "linked_kmeans_patterns"
    ].fillna("").str.len().gt(0).all()
)

check(
    "Anomaly comparison counts do not exceed group sizes",
    candidates[
        "dbscan_candidate_complaints"
    ].le(
        candidates["incoming_reports"]
    ).all()
)

check(
    "Historical DBSCAN context unchanged",
    fused["dbscan_status"].tolist()
    == source["dbscan_status"].tolist()
)

check(
    "Historical K-Means context unchanged",
    fused["kmeans_reporting_profile"].tolist()
    == source["kmeans_reporting_profile"].tolist()
)

check(
    "Seasonal support has nullable boolean dtype",
    str(
        fused["anomaly_seasonal_support"].dtype
    ) == "boolean"
)

check(
    "Broad support has nullable boolean dtype",
    str(
        fused["anomaly_broad_support"].dtype
    ) == "boolean"
)

check(
    "Count and maximum columns are nullable integers",
    all(
        str(fused[column].dtype) == "Int64"
        for column in [
            "anomaly_area_category_reports",
            "anomaly_seasonal_max",
            "anomaly_broad_max"
        ]
    )
)

check(
    "Historical mean has nullable float dtype",
    str(
        fused[
            "anomaly_historical_daily_mean"
        ].dtype
    ) == "Float64"
)

check(
    "Mixed input retains 500 records",
    len(mixed_fused) == 500
)

check(
    "Mixed input selects 495 records",
    mixed_fixed["stats"]["selected_records"] == 495
)

check(
    "Five date exclusions preserved",
    len(excluded) == 5
)

check(
    "Date-excluded records have no anomaly candidates",
    not excluded["anomaly_candidate"].any()
)

check(
    "Date-excluded evidence is missing",
    excluded[evidence_columns].isna().all().all()
)

check(
    "Geographic review excludes one record",
    quality_fixed["stats"][
        "anomaly_eligible_records"
    ] == 499
)

check(
    "Geographically excluded record cannot be a candidate",
    len(quality_excluded) == 1
    and not quality_excluded[
        "anomaly_candidate"
    ].any()
)

check(
    "Geographically excluded evidence is missing",
    quality_excluded[
        evidence_columns
    ].isna().all().all()
)

check(
    "Single-area K-Means declines fitting safely",
    small_fixed["stats"][
        "kmeans_status"
    ] == "insufficient_areas"
)

check(
    "Single-area DBSCAN still processes 25 reports",
    small_fixed[
        "kmeans_dbscan_output"
    ]["dbscan_output"]["stats"][
        "total_records"
    ] == 25
)

check(
    "Empty input handled safely",
    empty_fixed["stats"]["input_records"] == 0
    and empty_fixed["combined_records"].empty
)

check(
    "Original input DataFrame unchanged",
    source.equals(source_before)
)

# --------------------------------------------------
# 8. Confirm historical and model files untouched
# --------------------------------------------------

root = Path(PROJECT_ROOT).resolve()

protected_paths = [
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1.csv.gz",
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1_manifest.json",
    root / "src/models/new_batch_area_kmeans.py",
    root / "src/models/new_batch_dbscan.py",
    root / "src/models/new_batch_anomaly_screen.py",
    root / "src/integration/combined_batch_intelligence.py",
    root / "src/integration/daily_batch_window.py"
]

all_protected_unchanged = all(
    path.is_file()
    and hashlib.sha256(
        path.read_bytes()
    ).hexdigest() == original_hashes[path]
    for path in protected_paths
)

check(
    "Historical snapshot and existing models unchanged",
    all_protected_unchanged
)

results_19_2c1 = pd.DataFrame(checks)

print("\n8. CORRECTIVE VALIDATION")

display(results_19_2c1)

passed = int(results_19_2c1["passed"].sum())
total = len(results_19_2c1)

print(f"\nChecks passed: {passed}/{total}")

failures = results_19_2c1.loc[
    ~results_19_2c1["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 9. Save corrected module only if tests pass
# --------------------------------------------------

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No corrected module saved.")

    for failure in failures:
        print("FAILED:", failure)

else:

    if module_path.exists():

        assert module_path.read_text(
            encoding="utf-8"
        ) == patched_source_19_2c1, (
            "Existing module has different content. "
            "It will not be overwritten."
        )

    else:

        module_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with module_path.open(
            "x",
            encoding="utf-8"
        ) as file:
            file.write(
                patched_source_19_2c1
            )

    print("\n9. VERIFIED CORRECTED MODULE")
    print(module_path)

    print("\nSTATUS: PASSED")
    print("PHASE 19.2C.1 THREE-MODEL INTEGRATION VERIFIED")
    print("PANDAS NULLABLE DTYPE ERROR CORRECTED")
    print("K-MEANS RESULTS PRESERVED")
    print("DBSCAN RESULTS PRESERVED")
    print("ANOMALY RESULTS PRESERVED")
    print("DATE AND GEOGRAPHY EXCLUSIONS VERIFIED")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import hashlib
import json
import sys
import pandas as pd
import numpy as np

print(" SMART ASSISTANT INVESTIGATION BRIEFS")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.integration.anomaly_snapshot_integration import (
    load_anomaly_snapshot,
    analyze_batch_with_anomaly_context
)

module_path = (
    root / "src/integration"
    / "smart_assistant_investigation_briefs.py"
)

output_dir = (
    root / "outputs/tests"
    / "phase_19_2d_investigation_briefs"
)

bundle_path = output_dir / "investigation_bundle.json"
map_path = output_dir / "spatial_candidates_map.geojson"

snapshot_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1.csv.gz"
)

manifest_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1_manifest.json"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

protected = [
    snapshot_path,
    manifest_path,
    input_path,
    root / "src/integration"
    / "anomaly_snapshot_integration.py",
    root / "src/integration"
    / "combined_batch_intelligence.py",
    root / "src/models/new_batch_area_kmeans.py",
    root / "src/models/new_batch_dbscan.py",
    root / "src/models/new_batch_anomaly_screen.py"
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

original_hashes = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Define the reusable explanation module
# --------------------------------------------------

module_source = dedent(r'''
"""Convert three-model output into investigation briefs.

Explanations are descriptive screening evidence,
not confirmed infrastructure faults, risk scores,
statistical significance or repair priorities.
"""

import math
import numpy as np
import pandas as pd


def _unique(values):
    return sorted({
        str(value).strip()
        for value in values.dropna()
        if str(value).strip()
    })


def _historical_relationship(group):

    statuses = group[
        "dbscan_status"
    ].astype(str)

    established = int(
        statuses.eq("established_cluster").sum()
    )

    outside = int(
        statuses.eq(
            "outside_existing_clusters"
        ).sum()
    )

    ambiguous = int(
        statuses.eq(
            "ambiguous_association"
        ).sum()
    )

    other = (
        len(group)
        - established
        - outside
        - ambiguous
    )

    if outside == len(group):
        relationship = (
            "entirely_outside_historical_associations"
        )

    elif established == len(group):
        relationship = (
            "entirely_associated_with_historical_clusters"
        )

    else:
        relationship = (
            "mixed_or_uncertain_historical_associations"
        )

    return {
        "relationship": relationship,
        "established": established,
        "outside": outside,
        "ambiguous": ambiguous,
        "other": other
    }


def _representative_location(group):
    """Return an actual submitted coordinate."""

    coordinates = group[[
        "latitude",
        "longitude"
    ]].apply(
        pd.to_numeric,
        errors="coerce"
    )

    if coordinates.isna().any().any():
        raise ValueError(
            "Invalid spatial candidate coordinates"
        )

    values = coordinates.to_numpy(
        dtype=float
    )

    if not np.isfinite(values).all():
        raise ValueError(
            "Non-finite candidate coordinates"
        )

    center_lat, center_lon = (
        values.mean(axis=0)
    )

    lat = np.radians(values[:, 0])
    lon = np.radians(values[:, 1])

    lat0 = math.radians(
        float(center_lat)
    )

    lon0 = math.radians(
        float(center_lon)
    )

    a = (
        np.sin((lat - lat0) / 2) ** 2
        + np.cos(lat0)
        * np.cos(lat)
        * np.sin((lon - lon0) / 2) ** 2
    )

    distances = (
        2
        * 6371008.8
        * np.arcsin(
            np.sqrt(np.clip(a, 0, 1))
        )
    )

    chosen = values[
        int(np.argmin(distances))
    ]

    distinct = len(
        np.unique(values, axis=0)
    )

    return (
        float(chosen[0]),
        float(chosen[1]),
        int(distinct)
    )


def build_investigation_briefs(
    integrated_result,
    synthetic_data=False
):
    """Build two types of candidate explanation.

    1. Area-category reporting-increase briefs.
    2. Local spatial concentration briefs.

    Also returns record audit and map-ready GeoJSON.
    """

    rows = integrated_result[
        "combined_records"
    ]

    anomaly_candidates = integrated_result[
        "anomaly_candidates"
    ]

    profiles = integrated_result[
        "kmeans_dbscan_output"
    ]["dbscan_output"]["candidate_profiles"]

    reporting_day = str(
        integrated_result[
            "stats"
        ]["reporting_day"]
    )

    required_columns = {
        "source_row",
        "complaint_id",
        "daily_window_status",
        "complaint_category",
        "latitude",
        "longitude",
        "anomaly_area_id",
        "anomaly_category",
        "anomaly_input_status",
        "anomaly_review_status",
        "anomaly_candidate",
        "new_kmeans_description",
        "new_dbscan_status",
        "new_dbscan_candidate_id",
        "dbscan_status"
    }

    missing = (
        required_columns
        - set(rows.columns)
    )

    if missing:
        raise ValueError(
            f"Missing record columns: {sorted(missing)}"
        )

    if rows["source_row"].duplicated().any():
        raise ValueError(
            "Duplicate source_row values"
        )

    if rows["anomaly_candidate"].isna().any():
        raise ValueError(
            "Missing anomaly candidate flags"
        )

    # ----------------------------------------------
    # 2. Area/category anomaly explanations
    # ----------------------------------------------

    anomaly_briefs = []

    for item in anomaly_candidates.itertuples(
        index=False
    ):

        area = str(item.area_id)
        category = str(
            item.complaint_category
        )

        group = rows.loc[
            rows["anomaly_area_id"].eq(area)
            & rows["anomaly_category"].eq(
                category
            )
            & rows["anomaly_input_status"].eq(
                "eligible"
            )
        ]

        count = len(group)

        if (
            count != int(item.incoming_reports)
            or not group[
                "anomaly_candidate"
            ].all()
        ):
            raise ValueError(
                f"Anomaly count mismatch: "
                f"{area}, {category}"
            )

        spatial_count = int(
            group["new_dbscan_status"]
            .eq("candidate_concentration")
            .sum()
        )

        patterns = _unique(
            group["new_kmeans_description"]
        )

        spatial_ids = _unique(
            group["new_dbscan_candidate_id"]
        )

        historical = (
            _historical_relationship(group)
        )

        explanation = (
            f"{count} {category} complaints "
            f"were submitted in {area} on "
            f"{reporting_day}. The count exceeds "
            f"the historical seasonal maximum "
            f"of {int(item.seasonal_daily_max)} "
            f"and all-weekday maximum of "
            f"{int(item.all_weekday_daily_max)}. "
            f"{spatial_count} complaints also "
            f"belong to new-batch spatial "
            f"concentrations. Reporting pattern: "
            f"{', '.join(patterns) if patterns else 'unavailable'}. "
            f"This is an exploratory reporting "
            f"increase, not a confirmed defect."
        )

        anomaly_briefs.append({
            "evidence_type":
                "area_category_reporting_increase",
            "area_id": area,
            "complaint_category": category,
            "reporting_day": reporting_day,
            "incoming_complaints": count,
            "seasonal_historical_max":
                int(item.seasonal_daily_max),
            "all_weekday_historical_max":
                int(item.all_weekday_daily_max),
            "kmeans_reporting_patterns":
                patterns,
            "complaints_in_spatial_candidates":
                spatial_count,
            "linked_spatial_candidate_ids":
                spatial_ids,
            "historical_dbscan_associations":
                historical,
            "screening_status":
                "cross_window_consistent_candidate",
            "explanation": explanation,
            "suggested_review": (
                "Check report descriptions, "
                "duplicate submissions, location "
                "accuracy, category definitions "
                "and recent operational records "
                "before considering an investigation."
            ),
            "synthetic_data": bool(
                synthetic_data
            )
        })

    # ----------------------------------------------
    # 3. Spatial concentration explanations
    # ----------------------------------------------

    spatial_briefs = []
    map_features = []

    for profile in profiles.itertuples(
        index=False
    ):

        candidate_id = str(
            profile.new_batch_candidate_id
        )

        group = rows.loc[
            rows["new_dbscan_status"].eq(
                "candidate_concentration"
            )
            & rows[
                "new_dbscan_candidate_id"
            ].eq(candidate_id)
        ]

        count = len(group)

        if count != int(
            profile.incoming_complaints
        ):
            raise ValueError(
                f"Spatial count mismatch: "
                f"{candidate_id}"
            )

        categories = _unique(
            group["complaint_category"]
        )

        if (
            len(categories) != 1
            or categories[0] != str(
                profile.complaint_category
            )
        ):
            raise ValueError(
                f"Mixed spatial categories: "
                f"{candidate_id}"
            )

        lat, lon, distinct = (
            _representative_location(group)
        )

        if distinct != int(
            profile.distinct_locations
        ):
            raise ValueError(
                "Spatial location count mismatch"
            )

        historical = (
            _historical_relationship(group)
        )

        anomaly_overlap = int(
            group["anomaly_candidate"].sum()
        )

        area_ids = _unique(
            group["anomaly_area_id"]
        )

        patterns = _unique(
            group["new_kmeans_description"]
        )

        linked_anomaly_groups = sorted({
            (
                f"{row.anomaly_area_id} / "
                f"{row.anomaly_category}"
            )
            for row in group.loc[
                group["anomaly_candidate"]
            ].itertuples(index=False)
        })

        explanation = (
            f"{count} {categories[0]} complaints "
            f"at {distinct} distinct submitted "
            f"coordinates form spatial candidate "
            f"{candidate_id}. Historical "
            f"association: {historical['outside']} "
            f"outside, {historical['established']} "
            f"established, "
            f"{historical['ambiguous']} ambiguous "
            f"and {historical['other']} other. "
            f"{anomaly_overlap} members also "
            f"belong to area-category reporting "
            f"increase candidates. The map pin "
            f"is a submitted complaint coordinate, "
            f"not a verified infrastructure fault."
        )

        spatial_briefs.append({
            "evidence_type":
                "new_batch_spatial_concentration",
            "candidate_id": candidate_id,
            "complaint_category":
                categories[0],
            "reporting_day": reporting_day,
            "incoming_complaints": count,
            "distinct_submitted_locations":
                distinct,
            "representative_latitude":
                round(lat, 7),
            "representative_longitude":
                round(lon, 7),
            "area_ids": area_ids,
            "kmeans_reporting_patterns":
                patterns,
            "complaints_in_anomaly_candidate_groups":
                anomaly_overlap,
            "linked_anomaly_area_categories":
                linked_anomaly_groups,
            "historical_dbscan_associations":
                historical,
            "explanation": explanation,
            "suggested_review": (
                "Review nearby report descriptions, "
                "duplicate coordinates and location "
                "accuracy. Verify conditions before "
                "any field action."
            ),
            "synthetic_data": bool(
                synthetic_data
            )
        })

        map_features.append({
            "type": "Feature",
            "id": candidate_id,
            "geometry": {
                "type": "Point",
                "coordinates": [lon, lat]
            },
            "properties": {
                "candidate_id": candidate_id,
                "complaint_category":
                    categories[0],
                "incoming_complaints": count,
                "distinct_locations": distinct,
                "historical_relationship":
                    historical["relationship"],
                "complaints_in_anomaly_candidate_groups":
                    anomaly_overlap,
                "synthetic_data": bool(
                    synthetic_data
                )
            }
        })

    # ----------------------------------------------
    # 4. Traceable status for every original record
    # ----------------------------------------------

    audit_rows = []

    for item in rows.itertuples(
        index=False
    ):

        audit_rows.append({
            "source_row": str(
                item.source_row
            ),
            "complaint_id": str(
                item.complaint_id
            ),
            "daily_window_status": str(
                item.daily_window_status
            ),
            "anomaly_input_status": str(
                item.anomaly_input_status
            ),
            "anomaly_review_status": str(
                item.anomaly_review_status
            ),
            "anomaly_candidate": bool(
                item.anomaly_candidate
            ),
            "new_dbscan_status": str(
                item.new_dbscan_status
            ),
            "new_dbscan_candidate_id": (
                None
                if pd.isna(
                    item.new_dbscan_candidate_id
                )
                else str(
                    item.new_dbscan_candidate_id
                )
            )
        })

    # ----------------------------------------------
    # 5. Reconciliation
    # ----------------------------------------------

    stats = {
        "input_records": len(rows),
        "reporting_day": reporting_day,
        "anomaly_candidate_groups":
            len(anomaly_briefs),
        "anomaly_candidate_complaints":
            int(
                rows["anomaly_candidate"].sum()
            ),
        "spatial_candidate_groups":
            len(spatial_briefs),
        "spatial_candidate_complaints":
            int(
                rows["new_dbscan_status"]
                .eq("candidate_concentration")
                .sum()
            ),
        "spatial_groups_entirely_outside_history":
            sum(
                brief[
                    "historical_dbscan_associations"
                ]["relationship"]
                == "entirely_outside_historical_associations"
                for brief in spatial_briefs
            ),
        "synthetic_data": bool(
            synthetic_data
        )
    }

    if (
        stats["anomaly_candidate_complaints"]
        != sum(
            item["incoming_complaints"]
            for item in anomaly_briefs
        )
    ):
        raise ValueError(
            "Anomaly brief totals do not reconcile"
        )

    if (
        stats["spatial_candidate_complaints"]
        != sum(
            item["incoming_complaints"]
            for item in spatial_briefs
        )
    ):
        raise ValueError(
            "Spatial brief totals do not reconcile"
        )

    return {
        "metadata": {
            "reporting_day": reporting_day,
            "synthetic_data": bool(
                synthetic_data
            ),
            "interpretation": (
                "Exploratory reporting evidence, "
                "not verified infrastructure faults, "
                "risk scores or automatic repair "
                "priorities."
            )
        },
        "stats": stats,
        "anomaly_briefs": anomaly_briefs,
        "spatial_briefs": spatial_briefs,
        "record_audit": audit_rows,
        "map_geojson": {
            "type": "FeatureCollection",
            "name":
                "new_batch_spatial_candidate_locations",
            "features": map_features
        }
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

build_investigation_briefs = namespace[
    "build_investigation_briefs"
]

# --------------------------------------------------
# 6. Load saved baseline and run existing pipeline
# --------------------------------------------------

print("\n1. LOADING HISTORICAL BASELINE")

baseline = load_anomaly_snapshot(
    snapshot_path,
    manifest_path
)

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

source_before = source.copy(deep=True)

with threadpool_limits(limits=1):

    integrated = analyze_batch_with_anomaly_context(
        source,
        baseline,
        "2026-11-09"
    )

fused_before = integrated[
    "combined_records"
].copy(deep=True)

print("\n2. GENERATING INVESTIGATION BRIEFS")

bundle_19_2d = build_investigation_briefs(
    integrated,
    synthetic_data=True
)

print(bundle_19_2d["stats"])

# --------------------------------------------------
# 7. Display anomaly and spatial brief previews
# --------------------------------------------------

anomaly_briefs_19_2d = pd.DataFrame(
    bundle_19_2d["anomaly_briefs"]
)

spatial_briefs_19_2d = pd.DataFrame(
    bundle_19_2d["spatial_briefs"]
)

print("\n3. ANOMALY INVESTIGATION BRIEFS")

display(
    anomaly_briefs_19_2d[[
        "area_id",
        "complaint_category",
        "incoming_complaints",
        "seasonal_historical_max",
        "all_weekday_historical_max",
        "complaints_in_spatial_candidates",
        "kmeans_reporting_patterns"
    ]].head(15)
)

print("\n4. SPATIAL INVESTIGATION BRIEFS")

display(
    spatial_briefs_19_2d[[
        "candidate_id",
        "complaint_category",
        "incoming_complaints",
        "distinct_submitted_locations",
        "complaints_in_anomaly_candidate_groups",
        "area_ids"
    ]].head(23)
)

# --------------------------------------------------
# 8. Important example: spatial != anomaly
# --------------------------------------------------

print("\n5. SPATIAL CONCENTRATION WITH SPARSE HISTORY")

sewer_example = next(
    (
        item for item in bundle_19_2d[
            "spatial_briefs"
        ]
        if item["candidate_id"] == "NB-00-004"
    ),
    None
)

if sewer_example:

    print(sewer_example["explanation"])

    print("\nLinked anomaly groups:")
    print(
        sewer_example[
            "linked_anomaly_area_categories"
        ]
    )

    print("\nSuggested review:")
    print(
        sewer_example["suggested_review"]
    )

# --------------------------------------------------
# 9. Test an empty output and malformed evidence
# --------------------------------------------------

empty_result = dict(integrated)

empty_result["combined_records"] = (
    integrated["combined_records"]
    .iloc[:0].copy()
)

empty_result["anomaly_candidates"] = (
    integrated["anomaly_candidates"]
    .iloc[:0].copy()
)

empty_km_db = dict(
    integrated["kmeans_dbscan_output"]
)

empty_spatial = dict(
    empty_km_db["dbscan_output"]
)

empty_spatial["candidate_profiles"] = (
    empty_spatial["candidate_profiles"]
    .iloc[:0].copy()
)

empty_km_db["dbscan_output"] = empty_spatial

empty_result[
    "kmeans_dbscan_output"
] = empty_km_db

empty_bundle = build_investigation_briefs(
    empty_result,
    synthetic_data=True
)

# Verify corrupt spatial evidence is rejected.
tampered = dict(integrated)

tampered_km_db = dict(
    integrated["kmeans_dbscan_output"]
)

tampered_spatial = dict(
    tampered_km_db["dbscan_output"]
)

tampered_profiles = (
    tampered_spatial["candidate_profiles"]
    .copy(deep=True)
)

tampered_profiles.loc[
    tampered_profiles.index[0],
    "incoming_complaints"
] = 999

tampered_spatial[
    "candidate_profiles"
] = tampered_profiles

tampered_km_db[
    "dbscan_output"
] = tampered_spatial

tampered[
    "kmeans_dbscan_output"
] = tampered_km_db

tamper_rejected = False

try:
    build_investigation_briefs(
        tampered,
        synthetic_data=True
    )
except ValueError:
    tamper_rejected = True

# --------------------------------------------------
# 10. Prepare JSON outputs in memory
# --------------------------------------------------

bundle_text = json.dumps(
    bundle_19_2d,
    indent=2,
    ensure_ascii=False,
    allow_nan=False
) + "\n"

map_text = json.dumps(
    bundle_19_2d["map_geojson"],
    indent=2,
    ensure_ascii=False,
    allow_nan=False
) + "\n"

# --------------------------------------------------
# 11. Validation
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

stats = bundle_19_2d["stats"]

check(
    "All 500 original records traceable",
    stats["input_records"] == 500
    and len(
        bundle_19_2d["record_audit"]
    ) == 500
)

check(
    "Original complaint identities preserved",
    [
        item["complaint_id"]
        for item in bundle_19_2d["record_audit"]
    ] == source["complaint_id"].astype(str).tolist()
)

check(
    "15 anomaly investigation briefs",
    stats["anomaly_candidate_groups"] == 15
)

check(
    "204 complaints in anomaly groups",
    stats["anomaly_candidate_complaints"] == 204
)

check(
    "23 spatial concentration briefs",
    stats["spatial_candidate_groups"] == 23
)

check(
    "228 complaints in spatial candidates",
    stats["spatial_candidate_complaints"] == 228
)

check(
    "Ten spatial candidates entirely outside historical associations",
    stats[
        "spatial_groups_entirely_outside_history"
    ] == 10
)

check(
    "Anomaly briefs contain both historical maxima",
    all(
        item["incoming_complaints"]
        > item["seasonal_historical_max"]
        and item["incoming_complaints"]
        > item["all_weekday_historical_max"]
        for item in bundle_19_2d["anomaly_briefs"]
    )
)

check(
    "Anomaly briefs include K-Means context",
    all(
        len(item["kmeans_reporting_patterns"]) > 0
        for item in bundle_19_2d["anomaly_briefs"]
    )
)

check(
    "Spatial briefs preserve their unique IDs",
    len({
        item["candidate_id"]
        for item in bundle_19_2d["spatial_briefs"]
    }) == 23
)

check(
    "Each spatial candidate remains category-specific",
    all(
        item["complaint_category"]
        for item in bundle_19_2d["spatial_briefs"]
    )
)

check(
    "Spatial-anomaly overlap never exceeds spatial count",
    all(
        item[
            "complaints_in_anomaly_candidate_groups"
        ] <= item["incoming_complaints"]
        for item in bundle_19_2d["spatial_briefs"]
    )
)

check(
    "Spatial-anomaly overlap never exceeds anomaly count",
    sum(
        item[
            "complaints_in_anomaly_candidate_groups"
        ]
        for item in bundle_19_2d["spatial_briefs"]
    ) <= 204
)

check(
    "NB-00-004 has 15 spatial complaints",
    sewer_example is not None
    and sewer_example["incoming_complaints"] == 15
)

check(
    "NB-00-004 is entirely outside historical associations",
    sewer_example is not None
    and sewer_example[
        "historical_dbscan_associations"
    ]["outside"] == 15
)

check(
    "NB-00-004 does not become an automatic anomaly",
    sewer_example is not None
    and sewer_example[
        "complaints_in_anomaly_candidate_groups"
    ] == 0
)

check(
    "Every brief explicitly identifies synthetic data",
    all(
        item["synthetic_data"]
        for item in (
            bundle_19_2d["anomaly_briefs"]
            + bundle_19_2d["spatial_briefs"]
        )
    )
)

check(
    "Map contains 23 point features",
    len(
        bundle_19_2d[
            "map_geojson"
        ]["features"]
    ) == 23
)

check(
    "Map candidate IDs reconcile with spatial briefs",
    {
        item["id"]
        for item in bundle_19_2d[
            "map_geojson"
        ]["features"]
    } == {
        item["candidate_id"]
        for item in bundle_19_2d["spatial_briefs"]
    }
)

check(
    "Map coordinates are valid longitude-latitude pairs",
    all(
        -180 <= item["geometry"][
            "coordinates"
        ][0] <= 180
        and -90 <= item["geometry"][
            "coordinates"
        ][1] <= 90
        for item in bundle_19_2d[
            "map_geojson"
        ]["features"]
    )
)

check(
    "Empty input produces safe empty briefs",
    len(empty_bundle["anomaly_briefs"]) == 0
    and len(empty_bundle["spatial_briefs"]) == 0
    and len(empty_bundle["record_audit"]) == 0
)

check(
    "Corrupt spatial candidate counts rejected",
    tamper_rejected
)

check(
    "Source DataFrame unchanged",
    source.equals(source_before)
)

check(
    "Integrated DataFrame unchanged",
    integrated["combined_records"].equals(
        fused_before
    )
)

check(
    "Both JSON outputs serialize and reload",
    len(
        json.loads(bundle_text)[
            "anomaly_briefs"
        ]
    ) == 15
    and len(
        json.loads(map_text)["features"]
    ) == 23
)

check(
    "Historical evidence and models unmodified",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == original_hashes[path]
        for path in protected
    )
)

results_19_2d = pd.DataFrame(checks)

print("\n6. PHASE 19.2D VALIDATION")

display(results_19_2d)

passed = int(results_19_2d["passed"].sum())
total = len(results_19_2d)

print(f"\nChecks passed: {passed}/{total}")

failures = results_19_2d.loc[
    ~results_19_2d["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 12. Save only if every validation passes
# --------------------------------------------------

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new module or output files saved.")

    for name in failures:
        print("FAILED:", name)

else:

    destinations = {
        module_path: module_source,
        bundle_path: bundle_text,
        map_path: map_text
    }

    # Check every destination before creating any.
    for path, content in destinations.items():

        if path.exists():

            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing file differs: {path}. "
                "No overwrite performed."
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():

            with path.open(
                "x",
                encoding="utf-8"
            ) as file:
                file.write(content)

    print("\n7. VERIFIED OUTPUT FILES")

    print("Reusable module:", module_path)
    print("Investigation bundle:", bundle_path)
    print("Spatial map GeoJSON:", map_path)

    print("\nSTATUS: PASSED")
    print("PHASE 19.2D INVESTIGATION BRIEFS VERIFIED")
    print("15 ANOMALY EXPLANATIONS PREPARED")
    print("23 SPATIAL EXPLANATIONS PREPARED")
    print("K-MEANS REPORTING CONTEXT PRESERVED")
    print("HISTORICAL HOTSPOT CONTEXT PRESERVED")
    print("500 RECORDS REMAIN TRACEABLE")
    print("MAP-READY SPATIAL POINTS PREPARED")
    print("NO AUTOMATIC REPAIR PRIORITIES")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

### Prepare the Dashboard Data Feed


Before creating the graphical interface, we'll combine the validated results into one dashboard-ready data structure containing area summaries, anomaly cards, spatial concentration cards, map features and record-level evidence.

One especially important feature is avoiding double-counting. The 204 anomaly-associated complaints and 228 spatial-candidate complaints overlap; they must not be displayed as 432 distinct complaints

```python

from pathlib import Path
from textwrap import dedent
from threadpoolctl import threadpool_limits
import copy
import hashlib
import json
import sys
import pandas as pd

print(" DASHBOARD INTELLIGENCE DATA FEED")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.integration.anomaly_snapshot_integration import (
    load_anomaly_snapshot,
    analyze_batch_with_anomaly_context
)

module_path = (
    root / "src/integration"
    / "dashboard_intelligence_feed.py"
)

output_dir = (
    root / "outputs/tests"
    / "phase_19_2e_dashboard_feed"
)

feed_path = output_dir / "dashboard_feed.json"
map_path = output_dir / "dashboard_map.geojson"

snapshot_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1.csv.gz"
)

manifest_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1_manifest.json"
)

input_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

bundle_path = (
    root / "outputs/tests"
    / "phase_19_2d_investigation_briefs"
    / "investigation_bundle.json"
)

protected = [
    snapshot_path,
    manifest_path,
    input_path,
    bundle_path,
    root / "src/integration"
    / "anomaly_snapshot_integration.py",
    root / "src/integration"
    / "smart_assistant_investigation_briefs.py",
    root / "src/models/new_batch_area_kmeans.py",
    root / "src/models/new_batch_dbscan.py",
    root / "src/models/new_batch_anomaly_screen.py"
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

hashes_before = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Reusable dashboard feed module
# --------------------------------------------------

module_source = dedent(r'''
"""Dashboard data contract for three-model intelligence.

All counts describe complaint reports, not verified
infrastructure faults or automatic repair priorities.
"""

import pandas as pd


def _text(value):
    if value is None or pd.isna(value):
        return None
    return str(value)


def _unique(values):
    return sorted({
        str(value).strip()
        for value in values.dropna()
        if str(value).strip()
    })


def build_dashboard_feed(integrated, bundle):
    """Build reusable JSON-compatible dashboard data."""

    records = integrated[
        "combined_records"
    ].copy()

    briefs = bundle["anomaly_briefs"]
    spatial_briefs = bundle["spatial_briefs"]
    geojson = bundle["map_geojson"]

    required = {
        "source_row",
        "complaint_id",
        "daily_window_status",
        "anomaly_area_id",
        "anomaly_category",
        "anomaly_input_status",
        "anomaly_review_status",
        "anomaly_candidate",
        "new_kmeans_description",
        "kmeans_reporting_profile",
        "new_dbscan_status",
        "new_dbscan_candidate_id",
        "dbscan_status"
    }

    missing = required - set(records.columns)

    if missing:
        raise ValueError(
            f"Missing dashboard columns: {sorted(missing)}"
        )

    if records["source_row"].duplicated().any():
        raise ValueError("Duplicate source_row values")

    record_audit = bundle["record_audit"]

    if len(record_audit) != len(records):
        raise ValueError(
            "Brief record audit count mismatch"
        )

    if [
        str(item["source_row"])
        for item in record_audit
    ] != records["source_row"].astype(str).tolist():
        raise ValueError(
            "Brief and integrated source rows differ"
        )

    anomaly_mask = (
        records["anomaly_candidate"]
        .fillna(False)
        .astype(bool)
    )

    spatial_mask = (
        records["new_dbscan_status"]
        .eq("candidate_concentration")
        .fillna(False)
    )

    eligible_mask = (
        records["anomaly_input_status"]
        .eq("eligible")
        .fillna(False)
    )

    both = int(
        (anomaly_mask & spatial_mask).sum()
    )

    anomaly_only = int(
        (anomaly_mask & ~spatial_mask).sum()
    )

    spatial_only = int(
        (spatial_mask & ~anomaly_mask).sum()
    )

    neither = int(
        (~anomaly_mask & ~spatial_mask).sum()
    )

    n_anomaly = int(anomaly_mask.sum())
    n_spatial = int(spatial_mask.sum())

    if n_anomaly != int(
        bundle["stats"]["anomaly_candidate_complaints"]
    ):
        raise ValueError(
            "Anomaly brief complaint total mismatch"
        )

    if n_spatial != int(
        bundle["stats"]["spatial_candidate_complaints"]
    ):
        raise ValueError(
            "Spatial brief complaint total mismatch"
        )

    if len(briefs) != int(
        bundle["stats"]["anomaly_candidate_groups"]
    ):
        raise ValueError(
            "Anomaly brief group count mismatch"
        )

    if len(spatial_briefs) != int(
        bundle["stats"]["spatial_candidate_groups"]
    ):
        raise ValueError(
            "Spatial brief group count mismatch"
        )

    if both != sum(
        int(item[
            "complaints_in_anomaly_candidate_groups"
        ])
        for item in spatial_briefs
    ):
        raise ValueError(
            "Anomaly-spatial overlap mismatch"
        )

    anomaly_keys = {
        (
            str(row.anomaly_area_id),
            str(row.anomaly_category)
        )
        for row in records.loc[
            anomaly_mask
        ].itertuples(index=False)
    }

    brief_keys = {
        (
            str(item["area_id"]),
            str(item["complaint_category"])
        )
        for item in briefs
    }

    if anomaly_keys != brief_keys:
        raise ValueError(
            "Anomaly brief identities do not reconcile"
        )

    candidate_ids = {
        str(value)
        for value in records.loc[
            spatial_mask,
            "new_dbscan_candidate_id"
        ].dropna()
    }

    spatial_ids = {
        str(item["candidate_id"])
        for item in spatial_briefs
    }

    map_features = geojson["features"]

    map_ids = {
        str(item["id"])
        for item in map_features
    }

    if (
        candidate_ids != spatial_ids
        or spatial_ids != map_ids
        or len(map_features) != len(spatial_ids)
    ):
        raise ValueError(
            "Spatial brief and map IDs disagree"
        )

    # ----------------------------------------------
    # Area-level K-Means reporting summaries
    # ----------------------------------------------

    area_rows = records.loc[
        eligible_mask
    ].copy()

    area_cards = []

    for area_id, group in area_rows.groupby(
        "anomaly_area_id",
        sort=True
    ):

        category_counts = (
            group["anomaly_category"]
            .astype(str)
            .value_counts()
        )

        breakdown = [
            {
                "complaint_category": category,
                "incoming_complaints": int(count)
            }
            for category, count in sorted(
                category_counts.items(),
                key=lambda item: (
                    -item[1], item[0]
                )
            )
        ]

        in_spatial = group[
            "new_dbscan_status"
        ].eq("candidate_concentration")

        area_cards.append({
            "area_id": str(area_id),
            "incoming_complaints": int(len(group)),
            "reporting_patterns": _unique(
                group["new_kmeans_description"]
            ),
            "historical_kmeans_profiles": _unique(
                group["kmeans_reporting_profile"]
            ),
            "category_breakdown": breakdown,
            "dominant_category": (
                breakdown[0]["complaint_category"]
                if breakdown else None
            ),
            "anomaly_candidate_complaints": int(
                group["anomaly_candidate"].sum()
            ),
            "spatial_candidate_complaints": int(
                in_spatial.sum()
            ),
            "spatial_candidate_ids": _unique(
                group.loc[
                    in_spatial,
                    "new_dbscan_candidate_id"
                ]
            ),
            "historical_established_complaints": int(
                group["dbscan_status"].eq(
                    "established_cluster"
                ).sum()
            ),
            "historical_outside_complaints": int(
                group["dbscan_status"].eq(
                    "outside_existing_clusters"
                ).sum()
            ),
            "historical_ambiguous_complaints": int(
                group["dbscan_status"].eq(
                    "ambiguous_association"
                ).sum()
            )
        })

    # ----------------------------------------------
    # Record-level dashboard table
    # ----------------------------------------------

    record_rows = []

    for row in records.itertuples(index=False):

        record_rows.append({
            "source_row": _text(row.source_row),
            "complaint_id": _text(row.complaint_id),
            "area_id": _text(row.anomaly_area_id),
            "complaint_category": _text(
                row.anomaly_category
            ),
            "reporting_day_status": _text(
                row.daily_window_status
            ),
            "anomaly_input_status": _text(
                row.anomaly_input_status
            ),
            "new_kmeans_pattern": _text(
                row.new_kmeans_description
            ),
            "historical_kmeans_profile": _text(
                row.kmeans_reporting_profile
            ),
            "new_dbscan_status": _text(
                row.new_dbscan_status
            ),
            "new_dbscan_candidate_id": _text(
                row.new_dbscan_candidate_id
            ),
            "historical_dbscan_status": _text(
                row.dbscan_status
            ),
            "anomaly_review_status": _text(
                row.anomaly_review_status
            ),
            "anomaly_candidate": bool(
                row.anomaly_candidate
            ),
            "spatial_candidate": (
                row.new_dbscan_status
                == "candidate_concentration"
            )
        })

    patterns = sorted({
        pattern
        for area in area_cards
        for pattern in area["reporting_patterns"]
    })

    summary = {
        "input_complaints": int(len(records)),
        "selected_day_complaints": int(
            records["daily_window_status"]
            .eq("selected_day").sum()
        ),
        "anomaly_eligible_complaints": int(
            eligible_mask.sum()
        ),
        "areas_with_eligible_reports":
            len(area_cards),
        "new_kmeans_patterns": len(patterns),
        "spatial_candidate_groups":
            len(spatial_briefs),
        "spatial_candidate_complaints": n_spatial,
        "anomaly_candidate_groups":
            len(briefs),
        "anomaly_candidate_complaints": n_anomaly,
        "complaints_in_both_signals": both,
        "complaints_anomaly_only": anomaly_only,
        "complaints_spatial_only": spatial_only,
        "complaints_in_neither_signal": neither,
        "distinct_complaints_in_either_signal":
            n_anomaly + n_spatial - both,
        "historical_baseline_areas": int(
            integrated["stats"][
                "historical_baseline_areas"
            ]
        )
    }

    if (
        both + anomaly_only
        + spatial_only + neither
        != len(records)
    ):
        raise ValueError(
            "Dashboard report partition mismatch"
        )

    if sum(
        area["incoming_complaints"]
        for area in area_cards
    ) != int(eligible_mask.sum()):
        raise ValueError(
            "Dashboard area totals mismatch"
        )

    return {
        "metadata": {
            "schema_version": "1.0",
            "reporting_day": str(
                integrated["stats"]["reporting_day"]
            ),
            "synthetic_data": bool(
                bundle["metadata"]["synthetic_data"]
            ),
            "evidence_policy": (
                "Screening candidates and reporting "
                "patterns only. No validated defects, "
                "risk scores, statistical significance "
                "claims or automatic repair priorities."
            ),
            "counting_policy": (
                "Anomaly and spatial complaint counts "
                "may overlap. Use distinct_complaints_"
                "in_either_signal for their union."
            )
        },
        "summary": summary,
        "filters": {
            "boroughs": sorted({
                str(area["area_id"]).split(
                    "_", 1
                )[0]
                for area in area_cards
            }),
            "categories": sorted({
                str(value)
                for value in area_rows[
                    "anomaly_category"
                ].dropna().unique()
            }),
            "kmeans_patterns": patterns,
            "anomaly_review_statuses": _unique(
                records["anomaly_review_status"]
            )
        },
        "area_cards": area_cards,
        "anomaly_cards": [
            {
                "id": (
                    f"ANOMALY::{item['area_id']}::"
                    f"{item['complaint_category']}"
                ),
                **item
            }
            for item in briefs
        ],
        "spatial_cards": [
            dict(item)
            for item in spatial_briefs
        ],
        "records": record_rows,
        "map_geojson": geojson
    }
''').lstrip()

compile(
    module_source,
    str(module_path),
    "exec"
)

namespace = {}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

build_dashboard_feed = namespace[
    "build_dashboard_feed"
]

# --------------------------------------------------
# 2. Load saved results and run integrated models
# --------------------------------------------------

print("\n1. LOADING VERIFIED DATA")

baseline = load_anomaly_snapshot(
    snapshot_path,
    manifest_path
)

source = pd.read_csv(
    input_path,
    dtype="string",
    keep_default_na=False
)

source_before = source.copy(deep=True)

with bundle_path.open(
    encoding="utf-8"
) as file:
    bundle = json.load(file)

with threadpool_limits(limits=1):

    integrated = analyze_batch_with_anomaly_context(
        source,
        baseline,
        "2026-11-09"
    )

integrated_before = (
    integrated["combined_records"]
    .copy(deep=True)
)

# --------------------------------------------------
# 3. Generate dashboard feed
# --------------------------------------------------

print("\n2. BUILDING DASHBOARD DATA FEED")

dashboard_feed_19_2e = build_dashboard_feed(
    integrated,
    bundle
)

summary = dashboard_feed_19_2e[
    "summary"
]

for name, value in summary.items():
    print(f"{name}: {value}")

print("\n3. AREA REPORTING PATTERNS")

display(
    pd.DataFrame(
        dashboard_feed_19_2e["area_cards"]
    )[[
        "area_id",
        "incoming_complaints",
        "reporting_patterns",
        "anomaly_candidate_complaints",
        "spatial_candidate_complaints",
        "dominant_category"
    ]]
)

# --------------------------------------------------
# 4. Display distinct complaint accounting
# --------------------------------------------------

print("\n4. COMPLAINT OVERLAP ACCOUNTING")

signal_accounting = pd.DataFrame([
    {
        "complaint_group": "Both signals",
        "distinct_complaints":
            summary["complaints_in_both_signals"]
    },
    {
        "complaint_group": "Anomaly only",
        "distinct_complaints":
            summary["complaints_anomaly_only"]
    },
    {
        "complaint_group": "Spatial only",
        "distinct_complaints":
            summary["complaints_spatial_only"]
    },
    {
        "complaint_group": "Neither signal",
        "distinct_complaints":
            summary["complaints_in_neither_signal"]
    }
])

display(signal_accounting)

# --------------------------------------------------
# 5. Validate dashboard integrity
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "All 500 source records included",
    summary["input_complaints"] == 500
)

check(
    "All 500 records belong to reporting day",
    summary["selected_day_complaints"] == 500
)

check(
    "All 500 anomaly input records eligible",
    summary["anomaly_eligible_complaints"] == 500
)

check(
    "20 geographic area cards created",
    len(
        dashboard_feed_19_2e["area_cards"]
    ) == 20
)

check(
    "Three new K-Means patterns represented",
    summary["new_kmeans_patterns"] == 3
)

check(
    "All area cards reconcile to 500 reports",
    sum(
        item["incoming_complaints"]
        for item in dashboard_feed_19_2e[
            "area_cards"
        ]
    ) == 500
)

check(
    "All 20 synthetic areas contain 25 reports",
    all(
        item["incoming_complaints"] == 25
        for item in dashboard_feed_19_2e[
            "area_cards"
        ]
    )
)

check(
    "23 spatial candidate cards retained",
    summary["spatial_candidate_groups"] == 23
    and len(
        dashboard_feed_19_2e["spatial_cards"]
    ) == 23
)

check(
    "228 complaints belong to spatial candidates",
    summary["spatial_candidate_complaints"] == 228
)

check(
    "15 anomaly candidate cards retained",
    summary["anomaly_candidate_groups"] == 15
    and len(
        dashboard_feed_19_2e["anomaly_cards"]
    ) == 15
)

check(
    "204 complaints belong to anomaly candidates",
    summary["anomaly_candidate_complaints"] == 204
)

check(
    "Signal groups partition all complaints",
    int(
        signal_accounting[
            "distinct_complaints"
        ].sum()
    ) == 500
)

check(
    "Distinct union uses overlap correction",
    summary[
        "distinct_complaints_in_either_signal"
    ]
    == (
        204 + 228
        - summary["complaints_in_both_signals"]
    )
)

check(
    "Overlap matches spatial investigation briefs",
    summary["complaints_in_both_signals"]
    == sum(
        item[
            "complaints_in_anomaly_candidate_groups"
        ]
        for item in bundle["spatial_briefs"]
    )
)

check(
    "All historical area coverage retained",
    summary["historical_baseline_areas"] == 238
)

check(
    "Dashboard has 500 record-level entries",
    len(
        dashboard_feed_19_2e["records"]
    ) == 500
)

check(
    "Record identities remain in original order",
    [
        item["source_row"]
        for item in dashboard_feed_19_2e["records"]
    ] == source["source_row"].astype(str).tolist()
)

check(
    "Dashboard retains 23 map features",
    len(
        dashboard_feed_19_2e[
            "map_geojson"
        ]["features"]
    ) == 23
)

check(
    "Map IDs match the spatial cards",
    {
        feature["id"]
        for feature in dashboard_feed_19_2e[
            "map_geojson"
        ]["features"]
    } == {
        item["candidate_id"]
        for item in dashboard_feed_19_2e[
            "spatial_cards"
        ]
    }
)

check(
    "Dashboard explicitly labels synthetic data",
    dashboard_feed_19_2e[
        "metadata"
    ]["synthetic_data"] is True
)

check(
    "Dashboard retains investigation explanations",
    all(
        item.get("explanation")
        and item.get("suggested_review")
        for item in (
            dashboard_feed_19_2e["anomaly_cards"]
            + dashboard_feed_19_2e[
                "spatial_cards"
            ]
        )
    )
)

# --------------------------------------------------
# 6. Reject inconsistent upstream evidence
# --------------------------------------------------

tampered_bundle = copy.deepcopy(bundle)

tampered_bundle["stats"][
    "anomaly_candidate_complaints"
] = 999

tamper_rejected = False

try:
    build_dashboard_feed(
        integrated,
        tampered_bundle
    )
except ValueError:
    tamper_rejected = True

check(
    "Inconsistent anomaly counts rejected",
    tamper_rejected
)

check(
    "Original source DataFrame unchanged",
    source.equals(source_before)
)

check(
    "Integrated model records unchanged",
    integrated["combined_records"].equals(
        integrated_before
    )
)

check(
    "Protected data and models unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == hashes_before[path]
        for path in protected
    )
)

# --------------------------------------------------
# 7. Serialize outputs and validate JSON
# --------------------------------------------------

feed_text = json.dumps(
    dashboard_feed_19_2e,
    indent=2,
    ensure_ascii=False,
    allow_nan=False
) + "\n"

map_text = json.dumps(
    dashboard_feed_19_2e["map_geojson"],
    indent=2,
    ensure_ascii=False,
    allow_nan=False
) + "\n"

check(
    "Dashboard JSON serializes and reloads",
    json.loads(feed_text)["summary"][
        "input_complaints"
    ] == 500
)

check(
    "GeoJSON serializes and reloads",
    len(
        json.loads(map_text)["features"]
    ) == 23
)

results_19_2e = pd.DataFrame(checks)

print("\n5. PHASE 19.2E VALIDATION")

display(results_19_2e)

passed = int(results_19_2e["passed"].sum())
total = len(results_19_2e)

print(f"\nChecks passed: {passed}/{total}")

failed = results_19_2e.loc[
    ~results_19_2e["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 8. Save only after successful validation
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No dashboard feed files saved.")

    for name in failed:
        print("FAILED:", name)

else:

    destinations = {
        module_path: module_source,
        feed_path: feed_text,
        map_path: map_text
    }

    # Check all existing targets before writing.
    for path, content in destinations.items():

        if path.exists():
            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing output differs: {path}. "
                "No overwrite performed."
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():
            with path.open(
                "x",
                encoding="utf-8"
            ) as file:
                file.write(content)

    print("\n6. VERIFIED DASHBOARD OUTPUTS")

    print("Reusable module:", module_path)
    print("Dashboard feed:", feed_path)
    print("Map GeoJSON:", map_path)

    print("\nSTATUS: PASSED")
    print("PHASE 19.2E DASHBOARD FEED VERIFIED")
    print("THREE-MODEL EVIDENCE PRESERVED")
    print("20 AREA CARDS PREPARED")
    print("15 ANOMALY CARDS PREPARED")
    print("23 SPATIAL CARDS PREPARED")
    print("COMPLAINT OVERLAP ACCOUNTED FOR")
    print("500 COMPLAINT RECORDS TRACEABLE")
    print("NO AUTOMATIC REPAIR PRIORITIES")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import ast
import hashlib
import importlib.util
import json
import sys
import pandas as pd

print(" STREAMLIT DASHBOARD GUI")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

app_path = (
    root / "dashboard"
    / "urban_pulse_dashboard.py"
)

feed_path = (
    root / "outputs/tests"
    / "phase_19_2e_dashboard_feed"
    / "dashboard_feed.json"
)

map_path = (
    root / "outputs/tests"
    / "phase_19_2e_dashboard_feed"
    / "dashboard_map.geojson"
)

protected = [
    feed_path,
    map_path,
    root / "src/integration"
    / "dashboard_intelligence_feed.py",
    root / "src/integration"
    / "anomaly_snapshot_integration.py",
    root / "src/models"
    / "new_batch_area_kmeans.py",
    root / "src/models"
    / "new_batch_dbscan.py",
    root / "src/models"
    / "new_batch_anomaly_screen.py"
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

hashes_before = {
    path: hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Build a standalone GUI that reads saved JSON
# --------------------------------------------------

app_source = dedent(r'''
"""Read-only Urban Infrastructure Pulse dashboard.

Launch from the project root:
    python -m streamlit run dashboard/urban_pulse_dashboard.py

This prototype displays saved synthetic demonstration
results. It does not retrain or modify historical models.
"""

import json
from pathlib import Path

import pandas as pd


PROJECT_ROOT = Path(__file__).resolve().parents[1]

FEED_PATH = (
    PROJECT_ROOT
    / "outputs"
    / "tests"
    / "phase_19_2e_dashboard_feed"
    / "dashboard_feed.json"
)


def load_feed(path=FEED_PATH):
    with Path(path).open(
        "r",
        encoding="utf-8"
    ) as file:
        return json.load(file)


def signal_type(row):
    anomaly = bool(
        row.get("anomaly_candidate", False)
    )
    spatial = bool(
        row.get("spatial_candidate", False)
    )

    if anomaly and spatial:
        return "Both signals"

    if anomaly:
        return "Anomaly only"

    if spatial:
        return "Spatial only"

    return "Neither"


def filter_records(
    records,
    borough="All",
    category="All",
    pattern="All",
    signal="All"
):
    selected = []

    for record in records:

        area_id = str(
            record.get("area_id") or ""
        )

        current_borough = (
            area_id.split("_", 1)[0]
        )

        if (
            borough != "All"
            and current_borough != borough
        ):
            continue

        if (
            category != "All"
            and record.get(
                "complaint_category"
            ) != category
        ):
            continue

        if (
            pattern != "All"
            and record.get(
                "new_kmeans_pattern"
            ) != pattern
        ):
            continue

        if (
            signal != "All"
            and signal_type(record) != signal
        ):
            continue

        selected.append(record)

    return selected


def build_filtered_view(feed, records):
    """Link selected reports to saved evidence cards."""

    anomaly_keys = {
        (
            str(item["area_id"]),
            str(item["complaint_category"])
        )
        for item in records
        if item["anomaly_candidate"]
    }

    spatial_ids = {
        str(item["new_dbscan_candidate_id"])
        for item in records
        if (
            item["spatial_candidate"]
            and item.get(
                "new_dbscan_candidate_id"
            ) is not None
        )
    }

    anomaly_cards = [
        item
        for item in feed["anomaly_cards"]
        if (
            str(item["area_id"]),
            str(item["complaint_category"])
        ) in anomaly_keys
    ]

    spatial_cards = [
        item
        for item in feed["spatial_cards"]
        if str(item["candidate_id"]) in spatial_ids
    ]

    map_features = [
        feature
        for feature in feed[
            "map_geojson"
        ]["features"]
        if str(feature["id"]) in spatial_ids
    ]

    both = sum(
        signal_type(item) == "Both signals"
        for item in records
    )

    anomaly_only = sum(
        signal_type(item) == "Anomaly only"
        for item in records
    )

    spatial_only = sum(
        signal_type(item) == "Spatial only"
        for item in records
    )

    neither = sum(
        signal_type(item) == "Neither"
        for item in records
    )

    summary = {
        "selected": len(records),
        "both": both,
        "anomaly_only": anomaly_only,
        "spatial_only": spatial_only,
        "neither": neither,
        "anomaly_reports":
            both + anomaly_only,
        "spatial_reports":
            both + spatial_only,
        "distinct_either":
            both + anomaly_only + spatial_only,
        "anomaly_groups":
            len(anomaly_cards),
        "spatial_groups":
            len(spatial_cards),
        "areas": len({
            str(item["area_id"])
            for item in records
        })
    }

    if sum([
        both,
        anomaly_only,
        spatial_only,
        neither
    ]) != len(records):
        raise ValueError(
            "Selected complaint counts disagree."
        )

    return {
        "records": records,
        "anomaly_cards": anomaly_cards,
        "spatial_cards": spatial_cards,
        "map_features": map_features,
        "summary": summary
    }


def validate_feed(feed):
    """Reject inconsistent dashboard source data."""

    required = {
        "metadata",
        "summary",
        "filters",
        "area_cards",
        "anomaly_cards",
        "spatial_cards",
        "records",
        "map_geojson"
    }

    if not required.issubset(feed):
        raise ValueError(
            "Dashboard feed is missing required sections."
        )

    records = feed["records"]
    summary = feed["summary"]

    if len(records) != summary[
        "input_complaints"
    ]:
        raise ValueError(
            "Input record count mismatch."
        )

    identities = [
        str(item["source_row"])
        for item in records
    ]

    if len(identities) != len(
        set(identities)
    ):
        raise ValueError(
            "Duplicate complaint source rows."
        )

    view = build_filtered_view(
        feed,
        records
    )

    counts = view["summary"]

    expected = {
        "anomaly_candidate_complaints":
            counts["anomaly_reports"],
        "spatial_candidate_complaints":
            counts["spatial_reports"],
        "complaints_in_both_signals":
            counts["both"],
        "complaints_anomaly_only":
            counts["anomaly_only"],
        "complaints_spatial_only":
            counts["spatial_only"],
        "complaints_in_neither_signal":
            counts["neither"],
        "distinct_complaints_in_either_signal":
            counts["distinct_either"]
    }

    for key, value in expected.items():
        if summary[key] != value:
            raise ValueError(
                f"Dashboard count mismatch: {key}"
            )

    if counts["anomaly_groups"] != summary[
        "anomaly_candidate_groups"
    ]:
        raise ValueError(
            "Anomaly group count mismatch."
        )

    if counts["spatial_groups"] != summary[
        "spatial_candidate_groups"
    ]:
        raise ValueError(
            "Spatial group count mismatch."
        )

    return view


def main():
    import streamlit as st

    st.set_page_config(
        page_title="Urban Infrastructure Pulse",
        page_icon="🏙️",
        layout="wide",
        initial_sidebar_state="expanded"
    )

    st.title("Urban Infrastructure Pulse")
    st.caption(
        "Smart infrastructure reporting intelligence"
    )

    if not FEED_PATH.is_file():
        st.error(
            "The validated dashboard feed was not found."
        )
        st.code(str(FEED_PATH))
        st.stop()

    try:
        feed = load_feed()
        validate_feed(feed)
    except (OSError, ValueError, KeyError) as exc:
        st.error(
            f"Invalid dashboard feed: {exc}"
        )
        st.stop()

    if feed["metadata"].get(
        "synthetic_data", False
    ):
        st.warning(
            "Synthetic demonstration data — "
            "not live NYC 311 incidents or "
            "verified infrastructure problems."
        )

    st.caption(
        "Reporting day: "
        + str(feed["metadata"]["reporting_day"])
        + " | Historical comparison: 2023–2025"
    )

    # ----------------------------------------------
    # Interactive filters
    # ----------------------------------------------

    st.sidebar.header("Explore complaints")

    borough = st.sidebar.selectbox(
        "Borough",
        ["All"] + feed["filters"]["boroughs"]
    )

    category = st.sidebar.selectbox(
        "Complaint category",
        ["All"] + feed["filters"]["categories"]
    )

    pattern = st.sidebar.selectbox(
        "K-Means reporting pattern",
        ["All"] + feed["filters"][
            "kmeans_patterns"
        ]
    )

    signal = st.sidebar.selectbox(
        "Investigation signal",
        [
            "All",
            "Both signals",
            "Anomaly only",
            "Spatial only",
            "Neither"
        ]
    )

    st.sidebar.caption(
        "Filters change the displayed report subset. "
        "Historical maxima and investigation "
        "explanations refer to the complete "
        "original reporting-day batch."
    )

    records = filter_records(
        feed["records"],
        borough=borough,
        category=category,
        pattern=pattern,
        signal=signal
    )

    view = build_filtered_view(
        feed, records
    )

    stats = view["summary"]

    # ----------------------------------------------
    # Overview metrics
    # ----------------------------------------------

    m1, m2, m3, m4, m5 = st.columns(5)

    m1.metric(
        "Selected reports",
        stats["selected"]
    )

    m2.metric(
        "Anomaly-associated",
        stats["anomaly_reports"]
    )

    m3.metric(
        "Spatial-associated",
        stats["spatial_reports"]
    )

    m4.metric(
        "Distinct with either",
        stats["distinct_either"]
    )

    m5.metric(
        "Areas represented",
        stats["areas"]
    )

    st.caption(
        "Anomaly and spatial totals overlap. "
        "'Distinct with either' counts each "
        "complaint once."
    )

    tabs = st.tabs([
        "Overview",
        "K-Means Areas",
        "Anomaly Reviews",
        "DBSCAN Map",
        "Complaint Records"
    ])

    # ----------------------------------------------
    # Tab 1: Overview
    # ----------------------------------------------

    with tabs[0]:

        st.subheader("Reporting overview")

        summary_frame = pd.DataFrame([
            {
                "Signal classification":
                    "Both signals",
                "Complaints": stats["both"]
            },
            {
                "Signal classification":
                    "Anomaly only",
                "Complaints":
                    stats["anomaly_only"]
            },
            {
                "Signal classification":
                    "Spatial only",
                "Complaints":
                    stats["spatial_only"]
            },
            {
                "Signal classification":
                    "Neither",
                "Complaints":
                    stats["neither"]
            }
        ])

        left, right = st.columns(2)

        with left:
            st.markdown(
                "**Distinct complaint accounting**"
            )
            st.dataframe(
                summary_frame,
                hide_index=True,
                use_container_width=True
            )

        with right:
            st.markdown(
                "**Complaints by category**"
            )

            if records:
                frame = pd.DataFrame(records)
                st.bar_chart(
                    frame[
                        "complaint_category"
                    ].value_counts()
                )
            else:
                st.info(
                    "No complaints match these filters."
                )

        if records:
            st.markdown(
                "**Complaints by borough–ZIP area**"
            )

            frame = pd.DataFrame(records)

            st.bar_chart(
                frame["area_id"]
                .value_counts()
                .sort_index()
            )

    # ----------------------------------------------
    # Tab 2: K-Means area patterns
    # ----------------------------------------------

    with tabs[1]:

        st.subheader(
            "K-Means area reporting patterns"
        )

        st.caption(
            "Patterns describe relative complaint "
            "category mixtures, not infrastructure "
            "condition or repair urgency."
        )

        if records:
            frame = pd.DataFrame(records)

            area_table = []

            for area_id, group in frame.groupby(
                "area_id", sort=True
            ):

                most_common = (
                    group["complaint_category"]
                    .value_counts()
                    .index[0]
                )

                patterns = sorted({
                    str(value)
                    for value in group[
                        "new_kmeans_pattern"
                    ].dropna()
                })

                area_table.append({
                    "Area": area_id,
                    "Selected reports":
                        len(group),
                    "K-Means pattern":
                        " | ".join(patterns),
                    "Most reported category":
                        most_common,
                    "Anomaly-associated":
                        int(group[
                            "anomaly_candidate"
                        ].sum()),
                    "Spatial-associated":
                        int(group[
                            "spatial_candidate"
                        ].sum())
                })

            st.dataframe(
                pd.DataFrame(area_table),
                hide_index=True,
                use_container_width=True
            )
        else:
            st.info(
                "No area patterns match the filters."
            )

    # ----------------------------------------------
    # Tab 3: Anomaly screening evidence
    # ----------------------------------------------

    with tabs[2]:

        st.subheader(
            "Historical reporting-increase reviews"
        )

        st.caption(
            "These are screening candidates, "
            "not statistically confirmed anomalies."
        )

        if not view["anomaly_cards"]:
            st.info(
                "No anomaly candidate groups match "
                "the selected filters."
            )

        for card in view["anomaly_cards"]:

            matched = sum(
                record["anomaly_candidate"]
                and record["area_id"]
                    == card["area_id"]
                and record["complaint_category"]
                    == card["complaint_category"]
                for record in records
            )

            heading = (
                f"{card['area_id']} — "
                f"{card['complaint_category']} "
                f"({matched} selected reports)"
            )

            with st.expander(heading):

                st.write(
                    card["explanation"]
                )

                st.markdown(
                    "**Historical comparison**"
                )

                c1, c2, c3 = st.columns(3)

                c1.metric(
                    "Full-batch reports",
                    card["incoming_complaints"]
                )

                c2.metric(
                    "Seasonal historical max",
                    card[
                        "seasonal_historical_max"
                    ]
                )

                c3.metric(
                    "All-Monday historical max",
                    card[
                        "all_weekday_historical_max"
                    ]
                )

                st.markdown(
                    "**Suggested verification**"
                )

                st.write(
                    card["suggested_review"]
                )

    # ----------------------------------------------
    # Tab 4: DBSCAN spatial concentrations
    # ----------------------------------------------

    with tabs[3]:

        st.subheader(
            "New-batch spatial concentrations"
        )

        st.caption(
            "Pins represent submitted complaint "
            "coordinates. They are not verified "
            "infrastructure fault locations."
        )

        if not view["map_features"]:
            st.info(
                "No spatial concentrations match "
                "the current filters."
            )

        else:
            points = []

            for feature in view[
                "map_features"
            ]:

                longitude, latitude = (
                    feature["geometry"][
                        "coordinates"
                    ]
                )

                points.append({
                    "latitude": latitude,
                    "longitude": longitude,
                    "candidate_id": feature["id"]
                })

            points_frame = pd.DataFrame(
                points
            )

            st.map(
                points_frame[[
                    "latitude",
                    "longitude"
                ]]
            )

            st.caption(
                "One representative point per "
                "spatial candidate; not every "
                "individual complaint location."
            )

        for card in view["spatial_cards"]:

            matched = sum(
                record["spatial_candidate"]
                and record[
                    "new_dbscan_candidate_id"
                ] == card["candidate_id"]
                for record in records
            )

            with st.expander(
                f"{card['candidate_id']} — "
                f"{card['complaint_category']} "
                f"({matched} selected reports)"
            ):

                st.write(
                    card["explanation"]
                )

                st.write(
                    "**Full-batch spatial reports:**",
                    card["incoming_complaints"]
                )

                st.write(
                    "**Distinct submitted locations:**",
                    card[
                        "distinct_submitted_locations"
                    ]
                )

                st.write(
                    "**Suggested verification:** "
                    + card["suggested_review"]
                )

    # ----------------------------------------------
    # Tab 5: Record-level evidence
    # ----------------------------------------------

    with tabs[4]:

        st.subheader(
            "Complaint-level evidence"
        )

        st.caption(
            "This table retains both new-batch "
            "signals and historical comparison "
            "statuses for traceability."
        )

        if not records:
            st.info(
                "No complaint records match the filters."
            )

        else:
            frame = pd.DataFrame(
                records
            )

            st.dataframe(
                frame,
                hide_index=True,
                use_container_width=True
            )

            st.download_button(
                "Export filtered complaints as CSV",
                data=frame.to_csv(
                    index=False
                ).encode("utf-8"),
                file_name=(
                    "urban_pulse_filtered_complaints.csv"
                ),
                mime="text/csv"
            )

    st.divider()

    st.caption(
        "Urban Infrastructure Pulse Intelligence "
        "| Synthetic-data pilot | "
        "Human verification required before "
        "operational action."
    )


if __name__ == "__main__":
    main()
''').lstrip()

ast.parse(app_source)

compiled_app = compile(
    app_source,
    str(app_path),
    "exec"
)

namespace = {
    "__name__": "_phase20_validation",
    "__file__": str(app_path)
}

exec(compiled_app, namespace)

load_feed = namespace["load_feed"]
validate_feed = namespace["validate_feed"]
filter_records = namespace["filter_records"]
build_filtered_view = namespace[
    "build_filtered_view"
]
signal_type = namespace["signal_type"]

# --------------------------------------------------
# 2. Load the real saved dashboard feed
# --------------------------------------------------

print("\n1. VALIDATING SAVED DASHBOARD DATA")

feed = load_feed(feed_path)
full_view = validate_feed(feed)

print("Reporting day:",
      feed["metadata"]["reporting_day"])

print("Synthetic data:",
      feed["metadata"]["synthetic_data"])

for name, value in full_view["summary"].items():
    print(f"{name}: {value}")

# --------------------------------------------------
# 3. Test interactive filter behavior
# --------------------------------------------------

print("\n2. TESTING DASHBOARD FILTERS")

rows = feed["records"]

bronx_rows = filter_records(
    rows,
    borough="BRONX"
)

bronx_sewer = filter_records(
    rows,
    borough="BRONX",
    category="SEWER"
)

both_rows = filter_records(
    rows,
    signal="Both signals"
)

anomaly_only_rows = filter_records(
    rows,
    signal="Anomaly only"
)

spatial_only_rows = filter_records(
    rows,
    signal="Spatial only"
)

neither_rows = filter_records(
    rows,
    signal="Neither"
)

queens_sewer_rows = [
    row
    for row in rows
    if (
        row["area_id"] == "QUEENS_11005"
        and row["complaint_category"] == "SEWER"
    )
]

queens_sewer_view = build_filtered_view(
    feed,
    queens_sewer_rows
)

road_pattern_rows = filter_records(
    rows,
    pattern="Road-heavy reporting"
)

# --------------------------------------------------
# 4. Validation checklist
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

summary = full_view["summary"]

check(
    "Dashboard Python source compiles",
    len(app_source) > 0
)

check(
    "All 500 complaint records load",
    summary["selected"] == 500
)

check(
    "15 anomaly groups represented",
    summary["anomaly_groups"] == 15
)

check(
    "23 spatial groups represented",
    summary["spatial_groups"] == 23
)

check(
    "204 anomaly-associated complaints",
    summary["anomaly_reports"] == 204
)

check(
    "228 spatial-associated complaints",
    summary["spatial_reports"] == 228
)

check(
    "156 complaints in both signals",
    summary["both"] == 156
)

check(
    "48 anomaly-only complaints",
    summary["anomaly_only"] == 48
)

check(
    "72 spatial-only complaints",
    summary["spatial_only"] == 72
)

check(
    "224 complaints in neither signal",
    summary["neither"] == 224
)

check(
    "276 distinct complaints in either signal",
    summary["distinct_either"] == 276
)

check(
    "All 20 reporting areas available",
    summary["areas"] == 20
)

check(
    "Borough filter selects 100 Bronx reports",
    len(bronx_rows) == 100
)

check(
    "Combined borough-category filter works",
    len(bronx_sewer) > 0
    and all(
        item["area_id"].startswith(
            "BRONX_"
        )
        and item["complaint_category"]
            == "SEWER"
        for item in bronx_sewer
    )
)

check(
    "Both-signal filter is correct",
    len(both_rows) == 156
)

check(
    "Anomaly-only filter is correct",
    len(anomaly_only_rows) == 48
)

check(
    "Spatial-only filter is correct",
    len(spatial_only_rows) == 72
)

check(
    "Neither-signal filter is correct",
    len(neither_rows) == 224
)

check(
    "All four signal filters partition 500",
    sum(map(len, [
        both_rows,
        anomaly_only_rows,
        spatial_only_rows,
        neither_rows
    ])) == 500
)

check(
    "Road-heavy reporting pattern filter works",
    len(road_pattern_rows) > 0
    and all(
        item["new_kmeans_pattern"]
        == "Road-heavy reporting"
        for item in road_pattern_rows
    )
)

check(
    "Sparse Queens sewer remains spatial-only",
    len(queens_sewer_rows) == 15
    and all(
        item["spatial_candidate"]
        and not item["anomaly_candidate"]
        for item in queens_sewer_rows
    )
)

check(
    "Sparse Queens sewer map candidate retained",
    "NB-00-004" in {
        feature["id"]
        for feature in queens_sewer_view[
            "map_features"
        ]
    }
)

check(
    "Sparse Queens sewer has no anomaly card",
    len(
        queens_sewer_view["anomaly_cards"]
    ) == 0
)

check(
    "All spatial cards have matching map points",
    len(
        full_view["map_features"]
    ) == 23
)

check(
    "Source data explicitly identified as synthetic",
    feed["metadata"]["synthetic_data"] is True
)

# Test an empty filter result.
empty_selection = filter_records(
    rows,
    borough="NONEXISTENT"
)

empty_view = build_filtered_view(
    feed,
    empty_selection
)

check(
    "Empty filter selection handled safely",
    empty_view["summary"]["selected"] == 0
    and not empty_view["anomaly_cards"]
    and not empty_view["map_features"]
)

# Check the original feed was not modified.
check(
    "Saved feed source unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == hashes_before[path]
        for path in protected
    )
)

results_20_1 = pd.DataFrame(checks)

print("\n3. PHASE 20.1 VALIDATION")

display(results_20_1)

passed = int(
    results_20_1["passed"].sum()
)
total = len(results_20_1)

print(f"\nChecks passed: {passed}/{total}")

failed = results_20_1.loc[
    ~results_20_1["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 5. Save the GUI only after validation
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("Dashboard GUI was not saved.")

    for item in failed:
        print("FAILED:", item)

else:

    if app_path.exists():
        assert app_path.read_text(
            encoding="utf-8"
        ) == app_source, (
            "An existing dashboard file differs. "
            "No overwrite performed."
        )
    else:
        app_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with app_path.open(
            "x",
            encoding="utf-8"
        ) as file:
            file.write(app_source)

    streamlit_installed = (
        importlib.util.find_spec(
            "streamlit"
        ) is not None
    )

    print("\n4. GUI FILE VERIFIED")
    print("Dashboard:", app_path)
    print("Streamlit installed:", streamlit_installed)

    print("\nSTATUS: PASSED")
    print("PHASE 20.1 DASHBOARD GUI CREATED")
    print("OVERVIEW METRICS IMPLEMENTED")
    print("BOROUGH AND CATEGORY FILTERS IMPLEMENTED")
    print("K-MEANS PATTERN VIEW IMPLEMENTED")
    print("ANOMALY INVESTIGATION CARDS IMPLEMENTED")
    print("SPATIAL CANDIDATE MAP IMPLEMENTED")
    print("COMPLAINT TABLE AND CSV EXPORT IMPLEMENTED")
    print("HISTORICAL DATA AND MODELS UNMODIFIED")

    if streamlit_installed:
        print("\nREADY TO LAUNCH FROM VS CODE TERMINAL")
        print(
            "python -m streamlit run "
            "dashboard/urban_pulse_dashboard.py"
        )
    else:
        print("\nSTREAMLIT PACKAGE NOT DETECTED")
        print(
            "Install in the active ml environment: "
            "python -m pip install streamlit"
        )
```

```python

from pathlib import Path
import ast
import hashlib
import json
import pandas as pd

print(" LIVE CSV UPLOAD INTEGRATION AUDIT")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

files = {
    "bulk_contract":
        root / "src/integration/bulk_complaint_contract.py",
    "bulk_processor":
        root / "src/integration/bulk_complaint_processor.py",
    "daily_window":
        root / "src/integration/daily_batch_window.py",
    "three_model_integration":
        root / "src/integration/anomaly_snapshot_integration.py",
    "investigation_briefs":
        root / "src/integration/smart_assistant_investigation_briefs.py",
    "dashboard_feed_builder":
        root / "src/integration/dashboard_intelligence_feed.py",
    "streamlit_gui":
        root / "dashboard/urban_pulse_dashboard.py",
    "upload_template":
        root / "outputs/templates/new_complaints_template.csv",
    "verified_500":
        root / "outputs/tests/phase_17_4c_backend_verified"
        / "verified_500_backend_results.csv",
    "dashboard_feed":
        root / "outputs/tests/phase_19_2e_dashboard_feed"
        / "dashboard_feed.json",
    "historical_snapshot":
        root / "outputs/tables/anomaly"
        / "historical_daily_counts_2023_2025_v1.csv.gz"
}

existing = {
    name: path
    for name, path in files.items()
    if path.is_file()
}

hashes_before = {
    name: hashlib.sha256(path.read_bytes()).hexdigest()
    for name, path in existing.items()
    if name != "historical_snapshot"
}

# --------------------------------------------------
# 1. File readiness
# --------------------------------------------------

print("\n1. INTEGRATION FILE READINESS")

readiness = pd.DataFrame([
    {
        "component": name,
        "available": path.is_file(),
        "size_kb": round(
            path.stat().st_size / 1024, 1
        ) if path.is_file() else None
    }
    for name, path in files.items()
])

display(readiness)

# --------------------------------------------------
# 2. Discover actual processor API without importing
#    or executing the backend module
# --------------------------------------------------

def inspect_python_api(path):

    source = path.read_text(
        encoding="utf-8-sig"
    )

    tree = ast.parse(
        source,
        filename=str(path)
    )

    entries = []

    for node in tree.body:

        if isinstance(
            node,
            (ast.FunctionDef, ast.AsyncFunctionDef)
        ):

            if node.name.startswith("_"):
                continue

            entries.append({
                "object": node.name,
                "type": "function",
                "signature": (
                    f"{node.name}("
                    f"{ast.unparse(node.args)})"
                ),
                "description": (
                    ast.get_docstring(node) or ""
                ).split("\n")[0][:160]
            })

        elif isinstance(node, ast.ClassDef):

            entries.append({
                "object": node.name,
                "type": "class",
                "signature": f"class {node.name}",
                "description": (
                    ast.get_docstring(node) or ""
                ).split("\n")[0][:160]
            })

            for method in node.body:

                if not isinstance(
                    method,
                    (
                        ast.FunctionDef,
                        ast.AsyncFunctionDef
                    )
                ):
                    continue

                if (
                    method.name.startswith("_")
                    and method.name != "__init__"
                ):
                    continue

                entries.append({
                    "object": (
                        f"{node.name}.{method.name}"
                    ),
                    "type": "method",
                    "signature": (
                        f"{method.name}("
                        f"{ast.unparse(method.args)})"
                    ),
                    "description": (
                        ast.get_docstring(method) or ""
                    ).split("\n")[0][:160]
                })

    return pd.DataFrame(entries), tree


print("\n2. BULK PROCESSOR PUBLIC API")

if files["bulk_processor"].is_file():

    processor_api, processor_tree = (
        inspect_python_api(
            files["bulk_processor"]
        )
    )

    display(processor_api.head(35))

else:

    processor_api = pd.DataFrame()

    print("Bulk processor file not found.")

print("\n3. BULK INPUT CONTRACT PUBLIC API")

if files["bulk_contract"].is_file():

    contract_api, contract_tree = (
        inspect_python_api(
            files["bulk_contract"]
        )
    )

    display(contract_api.head(35))

else:

    contract_api = pd.DataFrame()

    print("Bulk contract file not found.")

# --------------------------------------------------
# 3. Identify important backend constants and
#    validation settings
# --------------------------------------------------

print("\n4. INPUT CONTRACT CONSTANTS")

if files["bulk_contract"].is_file():

    contract_source = files[
        "bulk_contract"
    ].read_text(
        encoding="utf-8-sig"
    )

    constants = []

    for node in contract_tree.body:

        if not isinstance(
            node,
            (ast.Assign, ast.AnnAssign)
        ):
            continue

        if isinstance(node, ast.Assign):
            targets = node.targets
        else:
            targets = [node.target]

        names = [
            target.id
            for target in targets
            if isinstance(target, ast.Name)
        ]

        for name in names:

            if any(
                word in name.upper()
                for word in [
                    "REQUIRED",
                    "COLUMN",
                    "FIELD",
                    "STATUS",
                    "CATEGORY",
                    "SCHEMA",
                    "ALIAS"
                ]
            ):

                value = (
                    ast.get_source_segment(
                        contract_source,
                        node.value
                    ) or ""
                )

                constants.append({
                    "constant": name,
                    "definition": value[:450]
                })

    if constants:
        display(pd.DataFrame(constants))
    else:
        print(
            "No matching top-level constants. "
            "Review public functions above."
        )

# --------------------------------------------------
# 4. Compare raw-upload template with verified
#    processed backend output
# --------------------------------------------------

print("\n5. UPLOAD TEMPLATE AND VERIFIED BACKEND SCHEMAS")

schema_tables = {}

for name in ["upload_template", "verified_500"]:

    path = files[name]

    if not path.is_file():
        continue

    sample = pd.read_csv(
        path,
        nrows=3,
        dtype="string",
        keep_default_na=False
    )

    schema_tables[name] = sample

    print(f"\n{name}")
    print("Columns:", len(sample.columns))
    print(sample.columns.tolist())

    print("Sample records:", len(sample))

    if not sample.empty:
        display(sample.head(2))

if (
    "upload_template" in schema_tables
    and "verified_500" in schema_tables
):

    template_columns = set(
        schema_tables[
            "upload_template"
        ].columns
    )

    backend_columns = set(
        schema_tables[
            "verified_500"
        ].columns
    )

    print("\n6. SCHEMA DIFFERENCES")

    print(
        "Template columns absent from backend output:",
        sorted(template_columns - backend_columns)
    )

    print(
        "Backend-added output columns:",
        sorted(backend_columns - template_columns)
    )

# --------------------------------------------------
# 5. Verify saved dashboard contract
# --------------------------------------------------

print("\n7. DASHBOARD FEED STRUCTURE")

if files["dashboard_feed"].is_file():

    with files["dashboard_feed"].open(
        encoding="utf-8"
    ) as file:
        feed = json.load(file)

    print("Top-level sections:", sorted(feed.keys()))

    print("Reporting day:",
          feed["metadata"].get("reporting_day"))

    print("Synthetic demonstration:",
          feed["metadata"].get("synthetic_data"))

    print("Saved complaints:",
          len(feed.get("records", [])))

    print("Area cards:",
          len(feed.get("area_cards", [])))

    print("Anomaly cards:",
          len(feed.get("anomaly_cards", [])))

    print("Spatial cards:",
          len(feed.get("spatial_cards", [])))

else:

    print("Dashboard feed missing.")

# --------------------------------------------------
# 6. Verify Phase 20.1 GUI is available
# --------------------------------------------------

print("\n8. STREAMLIT GUI READINESS")

gui_path = files["streamlit_gui"]

if gui_path.is_file():

    gui_source = gui_path.read_text(
        encoding="utf-8"
    )

    ast.parse(gui_source)

    print("Dashboard file:", gui_path)

    print(
        "Uses Streamlit:",
        "import streamlit" in gui_source
    )

    print(
        "Contains filter interface:",
        "filter_records" in gui_source
    )

    print(
        "Contains anomaly view:",
        "Anomaly Reviews" in gui_source
    )

    print(
        "Contains DBSCAN map:",
        "DBSCAN Map" in gui_source
    )

else:

    print(
        "Phase 20.1 GUI file not yet present. "
        "Finish that stage before attaching uploads."
    )

# --------------------------------------------------
# 7. Final read-only audit checks
# --------------------------------------------------

checks = []

def check(name, condition):

    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Bulk input contract available",
    files["bulk_contract"].is_file()
)

check(
    "Bulk processor available",
    files["bulk_processor"].is_file()
)

check(
    "Bulk processor API discovered",
    len(processor_api) > 0
)

check(
    "Bulk contract API discovered",
    len(contract_api) > 0
)

check(
    "Upload CSV template available",
    files["upload_template"].is_file()
)

check(
    "Verified processed example available",
    files["verified_500"].is_file()
)

check(
    "Three-model integration available",
    files["three_model_integration"].is_file()
)

check(
    "Investigation brief builder available",
    files["investigation_briefs"].is_file()
)

check(
    "Dashboard data builder available",
    files["dashboard_feed_builder"].is_file()
)

check(
    "Phase 20.1 Streamlit GUI available",
    gui_path.is_file()
)

check(
    "Saved dashboard feed available",
    files["dashboard_feed"].is_file()
)

check(
    "Historical snapshot available",
    files["historical_snapshot"].is_file()
)

check(
    "All inspected files unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest() == hashes_before[name]
        for name, path in existing.items()
        if name != "historical_snapshot"
    )
)

results_20_2a = pd.DataFrame(checks)

print("\n9. PHASE 20.2A VALIDATION")

display(results_20_2a)

passed = int(results_20_2a["passed"].sum())
total = len(results_20_2a)

print(f"\nChecks passed: {passed}/{total}")

if passed == total:

    print("\nSTATUS: UPLOAD INTEGRATION CONTRACT READY")
    print("BACKEND AND GUI COMPONENTS LOCATED")

else:

    print("\nSTATUS: INTEGRATION PREREQUISITES NEED REVIEW")

print("NO FILES CREATED OR MODIFIED")
print("NO MODELS RETRAINED")
print("NO COMPLAINTS PROCESSED")
```

```python

from pathlib import Path
from textwrap import dedent
from io import BytesIO
import ast
import hashlib
import inspect
import sys
import pandas as pd

print(" STREAMLIT CSV UPLOAD VALIDATION")
print("=" * 75)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

from src.integration.bulk_complaint_contract import (
    INPUT_COLUMNS,
    REQUIRED_COLUMNS,
    read_batch_csv,
    validate_batch
)

from src.integration.bulk_complaint_processor import (
    process_complaint_batch
)

adapter_path = (
    root / "src/integration/upload_preflight.py"
)

page_path = (
    root / "dashboard/pages/1_Upload_Complaints.py"
)

template_path = (
    root / "outputs/templates/new_complaints_template.csv"
)

verified_path = (
    root / "outputs/tests/phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

taxonomy_path = (
    root / "outputs/tables/dbscan"
    / "dbscan_action_taxonomy_v1.csv"
)

dashboard_path = (
    root / "dashboard/urban_pulse_dashboard.py"
)

protected = [
    template_path,
    verified_path,
    taxonomy_path,
    dashboard_path,
    root / "src/integration/bulk_complaint_contract.py",
    root / "src/integration/bulk_complaint_processor.py"
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

hashes_before = {
    str(path): hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

categories = sorted({
    str(value).strip().upper()
    for value in pd.read_csv(taxonomy_path)[
        "complaint_category"
    ].dropna()
})

assert len(categories) == 15

print("\n1. VERIFIED UPLOAD CONTRACT")
print("Input columns:", list(INPUT_COLUMNS))
print("Required columns:", list(REQUIRED_COLUMNS))
print("Supported categories:", len(categories))

# --------------------------------------------------
# 1. Reusable CSV-upload preflight module
# --------------------------------------------------

adapter_source = dedent(r'''
"""Safe in-memory upload validation for NYC 311 batches.

Uses the existing bulk complaint contract.
This module does not run models or modify files.
"""

from io import BytesIO

import pandas as pd

from src.integration.bulk_complaint_contract import (
    REQUIRED_COLUMNS,
    read_batch_csv,
    validate_batch
)


MAX_UPLOAD_BYTES = 10 * 1024 * 1024
MAX_UPLOAD_ROWS = 5000


def _find_validation_frame(value, expected_rows):
    """Find row-level results in the contract output."""

    if isinstance(value, pd.DataFrame):
        if len(value) == expected_rows:
            return value.copy()
        return None

    if isinstance(value, dict):
        for item in value.values():
            found = _find_validation_frame(
                item, expected_rows
            )
            if found is not None:
                return found

    if isinstance(value, (tuple, list)):

        if (
            len(value) == expected_rows
            and all(
                isinstance(item, dict)
                for item in value
            )
        ):
            return pd.DataFrame(value)

        for item in value:
            found = _find_validation_frame(
                item, expected_rows
            )
            if found is not None:
                return found

    return None


def run_upload_preflight(
    csv_bytes,
    supported_categories,
    max_bytes=MAX_UPLOAD_BYTES,
    max_rows=MAX_UPLOAD_ROWS
):
    """Read and validate CSV, without processing it."""

    if not isinstance(
        csv_bytes, (bytes, bytearray)
    ):
        raise TypeError(
            "Uploaded CSV must be provided as bytes."
        )

    if len(csv_bytes) == 0:
        raise ValueError("Uploaded CSV is empty.")

    if len(csv_bytes) > max_bytes:
        raise ValueError(
            "Uploaded CSV exceeds the size limit."
        )

    frame = read_batch_csv(
        BytesIO(csv_bytes)
    )

    if not isinstance(frame, pd.DataFrame):
        raise ValueError(
            "CSV reader did not return a DataFrame."
        )

    if frame.columns.duplicated().any():
        raise ValueError(
            "Duplicate CSV column names."
        )

    missing = [
        column
        for column in REQUIRED_COLUMNS
        if column not in frame.columns
    ]

    if missing:
        raise ValueError(
            f"Missing required CSV columns: {missing}"
        )

    if frame.empty:
        raise ValueError(
            "CSV has no complaint rows."
        )

    if len(frame) > max_rows:
        raise ValueError(
            f"Maximum {max_rows} complaints per upload."
        )

    # The established project contract is
    # the sole source of row validation.
    validation = validate_batch(
        frame.copy(deep=True),
        supported_categories
    )

    results = _find_validation_frame(
        validation,
        len(frame)
    )

    if results is None:
        raise ValueError(
            "Could not identify one validation result "
            "per uploaded row. Review the bulk "
            "contract return structure."
        )

    status_column = next(
        (
            column for column in [
                "validation_status",
                "status"
            ]
            if column in results.columns
        ),
        None
    )

    if status_column is not None:
        status_counts = {
            str(key): int(value)
            for key, value in results[
                status_column
            ].astype(str).value_counts().items()
        }
    else:
        status_counts = {}

    return {
        "input_records": frame,
        "validation_results": results,
        "row_count": int(len(frame)),
        "status_column": status_column,
        "status_counts": status_counts,
        "model_processing_performed": False,
        "historical_models_modified": False
    }
''').lstrip()

compile(
    adapter_source,
    str(adapter_path),
    "exec"
)

adapter_namespace = {
    "__name__": "_upload_preflight_validation"
}

exec(
    compile(
        adapter_source,
        str(adapter_path),
        "exec"
    ),
    adapter_namespace
)

run_upload_preflight = adapter_namespace[
    "run_upload_preflight"
]

# --------------------------------------------------
# 2. Create a standalone Streamlit upload page
# --------------------------------------------------

page_source = dedent(r'''
"""CSV upload and validation page for Urban Pulse."""

from pathlib import Path
import sys

import pandas as pd
import streamlit as st


PROJECT_ROOT = Path(__file__).resolve().parents[2]

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.integration.upload_preflight import (
    MAX_UPLOAD_BYTES,
    MAX_UPLOAD_ROWS,
    run_upload_preflight
)


TEMPLATE_PATH = (
    PROJECT_ROOT
    / "outputs/templates/new_complaints_template.csv"
)

TAXONOMY_PATH = (
    PROJECT_ROOT
    / "outputs/tables/dbscan/dbscan_action_taxonomy_v1.csv"
)


st.set_page_config(
    page_title="Upload Complaints | Urban Pulse",
    page_icon="📄",
    layout="wide"
)

st.title("Upload New Complaints")

st.caption(
    "Phase 20.2 — CSV upload validation"
)

st.info(
    "This page checks uploaded complaint data "
    "using the existing project validation contract. "
    "K-Means, DBSCAN and anomaly screening are "
    "not run during this preflight stage."
)

if not TEMPLATE_PATH.is_file():
    st.error(
        "The upload template is missing."
    )
    st.stop()

if not TAXONOMY_PATH.is_file():
    st.error(
        "The category taxonomy is missing."
    )
    st.stop()

categories = sorted({
    str(value).strip().upper()
    for value in pd.read_csv(
        TAXONOMY_PATH
    )["complaint_category"].dropna()
})

st.download_button(
    "Download CSV Template",
    data=TEMPLATE_PATH.read_bytes(),
    file_name="new_complaints_template.csv",
    mime="text/csv"
)

st.markdown(
    "Upload a CSV containing complaint IDs, "
    "timestamps, coordinates and categories."
)

st.caption(
    f"Maximum file size: "
    f"{MAX_UPLOAD_BYTES // (1024 * 1024)} MB. "
    f"Maximum complaint rows: {MAX_UPLOAD_ROWS:,}."
)

uploaded = st.file_uploader(
    "Choose a new complaints CSV",
    type=["csv"],
    accept_multiple_files=False
)

if uploaded is not None:

    st.write(
        "**Uploaded file:**",
        uploaded.name
    )

    if st.button(
        "Validate Uploaded Complaints",
        type="primary"
    ):

        try:
            result = run_upload_preflight(
                uploaded.getvalue(),
                categories
            )

        except Exception as exc:
            st.error(
                "Upload validation could not "
                f"be completed: {exc}"
            )

        else:
            st.success(
                "CSV validation finished. "
                "No models were executed."
            )

            st.metric(
                "Submitted complaint rows",
                result["row_count"]
            )

            if result["status_counts"]:

                st.subheader(
                    "Validation status breakdown"
                )

                counts = pd.DataFrame([
                    {
                        "Validation status": key,
                        "Complaints": value
                    }
                    for key, value
                    in result[
                        "status_counts"
                    ].items()
                ])

                st.dataframe(
                    counts,
                    hide_index=True,
                    use_container_width=True
                )

            st.subheader(
                "Row-level validation results"
            )

            st.dataframe(
                result["validation_results"],
                hide_index=True,
                use_container_width=True
            )

            csv_output = (
                result["validation_results"]
                .to_csv(index=False)
                .encode("utf-8")
            )

            st.download_button(
                "Download Validation Results",
                data=csv_output,
                file_name=(
                    "upload_validation_results.csv"
                ),
                mime="text/csv"
            )

            st.warning(
                "Validation is not geographic model "
                "processing. These complaints have not "
                "yet been assigned K-Means patterns, "
                "DBSCAN concentrations or anomaly flags."
            )

st.divider()

st.caption(
    "Urban Infrastructure Pulse Intelligence "
    "| Upload validation only | "
    "No historical models modified"
)
''').lstrip()

ast.parse(page_source)

# --------------------------------------------------
# 3. Test actual input through existing contract
# --------------------------------------------------

print("\n2. TESTING REAL CSV UPLOAD CONTRACT")

template_result = run_upload_preflight(
    template_path.read_bytes(),
    categories
)

print(
    "Template records:",
    template_result["row_count"]
)

print(
    "Template validation statuses:",
    template_result["status_counts"]
)

display(
    template_result[
        "validation_results"
    ].head()
)

# Reconstruct the six-column raw input from
# previously verified synthetic backend records.
verified = pd.read_csv(
    verified_path,
    dtype="string",
    keep_default_na=False
)

raw_500 = verified.loc[
    :, list(INPUT_COLUMNS)
].copy()

raw_500_snapshot = raw_500.copy(deep=True)

csv_500 = raw_500.to_csv(
    index=False
).encode("utf-8")

batch_result = run_upload_preflight(
    csv_500,
    categories
)

print("\n3. TESTING 500-COMPLAINT UPLOAD")

print("Rows read:", batch_result["row_count"])

print(
    "Validation statuses:",
    batch_result["status_counts"]
)

display(
    batch_result[
        "validation_results"
    ].head(5)
)

# --------------------------------------------------
# 4. Test invalid and missing-file safeguards
# --------------------------------------------------

def is_rejected(data):
    try:
        run_upload_preflight(
            data,
            categories
        )
    except (
        ValueError,
        TypeError,
        pd.errors.EmptyDataError
    ):
        return True
    return False

empty_rejected = is_rejected(b"")

missing_column_bytes = (
    b"complaint_id,created_date\n"
    b"BAD-001,2026-11-09\n"
)

missing_columns_rejected = is_rejected(
    missing_column_bytes
)

oversized_rejected = is_rejected(
    b"x" * (10 * 1024 * 1024 + 1)
)

# --------------------------------------------------
# 5. Discover exact processor signature
# --------------------------------------------------

print("\n4. COMPLETE BULK PROCESSOR SIGNATURE")

processor_signature = inspect.signature(
    process_complaint_batch
)

print(
    "process_complaint_batch",
    processor_signature
)

parameters = pd.DataFrame([
    {
        "parameter": name,
        "kind": str(parameter.kind),
        "required": (
            parameter.default
            is inspect.Parameter.empty
        ),
        "default": (
            None
            if parameter.default
            is inspect.Parameter.empty
            else repr(parameter.default)[:150]
        )
    }
    for name, parameter
    in processor_signature.parameters.items()
])

display(parameters)

# --------------------------------------------------
# 6. Validate before saving new files
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Upload adapter compiles",
    bool(adapter_source)
)

check(
    "Streamlit page compiles",
    bool(page_source)
)

check(
    "Project upload template accepted",
    template_result["row_count"] == 2
)

check(
    "Template has row-level validation results",
    len(
        template_result[
            "validation_results"
        ]
    ) == 2
)

check(
    "Verified synthetic upload has 500 rows",
    batch_result["row_count"] == 500
)

check(
    "500 rows receive validation results",
    len(
        batch_result[
            "validation_results"
        ]
    ) == 500
)

check(
    "Required CSV fields preserved",
    set(REQUIRED_COLUMNS).issubset(
        batch_result[
            "input_records"
        ].columns
    )
)

check(
    "Original uploaded text preserved",
    batch_result[
        "input_records"
    ]["complaint_id"].astype(str).tolist()
    == raw_500["complaint_id"].astype(str).tolist()
)

check(
    "Original ZIP text preserved",
    batch_result[
        "input_records"
    ]["incident_zip"].astype(str).tolist()
    == raw_500["incident_zip"].astype(str).tolist()
)

check(
    "Empty upload rejected",
    empty_rejected
)

check(
    "Missing mandatory columns rejected",
    missing_columns_rejected
)

check(
    "Oversized CSV rejected",
    oversized_rejected
)

check(
    "Upload validation does not run models",
    batch_result[
        "model_processing_performed"
    ] is False
)

check(
    "Existing input DataFrame unchanged",
    raw_500.equals(raw_500_snapshot)
)

check(
    "Backend processor signature identified",
    "frame" in processor_signature.parameters
)

check(
    "Historical model artifacts untouched",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest()
        == hashes_before[str(path)]
        for path in protected
    )
)

results_20_2b = pd.DataFrame(checks)

print("\n5. PHASE 20.2B VALIDATION")

display(results_20_2b)

passed = int(
    results_20_2b["passed"].sum()
)

total = len(results_20_2b)

print(f"\nChecks passed: {passed}/{total}")

failures = results_20_2b.loc[
    ~results_20_2b["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 7. Save only after all tests pass
# --------------------------------------------------

if failures:

    print("\nSTATUS: NEEDS REVIEW")
    print("No upload GUI files saved.")

    for failure in failures:
        print("FAILED:", failure)

else:

    destinations = {
        adapter_path: adapter_source,
        page_path: page_source
    }

    for path, content in destinations.items():

        if path.exists():
            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing file differs: {path}. "
                "No overwrite performed."
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():
            with path.open(
                "x",
                encoding="utf-8"
            ) as file:
                file.write(content)

    print("\n6. VERIFIED UPLOAD GUI FILES")

    print("Upload adapter:", adapter_path)
    print("Streamlit upload page:", page_path)

    print("\nSTATUS: PASSED")
    print("PHASE 20.2B CSV UPLOAD VALIDATION READY")
    print("EXISTING BULK CONTRACT REUSED")
    print("UPLOAD SIZE AND COLUMN SAFEGUARDS VERIFIED")
    print("500-ROW INPUT CONTRACT TESTED")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")
    print("LIVE MODEL PROCESSING NOT YET ENABLED")
```

```python

from pathlib import Path
import ast
import inspect
import json
import os
import textwrap

from src.integration.bulk_complaint_processor import (
    process_complaint_batch
)

print("BACKEND ADAPTER DISCOVERY")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

# --------------------------------------------------
# 1. Identify exactly how adapter is called
# --------------------------------------------------

source = textwrap.dedent(
    inspect.getsource(process_complaint_batch)
)

tree = ast.parse(source)

function = tree.body[0]

print("\n1. ADAPTER CALLS INSIDE BULK PROCESSOR")

adapter_calls = []

for node in ast.walk(function):

    if not isinstance(node, ast.Call):
        continue

    target = node.func

    is_adapter_call = (
        isinstance(target, ast.Name)
        and target.id == "adapter"
    )

    is_adapter_method = (
        isinstance(target, ast.Attribute)
        and isinstance(target.value, ast.Name)
        and target.value.id == "adapter"
    )

    if is_adapter_call or is_adapter_method:

        adapter_calls.append({
            "line": node.lineno,
            "expression": ast.unparse(node)[:550]
        })

for call in adapter_calls:
    print(
        f"Line {call['line']}: "
        f"{call['expression']}"
    )

if not adapter_calls:
    print(
        "No direct adapter calls found. "
        "Checking other references below."
    )

# --------------------------------------------------
# 2. Print only relevant adapter source lines
# --------------------------------------------------

print("\n2. ADAPTER IMPLEMENTATION CONTEXT")

source_lines = source.splitlines()

relevant_lines = set()

for index, line in enumerate(source_lines):

    if (
        "adapter" in line
        or "backend_evaluations" in line
        or "responses_by_source_row" in line
    ):

        for position in range(
            max(0, index - 3),
            min(len(source_lines), index + 7)
        ):
            relevant_lines.add(position)

for index in sorted(relevant_lines):

    print(
        f"{index + 1:03d}: "
        f"{source_lines[index]}"
    )

# --------------------------------------------------
# 3. Identify processor return fields
# --------------------------------------------------

print("\n3. PROCESSOR RETURN CONTRACT")

return_nodes = [
    node
    for node in ast.walk(function)
    if isinstance(node, ast.Return)
]

for number, node in enumerate(
    return_nodes,
    start=1
):

    value = node.value

    if isinstance(value, ast.Dict):

        keys = []

        for key in value.keys:

            if isinstance(key, ast.Constant):
                keys.append(str(key.value))
            else:
                keys.append(ast.unparse(key))

        print(
            f"Return {number} dictionary keys:",
            keys
        )

    else:

        print(
            f"Return {number}:",
            ast.unparse(value)[:350]
            if value is not None
            else "None"
        )

# --------------------------------------------------
# 4. Search actual project call sites
# --------------------------------------------------

print("\n4. EXISTING PROCESSOR USAGE EXAMPLES")

search_roots = [
    root / "src",
    root / "notebooks",
    root / "dashboard"
]

skip_directories = {
    ".git",
    ".venv",
    "venv",
    "__pycache__",
    ".ipynb_checkpoints",
    "node_modules",
    "site-packages"
}

examples = []
MAX_EXAMPLES = 8

def inspect_code_lines(code, location):

    lines = code.splitlines()

    for index, line in enumerate(lines):

        if "process_complaint_batch(" not in line:
            continue

        # Ignore its function definition.
        if line.lstrip().startswith(
            ("def ", "async def ")
        ):
            continue

        start = max(0, index - 8)
        end = min(
            len(lines),
            index + 16
        )

        excerpt = "\n".join(
            f"{position + 1:03d}: {lines[position]}"
            for position in range(start, end)
        )

        examples.append({
            "location": location,
            "excerpt": excerpt
        })

        if len(examples) >= MAX_EXAMPLES:
            return

for search_root in search_roots:

    if not search_root.is_dir():
        continue

    if len(examples) >= MAX_EXAMPLES:
        break

    for current_dir, directories, filenames in os.walk(
        search_root
    ):

        directories[:] = sorted(
            name for name in directories
            if name not in skip_directories
        )

        for filename in sorted(filenames):

            if len(examples) >= MAX_EXAMPLES:
                break

            path = Path(current_dir) / filename

            if path.suffix not in {
                ".py", ".ipynb"
            }:
                continue

            try:

                if path.suffix == ".py":

                    code = path.read_text(
                        encoding="utf-8-sig"
                    )

                    inspect_code_lines(
                        code,
                        str(path.relative_to(root))
                    )

                else:

                    notebook = json.loads(
                        path.read_text(
                            encoding="utf-8"
                        )
                    )

                    for cell_index, cell in enumerate(
                        notebook.get("cells", [])
                    ):

                        if cell.get(
                            "cell_type"
                        ) != "code":
                            continue

                        content = cell.get(
                            "source", []
                        )

                        code = (
                            "".join(content)
                            if isinstance(content, list)
                            else str(content)
                        )

                        if (
                            "process_complaint_batch("
                            not in code
                        ):
                            continue

                        inspect_code_lines(
                            code,
                            (
                                str(path.relative_to(root))
                                + f" | cell {cell_index}"
                            )
                        )

                        if len(examples) >= MAX_EXAMPLES:
                            break

            except (
                OSError,
                UnicodeError,
                json.JSONDecodeError
            ):
                continue

for number, example in enumerate(
    examples,
    start=1
):

    print(f"\nEXAMPLE {number}")
    print("Location:", example["location"])
    print(example["excerpt"])

if not examples:
    print(
        "No previous processor invocation found "
        "in src, notebooks or dashboard."
    )

# --------------------------------------------------
# 5. Check currently loaded adapter variables
# --------------------------------------------------

print("\n5. POSSIBLE EXISTING NOTEBOOK ADAPTERS")

adapter_variables = []

for name, value in list(globals().items()):

    if name.startswith("_"):
        continue

    if "adapter" not in name.lower():
        continue

    if inspect.ismodule(value):
        continue

    adapter_variables.append({
        "variable": name,
        "python_type": type(value).__name__,
        "callable": callable(value)
    })

if adapter_variables:

    for item in adapter_variables:

        print(
            f"{item['variable']} | "
            f"type={item['python_type']} | "
            f"callable={item['callable']}"
        )

else:

    print(
        "No adapter-named object currently "
        "loaded in notebook memory."
    )

# --------------------------------------------------
# 6. Completion
# --------------------------------------------------

print("\n6. AUDIT SUMMARY")

print(
    "Direct adapter calls:",
    len(adapter_calls)
)

print(
    "Existing processor usage examples:",
    len(examples)
)

print(
    "Adapter-related notebook variables:",
    len(adapter_variables)
)

print("\nSTATUS: ADAPTER DISCOVERY COMPLETED")
print("NO COMPLAINTS PROCESSED")
print("NO FILES MODIFIED")
print("NO MODELS RETRAINED")
```

```python

from pathlib import Path
import ast
import inspect
import json
import textwrap

from src.integration.bulk_complaint_processor import (
    process_complaint_batch
)

print(" FOCUSED ADAPTER INSPECTION")
print("=" * 65)

root = Path(PROJECT_ROOT).resolve()

# 1. Processor return structure
tree = ast.parse(
    textwrap.dedent(
        inspect.getsource(process_complaint_batch)
    )
)

print("\n1. PROCESSOR RETURN STRUCTURE")

for node in ast.walk(tree):
    if not isinstance(node, ast.Return):
        continue

    if isinstance(node.value, ast.Dict):
        print(
            "Return keys:",
            [
                ast.literal_eval(k)
                if isinstance(k, ast.Constant)
                else ast.unparse(k)
                for k in node.value.keys
            ]
        )
    elif node.value is not None:
        print(
            "Return expression:",
            ast.unparse(node.value)[:350]
        )

# 2. Find actual adapter classes in src
print("\n2. EXISTING ADAPTER CLASSES")

adapter_classes = []

for path in sorted((root / "src").rglob("*.py")):

    if "__pycache__" in path.parts:
        continue

    try:
        source = path.read_text(
            encoding="utf-8-sig"
        )
        module_tree = ast.parse(source)
    except (OSError, UnicodeError, SyntaxError):
        continue

    for cls in ast.walk(module_tree):
        if not isinstance(cls, ast.ClassDef):
            continue

        methods = {
            item.name: item
            for item in cls.body
            if isinstance(
                item,
                (ast.FunctionDef, ast.AsyncFunctionDef)
            )
        }

        if "evaluate" not in methods:
            continue

        constructor = methods.get("__init__")
        evaluator = methods["evaluate"]

        info = {
            "class": cls.name,
            "file": str(path.relative_to(root)),
            "constructor": (
                ast.unparse(constructor.args)
                if constructor else "default"
            ),
            "evaluate": ast.unparse(evaluator.args)
        }

        adapter_classes.append(info)

for info in adapter_classes[:10]:
    print("\nClass:", info["class"])
    print("File:", info["file"])
    print("Constructor:", info["constructor"])
    print("Evaluate:", info["evaluate"])

print(
    "\nAdapter classes discovered:",
    len(adapter_classes)
)

# 3. Existing adapter objects in notebook memory
print("\n3. LOADED NOTEBOOK ADAPTER OBJECTS")

objects_found = 0

for name, obj in list(globals().items()):

    if name.startswith("_") or inspect.isclass(obj):
        continue

    if not callable(
        getattr(type(obj), "evaluate", None)
    ):
        continue

    print("Variable:", name)
    print("Type:", type(obj).__name__)
    print("Module:", type(obj).__module__)

    try:
        print(
            "Evaluate signature:",
            inspect.signature(obj.evaluate)
        )
    except (ValueError, TypeError):
        pass

    objects_found += 1

    if objects_found >= 5:
        break

if objects_found == 0:
    print("No loaded adapter instance found.")

# 4. Find existing processor calls in project code
print("\n4. EXISTING PROCESSOR INVOCATIONS")

call_sites = []

search_paths = (
    list((root / "src").rglob("*.py"))
    + list((root / "notebooks").glob("*.ipynb"))
)

for path in search_paths:

    if "__pycache__" in path.parts:
        continue

    try:
        if path.suffix == ".ipynb":
            notebook = json.loads(
                path.read_text(encoding="utf-8")
            )

            code_units = [
                "".join(cell.get("source", []))
                for cell in notebook.get("cells", [])
                if cell.get("cell_type") == "code"
            ]
        else:
            code_units = [
                path.read_text(encoding="utf-8-sig")
            ]

        for index, code in enumerate(code_units):

            if "process_complaint_batch(" not in code:
                continue

            try:
                code_tree = ast.parse(code)
            except SyntaxError:
                continue

            for node in ast.walk(code_tree):

                if not isinstance(node, ast.Call):
                    continue

                func = node.func

                if not (
                    isinstance(func, ast.Name)
                    and func.id == "process_complaint_batch"
                ):
                    continue

                call_sites.append({
                    "file": str(path.relative_to(root)),
                    "cell": index if path.suffix == ".ipynb"
                            else None,
                    "call": ast.unparse(node)[:650]
                })

    except (OSError, UnicodeError, json.JSONDecodeError):
        continue

for item in call_sites[:6]:

    print("\nFile:", item["file"])

    if item["cell"] is not None:
        print("Notebook cell index:", item["cell"])

    print("Call:", item["call"])

print("\nExisting calls discovered:", len(call_sites))

# 5. Completion
print("\n5. RESULT")
print("STATUS: ADAPTER CONTRACT INSPECTED")
print("NO FILES CREATED OR MODIFIED")
print("NO COMPLAINTS PROCESSED")
print("NO HISTORICAL MODELS MODIFIED")
```

```python

from pathlib import Path
import inspect
import json
import re

print(" ADAPTER INITIALIZATION AUDIT")
print("=" * 68)

root = Path(PROJECT_ROOT).resolve()

assert "adapter" in globals(), (
    "The working adapter is not loaded. "
    "Run the notebook setup cell first."
)

# --------------------------------------------------
# 1. Inspect actual adapter dependency hierarchy
# --------------------------------------------------

print("\n1. WORKING ADAPTER DEPENDENCY TREE")

visited = set()
discovered_classes = {}

def inspect_object(obj, label, depth=0):

    if depth > 7 or len(visited) >= 45:
        return

    object_id = id(obj)

    if object_id in visited:
        return

    visited.add(object_id)

    cls = type(obj)
    module = cls.__module__

    prefix = "  " * depth

    print(
        f"{prefix}{label}: "
        f"{module}.{cls.__name__}"
    )

    if not module.startswith("src."):
        return

    discovered_classes[
        (module, cls.__name__)
    ] = cls

    try:
        attributes = vars(obj)
    except TypeError:
        return

    for name, value in attributes.items():

        if name.startswith("__"):
            continue

        value_type = type(value)
        value_module = value_type.__module__

        if value_module.startswith("src."):

            inspect_object(
                value,
                name,
                depth + 1
            )

        elif isinstance(value, Path):

            print(
                f"{prefix}  {name}: "
                f"Path({value})"
            )

        elif isinstance(
            value,
            (str, int, float, bool, type(None))
        ):

            # Show only short non-sensitive
            # configuration values.
            if any(
                term in name.lower()
                for term in [
                    "path",
                    "file",
                    "tolerance",
                    "radius",
                    "threshold"
                ]
            ):

                print(
                    f"{prefix}  {name}: "
                    f"{str(value)[:150]}"
                )

            else:

                print(
                    f"{prefix}  {name}: "
                    f"{value_type.__name__}"
                )

        else:

            shape = getattr(
                value, "shape", None
            )

            detail = (
                f" shape={shape}"
                if shape is not None else ""
            )

            print(
                f"{prefix}  {name}: "
                f"{value_module}."
                f"{value_type.__name__}"
                f"{detail}"
            )

inspect_object(
    adapter,
    "adapter"
)

# --------------------------------------------------
# 2. Inspect constructors of actual dependencies
# --------------------------------------------------

print("\n2. SERVICE CONSTRUCTORS")

for (module, class_name), cls in (
    discovered_classes.items()
):

    print(f"\n{module}.{class_name}")

    try:
        print(
            "Constructor signature:",
            inspect.signature(cls)
        )

        lines = inspect.getsource(
            cls.__init__
        ).splitlines()

        # Constructor only, not full service code.
        for line in lines[:28]:
            print(line)

        if len(lines) > 28:
            print(
                f"... {len(lines) - 28} "
                "additional constructor lines"
            )

    except (
        OSError,
        TypeError,
        ValueError
    ) as error:

        print(
            "Source unavailable:",
            type(error).__name__
        )

# --------------------------------------------------
# 3. Find the notebook's original initialization
# --------------------------------------------------

print("\n3. ORIGINAL NOTEBOOK ADAPTER SETUP")

notebook_path = (
    root / "notebooks"
    / "06_smart_assistant_bulk_intelligence.ipynb"
)

assert notebook_path.is_file()

notebook = json.loads(
    notebook_path.read_text(
        encoding="utf-8"
    )
)

setup_patterns = [
    r"\badapter\s*=",
    r"InfrastructureGUIAdapter\s*\(",
    r"CombinedInfrastructureInsightService\s*\(",
    r"ZCTAAwareCombinedInsightService\s*\(",
    r"ValidatedNewComplaintService\s*\(",
    r"NewComplaintInsightService\s*\(",
    r"ZCTACoordinateValidator\s*\("
]

matches = []

for cell_index, cell in enumerate(
    notebook.get("cells", [])
):

    if cell.get("cell_type") != "code":
        continue

    raw_source = cell.get(
        "source", []
    )

    code = (
        "".join(raw_source)
        if isinstance(raw_source, list)
        else str(raw_source)
    )

    lines = code.splitlines()

    positions = [
        i
        for i, line in enumerate(lines)
        if any(
            re.search(pattern, line)
            for pattern in setup_patterns
        )
    ]

    if not positions:
        continue

    # Prefer original setup cells instead of
    # long downstream processing examples.
    score = sum(
        2 if "InfrastructureGUIAdapter(" in lines[i]
        else 1
        for i in positions
    )

    matches.append(
        (score, cell_index, lines, positions)
    )

matches.sort(
    key=lambda item: (
        -item[0],
        item[1]
    )
)

print(
    "Notebook cells with adapter setup references:",
    len(matches)
)

for score, cell_index, lines, positions in matches[:5]:

    print(
        f"\n--- NOTEBOOK CELL {cell_index} ---"
    )

    selected = set()

    for pos in positions:

        selected.update(
            range(
                max(0, pos - 7),
                min(len(lines), pos + 11)
            )
        )

    # Keep the report small enough to share.
    selected = sorted(selected)[:65]

    for index in selected:

        print(
            f"{index + 1:03d}: "
            f"{lines[index][:190]}"
        )

    if len(selected) == 65:
        print("... excerpt limited to 65 lines")

# --------------------------------------------------
# 4. Show relevant loaded configuration types
# --------------------------------------------------

print("\n4. REUSABLE SETUP VARIABLES")

interesting_names = [
    "adapter",
    "insight_service",
    "combined_service",
    "validated_service",
    "zcta_validator",
    "predictor",
    "engine",
    "payload_builder",
    "kmeans_lookup"
]

for name in interesting_names:

    if name in globals():

        value = globals()[name]

        print(
            name,
            "=>",
            type(value).__module__,
            type(value).__name__
        )

# --------------------------------------------------
# 5. Final status
# --------------------------------------------------

print("\n5. AUDIT RESULT")

print(
    "Working adapter:",
    type(adapter).__name__
)

print(
    "Project service classes discovered:",
    len(discovered_classes)
)

print(
    "Notebook initialization references:",
    len(matches)
)

print("\nSTATUS: ADAPTER INITIALIZATION INSPECTED")
print("NO COMPLAINTS PROCESSED")
print("NO FILES MODIFIED")
print("NO HISTORICAL MODELS MODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import ast
import json
import sys
import pandas as pd
from threadpoolctl import threadpool_limits

print("LIVE CSV UPLOAD PIPELINE")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

bridge_path = (
    root / "src/integration"
    / "live_uploaded_batch_pipeline.py"
)

page_path = (
    root / "dashboard/pages"
    / "2_Process_New_Batch.py"
)

verified_path = (
    root / "outputs/tests"
    / "phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

snapshot_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1.csv.gz"
)

manifest_path = (
    root / "outputs/tables/anomaly"
    / "historical_daily_counts_2023_2025_v1_manifest.json"
)

protected = [
    verified_path,
    snapshot_path,
    manifest_path,
    root / "src/ui/service_factory.py",
    root / "src/integration/bulk_complaint_processor.py",
    root / "src/integration/upload_preflight.py",
    root / "src/integration/anomaly_snapshot_integration.py",
    root / "src/integration/smart_assistant_investigation_briefs.py",
    root / "src/integration/dashboard_intelligence_feed.py",
    root / "dashboard/urban_pulse_dashboard.py",
    root / "dashboard/pages/1_Upload_Complaints.py"
]

for path in protected:
    assert path.is_file(), f"Missing prerequisite: {path}"

signatures_before = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

# --------------------------------------------------
# 1. Build reusable upload-processing bridge
# --------------------------------------------------

bridge_source = dedent(r'''
"""Process uploaded complaints using existing models.

Historical models and source files remain read-only.
Uploaded results remain in memory until exported.
"""

from pathlib import Path

import pandas as pd

from src.integration.upload_preflight import (
    run_upload_preflight
)

from src.integration.bulk_complaint_processor import (
    process_complaint_batch
)

from src.integration.daily_batch_window import (
    select_daily_complaints
)

from src.integration.anomaly_snapshot_integration import (
    load_anomaly_snapshot,
    analyze_batch_with_anomaly_context
)

from src.integration.smart_assistant_investigation_briefs import (
    build_investigation_briefs
)

from src.integration.dashboard_intelligence_feed import (
    build_dashboard_feed
)

from src.ui.service_factory import get_gui_adapter


def load_upload_baseline(project_root):

    root = Path(project_root).resolve()

    return load_anomaly_snapshot(
        root / "outputs/tables/anomaly"
        / "historical_daily_counts_2023_2025_v1.csv.gz",

        root / "outputs/tables/anomaly"
        / "historical_daily_counts_2023_2025_v1_manifest.json"
    )


def build_verified_day_feed(
    processed_records,
    baseline,
    reporting_day,
    synthetic_data=True
):
    """Analyze only geographically verified selected-day rows."""

    if not isinstance(
        processed_records,
        pd.DataFrame
    ):
        raise TypeError(
            "Backend records must be a DataFrame."
        )

    required = {
        "source_row",
        "complaint_id",
        "created_date",
        "borough",
        "incident_zip",
        "complaint_category",
        "latitude",
        "longitude",
        "processing_status",
        "zip_validation_status"
    }

    missing = required - set(
        processed_records.columns
    )

    if missing:
        raise ValueError(
            f"Backend missing columns: {sorted(missing)}"
        )

    if processed_records[
        "source_row"
    ].duplicated().any():
        raise ValueError(
            "Duplicate backend source_row values."
        )

    quality_ok = (
        processed_records[
            "processing_status"
        ].astype(str).eq("processed")
        &
        processed_records[
            "zip_validation_status"
        ].astype(str).eq("zcta_consistent")
    )

    verified = processed_records.loc[
        quality_ok
    ].copy()

    window = select_daily_complaints(
        verified,
        reporting_day
    )

    selected = window[
        "selected_records"
    ].copy()

    details = {
        "backend_records":
            int(len(processed_records)),
        "geographically_verified":
            int(len(verified)),
        "excluded_by_backend_quality":
            int((~quality_ok).sum()),
        "excluded_by_reporting_day":
            int(len(verified) - len(selected)),
        "selected_for_new_models":
            int(len(selected))
    }

    if selected.empty:
        return {
            "feed": None,
            "briefs": None,
            "integrated": None,
            "selection_stats": details,
            "status":
                "no_eligible_reporting_day_records"
        }

    integrated = analyze_batch_with_anomaly_context(
        selected,
        baseline,
        reporting_day
    )

    briefs = build_investigation_briefs(
        integrated,
        synthetic_data=synthetic_data
    )

    feed = build_dashboard_feed(
        integrated,
        briefs
    )

    feed["metadata"]["input_origin"] = (
        "uploaded_csv"
    )

    feed["metadata"]["data_verified_in_field"] = (
        False
    )

    feed["metadata"]["upload_note"] = (
        "Uploaded reports were geographically "
        "screened. No infrastructure fault "
        "has been independently verified."
    )

    if feed["summary"][
        "input_complaints"
    ] != len(selected):
        raise AssertionError(
            "Dashboard record count mismatch."
        )

    if len(
        briefs["record_audit"]
    ) != len(selected):
        raise AssertionError(
            "Investigation audit lost records."
        )

    return {
        "feed": feed,
        "briefs": briefs,
        "integrated": integrated,
        "selection_stats": details,
        "status": "analyzed"
    }


def process_uploaded_csv_bytes(
    csv_bytes,
    reporting_day,
    project_root,
    synthetic_data=True,
    adapter=None,
    baseline=None
):
    """Run validation, backend and three-model analysis.

    The backend audit retains all uploaded rows.
    The dashboard contains only verified rows
    belonging to the selected reporting day.
    """

    if baseline is None:
        baseline = load_upload_baseline(
            project_root
        )

    categories = baseline[
        "supported_categories"
    ]

    preflight = run_upload_preflight(
        csv_bytes,
        categories
    )

    original = preflight[
        "input_records"
    ].copy(deep=True)

    if adapter is None:
        adapter = get_gui_adapter()

    backend = process_complaint_batch(
        original,
        categories,
        adapter,
        use_cache=True
    )

    if (
        not isinstance(backend, dict)
        or "records" not in backend
    ):
        raise ValueError(
            "Unexpected bulk backend response."
        )

    rows = backend["records"]

    if isinstance(rows, pd.DataFrame):
        rows = rows.copy(deep=True)

    elif isinstance(rows, (list, tuple)):
        rows = pd.DataFrame(list(rows))

    else:
        raise TypeError(
            "Unexpected backend records container."
        )

    if len(rows) != len(original):
        raise ValueError(
            "Backend dropped uploaded rows."
        )

    analysis = build_verified_day_feed(
        rows,
        baseline,
        reporting_day,
        synthetic_data=synthetic_data
    )

    return {
        **analysis,
        "uploaded_rows": int(len(original)),
        "backend_records": rows,
        "backend_stats": backend.get(
            "stats", {}
        ),
        "preflight_status_counts":
            preflight["status_counts"]
    }
''').lstrip()

# --------------------------------------------------
# 2. Build Streamlit processing page
# --------------------------------------------------

page_source = dedent(r'''
"""Process uploaded complaints with existing backend."""

from datetime import date
from pathlib import Path
import hashlib
import json
import sys

import pandas as pd
import streamlit as st


ROOT = Path(__file__).resolve().parents[2]

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.integration.live_uploaded_batch_pipeline import (
    process_uploaded_csv_bytes,
    load_upload_baseline
)

from src.ui.service_factory import (
    get_gui_adapter
)

from dashboard.urban_pulse_dashboard import (
    validate_feed,
    filter_records,
    build_filtered_view
)


st.set_page_config(
    page_title="Process New Batch | Urban Pulse",
    page_icon="🏙️",
    layout="wide"
)

st.title("Process New Complaint Batch")

st.caption(
    "Upload CSV → Geographic backend → "
    "K-Means + DBSCAN + Anomaly screening"
)

st.warning(
    "Reporting evidence only. Results do not "
    "confirm infrastructure defects or "
    "determine repair priority."
)


@st.cache_resource(
    show_spinner="Loading historical services..."
)
def existing_adapter():
    return get_gui_adapter()


@st.cache_resource(
    show_spinner="Loading historical daily baseline..."
)
def existing_baseline():
    return load_upload_baseline(ROOT)


uploaded_file = st.file_uploader(
    "Choose complaint CSV",
    type=["csv"],
    key="batch_csv"
)

reporting_day = st.date_input(
    "NYC reporting day",
    value=date.today(),
    key="batch_day"
)

synthetic = st.checkbox(
    "Synthetic / test data",
    value=True,
    help=(
        "Leave selected for demonstrations. "
        "Uploaded real-world reports are not "
        "independently verified."
    )
)

st.caption(
    "Maximum upload: 10 MB and 5,000 rows. "
    "Historical models remain read-only."
)

upload_key = None

if uploaded_file is not None:

    payload = uploaded_file.getvalue()

    upload_key = (
        hashlib.sha256(payload).hexdigest(),
        reporting_day.isoformat(),
        bool(synthetic)
    )

if st.button(
    "Process Uploaded Complaints",
    type="primary",
    disabled=uploaded_file is None
):

    try:

        with st.spinner(
            "Validating, geographically verifying "
            "and analyzing complaints..."
        ):

            result = process_uploaded_csv_bytes(
                payload,
                reporting_day.isoformat(),
                ROOT,
                synthetic_data=synthetic,
                adapter=existing_adapter(),
                baseline=existing_baseline()
            )

        st.session_state[
            "urban_pulse_live_result"
        ] = result

        st.session_state[
            "urban_pulse_live_key"
        ] = upload_key

    except Exception as exc:

        st.session_state.pop(
            "urban_pulse_live_result",
            None
        )

        st.session_state.pop(
            "urban_pulse_live_key",
            None
        )

        st.error(
            f"Batch processing failed: "
            f"{type(exc).__name__}: {exc}"
        )


result = st.session_state.get(
    "urban_pulse_live_result"
)

if (
    result is None
    or st.session_state.get(
        "urban_pulse_live_key"
    ) != upload_key
):

    st.info(
        "Upload a CSV and press "
        "Process Uploaded Complaints."
    )

    st.stop()


st.success(
    "Backend processing completed."
)

st.subheader("Processing audit")

metrics = result["selection_stats"]

columns = st.columns(4)

columns[0].metric(
    "Uploaded",
    result["uploaded_rows"]
)

columns[1].metric(
    "Geographically verified",
    metrics["geographically_verified"]
)

columns[2].metric(
    "Selected reporting day",
    metrics["selected_for_new_models"]
)

columns[3].metric(
    "Excluded",
    (
        metrics["excluded_by_backend_quality"]
        + metrics["excluded_by_reporting_day"]
    )
)

st.caption(
    "All uploaded rows remain available in the "
    "backend audit. Only geographically verified "
    "reports from the selected NYC reporting day "
    "enter the new-batch models."
)

with st.expander(
    "Backend audit — including rejected records"
):

    st.dataframe(
        result["backend_records"],
        use_container_width=True,
        hide_index=True
    )

    st.download_button(
        "Download backend audit CSV",
        result["backend_records"]
        .to_csv(index=False)
        .encode("utf-8"),
        "uploaded_batch_backend_audit.csv",
        "text/csv"
    )


feed = result["feed"]

if feed is None:

    st.info(
        "No geographically verified complaints "
        "matched the selected reporting day. "
        "No model conclusions were generated."
    )

    st.stop()


try:
    validate_feed(feed)

except (ValueError, KeyError, TypeError) as exc:

    st.error(
        f"Dashboard feed validation failed: {exc}"
    )

    st.stop()


st.subheader("Verified-day intelligence")

filters = feed["filters"]

f1, f2, f3 = st.columns(3)

borough = f1.selectbox(
    "Borough",
    ["All"] + filters["boroughs"]
)

category = f2.selectbox(
    "Category",
    ["All"] + filters["categories"]
)

signal = f3.selectbox(
    "Signal",
    [
        "All",
        "Both signals",
        "Anomaly only",
        "Spatial only",
        "Neither"
    ]
)

records = filter_records(
    feed["records"],
    borough=borough,
    category=category,
    signal=signal
)

view = build_filtered_view(
    feed,
    records
)

summary = view["summary"]

a, b, c, d = st.columns(4)

a.metric(
    "Filtered reports",
    summary["selected"]
)

b.metric(
    "Anomaly-associated",
    summary["anomaly_reports"]
)

c.metric(
    "Spatial-associated",
    summary["spatial_reports"]
)

d.metric(
    "Distinct with either",
    summary["distinct_either"]
)

st.caption(
    "Anomaly and spatial totals overlap. "
    "The explanations and area cards refer "
    "to the complete processed reporting-day batch."
)

tabs = st.tabs([
    "Area patterns",
    "Anomaly reviews",
    "Spatial map",
    "Records"
])

with tabs[0]:

    areas = pd.DataFrame(
        feed["area_cards"]
    )

    selected_areas = {
        row["area_id"]
        for row in records
    }

    if not areas.empty:

        st.dataframe(
            areas.loc[
                areas["area_id"].isin(
                    selected_areas
                )
            ],
            hide_index=True,
            use_container_width=True
        )

    else:
        st.info(
            "No area reporting patterns available."
        )


with tabs[1]:

    if not view["anomaly_cards"]:
        st.info(
            "No anomaly candidates match the filters."
        )

    for card in view["anomaly_cards"]:

        with st.expander(
            f"{card['area_id']} — "
            f"{card['complaint_category']}"
        ):

            st.write(
                card["explanation"]
            )

            st.caption(
                card["suggested_review"]
            )


with tabs[2]:

    points = []

    for feature in view["map_features"]:

        longitude, latitude = (
            feature["geometry"]["coordinates"]
        )

        points.append({
            "latitude": latitude,
            "longitude": longitude,
            "candidate_id": feature["id"]
        })

    if points:

        st.map(
            pd.DataFrame(points)[[
                "latitude",
                "longitude"
            ]]
        )

    else:

        st.info(
            "No spatial candidate locations "
            "match the current filters."
        )

    for card in view["spatial_cards"]:

        with st.expander(
            f"{card['candidate_id']} — "
            f"{card['complaint_category']}"
        ):

            st.write(
                card["explanation"]
            )

            st.caption(
                card["suggested_review"]
            )


with tabs[3]:

    if records:

        st.dataframe(
            pd.DataFrame(records),
            hide_index=True,
            use_container_width=True
        )

    else:
        st.info(
            "No complaint records match the filters."
        )


feed_bytes = json.dumps(
    feed,
    indent=2,
    ensure_ascii=False,
    allow_nan=False
).encode("utf-8")

st.download_button(
    "Download dashboard feed JSON",
    feed_bytes,
    "uploaded_batch_dashboard_feed.json",
    "application/json"
)

st.caption(
    "Uploaded input and generated outputs "
    "remain in this Streamlit session unless "
    "explicitly downloaded. Historical models "
    "are not modified."
)
''').lstrip()

ast.parse(bridge_source)
ast.parse(page_source)

print("\n1. PYTHON SOURCE VALIDATION PASSED")

# --------------------------------------------------
# 3. Test bridge in memory before saving
# --------------------------------------------------

namespace = {
    "__name__": "_phase20_2d_validation",
    "__file__": str(bridge_path)
}

exec(
    compile(
        bridge_source,
        str(bridge_path),
        "exec"
    ),
    namespace
)

run_upload = namespace[
    "process_uploaded_csv_bytes"
]

build_verified = namespace[
    "build_verified_day_feed"
]

load_baseline = namespace[
    "load_upload_baseline"
]

from src.integration.bulk_complaint_contract import (
    INPUT_COLUMNS
)

from src.ui.service_factory import (
    get_gui_adapter
)

from dashboard.urban_pulse_dashboard import (
    validate_feed,
    build_filtered_view
)

# Recreate raw CSV from previously verified records.
verified_csv = pd.read_csv(
    verified_path,
    dtype="string",
    keep_default_na=False
)

raw_csv = verified_csv.loc[
    :, list(INPUT_COLUMNS)
].to_csv(
    index=False
).encode("utf-8")

print("\n2. EXECUTING REAL 500-RECORD UPLOAD")

baseline = load_baseline(root)

adapter = get_gui_adapter()

with threadpool_limits(limits=1):

    result = run_upload(
        raw_csv,
        "2026-11-09",
        root,
        synthetic_data=True,
        adapter=adapter,
        baseline=baseline
    )

print("\n3. BACKEND SELECTION SUMMARY")

for name, value in result[
    "selection_stats"
].items():

    print(f"{name}: {value}")

feed = result["feed"]

if feed is not None:

    print("\n4. THREE-MODEL DASHBOARD SUMMARY")

    for name, value in feed[
        "summary"
    ].items():

        print(f"{name}: {value}")

# --------------------------------------------------
# 4. Regression checks
# --------------------------------------------------

checks = []

def check(name, condition):

    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "Backend bridge compiles",
    len(bridge_source) > 0
)

check(
    "Streamlit page compiles",
    len(page_source) > 0
)

check(
    "500 uploaded records processed",
    result["uploaded_rows"] == 500
)

check(
    "500 backend audit records retained",
    len(
        result["backend_records"]
    ) == 500
)

check(
    "500 geographically verified",
    result["selection_stats"][
        "geographically_verified"
    ] == 500
)

check(
    "500 records selected for reporting day",
    result["selection_stats"][
        "selected_for_new_models"
    ] == 500
)

check(
    "Dashboard feed generated",
    feed is not None
)

if feed is not None:

    validate_feed(feed)

    summary = feed["summary"]

    check(
        "500 dashboard complaint records",
        len(feed["records"]) == 500
    )

    check(
        "20 area cards",
        len(feed["area_cards"]) == 20
    )

    check(
        "Three K-Means reporting patterns",
        summary["new_kmeans_patterns"] == 3
    )

    check(
        "23 DBSCAN spatial candidates",
        summary[
            "spatial_candidate_groups"
        ] == 23
    )

    check(
        "228 spatial candidate complaints",
        summary[
            "spatial_candidate_complaints"
        ] == 228
    )

    check(
        "15 anomaly candidate groups",
        summary[
            "anomaly_candidate_groups"
        ] == 15
    )

    check(
        "204 anomaly-associated complaints",
        summary[
            "anomaly_candidate_complaints"
        ] == 204
    )

    check(
        "156 overlapping complaints",
        summary[
            "complaints_in_both_signals"
        ] == 156
    )

    check(
        "276 distinct complaints with either signal",
        summary[
            "distinct_complaints_in_either_signal"
        ] == 276
    )

    check(
        "23 spatial map features",
        len(
            feed["map_geojson"]["features"]
        ) == 23
    )

    check(
        "Dashboard selection reconciles",
        build_filtered_view(
            feed,
            feed["records"]
        )["summary"]["selected"] == 500
    )

    check(
        "Synthetic upload labeled",
        feed["metadata"][
            "synthetic_data"
        ] is True
    )

    check(
        "Uploaded-data origin documented",
        feed["metadata"][
            "input_origin"
        ] == "uploaded_csv"
    )

# --------------------------------------------------
# 5. Test geographic exclusion without
#    rerunning the full historical backend
# --------------------------------------------------

print("\n5. GEOGRAPHIC EXCLUSION TEST")

rejected = result[
    "backend_records"
].copy(deep=True)

rejected.loc[
    rejected.index[0],
    "zip_validation_status"
] = "review_required"

with threadpool_limits(limits=1):

    partial = build_verified(
        rejected,
        baseline,
        "2026-11-09",
        synthetic_data=True
    )

check(
    "Geographically unverified row excluded",
    partial["selection_stats"][
        "excluded_by_backend_quality"
    ] == 1
)

check(
    "499 verified reports remain",
    partial["selection_stats"][
        "selected_for_new_models"
    ] == 499
)

check(
    "Backend audit retains rejected row",
    len(rejected) == 500
)

# --------------------------------------------------
# 6. Test no matching reporting day
# --------------------------------------------------

print("\n6. NO-MATCHING-DAY SAFEGUARD")

no_day = build_verified(
    result["backend_records"],
    baseline,
    "2026-11-10",
    synthetic_data=True
)

check(
    "No matching day generates no dashboard feed",
    no_day["feed"] is None
)

check(
    "No matching day produces zero model inputs",
    no_day["selection_stats"][
        "selected_for_new_models"
    ] == 0
)

# --------------------------------------------------
# 7. Verify existing files untouched
# --------------------------------------------------

signatures_after = {
    str(path): (
        path.stat().st_size,
        path.stat().st_mtime_ns
    )
    for path in protected
}

check(
    "Existing models, dashboard and data unchanged",
    signatures_before == signatures_after
)

results_20_2d = pd.DataFrame(checks)

print("\n7. PHASE 20.2D VALIDATION")

display(results_20_2d)

passed = int(
    results_20_2d["passed"].sum()
)

total = len(results_20_2d)

print(f"\nChecks passed: {passed}/{total}")

failed = results_20_2d.loc[
    ~results_20_2d["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 8. Save only after successful validation
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new integration files saved.")

    for name in failed:
        print("FAILED:", name)

else:

    destinations = {
        bridge_path: bridge_source,
        page_path: page_source
    }

    for path, content in destinations.items():

        if path.exists():

            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing file differs: {path}. "
                "No overwrite performed."
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():

            with path.open(
                "x",
                encoding="utf-8"
            ) as file:

                file.write(content)

    print("\n8. VERIFIED INTEGRATION FILES")

    print("Live backend bridge:", bridge_path)
    print("New Streamlit page:", page_path)

    print("\nSTATUS: PASSED")
    print("PHASE 20.2D LIVE UPLOAD PIPELINE VERIFIED")
    print("ORIGINAL SERVICE FACTORY REUSED")
    print("GEOGRAPHIC VALIDATION PRESERVED")
    print("K-MEANS, DBSCAN AND ANOMALY INTEGRATED")
    print("INVALID AND OFF-DAY RECORDS EXCLUDED")
    print("EXISTING DEMONSTRATION DASHBOARD PRESERVED")
    print("NO HISTORICAL MODELS MODIFIED")
```

```python

from pathlib import Path
from io import BytesIO
import hashlib
import importlib.util
import pandas as pd

from src.integration.bulk_complaint_contract import (
    INPUT_COLUMNS,
    read_batch_csv
)

print(" STREAMLIT GUI ACCEPTANCE PREPARATION")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

verified_path = (
    root / "outputs/tests/phase_17_4c_backend_verified"
    / "verified_500_backend_results.csv"
)

dashboard_path = (
    root / "dashboard/urban_pulse_dashboard.py"
)

processing_page = (
    root / "dashboard/pages/2_Process_New_Batch.py"
)

pipeline_path = (
    root / "src/integration/live_uploaded_batch_pipeline.py"
)

output_dir = (
    root / "outputs/tests/phase_20_2e_gui_acceptance"
)

clean_path = output_dir / "gui_test_clean_500.csv"
mixed_path = output_dir / "gui_test_mixed_500.csv"

protected = [
    verified_path,
    dashboard_path,
    processing_page,
    pipeline_path
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

original_hashes = {
    str(path): hashlib.sha256(path.read_bytes()).hexdigest()
    for path in protected
}

# 1. Reconstruct the original six-column CSV.
verified = pd.read_csv(
    verified_path,
    dtype="string",
    keep_default_na=False
)

columns = list(INPUT_COLUMNS)

assert len(columns) == 6
assert set(columns).issubset(verified.columns)

clean = verified[columns].copy()

assert len(clean) == 500
assert clean["complaint_id"].nunique() == 500

# 2. Create a controlled invalid/mixed-day test.
mixed = clean.copy(deep=True)

mixed.loc[mixed.index[0], "incident_zip"] = "99999"
mixed.loc[mixed.index[1], "created_date"] = "invalid-date"
mixed.loc[mixed.index[2], "created_date"] = "2026-11-10"

# 3. Serialize without changing original records.
clean_bytes = clean.to_csv(
    index=False
).encode("utf-8")

mixed_bytes = mixed.to_csv(
    index=False
).encode("utf-8")

# 4. Ensure both CSVs can be opened by the
# existing project CSV reader.
clean_read = read_batch_csv(
    BytesIO(clean_bytes)
)

mixed_read = read_batch_csv(
    BytesIO(mixed_bytes)
)

checks = {
    "Clean upload contains 500 rows":
        len(clean_read) == 500,

    "Mixed upload contains 500 rows":
        len(mixed_read) == 500,

    "Exactly six input fields used":
        list(clean_read.columns) == columns,

    "All original complaint IDs retained":
        clean_read["complaint_id"].astype(str).tolist()
        == clean["complaint_id"].astype(str).tolist(),

    "Mixed upload contains invalid ZIP probe":
        (
            mixed_read.iloc[0]["incident_zip"]
            == "99999"
        ),

    "Mixed upload contains invalid date probe":
        (
            mixed_read.iloc[1]["created_date"]
            == "invalid-date"
        ),

    "Mixed upload contains off-day probe":
        (
            mixed_read.iloc[2]["created_date"]
            == "2026-11-10"
        ),

    "Streamlit processing page exists":
        processing_page.is_file(),

    "Existing upload pipeline exists":
        pipeline_path.is_file(),

    "Historical and GUI source files unchanged":
        all(
            hashlib.sha256(path.read_bytes()).hexdigest()
            == original_hashes[str(path)]
            for path in protected
        )
}

results = pd.DataFrame([
    {"test": name, "passed": bool(value)}
    for name, value in checks.items()
])

print("\n1. ACCEPTANCE PREPARATION VALIDATION")
display(results)

passed = int(results["passed"].sum())
total = len(results)

print(f"\nChecks passed: {passed}/{total}")

if passed == total:

    def save_without_overwrite(path, content):
        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if path.exists():
            assert path.read_bytes() == content, (
                f"Existing test file differs: {path}"
            )
        else:
            with path.open("xb") as file:
                file.write(content)

    save_without_overwrite(
        clean_path, clean_bytes
    )

    save_without_overwrite(
        mixed_path, mixed_bytes
    )

    print("\n2. GUI TEST FILES")
    print("Clean 500:", clean_path)
    print("Mixed 500:", mixed_path)

    print(
        "\nStreamlit installed:",
        importlib.util.find_spec("streamlit")
        is not None
    )

    print("\nSTATUS: GUI ACCEPTANCE FILES READY")
    print("NO HISTORICAL DATA OR MODELS MODIFIED")

else:
    print("\nSTATUS: NEEDS REVIEW")
    print("NO TEST FILES SAVED")
```

```python

from pathlib import Path
from textwrap import dedent
import ast
import copy
import hashlib
import json
import sys
import pandas as pd

print(" INVESTIGATION RECOMMENDATIONS")
print("=" * 72)

root = Path(PROJECT_ROOT).resolve()

module_path = (
    root / "src/integration"
    / "investigation_review_queue.py"
)

page_path = (
    root / "dashboard/pages"
    / "3_Investigation_Recommendations.py"
)

feed_path = (
    root / "outputs/tests"
    / "phase_19_2e_dashboard_feed"
    / "dashboard_feed.json"
)

protected = [
    feed_path,
    root / "dashboard/urban_pulse_dashboard.py",
    root / "dashboard/pages/2_Process_New_Batch.py",
    root / "src/integration/live_uploaded_batch_pipeline.py",
    root / "src/integration/dashboard_intelligence_feed.py",
    root / "src/integration/smart_assistant_investigation_briefs.py"
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

before_hashes = {
    str(path): hashlib.sha256(
        path.read_bytes()
    ).hexdigest()
    for path in protected
}

# --------------------------------------------------
# 1. Build human-readable recommendation logic
# --------------------------------------------------

module_source = dedent(r'''
"""Evidence-led human investigation review queue.

The ranking is a transparent review-order heuristic,
NOT an infrastructure risk or repair-priority score.
"""


def display_location(area_id):
    text = str(area_id)

    if "_" not in text:
        return text

    borough, zip_code = text.rsplit("_", 1)

    return (
        borough.title()
        + " — ZIP "
        + zip_code
    )


def suggested_action(category, spatial=False):

    name = str(category).upper()

    if "TRAFFIC SIGNAL" in name:

        action = (
            "Identify the relevant intersection "
            "from the original complaint details. "
            "Compare reports to determine whether "
            "they describe the same signal issue, "
            "review previous signal maintenance "
            "and service records, and verify "
            "operation at the reported location."
        )

    elif "STREET LIGHT" in name:

        action = (
            "Compare reported locations and any "
            "available pole or fixture identifiers. "
            "Check whether reports describe the "
            "same outage, review recent lighting "
            "maintenance records, and verify "
            "the reported condition."
        )

    elif "SEWER" in name:

        action = (
            "Review the sewer complaint descriptions "
            "for a possible shared blockage, drainage "
            "or infrastructure issue. Compare nearby "
            "locations, examine recent sewer "
            "maintenance and cleaning history, "
            "and verify conditions before field action."
        )

    elif (
        "STREET CONDITION" in name
        or "SIDEWALK" in name
    ):

        action = (
            "Compare report descriptions and "
            "locations to identify repeated damage "
            "or reports about the same road or "
            "sidewalk segment. Review repair and "
            "maintenance history, then consider "
            "a site inspection to verify the issue."
        )

    else:

        action = (
            "Review the original complaints, "
            "compare locations and descriptions "
            "for a possible common cause, "
            "examine recent maintenance records, "
            "and verify the reported conditions."
        )

    if spatial:

        action += (
            " Pay particular attention to the "
            "newly identified spatial grouping; "
            "a concentration does not itself "
            "confirm one shared defect."
        )

    return action


def build_review_queue(feed):
    """Generate ranked, evidence-grounded review cards."""

    required = {
        "metadata",
        "summary",
        "anomaly_cards",
        "spatial_cards",
        "area_cards",
        "map_geojson"
    }

    if not required.issubset(feed):
        raise ValueError(
            "Dashboard feed missing required sections."
        )

    summary = feed["summary"]
    anomalies = feed["anomaly_cards"]
    spatial = feed["spatial_cards"]

    if len(anomalies) != int(
        summary["anomaly_candidate_groups"]
    ):
        raise ValueError(
            "Anomaly group count mismatch."
        )

    if len(spatial) != int(
        summary["spatial_candidate_groups"]
    ):
        raise ValueError(
            "Spatial group count mismatch."
        )

    spatial_ids = [
        str(item["candidate_id"])
        for item in spatial
    ]

    if len(set(spatial_ids)) != len(spatial_ids):
        raise ValueError(
            "Duplicate spatial candidate IDs."
        )

    features = feed["map_geojson"]["features"]

    if set(spatial_ids) != {
        str(item["id"])
        for item in features
    }:
        raise ValueError(
            "Map candidate identities disagree."
        )

    queue = []

    # Area/category historical reporting increases.
    for item in anomalies:

        area_id = str(item["area_id"])

        category = str(
            item["complaint_category"]
        )

        count = int(
            item["incoming_complaints"]
        )

        seasonal = int(
            item["seasonal_historical_max"]
        )

        broad = int(
            item["all_weekday_historical_max"]
        )

        overlap = int(
            item["complaints_in_spatial_candidates"]
        )

        if (
            overlap < 0
            or overlap > count
            or count <= seasonal
            or count <= broad
        ):
            raise ValueError(
                "Invalid anomaly comparison evidence."
            )

        has_spatial = overlap > 0

        if has_spatial:

            why = (
                f"{count} {category.lower()} reports "
                f"were received in {display_location(area_id)}. "
                f"The historical seasonal daily "
                f"maximum was {seasonal}, and the "
                f"all-weekday daily maximum was "
                f"{broad}. {overlap} of the new "
                f"reports also belong to one or "
                f"more spatial concentrations. "
                f"Both reporting-volume and "
                f"spatial screening signals "
                f"support reviewing this area."
            )

        else:

            why = (
                f"{count} {category.lower()} reports "
                f"were received in {display_location(area_id)}, "
                f"compared with historical daily "
                f"maxima of {seasonal} (seasonal) "
                f"and {broad} (all-weekday). "
                f"No new spatial concentration "
                f"is linked to these reports."
            )

        queue.append({
            "kind": "area_category",
            "tier": (
                1 if has_spatial else 2
            ),
            "review_label": (
                "Review first — two screening signals"
                if has_spatial
                else "Review reporting increase"
            ),
            "area_ids": [area_id],
            "location": display_location(
                area_id
            ),
            "category": category,
            "incoming_reports": count,
            "seasonal_max": seasonal,
            "all_weekday_max": broad,
            "historical_excess": (
                count - max(seasonal, broad)
            ),
            "spatial_overlap_reports": overlap,
            "candidate_id": None,
            "linked_spatial_ids": [
                str(value)
                for value in item.get(
                    "linked_spatial_candidate_ids", []
                )
            ],
            "kmeans_patterns": [
                str(value)
                for value in item.get(
                    "kmeans_reporting_patterns", []
                )
            ],
            "why": why,
            "suggested_action":
                suggested_action(
                    category,
                    spatial=has_spatial
                ),
            "evidence_note": (
                "Cross-window consistent historical "
                "screening candidate. Not a "
                "statistically confirmed anomaly."
            )
        })

    # Spatial findings without linked anomaly records.
    for item in spatial:

        count = int(
            item["incoming_complaints"]
        )

        overlap = int(
            item[
                "complaints_in_anomaly_candidate_groups"
            ]
        )

        if overlap < 0 or overlap > count:
            raise ValueError(
                "Invalid spatial overlap count."
            )

        if overlap != 0:
            continue

        candidate_id = str(
            item["candidate_id"]
        )

        category = str(
            item["complaint_category"]
        )

        areas = [
            str(value)
            for value in item["area_ids"]
        ]

        locations = ", ".join(
            display_location(area)
            for area in areas
        )

        distinct = int(
            item["distinct_submitted_locations"]
        )

        historical = item[
            "historical_dbscan_associations"
        ]

        outside = int(
            historical["outside"]
        )

        why = (
            f"{count} {category.lower()} reports "
            f"at {distinct} distinct submitted "
            f"coordinates form spatial candidate "
            f"{candidate_id} in {locations}. "
            f"{outside} reports are outside "
            f"established historical DBSCAN "
            f"associations. No qualifying "
            f"area/category reporting-increase "
            f"candidate is linked to this spatial "
            f"group. The concentration merits "
            f"verification, but is not a "
            f"confirmed infrastructure incident."
        )

        queue.append({
            "kind": "spatial_only",
            "tier": 3,
            "review_label":
                "Review spatial concentration",
            "area_ids": areas,
            "location": locations,
            "category": category,
            "incoming_reports": count,
            "seasonal_max": None,
            "all_weekday_max": None,
            "historical_excess": 0,
            "spatial_overlap_reports": 0,
            "candidate_id": candidate_id,
            "linked_spatial_ids": [
                candidate_id
            ],
            "kmeans_patterns": [
                str(value)
                for value in item.get(
                    "kmeans_reporting_patterns", []
                )
            ],
            "why": why,
            "suggested_action":
                suggested_action(
                    category,
                    spatial=True
                ),
            "evidence_note": (
                "Spatial screening only. "
                "This is not a historical-count "
                "anomaly conclusion."
            )
        })

    # Ranking is for human review, not field urgency.
    queue.sort(
        key=lambda item: (
            item["tier"],
            -item["historical_excess"],
            -item["spatial_overlap_reports"],
            -item["incoming_reports"],
            item["location"],
            item["category"]
        )
    )

    for index, item in enumerate(
        queue,
        start=1
    ):
        item["review_order"] = index

    first = sum(
        item["tier"] == 1
        for item in queue
    )

    historical_only = sum(
        item["tier"] == 2
        for item in queue
    )

    spatial_only = sum(
        item["tier"] == 3
        for item in queue
    )

    # Readable area summaries; never show raw
    # category_breakdown dictionaries as objects.
    area_table = []

    for area in feed["area_cards"]:

        area_table.append({
            "area_id": str(
                area["area_id"]
            ),
            "location": display_location(
                area["area_id"]
            ),
            "reports": int(
                area["incoming_complaints"]
            ),
            "reporting_pattern": ", ".join(
                map(
                    str,
                    area.get(
                        "reporting_patterns", []
                    )
                )
            ),
            "dominant_category": str(
                area["dominant_category"]
            ),
            "anomaly_associated": int(
                area["anomaly_candidate_complaints"]
            ),
            "spatial_associated": int(
                area["spatial_candidate_complaints"]
            )
        })

    return {
        "metadata": {
            "reporting_day": feed[
                "metadata"
            ]["reporting_day"],
            "synthetic_data": bool(
                feed["metadata"].get(
                    "synthetic_data", False
                )
            ),
            "review_policy": (
                "Human screening order only. "
                "Not a risk score, verified defect, "
                "statistical significance claim "
                "or repair dispatch instruction."
            )
        },
        "summary": {
            "reports": int(
                summary["input_complaints"]
            ),
            "distinct_with_signals": int(
                summary[
                    "distinct_complaints_in_either_signal"
                ]
            ),
            "review_first_groups": first,
            "historical_only_groups":
                historical_only,
            "spatial_only_groups": spatial_only,
            "review_items": len(queue),
            "all_spatial_candidates":
                len(spatial)
        },
        "queue": queue,
        "area_table": area_table,
        "spatial_cards": spatial,
        "map_geojson": feed["map_geojson"]
    }
''').lstrip()

# --------------------------------------------------
# 2. Create simplified recommendations GUI page
# --------------------------------------------------

page_source = dedent(r'''
"""Human-readable investigation recommendations page."""

from pathlib import Path
import json
import sys

import pandas as pd
import streamlit as st


ROOT = Path(__file__).resolve().parents[2]

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.integration.investigation_review_queue import (
    build_review_queue
)


DEMO_FEED = (
    ROOT
    / "outputs/tests/phase_19_2e_dashboard_feed"
    / "dashboard_feed.json"
)


st.set_page_config(
    page_title="Investigation Recommendations",
    page_icon="🔎",
    layout="wide"
)

st.title("🔎 Where Should We Investigate First?")

st.caption(
    "Urban Infrastructure Pulse — "
    "explainable investigation recommendations"
)

live = st.session_state.get(
    "urban_pulse_live_result"
)

if (
    isinstance(live, dict)
    and isinstance(live.get("feed"), dict)
):

    feed = live["feed"]

    st.info(
        "Showing your most recently processed "
        "uploaded batch from this browser session."
    )

else:

    if not DEMO_FEED.is_file():

        st.error(
            "No completed upload or demonstration "
            "feed is available."
        )

        st.stop()

    with DEMO_FEED.open(
        encoding="utf-8"
    ) as file:

        feed = json.load(file)

    st.info(
        "Showing the saved 500-complaint "
        "synthetic demonstration. Process a "
        "new batch to view uploaded results."
    )


try:

    result = build_review_queue(
        feed
    )

except (ValueError, KeyError, TypeError) as exc:

    st.error(
        f"Could not build recommendations: {exc}"
    )

    st.stop()


if result["metadata"]["synthetic_data"]:

    st.warning(
        "Synthetic demonstration data. "
        "These are test recommendations, "
        "not confirmed NYC incidents."
    )

else:

    st.warning(
        "Uploaded complaint data is unverified "
        "in the field. Screening results are "
        "not confirmed faults or repair priorities."
    )


st.markdown(
    "**What this page answers:** "
    "Which places have noteworthy reporting "
    "evidence, why should they be reviewed, "
    "and what should investigators check?"
)

st.caption(
    "Review order is not a measure of danger. "
    "It considers historical reporting increases "
    "and new spatial concentrations; emergency "
    "severity requires independent assessment."
)

summary = result["summary"]

m1, m2, m3, m4 = st.columns(4)

m1.metric(
    "Analyzed reports",
    summary["reports"]
)

m2.metric(
    "Review first groups",
    summary["review_first_groups"]
)

m3.metric(
    "Spatial-only reviews",
    summary["spatial_only_groups"]
)

m4.metric(
    "Distinct reports flagged",
    summary["distinct_with_signals"]
)


st.divider()

# Filters only change displayed cards;
# rankings remain based on the complete batch.

queue = result["queue"]

boroughs = sorted({
    str(area).rsplit("_", 1)[0]
    for item in queue
    for area in item["area_ids"]
})

categories = sorted({
    item["category"]
    for item in queue
})

f1, f2 = st.columns(2)

borough = f1.selectbox(
    "Filter borough",
    ["All"] + boroughs,
    key="recommendation_borough"
)

category = f2.selectbox(
    "Filter complaint category",
    ["All"] + categories,
    key="recommendation_category"
)


def is_visible(item):

    if (
        category != "All"
        and item["category"] != category
    ):
        return False

    if borough != "All":

        if not any(
            area.startswith(
                borough + "_"
            )
            for area in item["area_ids"]
        ):
            return False

    return True


visible = [
    item
    for item in queue
    if is_visible(item)
]


def show_recommendation(item):

    with st.container(border=True):

        st.markdown(
            f"### {item['review_order']}. "
            f"{item['location']}"
        )

        st.markdown(
            f"**Complaint type:** {item['category']}"
        )

        st.caption(
            item["review_label"]
        )

        if item["kmeans_patterns"]:

            st.write(
                "**Area reporting pattern:**",
                ", ".join(
                    item["kmeans_patterns"]
                )
            )

        st.markdown("**Why investigate here?**")

        st.write(
            item["why"]
        )

        st.markdown("**Suggested action**")

        st.write(
            item["suggested_action"]
        )

        st.caption(
            "Evidence limitation: "
            + item["evidence_note"]
        )


tabs = st.tabs([
    "Investigate first",
    "Other review findings",
    "Spatial map",
    "Area overview"
])


with tabs[0]:

    st.subheader(
        "Places to review first"
    )

    st.write(
        "These areas show both an elevated "
        "historical reporting-count signal "
        "and new spatial concentration evidence."
    )

    first_items = [
        item for item in visible
        if item["tier"] == 1
    ]

    if not first_items:

        st.info(
            "No two-signal reviews match "
            "the selected filters."
        )

    else:

        for item in first_items[:3]:

            show_recommendation(item)

        remaining = first_items[3:]

        if remaining:

            with st.expander(
                f"View {len(remaining)} "
                "more investigation recommendations"
            ):

                for item in remaining:

                    show_recommendation(item)


with tabs[1]:

    st.subheader(
        "Other findings worth reviewing"
    )

    st.write(
        "These findings are based on a "
        "historical reporting increase alone "
        "or a spatial concentration without "
        "a linked historical-count anomaly."
    )

    other_items = [
        item for item in visible
        if item["tier"] != 1
    ]

    if not other_items:

        st.info(
            "No additional findings match "
            "the selected filters."
        )

    for item in other_items:

        show_recommendation(item)


with tabs[2]:

    st.subheader(
        "New spatial concentration locations"
    )

    st.caption(
        "Map pins are representative submitted "
        "complaint coordinates, not verified "
        "fault sites or intersections."
    )

    matching_spatial_ids = {
        str(item["candidate_id"])
        for item in result["spatial_cards"]

        if (
            category == "All"
            or item["complaint_category"]
                == category
        )

        and (
            borough == "All"
            or any(
                str(area).startswith(
                    borough + "_"
                )
                for area in item["area_ids"]
            )
        )
    }

    points = []

    for feature in result[
        "map_geojson"
    ]["features"]:

        if str(feature["id"]) not in (
            matching_spatial_ids
        ):
            continue

        lon, lat = (
            feature["geometry"][
                "coordinates"
            ]
        )

        points.append({
            "latitude": lat,
            "longitude": lon,
            "candidate_id": feature["id"]
        })

    if points:

        st.map(
            pd.DataFrame(points)[[
                "latitude",
                "longitude"
            ]]
        )

        st.dataframe(
            pd.DataFrame(points),
            hide_index=True,
            use_container_width=True
        )

    else:

        st.info(
            "No candidate map locations "
            "match these filters."
        )


with tabs[3]:

    st.subheader(
        "Area reporting patterns"
    )

    st.write(
        "The following is a simplified "
        "K-Means and model-evidence overview."
    )

    frame = pd.DataFrame(
        result["area_table"]
    )

    if borough != "All":

        frame = frame.loc[
            frame["area_id"].str.startswith(
                borough + "_"
            )
        ]

    if category != "All":

        frame = frame.loc[
            frame["dominant_category"]
            == category
        ]

    st.dataframe(
        frame,
        hide_index=True,
        use_container_width=True
    )


st.divider()

st.subheader("Export investigation recommendations")

export_rows = [
    {
        "review_order": item["review_order"],
        "review_label": item["review_label"],
        "location": item["location"],
        "complaint_category": item["category"],
        "incoming_reports": item["incoming_reports"],
        "why_investigate": item["why"],
        "suggested_action": item["suggested_action"],
        "evidence_note": item["evidence_note"]
    }
    for item in visible
]

if export_rows:

    export_csv = pd.DataFrame(
        export_rows
    ).to_csv(
        index=False
    ).encode("utf-8")

    st.download_button(
        "Download Investigation Recommendations CSV",
        data=export_csv,
        file_name=(
            "urban_pulse_investigation_"
            "recommendations.csv"
        ),
        mime="text/csv"
    )


st.caption(
    "Suggestions are for human verification. "
    "Do not dispatch repairs or claim confirmed "
    "faults based solely on these screens."
)
''').lstrip()

# --------------------------------------------------
# 3. Compile both modules without saving
# --------------------------------------------------

compile(
    module_source,
    str(module_path),
    "exec"
)

ast.parse(page_source)

namespace = {
    "__name__": "_review_queue_validation",
    "__file__": str(module_path)
}

exec(
    compile(
        module_source,
        str(module_path),
        "exec"
    ),
    namespace
)

build_queue = namespace[
    "build_review_queue"
]

# --------------------------------------------------
# 4. Test against saved validated dashboard feed
# --------------------------------------------------

print("\n1. LOADING VERIFIED 500-COMPLAINT FEED")

with feed_path.open(
    encoding="utf-8"
) as file:

    feed = json.load(file)

source_snapshot = copy.deepcopy(feed)

queue_result = build_queue(feed)

print("\n2. RECOMMENDATION SUMMARY")

for key, value in queue_result[
    "summary"
].items():

    print(f"{key}: {value}")

# --------------------------------------------------
# 5. Show top recommendations
# --------------------------------------------------

print("\n3. INVESTIGATE FIRST — PREVIEW")

preview = pd.DataFrame([
    {
        "review_order": item["review_order"],
        "location": item["location"],
        "category": item["category"],
        "incoming_reports": item[
            "incoming_reports"
        ],
        "historical_excess": item[
            "historical_excess"
        ],
        "spatial_overlap": item[
            "spatial_overlap_reports"
        ],
        "review_label": item["review_label"]
    }

    for item in queue_result["queue"]
])

display(
    preview.head(15)
)

print("\n4. EXAMPLE INVESTIGATION RECOMMENDATION")

first = queue_result["queue"][0]

print("PLACE:", first["location"])
print("CATEGORY:", first["category"])

print("\nWHY INVESTIGATE?")
print(first["why"])

print("\nSUGGESTED ACTION")
print(first["suggested_action"])

# --------------------------------------------------
# 6. Verify key evidence and safeguards
# --------------------------------------------------

manhattan = next(
    item
    for item in queue_result["queue"]
    if (
        "MANHATTAN_10003" in item["area_ids"]
        and item["category"] == "SEWER"
    )
)

queens = next(
    item
    for item in queue_result["queue"]
    if item["candidate_id"] == "NB-00-004"
)

tampered = copy.deepcopy(feed)

tampered["summary"][
    "anomaly_candidate_groups"
] = 999

tamper_rejected = False

try:

    build_queue(tampered)

except ValueError:

    tamper_rejected = True

checks = []

def check(name, condition):

    checks.append({
        "test": name,
        "passed": bool(condition)
    })


check(
    "Review queue module compiles",
    bool(module_source)
)

check(
    "Streamlit recommendations page compiles",
    bool(page_source)
)

check(
    "500 reports retained in summary",
    queue_result["summary"]["reports"] == 500
)

check(
    "276 unique complaints flagged",
    queue_result["summary"][
        "distinct_with_signals"
    ] == 276
)

check(
    "15 two-signal reviews prepared",
    queue_result["summary"][
        "review_first_groups"
    ] == 15
)

check(
    "Eight spatial-only reviews prepared",
    queue_result["summary"][
        "spatial_only_groups"
    ] == 8
)

check(
    "23 investigation items created",
    queue_result["summary"][
        "review_items"
    ] == 23
)

check(
    "All 23 spatial candidates preserved",
    queue_result["summary"][
        "all_spatial_candidates"
    ] == 23
)

check(
    "Review order numbers unique and sequential",
    [
        item["review_order"]
        for item in queue_result["queue"]
    ] == list(
        range(
            1,
            len(queue_result["queue"]) + 1
        )
    )
)

check(
    "Manhattan sewer evidence correct",
    manhattan["incoming_reports"] == 15
    and manhattan["seasonal_max"] == 2
    and manhattan["all_weekday_max"] == 2
    and manhattan["spatial_overlap_reports"] == 15
)

check(
    "Queens spatial-only candidate retained",
    queens["tier"] == 3
    and queens["incoming_reports"] == 15
)

check(
    "Every recommendation explains why",
    all(
        item["why"].strip()
        for item in queue_result["queue"]
    )
)

check(
    "Every recommendation suggests an action",
    all(
        item["suggested_action"].strip()
        for item in queue_result["queue"]
    )
)

check(
    "No invented intersection-specific location",
    all(
        item["location"] == namespace[
            "display_location"
        ](item["area_ids"][0])
        for item in queue_result["queue"]
        if len(item["area_ids"]) == 1
    )
)

check(
    "Readable area table contains 20 areas",
    len(
        queue_result["area_table"]
    ) == 20
)

check(
    "Inconsistent upstream counts rejected",
    tamper_rejected
)

check(
    "Source dashboard data unchanged",
    feed == source_snapshot
)

check(
    "Original application and data files unchanged",
    all(
        hashlib.sha256(
            path.read_bytes()
        ).hexdigest()
        == before_hashes[str(path)]
        for path in protected
    )
)

results = pd.DataFrame(checks)

print("\n5. PHASE 20.3A VALIDATION")

display(results)

passed = int(results["passed"].sum())
total = len(results)

print(f"\nChecks passed: {passed}/{total}")

failed = results.loc[
    ~results["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 7. Save only if all checks pass
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new files saved.")

    for name in failed:

        print("FAILED:", name)

else:

    destinations = {
        module_path: module_source,
        page_path: page_source
    }

    for path, content in destinations.items():

        if path.exists():

            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Different file already exists: {path}. "
                "No overwrite performed."
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():

            with path.open(
                "x",
                encoding="utf-8"
            ) as file:

                file.write(content)

    print("\n6. VERIFIED FILES")

    print("Review queue module:", module_path)
    print("New Streamlit page:", page_path)

    print("\nSTATUS: PASSED")
    print("PHASE 20.3A INVESTIGATION RECOMMENDATIONS READY")
    print("EVIDENCE-LED HUMAN REVIEW ORDER CREATED")
    print("WHY-INVESTIGATE EXPLANATIONS CREATED")
    print("CATEGORY-SPECIFIC ACTIONS CREATED")
    print("SPATIAL-ONLY REVIEWS PRESERVED")
    print("NO HISTORICAL MODELS MODIFIED")
```

```python

from pathlib import Path
from textwrap import dedent
import ast
import hashlib
import json
import sys
import pandas as pd
import streamlit as st

print(" RECOMMENDATIONS-FIRST DASHBOARD")
print("=" * 70)

root = Path(PROJECT_ROOT).resolve()

if str(root) not in sys.path:
    sys.path.insert(0, str(root))

launcher_path = (
    root / "dashboard/urban_pulse_app.py"
)

explorer_path = (
    root / "dashboard/pages/4_Model_Explorer.py"
)

original_app = (
    root / "dashboard/urban_pulse_dashboard.py"
)

recommendations_page = (
    root / "dashboard/pages"
    / "3_Investigation_Recommendations.py"
)

upload_page = (
    root / "dashboard/pages/1_Upload_Complaints.py"
)

process_page = (
    root / "dashboard/pages/2_Process_New_Batch.py"
)

review_module = (
    root / "src/integration"
    / "investigation_review_queue.py"
)

demo_feed = (
    root / "outputs/tests"
    / "phase_19_2e_dashboard_feed"
    / "dashboard_feed.json"
)

protected = [
    original_app,
    recommendations_page,
    upload_page,
    process_page,
    review_module,
    demo_feed
]

for path in protected:
    assert path.is_file(), f"Missing: {path}"

hashes_before = {
    str(p): hashlib.sha256(
        p.read_bytes()
    ).hexdigest()
    for p in protected
}

# --------------------------------------------------
# 1. New main application entry point
# --------------------------------------------------

launcher_source = dedent(r'''
"""Urban Infrastructure Pulse — primary application.

Launch:
python -m streamlit run dashboard/urban_pulse_app.py

Investigation recommendations are the default page.
"""

from pathlib import Path

import streamlit as st


HERE = Path(__file__).resolve().parent

st.set_page_config(
    page_title="Urban Infrastructure Pulse",
    page_icon="🏙️",
    layout="wide",
    initial_sidebar_state="expanded"
)

pages = {
    "Investigation Intelligence": [
        st.Page(
            str(
                HERE / "pages"
                / "3_Investigation_Recommendations.py"
            ),
            title="Investigation Recommendations",
            icon="🔎",
            default=True
        ),
        st.Page(
            str(
                HERE / "pages"
                / "4_Model_Explorer.py"
            ),
            title="Model Explorer",
            icon="📊"
        )
    ],

    "Complaint Processing": [
        st.Page(
            str(
                HERE / "pages"
                / "1_Upload_Complaints.py"
            ),
            title="Validate CSV Upload",
            icon="📄"
        ),
        st.Page(
            str(
                HERE / "pages"
                / "2_Process_New_Batch.py"
            ),
            title="Process New Batch",
            icon="⚙️"
        )
    ]
}

selected_page = st.navigation(
    pages,
    position="sidebar"
)

selected_page.run()
''').lstrip()

# --------------------------------------------------
# 2. Preserve original dashboard as Model Explorer
# --------------------------------------------------

explorer_source = dedent(r'''
"""Access the original model-exploration dashboard."""

from pathlib import Path
import sys


ROOT = Path(__file__).resolve().parents[2]

if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from dashboard.urban_pulse_dashboard import main


main()
''').lstrip()

ast.parse(launcher_source)
ast.parse(explorer_source)

print("\n1. SOURCE VALIDATION PASSED")
print("Streamlit version:", st.__version__)

# --------------------------------------------------
# 3. Verify existing recommendation data
# --------------------------------------------------

from src.integration.investigation_review_queue import (
    build_review_queue
)

with demo_feed.open(
    encoding="utf-8"
) as file:
    feed = json.load(file)

review = build_review_queue(feed)

print("\n2. CURRENT DEMONSTRATION SUMMARY")

for key, value in review["summary"].items():
    print(f"{key}: {value}")

# --------------------------------------------------
# 4. Validate new navigation design
# --------------------------------------------------

checks = []

def check(name, condition):
    checks.append({
        "test": name,
        "passed": bool(condition)
    })

check(
    "New launcher compiles",
    bool(compile(
        launcher_source,
        str(launcher_path),
        "exec"
    ))
)

check(
    "Model Explorer wrapper compiles",
    bool(compile(
        explorer_source,
        str(explorer_path),
        "exec"
    ))
)

check(
    "Streamlit supports modern navigation",
    hasattr(st, "navigation")
    and hasattr(st, "Page")
)

check(
    "Recommendations are the default page",
    "default=True" in launcher_source
    and "3_Investigation_Recommendations.py"
    in launcher_source
)

check(
    "Original dashboard remains accessible",
    "4_Model_Explorer.py" in launcher_source
    and "from dashboard.urban_pulse_dashboard import main"
    in explorer_source
)

check(
    "CSV validation page retained",
    "1_Upload_Complaints.py"
    in launcher_source
)

check(
    "Live processing page retained",
    "2_Process_New_Batch.py"
    in launcher_source
)

check(
    "500 synthetic demo complaints available",
    review["summary"]["reports"] == 500
)

check(
    "15 review-first groups preserved",
    review["summary"][
        "review_first_groups"
    ] == 15
)

check(
    "Eight spatial-only reviews preserved",
    review["summary"][
        "spatial_only_groups"
    ] == 8
)

check(
    "23 investigation items preserved",
    review["summary"]["review_items"] == 23
)

check(
    "Every review has suggested action",
    all(
        item["suggested_action"].strip()
        for item in review["queue"]
    )
)

check(
    "All original source files unchanged",
    all(
        hashlib.sha256(
            p.read_bytes()
        ).hexdigest() == hashes_before[str(p)]
        for p in protected
    )
)

results = pd.DataFrame(checks)

print("\n3. VALIDATION CHECKLIST")
display(results)

passed = int(results["passed"].sum())
total = len(results)

print(f"\nChecks passed: {passed}/{total}")

failed = results.loc[
    ~results["passed"],
    "test"
].tolist()

# --------------------------------------------------
# 5. Save only after checks pass
# --------------------------------------------------

if failed:

    print("\nSTATUS: NEEDS REVIEW")
    print("No new application files saved.")

    for name in failed:
        print("FAILED:", name)

else:

    destinations = {
        launcher_path: launcher_source,
        explorer_path: explorer_source
    }

    # Refuse to overwrite different existing files.
    for path, content in destinations.items():

        if path.exists():
            assert path.read_text(
                encoding="utf-8"
            ) == content, (
                f"Existing file differs: {path}"
            )

    for path, content in destinations.items():

        path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if not path.exists():
            with path.open(
                "x",
                encoding="utf-8"
            ) as file:
                file.write(content)

    print("\n4. NEW APPLICATION FILES")

    print("Main launcher:", launcher_path)
    print("Model Explorer:", explorer_path)

    print("\nSTATUS: PASSED")
    print("RECOMMENDATIONS SET AS DEFAULT PAGE")
    print("ORIGINAL MODEL EXPLORER PRESERVED")
    print("CSV VALIDATION AND PROCESSING RETAINED")
    print("NO HISTORICAL MODELS MODIFIED")

    print("\nNEW LAUNCH COMMAND:")
    print(
        "python -m streamlit run "
        "dashboard/urban_pulse_app.py"
    )
```
